```python
# =====================================================================
# STEP 8
# PUBLICATION-READY RESULTS PACKAGE
#
# PURPOSE
# -------
# Convert frozen Steps 1-7 outputs into:
#
# 1. Final point-forecast benchmark table
# 2. Statistical-significance table
# 3. Feature-ablation table
# 4. Conformal uncertainty table
# 5. External BPDB validation table
# 6. Temporal robustness table
# 7. Evening_Peak sensitivity table
# 8. Experimental-design summary
# 9. Publication-ready figures
# 10. Figure manifest
# 11. Exact claims registry
# 12. Results + Discussion manuscript outline
# 13. Excel workbook containing all final tables
#
#
# IMPORTANT
# ---------
# NO model tuning is performed here.
# NO test-based model modification is performed here.
# NO new architecture is selected here.
#
# Steps 1-7 remain frozen.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import re
import json
import math
import warnings

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display


warnings.filterwarnings("ignore")


pd.set_option(
    "display.max_columns",
    300
)

pd.set_option(
    "display.max_rows",
    400
)

pd.set_option(
    "display.width",
    300
)


print("=" * 125)

print(
    "STEP 8: PUBLICATION-READY RESULTS PACKAGE"
)

print("=" * 125)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount(
    "/content/drive"
)


print(
    "\nGoogle Drive mounted successfully."
)


# =====================================================================
# 2. PROJECT DIRECTORIES
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)


ML_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "ml_baselines"
)


DEEP_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "deep_sequence"
)


CONFORMAL_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "adaptive_conformal"
)


EXTERNAL_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "external_bpdb_validation"
)


STEP7_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "statistical_ablation_robustness"
)


STEP6_FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "external_bpdb_validation"
)


STEP7_FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "statistical_ablation_robustness"
)


PUBLICATION_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "publication_package"
)


PUBLICATION_FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "publication_ready"
)


PUBLICATION_RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


PUBLICATION_FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print(
    "\nPublication result directory:"
)

print(
    PUBLICATION_RESULT_DIR
)


print(
    "\nPublication figure directory:"
)

print(
    PUBLICATION_FIGURE_DIR
)


# =====================================================================
# 3. INPUT FILES
# =====================================================================

TRAIN_FILE = (
    FORECAST_DIR
    / "train_2015_2022.csv"
)


VALIDATION_FILE = (
    FORECAST_DIR
    / "validation_2023.csv"
)


TEST_FILE = (
    FORECAST_DIR
    / "test_2024_2026.csv"
)


ML_PREDICTION_FILE = (
    ML_RESULT_DIR
    / "all_test_predictions.csv"
)


DEEP_PREDICTION_FILE = (
    DEEP_RESULT_DIR
    / "deep_test_predictions.csv"
)


CONFORMAL_PRIMARY_FILE = (
    CONFORMAL_RESULT_DIR
    / "primary_90percent_comparison.csv"
)


CONFORMAL_WINDOW_FILE = (
    CONFORMAL_RESULT_DIR
    / "window_sensitivity_90percent.csv"
)


CONFORMAL_REGIME_FILE = (
    CONFORMAL_RESULT_DIR
    / "primary_method_regime_results.csv"
)


CONFORMAL_HOURLY_FILE = (
    CONFORMAL_RESULT_DIR
    / "primary_method_hourly_coverage.csv"
)


STEP5B_SUMMARY_FILE = (
    CONFORMAL_RESULT_DIR
    / "step5b_summary.json"
)


EXTERNAL_COMPARISON_FILE = (
    EXTERNAL_RESULT_DIR
    / "external_validation_comparison.csv"
)


EXTERNAL_YEARLY_FILE = (
    EXTERNAL_RESULT_DIR
    / "external_forecast_by_year.csv"
)


SOURCE_YEARLY_FILE = (
    EXTERNAL_RESULT_DIR
    / "pgcb_bpdb_source_agreement_by_year.csv"
)


STEP6_SUMMARY_FILE = (
    EXTERNAL_RESULT_DIR
    / "step6_summary.json"
)


STATISTICAL_FILE = (
    STEP7_RESULT_DIR
    / "paired_statistical_significance.csv"
)


ABLATION_FILE = (
    STEP7_RESULT_DIR
    / "ridge_feature_group_ablation.csv"
)


TEMPORAL_FILE = (
    STEP7_RESULT_DIR
    / "ridge_temporal_skill_vs_d1.csv"
)


MONTHLY_FILE = (
    STEP7_RESULT_DIR
    / "monthly_robustness.csv"
)


EVENING_SENSITIVITY_FILE = (
    STEP7_RESULT_DIR
    / "bpdb_evening_peak_timing_sensitivity.csv"
)


STEP7_SUMMARY_FILE = (
    STEP7_RESULT_DIR
    / "step7_summary.json"
)


# =====================================================================
# 4. CORE FILE CHECK
# =====================================================================

CORE_FILES = [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE,
    ML_PREDICTION_FILE,
    STATISTICAL_FILE,
    ABLATION_FILE,
    TEMPORAL_FILE,
    EVENING_SENSITIVITY_FILE,
    EXTERNAL_COMPARISON_FILE,
    STEP6_SUMMARY_FILE,
    STEP7_SUMMARY_FILE
]


print("\n" + "=" * 125)

print(
    "CORE FILE CHECK"
)

print("=" * 125)


for path in CORE_FILES:

    print(
        path,
        "->",
        path.exists()
    )


    if not path.exists():

        raise FileNotFoundError(
            f"""
Required Step 8 input missing:

{path}
"""
        )


print(
    "\nAll required Step 8 core inputs found."
)


# =====================================================================
# 5. OPTIONAL FILE CHECK
# =====================================================================

OPTIONAL_FILES = [

    DEEP_PREDICTION_FILE,
    CONFORMAL_PRIMARY_FILE,
    CONFORMAL_WINDOW_FILE,
    CONFORMAL_REGIME_FILE,
    CONFORMAL_HOURLY_FILE,
    STEP5B_SUMMARY_FILE,
    EXTERNAL_YEARLY_FILE,
    SOURCE_YEARLY_FILE,
    MONTHLY_FILE
]


print("\nOptional files:")


for path in OPTIONAL_FILES:

    print(
        path.name,
        "->",
        path.exists()
    )


# =====================================================================
# 6. LOAD FORECASTING DATA
# =====================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)


validation_df = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)


test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for dataframe in [

    train_df,
    validation_df,
    test_df
]:

    dataframe[
        "target_date"
    ] = pd.to_datetime(

        dataframe[
            "target_date"
        ],

        errors="coerce"
    )


    dataframe.sort_values(
        "target_date",
        inplace=True
    )


    dataframe.reset_index(
        drop=True,
        inplace=True
    )


# =====================================================================
# 7. HOURS / TARGETS
# =====================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


Y_TEST = (

    test_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


assert (
    Y_TEST.shape[1]
    ==
    23
)


# =====================================================================
# 8. METRIC FUNCTIONS
# =====================================================================

def mae(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )


    predicted = np.asarray(
        predicted,
        dtype=float
    )


    return float(

        np.mean(

            np.abs(
                actual
                -
                predicted
            )
        )
    )


def rmse(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )


    predicted = np.asarray(
        predicted,
        dtype=float
    )


    return float(

        np.sqrt(

            np.mean(

                (
                    actual
                    -
                    predicted
                )
                ** 2
            )
        )
    )


def mape(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )


    predicted = np.asarray(
        predicted,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            actual
        ),

        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    actual
                    -
                    predicted
                )

                /

                denominator
            )
        )
    )


def smape(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )


    predicted = np.asarray(
        predicted,
        dtype=float
    )


    denominator = (

        np.abs(
            actual
        )

        +

        np.abs(
            predicted
        )
    )


    valid = (
        denominator > 0
    )


    return float(

        100

        *

        np.mean(

            2

            *

            np.abs(

                actual[
                    valid
                ]

                -

                predicted[
                    valid
                ]
            )

            /

            denominator[
                valid
            ]
        )
    )


def metrics(
    actual,
    predicted
):

    return {

        "MAE_MW":
            mae(
                actual,
                predicted
            ),

        "RMSE_MW":
            rmse(
                actual,
                predicted
            ),

        "MAPE_percent":
            mape(
                actual,
                predicted
            ),

        "sMAPE_percent":
            smape(
                actual,
                predicted
            )
    }


# =====================================================================
# 9. SAVED-PREDICTION HELPERS
# =====================================================================

def prepare_prediction_source(
    file_path
):

    if not file_path.exists():

        return None


    dataframe = pd.read_csv(
        file_path,
        low_memory=False
    )


    date_column = None


    for candidate in [

        "target_date",
        "date"
    ]:

        if candidate in dataframe.columns:

            date_column = candidate

            break


    if (
        date_column
        is None
    ):

        return None


    dataframe[
        date_column
    ] = pd.to_datetime(

        dataframe[
            date_column
        ],

        errors="coerce"
    )


    dataframe = (

        dataframe

        .dropna(
            subset=[
                date_column
            ]
        )

        .drop_duplicates(
            subset=[
                date_column
            ]
        )

        .set_index(
            date_column
        )
    )


    test_dates = pd.DatetimeIndex(

        test_df[
            "target_date"
        ]
    )


    if len(

        test_dates.difference(
            dataframe.index
        )

    ) > 0:

        return None


    dataframe = dataframe.reindex(
        test_dates
    )


    return dataframe


def extract_prediction_matrix(
    dataframe,
    keywords
):

    if dataframe is None:

        return None


    selected_columns = []


    for hour in HOURS:

        pattern = re.compile(

            rf"h0?{hour}$",

            flags=re.IGNORECASE
        )


        candidates = []


        for column in dataframe.columns:

            lower = str(
                column
            ).lower()


            keyword_match = any(

                keyword.lower()
                in
                lower

                for keyword in keywords
            )


            if (

                keyword_match

                and

                pattern.search(
                    lower
                )

                and

                "target"
                not in
                lower

                and

                "actual"
                not in
                lower
            ):

                candidates.append(
                    column
                )


        if len(
            candidates
        ) == 0:

            return None


        candidates = sorted(

            candidates,

            key=lambda column: (

                0
                if "pred"
                in
                str(column).lower()
                else
                1,

                len(
                    str(column)
                )
            )
        )


        selected_columns.append(
            candidates[0]
        )


    if len(
        selected_columns
    ) != 23:

        return None


    matrix = (

        dataframe[
            selected_columns
        ]

        .to_numpy(
            dtype=float
        )
    )


    if matrix.shape != (
        len(
            test_df
        ),
        23
    ):

        return None


    if not np.isfinite(
        matrix
    ).all():

        return None


    return matrix


# =====================================================================
# 10. LOAD SAVED PREDICTION SOURCES
# =====================================================================

ml_source = prepare_prediction_source(
    ML_PREDICTION_FILE
)


deep_source = prepare_prediction_source(
    DEEP_PREDICTION_FILE
)


# =====================================================================
# 11. BUILD MODEL PREDICTION DICTIONARY
# =====================================================================

MODEL_PREDICTIONS = {}


ridge_pred = extract_prediction_matrix(

    ml_source,

    [
        "ridge"
    ]
)


if ridge_pred is None:

    raise RuntimeError(
        """
Frozen Ridge predictions could not be extracted.
"""
    )


MODEL_PREDICTIONS[
    "Ridge"
] = ridge_pred


# ---------------------------------------------------------------------
# Naive baselines
# ---------------------------------------------------------------------

D1_COLUMNS = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


D7_COLUMNS = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


MODEL_PREDICTIONS[
    "D1"
] = (

    test_df[
        D1_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


MODEL_PREDICTIONS[
    "D7"
] = (

    test_df[
        D7_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


# ---------------------------------------------------------------------
# ML / DL comparators
# ---------------------------------------------------------------------

model_detection = [

    (
        "Extra Trees",
        ml_source,
        [
            "extra",
            "extratrees",
            "extra_trees"
        ]
    ),

    (
        "XGBoost",
        ml_source,
        [
            "xgb",
            "xgboost"
        ]
    ),

    (
        "LSTM",
        deep_source,
        [
            "lstm"
        ]
    ),

    (
        "GRU",
        deep_source,
        [
            "gru"
        ]
    ),

    (
        "Transformer",
        deep_source,
        [
            "transformer"
        ]
    ),

    (
        "Temporal CNN",
        deep_source,
        [
            "cnn",
            "temporalcnn",
            "temporal_cnn"
        ]
    )
]


for (
    model_name,
    source,
    keywords
) in model_detection:

    matrix = extract_prediction_matrix(

        source,

        keywords
    )


    if matrix is not None:

        MODEL_PREDICTIONS[
            model_name
        ] = matrix


print(
    "\nDetected models:"
)

print(
    list(
        MODEL_PREDICTIONS.keys()
    )
)


# =====================================================================
# 12. TABLE 1
# EXPERIMENTAL DESIGN
# =====================================================================

TABLE1 = pd.DataFrame(
    [

        {
            "Component":
                "Target",

            "Specification":
                (
                    "Next-day national reported demand "
                    "profile at integer-hour slots 01:00-23:00"
                )
        },

        {
            "Component":
                "Forecast outputs",

            "Specification":
                "23 slots per target day"
        },

        {
            "Component":
                "Training period",

            "Specification":
                (
                    f"{train_df['target_date'].min().date()} "
                    f"to {train_df['target_date'].max().date()}"
                )
        },

        {
            "Component":
                "Training samples",

            "Specification":
                str(
                    len(
                        train_df
                    )
                )
        },

        {
            "Component":
                "Validation period",

            "Specification":
                (
                    f"{validation_df['target_date'].min().date()} "
                    f"to {validation_df['target_date'].max().date()}"
                )
        },

        {
            "Component":
                "Validation samples",

            "Specification":
                str(
                    len(
                        validation_df
                    )
                )
        },

        {
            "Component":
                "Locked test period",

            "Specification":
                (
                    f"{test_df['target_date'].min().date()} "
                    f"to {test_df['target_date'].max().date()}"
                )
        },

        {
            "Component":
                "Test samples",

            "Specification":
                str(
                    len(
                        test_df
                    )
                )
        },

        {
            "Component":
                "Primary point model",

            "Specification":
                "Ridge regression, alpha = 10"
        },

        {
            "Component":
                "Primary uncertainty method",

            "Specification":
                (
                    "Prequential adaptive trend-regime "
                    "normalized conformal, 180-day window"
                )
        },

        {
            "Component":
                "External validation source",

            "Specification":
                (
                    "BPDB Version 2 daily evening-peak demand, "
                    "2019-11-21 to 2024-12-30"
                )
        }
    ]
)


# =====================================================================
# 13. TABLE 2
# POINT-FORECAST BENCHMARK
# =====================================================================

model_rows = []


D1_MAE = mae(

    Y_TEST,

    MODEL_PREDICTIONS[
        "D1"
    ]
)


for model_name, prediction in (
    MODEL_PREDICTIONS.items()
):

    row_metrics = metrics(

        Y_TEST,

        prediction
    )


    skill_vs_d1 = (

        100

        *
        (
            D1_MAE

            -

            row_metrics[
                "MAE_MW"
            ]
        )

        /
        D1_MAE
    )


    if model_name == "Ridge":

        role = (
            "Primary model; validation-selected"
        )


    elif model_name == "D1":

        role = (
            "Primary naive baseline"
        )


    elif model_name == "D7":

        role = (
            "Weekly naive baseline"
        )


    elif model_name == "LSTM":

        role = (
            "Validation-selected deep comparator"
        )


    else:

        role = (
            "Comparator"
        )


    model_rows.append(
        {

            "Model":
                model_name,

            "Role":
                role,

            "MAE_MW":
                row_metrics[
                    "MAE_MW"
                ],

            "RMSE_MW":
                row_metrics[
                    "RMSE_MW"
                ],

            "MAPE_percent":
                row_metrics[
                    "MAPE_percent"
                ],

            "sMAPE_percent":
                row_metrics[
                    "sMAPE_percent"
                ],

            "Skill_vs_D1_percent":
                skill_vs_d1
        }
    )


TABLE2 = pd.DataFrame(
    model_rows
)


TABLE2 = (

    TABLE2

    .sort_values(
        "MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


# =====================================================================
# 14. LOAD STEP 7 STATISTICAL RESULTS
# =====================================================================

TABLE3 = pd.read_csv(
    STATISTICAL_FILE
)


# Replace numerical p=0 display with publication-safe notation later.

TABLE3[
    "HAC7_significance"
] = TABLE3[
    "HAC7_p_Holm"
].apply(

    lambda value:

        "<0.001"

        if (
            pd.notna(
                value
            )

            and

            value < 0.001
        )

        else

        (
            f"{value:.3f}"
            if pd.notna(
                value
            )
            else
            ""
        )
)


# =====================================================================
# 15. LOAD FEATURE ABLATION
# =====================================================================

TABLE4 = pd.read_csv(
    ABLATION_FILE
)


# =====================================================================
# 16. LOAD CONFORMAL RESULTS
# =====================================================================

if CONFORMAL_PRIMARY_FILE.exists():

    TABLE5 = pd.read_csv(
        CONFORMAL_PRIMARY_FILE
    )


else:

    # Verified frozen Step 5B values.
    TABLE5 = pd.DataFrame(
        [

            {
                "Method":
                    "Static Hour-Normalized",

                "Marginal_Coverage_percent":
                    82.5719,

                "Coverage_Gap_pp":
                    -7.4281,

                "MPIW_MW":
                    1487.8746,

                "Winkler":
                    3014.1161
            },

            {
                "Method":
                    "Static Trend-Regime Hour-Normalized",

                "Marginal_Coverage_percent":
                    83.4120,

                "Coverage_Gap_pp":
                    -6.5880,

                "MPIW_MW":
                    1524.6694,

                "Winkler":
                    2998.0089
            },

            {
                "Method":
                    "Adaptive Hour-Normalized W180",

                "Marginal_Coverage_percent":
                    89.2010,

                "Coverage_Gap_pp":
                    -0.7990,

                "MPIW_MW":
                    1938.2062,

                "Winkler":
                    2950.6400
            },

            {
                "Method":
                    "Adaptive Trend-Regime Hour-Normalized W180",

                "Marginal_Coverage_percent":
                    90.3538,

                "Coverage_Gap_pp":
                    0.3538,

                "MPIW_MW":
                    2042.4235,

                "Winkler":
                    2966.0381
            }
        ]
    )


# =====================================================================
# 17. LOAD EXTERNAL VALIDATION
# =====================================================================

TABLE6 = pd.read_csv(
    EXTERNAL_COMPARISON_FILE
)


# =====================================================================
# 18. LOAD TEMPORAL ROBUSTNESS
# =====================================================================

TABLE7 = pd.read_csv(
    TEMPORAL_FILE
)


# =====================================================================
# 19. LOAD EVENING_PEAK SENSITIVITY
# =====================================================================

TABLE8 = pd.read_csv(
    EVENING_SENSITIVITY_FILE
)


# =====================================================================
# 20. LOAD SUMMARY JSON FILES
# =====================================================================

with open(
    STEP6_SUMMARY_FILE,
    "r"
) as file:

    STEP6_SUMMARY = json.load(
        file
    )


with open(
    STEP7_SUMMARY_FILE,
    "r"
) as file:

    STEP7_SUMMARY = json.load(
        file
    )


if STEP5B_SUMMARY_FILE.exists():

    with open(
        STEP5B_SUMMARY_FILE,
        "r"
    ) as file:

        STEP5B_SUMMARY = json.load(
            file
        )


else:

    STEP5B_SUMMARY = {}


# =====================================================================
# 21. PRIMARY FROZEN VALUES
# =====================================================================

RIDGE_ROW = (

    TABLE2.loc[

        TABLE2[
            "Model"
        ]
        ==
        "Ridge"
    ]

    .iloc[0]
)


D1_ROW = (

    TABLE2.loc[

        TABLE2[
            "Model"
        ]
        ==
        "D1"
    ]

    .iloc[0]
)


RIDGE_D1_STAT_ROW = (

    TABLE3.loc[

        TABLE3[
            "Comparator"
        ]
        ==
        "D1"
    ]

    .iloc[0]
)


PRIMARY_ABLATION = (

    TABLE4.loc[

        TABLE4[
            "Removed_Group"
        ]
        !=
        "NONE"
    ]

    .sort_values(
        "Delta_Test_MAE_vs_Full_MW",

        ascending=False
    )

    .iloc[0]
)


# =====================================================================
# 22. EXTERNAL PRIMARY VALUES
# =====================================================================

SOURCE_ROW = (

    TABLE6.loc[

        TABLE6[
            "comparison"
        ]
        ==
        "BPDB_vs_PGCB_Explicit_EveningPeak"
    ]

    .iloc[0]
)


EXTERNAL_FORECAST_ROW = (

    TABLE6.loc[

        TABLE6[
            "comparison"
        ]
        ==
        "BPDB_vs_Ridge_17to22_PredictedPeak"
    ]

    .iloc[0]
)


# =====================================================================
# 23. TEMPORAL PRIMARY VALUES
# =====================================================================

TEMPORAL_YEAR_ROWS = (

    TABLE7.loc[

        TABLE7[
            "Period"
        ]
        !=
        "ALL_TEST_2024_2026"
    ]

    .copy()
)


minimum_temporal_skill = float(

    TEMPORAL_YEAR_ROWS[
        "Ridge_Skill_vs_D1_percent"
    ]
    .min()
)


maximum_temporal_skill = float(

    TEMPORAL_YEAR_ROWS[
        "Ridge_Skill_vs_D1_percent"
    ]
    .max()
)


# =====================================================================
# 24. SENSITIVITY PRIMARY VALUES
# =====================================================================

SENS_ALL = (

    TABLE8.loc[

        TABLE8[
            "Definition"
        ]
        ==
        "All_Explicit_EveningPeak"
    ]

    .iloc[0]
)


SENS_RESTRICTED = (

    TABLE8.loc[

        TABLE8[
            "Definition"
        ]
        ==
        "Explicit_Exclude_01_and_12"
    ]

    .iloc[0]
)


# =====================================================================
# 25. FINAL SUMMARY TABLE
# =====================================================================

TABLE9 = pd.DataFrame(
    [

        {
            "Result":
                "Frozen Ridge test MAE",

            "Value":
                f"{RIDGE_ROW['MAE_MW']:.2f} MW"
        },

        {
            "Result":
                "Frozen Ridge test RMSE",

            "Value":
                f"{RIDGE_ROW['RMSE_MW']:.2f} MW"
        },

        {
            "Result":
                "Frozen Ridge test MAPE",

            "Value":
                f"{RIDGE_ROW['MAPE_percent']:.3f}%"
        },

        {
            "Result":
                "Ridge skill over D1",

            "Value":
                f"{RIDGE_D1_STAT_ROW['Ridge_Skill_percent']:.2f}%"
        },

        {
            "Result":
                "Mean daily MAE improvement over D1",

            "Value":
                (
                    f"{RIDGE_D1_STAT_ROW['Mean_Daily_MAE_Improvement_MW']:.2f} MW"
                )
        },

        {
            "Result":
                "7-day bootstrap 95% CI",

            "Value":
                (
                    f"[{RIDGE_D1_STAT_ROW['Bootstrap7_CI_Lower_MW']:.2f}, "
                    f"{RIDGE_D1_STAT_ROW['Bootstrap7_CI_Upper_MW']:.2f}] MW"
                )
        },

        {
            "Result":
                "Primary adaptive conformal marginal coverage",

            "Value":
                "90.354%"
        },

        {
            "Result":
                "Primary adaptive conformal coverage gap",

            "Value":
                "+0.354 percentage points"
        },

        {
            "Result":
                "BPDB-PGCB source correlation",

            "Value":
                f"{SOURCE_ROW['Pearson_r']:.6f}"
        },

        {
            "Result":
                "BPDB-PGCB source MAE",

            "Value":
                f"{SOURCE_ROW['MAE_MW']:.2f} MW"
        },

        {
            "Result":
                "2024 BPDB external forecast correlation",

            "Value":
                f"{EXTERNAL_FORECAST_ROW['Pearson_r']:.6f}"
        },

        {
            "Result":
                "2024 BPDB external forecast MAPE",

            "Value":
                f"{EXTERNAL_FORECAST_ROW['MAPE_percent']:.3f}%"
        },

        {
            "Result":
                "Most damaging feature-group removal",

            "Value":
                (
                    f"{PRIMARY_ABLATION['Removed_Group']} "
                    f"(+{PRIMARY_ABLATION['Delta_Test_MAE_vs_Full_MW']:.2f} MW)"
                )
        },

        {
            "Result":
                "Year-specific Ridge skill range",

            "Value":
                (
                    f"{minimum_temporal_skill:.2f}% "
                    f"to {maximum_temporal_skill:.2f}%"
                )
        }
    ]
)


# =====================================================================
# 26. DISPLAY FINAL TABLES
# =====================================================================

print("\n" + "=" * 125)

print(
    "TABLE 1 — EXPERIMENTAL DESIGN"
)

print("=" * 125)

display(
    TABLE1
)


print("\n" + "=" * 125)

print(
    "TABLE 2 — POINT FORECASTING PERFORMANCE"
)

print("=" * 125)

display(
    TABLE2.round(
        4
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 3 — STATISTICAL SIGNIFICANCE"
)

print("=" * 125)

display(
    TABLE3.round(
        4
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 4 — FEATURE-GROUP ABLATION"
)

print("=" * 125)

display(
    TABLE4.round(
        4
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 5 — CONFORMAL UNCERTAINTY"
)

print("=" * 125)

display(
    TABLE5
)


print("\n" + "=" * 125)

print(
    "TABLE 6 — EXTERNAL BPDB VALIDATION"
)

print("=" * 125)

display(
    TABLE6.round(
        4
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 7 — TEMPORAL ROBUSTNESS"
)

print("=" * 125)

display(
    TABLE7.round(
        4
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 8 — EVENING_PEAK SENSITIVITY"
)

print("=" * 125)

display(
    TABLE8.round(
        5
    )
)


print("\n" + "=" * 125)

print(
    "TABLE 9 — FINAL RESULTS AT A GLANCE"
)

print("=" * 125)

display(
    TABLE9
)


# =====================================================================
# 27. SAVE INDIVIDUAL PUBLICATION TABLES
# =====================================================================

TABLE_FILES = {}


for number, (
    name,
    dataframe
) in enumerate(
    [

        (
            "experimental_design",
            TABLE1
        ),

        (
            "point_forecast_benchmark",
            TABLE2
        ),

        (
            "statistical_significance",
            TABLE3
        ),

        (
            "feature_ablation",
            TABLE4
        ),

        (
            "conformal_uncertainty",
            TABLE5
        ),

        (
            "external_bpdb_validation",
            TABLE6
        ),

        (
            "temporal_robustness",
            TABLE7
        ),

        (
            "evening_peak_sensitivity",
            TABLE8
        ),

        (
            "key_results_summary",
            TABLE9
        )
    ],

    start=1
):

    output_file = (

        PUBLICATION_RESULT_DIR
        /
        f"Table_{number:02d}_{name}.csv"
    )


    dataframe.to_csv(
        output_file,
        index=False
    )


    TABLE_FILES[
        name
    ] = output_file


# =====================================================================
# 28. CREATE MASTER EXCEL WORKBOOK
# =====================================================================

EXCEL_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "publication_ready_results_tables.xlsx"
)


with pd.ExcelWriter(

    EXCEL_FILE,

    engine="openpyxl"
) as writer:

    TABLE1.to_excel(
        writer,
        sheet_name=
            "T1_Design",
        index=False
    )


    TABLE2.to_excel(
        writer,
        sheet_name=
            "T2_PointModels",
        index=False
    )


    TABLE3.to_excel(
        writer,
        sheet_name=
            "T3_Statistics",
        index=False
    )


    TABLE4.to_excel(
        writer,
        sheet_name=
            "T4_Ablation",
        index=False
    )


    TABLE5.to_excel(
        writer,
        sheet_name=
            "T5_Uncertainty",
        index=False
    )


    TABLE6.to_excel(
        writer,
        sheet_name=
            "T6_External",
        index=False
    )


    TABLE7.to_excel(
        writer,
        sheet_name=
            "T7_Temporal",
        index=False
    )


    TABLE8.to_excel(
        writer,
        sheet_name=
            "T8_Sensitivity",
        index=False
    )


    TABLE9.to_excel(
        writer,
        sheet_name=
            "T9_KeyResults",
        index=False
    )


print(
    "\nExcel workbook created:"
)

print(
    EXCEL_FILE
)


# =====================================================================
# 29. FIGURE 1
# POINT-FORECAST MODEL COMPARISON
# =====================================================================

plot_table = (

    TABLE2

    .sort_values(
        "MAE_MW",

        ascending=True
    )

    .copy()
)


plt.figure(
    figsize=(
        11,
        7
    )
)


plt.barh(

    plot_table[
        "Model"
    ],

    plot_table[
        "MAE_MW"
    ]
)


plt.xlabel(
    "Test MAE (MW)"
)


plt.ylabel(
    "Model"
)


plt.title(
    "Point-Forecasting Performance on the Locked Test Period"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


FIG1 = (

    PUBLICATION_FIGURE_DIR
    /
    "Figure_01_Point_Model_MAE.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 30. FIGURE 2
# RIDGE VS D1 TEMPORAL PERFORMANCE
# =====================================================================

temporal_plot = (

    TABLE7.loc[

        TABLE7[
            "Period"
        ]
        !=
        "ALL_TEST_2024_2026"
    ]

    .copy()
)


x = np.arange(
    len(
        temporal_plot
    )
)


width = 0.35


plt.figure(
    figsize=(
        11,
        6
    )
)


plt.bar(

    x
    -
    width
    /
    2,

    temporal_plot[
        "Ridge_MAE_MW"
    ],

    width=
        width,

    label=
        "Ridge"
)


plt.bar(

    x
    +
    width
    /
    2,

    temporal_plot[
        "D1_MAE_MW"
    ],

    width=
        width,

    label=
        "Previous-day baseline"
)


plt.xticks(

    x,

    temporal_plot[
        "Period"
    ],

    rotation=15
)


plt.ylabel(
    "MAE (MW)"
)


plt.xlabel(
    "Evaluation period"
)


plt.title(
    "Temporal Robustness of Ridge Forecasting"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG2 = (

    PUBLICATION_FIGURE_DIR
    /
    "Figure_02_Temporal_Robustness.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 31. FIGURE 3
# RIDGE SIGNIFICANCE VS COMPARATORS
# =====================================================================

stat_plot = (

    TABLE3

    .sort_values(
        "Mean_Daily_MAE_Improvement_MW"
    )

    .copy()
)


centers = (

    stat_plot[
        "Mean_Daily_MAE_Improvement_MW"
    ]
    .to_numpy()
)


lower = (

    centers

    -

    stat_plot[
        "Bootstrap7_CI_Lower_MW"
    ]
    .to_numpy()
)


upper = (

    stat_plot[
        "Bootstrap7_CI_Upper_MW"
    ]
    .to_numpy()

    -

    centers
)


positions = np.arange(
    len(
        stat_plot
    )
)


plt.figure(
    figsize=(
        11,
        7
    )
)


plt.errorbar(

    centers,

    positions,

    xerr=
        np.vstack(
            [
                lower,
                upper
            ]
        ),

    fmt="o",

    capsize=4
)


plt.yticks(

    positions,

    stat_plot[
        "Comparator"
    ]
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Daily MAE Improvement of Ridge (MW)"
)


plt.ylabel(
    "Comparator"
)


plt.title(
    "Paired 7-Day Block-Bootstrap Confidence Intervals"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


FIG3 = (

    PUBLICATION_FIGURE_DIR
    /
    "Figure_03_Statistical_Significance.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 32. FIGURE 4
# FEATURE ABLATION
# =====================================================================

ablation_plot = (

    TABLE4.loc[

        TABLE4[
            "Removed_Group"
        ]
        !=
        "NONE"
    ]

    .sort_values(
        "Delta_Test_MAE_vs_Full_MW"
    )

    .copy()
)


plt.figure(
    figsize=(
        11,
        7
    )
)


plt.barh(

    ablation_plot[
        "Removed_Group"
    ],

    ablation_plot[
        "Delta_Test_MAE_vs_Full_MW"
    ]
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Change in Test MAE Relative to Full Model (MW)"
)


plt.ylabel(
    "Removed Feature Group"
)


plt.title(
    "Leave-One-Feature-Group-Out Ablation"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


FIG4 = (

    PUBLICATION_FIGURE_DIR
    /
    "Figure_04_Feature_Ablation.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 33. FIGURE 5
# EXTERNAL VALIDATION SUMMARY
# =====================================================================

external_plot = (

    TABLE6.loc[

        TABLE6[
            "comparison"
        ]
        .isin(
            [

                "BPDB_vs_PGCB_Explicit_EveningPeak",

                "BPDB_vs_PGCB_17to22_Max",

                "BPDB_vs_Ridge_17to22_PredictedPeak"
            ]
        )
    ]

    .copy()
)


label_map = {

    "BPDB_vs_PGCB_Explicit_EveningPeak":
        "PGCB explicit peak",

    "BPDB_vs_PGCB_17to22_Max":
        "PGCB 17-22 maximum",

    "BPDB_vs_Ridge_17to22_PredictedPeak":
        "Ridge forecast"
}


external_plot[
    "Short_Label"
] = (

    external_plot[
        "comparison"
    ]
    .map(
        label_map
    )
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(

    external_plot[
        "Short_Label"
    ],

    external_plot[
        "MAE_MW"
    ]
)


plt.ylabel(
    "MAE Against BPDB (MW)"
)


plt.xlabel(
    "Comparison"
)


plt.title(
    "Cross-Source and External Forecast Validation"
)


plt.xticks(
    rotation=20,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG5 = (

    PUBLICATION_FIGURE_DIR
    /
    "Figure_05_External_Validation.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 34. FIGURE MANIFEST
# =====================================================================

FIGURE_CANDIDATES = [

    (
        "Figure 1",
        FIG1,
        "Locked-test MAE comparison across baselines, ML and DL models."
    ),

    (
        "Figure 2",
        FIG2,
        "Year-specific Ridge versus previous-day baseline performance."
    ),

    (
        "Figure 3",
        FIG3,
        "Paired block-bootstrap confidence intervals for Ridge improvement."
    ),

    (
        "Figure 4",
        FIG4,
        "Leave-one-feature-group-out Ridge ablation."
    ),

    (
        "Figure 5",
        FIG5,
        "BPDB cross-source and external forecast validation."
    ),

    (
        "Supplementary Figure S1",
        STEP6_FIGURE_DIR
        /
        "01_bpdb_vs_pgcb_evening_peak.png",

        "Scatter plot showing PGCB-BPDB source agreement."
    ),

    (
        "Supplementary Figure S2",
        STEP6_FIGURE_DIR
        /
        "02_bpdb_vs_ridge_evening_forecast.png",

        "BPDB versus Ridge-predicted evening peak."
    ),

    (
        "Supplementary Figure S3",
        STEP6_FIGURE_DIR
        /
        "03_2024_bpdb_ridge_timeseries.png",

        "2024 BPDB and Ridge evening-demand time series."
    ),

    (
        "Supplementary Figure S4",
        STEP7_FIGURE_DIR
        /
        "04_monthly_ridge_skill.png",

        "Month-of-year forecasting robustness."
    ),

    (
        "Supplementary Figure S5",
        STEP7_FIGURE_DIR
        /
        "05_evening_peak_timing_sensitivity.png",

        "Evening_Peak timing sensitivity analysis."
    ),

    (
        "Supplementary Figure S6",
        STEP7_FIGURE_DIR
        /
        "06_rolling_ridge_improvement_vs_d1.png",

        "Rolling temporal stability of Ridge improvement."
    )
]


figure_rows = []


for (
    figure_name,
    path,
    purpose
) in FIGURE_CANDIDATES:

    figure_rows.append(
        {

            "Figure":
                figure_name,

            "File":
                str(
                    path
                ),

            "Exists":
                path.exists(),

            "Recommended_Use":
                purpose
        }
    )


FIGURE_MANIFEST = pd.DataFrame(
    figure_rows
)


FIGURE_MANIFEST_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "figure_manifest.csv"
)


FIGURE_MANIFEST.to_csv(
    FIGURE_MANIFEST_FILE,
    index=False
)


print("\n" + "=" * 125)

print(
    "FIGURE MANIFEST"
)

print("=" * 125)


display(
    FIGURE_MANIFEST
)


# =====================================================================
# 35. EXACT CLAIMS REGISTRY
#
# These are manuscript-safe claims.
# =====================================================================

ridge_p_statement = (

    "<0.001"

    if (
        RIDGE_D1_STAT_ROW[
            "HAC7_p_Holm"
        ]
        <
        0.001
    )

    else

    f"{RIDGE_D1_STAT_ROW['HAC7_p_Holm']:.3f}"
)


CLAIMS = f"""
# FROZEN MANUSCRIPT CLAIMS

## A. Data and forecast definition

1. The primary task is next-day national reported electricity-demand
   profile forecasting over 23 observed integer-hour slots (01:00-23:00).

2. The final evaluation uses a chronological split:
   training through 2022, validation in 2023, and a locked test period
   from 2024-01-01 through 2026-03-08.

3. The word "consumption" should not be used as an exact synonym for
   the target. The safer terminology is "reported system demand",
   "electricity demand", or "national load".

## B. Point forecasting

4. Ridge regression with alpha=10 was selected using the validation
   period and retained as the frozen primary point forecaster.

5. On the locked test period, Ridge achieved:
   MAE = {RIDGE_ROW['MAE_MW']:.2f} MW,
   RMSE = {RIDGE_ROW['RMSE_MW']:.2f} MW,
   MAPE = {RIDGE_ROW['MAPE_percent']:.3f}%,
   sMAPE = {RIDGE_ROW['sMAPE_percent']:.3f}%.

6. The previous-day baseline achieved an MAE of
   {D1_ROW['MAE_MW']:.2f} MW.

7. Ridge therefore provided
   {RIDGE_D1_STAT_ROW['Ridge_Skill_percent']:.2f}%
   MAE skill relative to the previous-day baseline.

## C. Statistical evidence

8. The mean daily MAE improvement of Ridge over the previous-day
   baseline was
   {RIDGE_D1_STAT_ROW['Mean_Daily_MAE_Improvement_MW']:.2f} MW.

9. The 7-day moving-block-bootstrap 95% confidence interval for this
   improvement was
   [{RIDGE_D1_STAT_ROW['Bootstrap7_CI_Lower_MW']:.2f},
    {RIDGE_D1_STAT_ROW['Bootstrap7_CI_Upper_MW']:.2f}] MW.

10. The corresponding 14-day block-bootstrap confidence interval was
    [{RIDGE_D1_STAT_ROW['Bootstrap14_CI_Lower_MW']:.2f},
     {RIDGE_D1_STAT_ROW['Bootstrap14_CI_Upper_MW']:.2f}] MW.

11. The HAC-based paired loss-difference test remained significant
    after Holm correction (p {ridge_p_statement}).

12. Do NOT report p=0.000 or p=0.
    Report p<0.001 when the numerical computation underflows.

## D. Feature ablation

13. The most damaging feature-group removal was
    {PRIMARY_ABLATION['Removed_Group']}.

14. Removing this group increased test MAE by
    {PRIMARY_ABLATION['Delta_Test_MAE_vs_Full_MW']:.2f} MW
    ({PRIMARY_ABLATION['Relative_Test_MAE_Change_percent']:.2f}%).

15. Removing Rolling28_STD produced a small post-hoc reduction in
    test MAE, but it performed worse on validation than the full model.
    Therefore this observation must NOT be used to redefine or replace
    the frozen primary model.

## E. Probabilistic forecasting

16. The primary uncertainty procedure is the pre-specified
    180-day prequential adaptive trend-regime normalized conformal
    approach.

17. At the nominal 90% level, marginal empirical coverage was
    approximately 90.354%, corresponding to a coverage gap of
    approximately +0.354 percentage points.

18. The interval result should be described as empirical calibration
    under prequential evaluation.

19. Do NOT claim an exact i.i.d. split-conformal coverage guarantee
    because the data are temporally dependent and potentially
    nonstationary.

20. For every test day t, the interval was constructed without using
    the outcome of day t. Once day t became observed, its residual
    could enter the calibration history for future days.

## F. Cross-source BPDB validation

21. The explicit PGCB Evening_Peak observations showed very strong
    agreement with BPDB generation-end evening demand:
    n = {int(SOURCE_ROW['n'])},
    r = {SOURCE_ROW['Pearson_r']:.6f},
    MAE = {SOURCE_ROW['MAE_MW']:.2f} MW,
    MAPE = {SOURCE_ROW['MAPE_percent']:.4f}%,
    bias = {SOURCE_ROW['Bias_MW']:.2f} MW.

22. This should be called "cross-source validation" or
    "independent data-source validation", not an independent
    physical ground truth.

23. The explicit Evening_Peak flag is a substantially closer mapping
    to BPDB evening demand than using a generic 23-slot daily maximum.

## G. External forecast validation

24. During the 2024 overlap with BPDB, the Ridge-derived 17:00-22:00
    predicted peak achieved:
    n = {int(EXTERNAL_FORECAST_ROW['n'])},
    r = {EXTERNAL_FORECAST_ROW['Pearson_r']:.6f},
    MAE = {EXTERNAL_FORECAST_ROW['MAE_MW']:.2f} MW,
    RMSE = {EXTERNAL_FORECAST_ROW['RMSE_MW']:.2f} MW,
    MAPE = {EXTERNAL_FORECAST_ROW['MAPE_percent']:.3f}%,
    bias = {EXTERNAL_FORECAST_ROW['Bias_MW']:.2f} MW.

25. This BPDB evening-peak metric is NOT directly equivalent to the
    primary 23-slot forecasting MAE because the BPDB peak can occur
    between integer-hour forecast slots.

## H. Temporal robustness

26. Ridge maintained an MAE skill over D1 between approximately
    {minimum_temporal_skill:.2f}% and
    {maximum_temporal_skill:.2f}% across the year-specific
    evaluation periods.

27. This supports temporal robustness rather than improvement
    concentrated in only one evaluation year.

## I. Evening_Peak sensitivity

28. Seven unusual explicit Evening_Peak records occurred at 01:00
    or 12:00.

29. Excluding those observations did not change the overlapping
    BPDB-PGCB result:
    all explicit MAE = {SENS_ALL['MAE_MW']:.3f} MW,
    restricted MAE = {SENS_RESTRICTED['MAE_MW']:.3f} MW.

30. Therefore the exceptionally strong cross-source agreement is
    not driven by those unusual timestamps.

## J. Claims to avoid

31. Do NOT claim that Ridge is universally superior to every possible
    forecasting architecture.

32. Do NOT claim that BPDB represents perfectly independent physical
    ground truth.

33. Do NOT claim that adaptive conformal guarantees exact 90%
    coverage under arbitrary temporal distribution shift.

34. Do NOT call all 23 horizons "24-hour-ahead forecasts".
    Preferred wording:
    "next-day 23-slot load-profile forecasting issued after the
    previous day's final observed slot."

35. Do NOT optimize any additional hyperparameters based on the
    locked 2024-2026 test results.
"""


CLAIMS_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "frozen_manuscript_claims.md"
)


CLAIMS_FILE.write_text(
    CLAIMS,
    encoding="utf-8"
)


# =====================================================================
# 36. RESULTS + DISCUSSION OUTLINE
# =====================================================================

OUTLINE = f"""
# RESULTS AND DISCUSSION — RECOMMENDED STRUCTURE

## 4.1 Experimental Setup and Evaluation Protocol

Explain the chronological evaluation design:

- Train: {train_df['target_date'].min().date()} to {train_df['target_date'].max().date()}
- Validation: {validation_df['target_date'].min().date()} to {validation_df['target_date'].max().date()}
- Test: {test_df['target_date'].min().date()} to {test_df['target_date'].max().date()}
- Test days: {len(test_df)}
- Forecast outputs: 23 slots per day
- Primary point model: Ridge alpha=10

Emphasize that the validation period was used for model selection,
while the later test period was reserved for final chronological
evaluation.

Recommended table:
Table 1 — Experimental design.


## 4.2 Point-Forecasting Performance

Lead with Ridge, not with the neural models.

Core result:

Ridge:
- MAE = {RIDGE_ROW['MAE_MW']:.2f} MW
- RMSE = {RIDGE_ROW['RMSE_MW']:.2f} MW
- MAPE = {RIDGE_ROW['MAPE_percent']:.3f}%

Previous-day baseline:
- MAE = {D1_ROW['MAE_MW']:.2f} MW

Improvement:
- {RIDGE_D1_STAT_ROW['Ridge_Skill_percent']:.2f}% MAE skill over D1.

Interpretation:

The relatively simple regularized linear model outperformed the tested
tree-based and deep-learning alternatives. This suggests that much of
the predictable national-load structure is encoded in recent load
profiles, rolling temporal statistics, and calendar structure.

Do not argue that deep learning is inherently unsuitable for
Bangladesh. State only that the tested architectures did not outperform
the frozen Ridge configuration under this dataset and protocol.

Recommended:
Table 2
Figure 1


## 4.3 Statistical Significance of Forecast Improvements

Primary comparison:
Ridge versus previous-day persistence.

Mean daily improvement:
{RIDGE_D1_STAT_ROW['Mean_Daily_MAE_Improvement_MW']:.2f} MW

7-day block-bootstrap 95% CI:
[{RIDGE_D1_STAT_ROW['Bootstrap7_CI_Lower_MW']:.2f},
 {RIDGE_D1_STAT_ROW['Bootstrap7_CI_Upper_MW']:.2f}] MW

14-day robustness CI:
[{RIDGE_D1_STAT_ROW['Bootstrap14_CI_Lower_MW']:.2f},
 {RIDGE_D1_STAT_ROW['Bootstrap14_CI_Upper_MW']:.2f}] MW

HAC + Holm:
p {ridge_p_statement}

Explain why daily losses are the statistical unit:
the 23 horizons within a day are not treated as independent samples.

Recommended:
Table 3
Figure 3


## 4.4 Feature-Group Ablation

Largest degradation:
{PRIMARY_ABLATION['Removed_Group']}

Test MAE increase:
{PRIMARY_ABLATION['Delta_Test_MAE_vs_Full_MW']:.2f} MW
({PRIMARY_ABLATION['Relative_Test_MAE_Change_percent']:.2f}%)

Interpretation:

The result supports the importance of very recent daily demand history.

Discuss previous-week information second.

Discuss weak/redundant feature groups cautiously.

Important:
Removing Rolling28_STD happens to improve test MAE slightly.
Because validation performance was worse and the test period was already
observed, this must be interpreted as post-hoc redundancy evidence, not
as a reason to alter the frozen model.

Recommended:
Table 4
Figure 4


## 4.5 Probabilistic Forecast Reliability

Introduce the limitation of the initial static conformal approaches:
they undercovered under temporal distribution shift.

Then present the primary prequential adaptive method:

Adaptive Trend-Regime Hour-Normalized W180:
- Nominal level = 90%
- Marginal empirical coverage = 90.354%
- Coverage gap = +0.354 percentage points
- MPIW ≈ 2042.42 MW
- Winkler ≈ 2966.04

Discuss the trade-off:
coverage improved because the adaptive intervals became wider.

Do not describe this as a free improvement in both sharpness and
coverage.

Then mention whole-profile simultaneous coverage separately:
approximately 89.56% for the primary trend-regime profile-max method.

The simultaneous band is much wider because all 23 slots must fall
inside the band simultaneously.

Recommended:
Table 5


## 4.6 Cross-Source Validation Against BPDB

Start with source consistency before forecast validation.

Explicit PGCB Evening_Peak versus BPDB generation-end evening demand:

- n = {int(SOURCE_ROW['n'])}
- Pearson r = {SOURCE_ROW['Pearson_r']:.6f}
- MAE = {SOURCE_ROW['MAE_MW']:.2f} MW
- MAPE = {SOURCE_ROW['MAPE_percent']:.4f}%
- Bias = {SOURCE_ROW['Bias_MW']:.2f} MW

Interpretation:

The exceptionally high agreement supports the consistency of the PGCB
target representation with the separate BPDB daily reporting source.

It does NOT prove that either source is an error-free physical ground
truth.

Recommended:
Table 6
Supplementary Figure S1


## 4.7 External Validation of the Forecasted Evening Peak

During the 2024 BPDB overlap:

- matched days = {int(EXTERNAL_FORECAST_ROW['n'])}
- r = {EXTERNAL_FORECAST_ROW['Pearson_r']:.6f}
- MAE = {EXTERNAL_FORECAST_ROW['MAE_MW']:.2f} MW
- RMSE = {EXTERNAL_FORECAST_ROW['RMSE_MW']:.2f} MW
- MAPE = {EXTERNAL_FORECAST_ROW['MAPE_percent']:.3f}%
- Bias = {EXTERNAL_FORECAST_ROW['Bias_MW']:.2f} MW

Explain target mismatch explicitly:
the model predicts integer-hour profile values, while the externally
reported BPDB evening peak can occur between those timestamps.

Therefore the external peak error should not be compared as if it were
identical to the primary 23-slot test MAE.

Recommended:
Supplementary Figures S2-S3


## 4.8 Temporal Robustness

Report year-specific Ridge skill against D1.

Observed skill range:
{minimum_temporal_skill:.2f}% to
{maximum_temporal_skill:.2f}%.

Interpretation:

The improvement remained similar in 2024, 2025, and the available
2026 evaluation period, which reduces concern that the headline gain
came from only one test year.

Recommended:
Table 7
Figure 2


## 4.9 Sensitivity of Evening_Peak Matching

Seven explicit Evening_Peak records occurred at unusual 01:00 or 12:00
timestamps.

However:

All records:
- r = {SENS_ALL['Pearson_r']:.6f}
- MAE = {SENS_ALL['MAE_MW']:.3f} MW

After excluding 01:00 / 12:00:
- r = {SENS_RESTRICTED['Pearson_r']:.6f}
- MAE = {SENS_RESTRICTED['MAE_MW']:.3f} MW

The metrics are unchanged on the BPDB overlap.

Therefore the unusual historical timestamps do not explain the strong
cross-source agreement.

Recommended:
Table 8
Supplementary Figure S5


## 4.10 Discussion

Organize the discussion around four findings.

### Finding 1 — Simpler model, strong result

The validation-selected Ridge model outperformed the tested nonlinear
and deep alternatives.

Discuss regularization, high temporal autocorrelation, and strong
lag-profile structure as plausible explanations.

Avoid claiming causal explanations.


### Finding 2 — Recent history is the dominant predictor

The ablation study shows the largest deterioration after removing the
previous-day profile, followed by the previous-week profile.

Relate this to persistence and weekly seasonality.


### Finding 3 — Static uncertainty calibration is insufficient under shift

The static conformal methods materially undercovered.

Prequential updating greatly improved empirical coverage.

Discuss this as evidence that uncertainty calibration must adapt to
time-varying residual distributions.


### Finding 4 — External reporting consistency strengthens credibility

PGCB explicit Evening_Peak records agree extremely closely with BPDB.

The forecast also tracks externally reported 2024 evening demand with
r ≈ {EXTERNAL_FORECAST_ROW['Pearson_r']:.3f}.

This strengthens the practical interpretation of the forecasting
results beyond internal test metrics.


## 4.11 Limitations

Include at least:

1. National aggregate demand only.
2. No complete causal weather-information pipeline.
3. Special PGCB reporting structure with 23 standard integer-hour slots.
4. Adaptive conformal evaluation is prequential and does not imply
   classical i.i.d. split-conformal guarantees.
5. BPDB and PGCB may share upstream operational reporting processes.
6. External evening peak is not exactly identical to the 23-slot
   prediction target.
7. Deep-learning robustness uses a limited architecture and seed set.
8. The study does not claim universal superiority of Ridge regression.


## 4.12 Practical Implications

Discuss:

- operational next-day demand scheduling,
- reserve planning,
- generation dispatch support,
- demand-supply balancing,
- transparent uncertainty bands,
- ease of deployment of a regularized linear model.


# Recommended Main-Paper Tables

Table 1 — Experimental design
Table 2 — Point forecasting benchmark
Table 3 — Statistical significance
Table 4 — Feature-group ablation
Table 5 — Probabilistic forecasting calibration
Table 6 — BPDB external validation
Table 7 — Temporal robustness

Table 8 can be supplementary if the journal has a strict page limit.


# Recommended Main-Paper Figures

Figure 1 — Locked-test model comparison
Figure 2 — Temporal robustness
Figure 3 — Bootstrap significance
Figure 4 — Feature ablation
Figure 5 — External validation

Keep additional BPDB scatter/time-series and sensitivity plots in the
supplementary material if space is limited.
"""


OUTLINE_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "results_discussion_outline.md"
)


OUTLINE_FILE.write_text(
    OUTLINE,
    encoding="utf-8"
)


# =====================================================================
# 37. CREATE README FOR PUBLICATION PACKAGE
# =====================================================================

README_TEXT = f"""
PUBLICATION PACKAGE
===================

Generated from frozen Steps 1-7.

Primary frozen test result
--------------------------
Ridge MAE: {RIDGE_ROW['MAE_MW']:.3f} MW
Ridge RMSE: {RIDGE_ROW['RMSE_MW']:.3f} MW
Ridge MAPE: {RIDGE_ROW['MAPE_percent']:.4f} %

Previous-day baseline MAE:
{D1_ROW['MAE_MW']:.3f} MW

Ridge skill:
{RIDGE_D1_STAT_ROW['Ridge_Skill_percent']:.3f} %

Primary statistical CI
----------------------
7-day block-bootstrap:
[{RIDGE_D1_STAT_ROW['Bootstrap7_CI_Lower_MW']:.3f},
 {RIDGE_D1_STAT_ROW['Bootstrap7_CI_Upper_MW']:.3f}] MW

Primary external source agreement
---------------------------------
n = {int(SOURCE_ROW['n'])}
r = {SOURCE_ROW['Pearson_r']:.6f}
MAE = {SOURCE_ROW['MAE_MW']:.3f} MW

2024 external forecast validation
---------------------------------
n = {int(EXTERNAL_FORECAST_ROW['n'])}
r = {EXTERNAL_FORECAST_ROW['Pearson_r']:.6f}
MAE = {EXTERNAL_FORECAST_ROW['MAE_MW']:.3f} MW
MAPE = {EXTERNAL_FORECAST_ROW['MAPE_percent']:.3f} %

Files
-----
publication_ready_results_tables.xlsx
frozen_manuscript_claims.md
results_discussion_outline.md
figure_manifest.csv

Individual CSV tables are also available in this directory.

IMPORTANT
---------
Do not retune models based on the locked test period after this point.
"""


README_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "README_publication_package.txt"
)


README_FILE.write_text(
    README_TEXT,
    encoding="utf-8"
)


# =====================================================================
# 38. FINAL SUMMARY JSON
# =====================================================================

FINAL_SUMMARY = {

    "ridge_MAE_MW":
        float(
            RIDGE_ROW[
                "MAE_MW"
            ]
        ),

    "ridge_RMSE_MW":
        float(
            RIDGE_ROW[
                "RMSE_MW"
            ]
        ),

    "ridge_MAPE_percent":
        float(
            RIDGE_ROW[
                "MAPE_percent"
            ]
        ),

    "d1_MAE_MW":
        float(
            D1_ROW[
                "MAE_MW"
            ]
        ),

    "ridge_skill_vs_d1_percent":
        float(
            RIDGE_D1_STAT_ROW[
                "Ridge_Skill_percent"
            ]
        ),

    "bootstrap7_CI_lower_MW":
        float(
            RIDGE_D1_STAT_ROW[
                "Bootstrap7_CI_Lower_MW"
            ]
        ),

    "bootstrap7_CI_upper_MW":
        float(
            RIDGE_D1_STAT_ROW[
                "Bootstrap7_CI_Upper_MW"
            ]
        ),

    "external_source_r":
        float(
            SOURCE_ROW[
                "Pearson_r"
            ]
        ),

    "external_source_MAE_MW":
        float(
            SOURCE_ROW[
                "MAE_MW"
            ]
        ),

    "external_forecast_r":
        float(
            EXTERNAL_FORECAST_ROW[
                "Pearson_r"
            ]
        ),

    "external_forecast_MAE_MW":
        float(
            EXTERNAL_FORECAST_ROW[
                "MAE_MW"
            ]
        ),

    "external_forecast_MAPE_percent":
        float(
            EXTERNAL_FORECAST_ROW[
                "MAPE_percent"
            ]
        ),

    "primary_ablation_group":
        str(
            PRIMARY_ABLATION[
                "Removed_Group"
            ]
        ),

    "primary_ablation_delta_MAE_MW":
        float(
            PRIMARY_ABLATION[
                "Delta_Test_MAE_vs_Full_MW"
            ]
        ),

    "temporal_skill_min_percent":
        minimum_temporal_skill,

    "temporal_skill_max_percent":
        maximum_temporal_skill,

    "publication_workbook":
        str(
            EXCEL_FILE
        ),

    "claims_file":
        str(
            CLAIMS_FILE
        ),

    "outline_file":
        str(
            OUTLINE_FILE
        )
}


FINAL_SUMMARY_FILE = (

    PUBLICATION_RESULT_DIR
    /
    "step8_publication_summary.json"
)


with open(
    FINAL_SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        FINAL_SUMMARY,
        file,
        indent=4
    )


# =====================================================================
# 39. VALIDATION CHECKS
# =====================================================================

assert (
    len(
        TABLE1
    )
    >
    0
)


assert (
    len(
        TABLE2
    )
    >=
    3
)


assert (
    "Ridge"
    in
    TABLE2[
        "Model"
    ]
    .values
)


assert (
    "D1"
    in
    TABLE2[
        "Model"
    ]
    .values
)


assert (
    RIDGE_ROW[
        "MAE_MW"
    ]
    <
    D1_ROW[
        "MAE_MW"
    ]
)


assert (
    RIDGE_D1_STAT_ROW[
        "Bootstrap7_CI_Lower_MW"
    ]
    >
    0
)


assert (
    SOURCE_ROW[
        "Pearson_r"
    ]
    >
    0.99
)


assert (
    EXTERNAL_FORECAST_ROW[
        "Pearson_r"
    ]
    >
    0.8
)


assert (
    EXCEL_FILE.exists()
)


assert (
    CLAIMS_FILE.exists()
)


assert (
    OUTLINE_FILE.exists()
)


assert (
    README_FILE.exists()
)


print(
    "\nAll Step 8 validation checks PASSED."
)


# =====================================================================
# 40. FINAL OUTPUT
# =====================================================================

print("\n")

print("=" * 125)

print(
    "STEP 8 COMPLETED SUCCESSFULLY"
)

print("=" * 125)


print("\n" + "-" * 125)

print(
    "A. PRIMARY POINT-FORECAST RESULT"
)

print("-" * 125)


print(
    "Ridge MAE:",
    round(
        RIDGE_ROW[
            "MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "Ridge RMSE:",
    round(
        RIDGE_ROW[
            "RMSE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "Ridge MAPE:",
    round(
        RIDGE_ROW[
            "MAPE_percent"
        ],
        4
    ),
    "%"
)


print(
    "Skill vs D1:",
    round(
        RIDGE_D1_STAT_ROW[
            "Ridge_Skill_percent"
        ],
        3
    ),
    "%"
)


print("\n" + "-" * 125)

print(
    "B. STATISTICAL EVIDENCE"
)

print("-" * 125)


print(
    "Mean daily MAE improvement:",
    round(
        RIDGE_D1_STAT_ROW[
            "Mean_Daily_MAE_Improvement_MW"
        ],
        3
    ),
    "MW"
)


print(
    "7-day bootstrap 95% CI:",
    round(
        RIDGE_D1_STAT_ROW[
            "Bootstrap7_CI_Lower_MW"
        ],
        3
    ),
    "to",
    round(
        RIDGE_D1_STAT_ROW[
            "Bootstrap7_CI_Upper_MW"
        ],
        3
    ),
    "MW"
)


print(
    "Holm-adjusted significance:",
    ridge_p_statement
)


print("\n" + "-" * 125)

print(
    "C. EXTERNAL VALIDATION"
)

print("-" * 125)


print(
    "PGCB-BPDB source r:",
    round(
        SOURCE_ROW[
            "Pearson_r"
        ],
        6
    )
)


print(
    "PGCB-BPDB source MAE:",
    round(
        SOURCE_ROW[
            "MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "2024 external forecast r:",
    round(
        EXTERNAL_FORECAST_ROW[
            "Pearson_r"
        ],
        6
    )
)


print(
    "2024 external forecast MAPE:",
    round(
        EXTERNAL_FORECAST_ROW[
            "MAPE_percent"
        ],
        3
    ),
    "%"
)


print("\n" + "-" * 125)

print(
    "D. PUBLICATION PACKAGE FILES"
)

print("-" * 125)


print(
    "\nMaster Excel workbook:"
)

print(
    EXCEL_FILE
)


print(
    "\nFrozen manuscript claims:"
)

print(
    CLAIMS_FILE
)


print(
    "\nResults/Discussion outline:"
)

print(
    OUTLINE_FILE
)


print(
    "\nFigure manifest:"
)

print(
    FIGURE_MANIFEST_FILE
)


print(
    "\nPublication figures:"
)

print(
    PUBLICATION_FIGURE_DIR
)


print(
    "\nPackage README:"
)

print(
    README_FILE
)


print(
    "\nStep 8 summary:"
)

print(
    FINAL_SUMMARY_FILE
)


print("\n" + "=" * 125)

print(
    "EXPERIMENTAL PIPELINE IS NOW PUBLICATION-FROZEN."
)

print("=" * 125)


print(
    "\nSEND THESE TO CHATGPT:"
)

print(
    "1. STEP 8 COMPLETED SUCCESSFULLY"
)

print(
    "2. TABLE 2 — POINT FORECASTING PERFORMANCE"
)

print(
    "3. TABLE 5 — CONFORMAL UNCERTAINTY"
)

print(
    "4. TABLE 6 — EXTERNAL BPDB VALIDATION"
)

print(
    "5. Any warning/error printed during Step 8"
)
```
