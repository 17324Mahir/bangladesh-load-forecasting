```python
# =====================================================================
# STEP 3
# MACHINE-LEARNING BASELINES FOR 23-SLOT NEXT-DAY LOAD FORECASTING
#
# Models:
#   1. Seasonal Naive D1
#   2. Seasonal Naive D7
#   3. Ridge Regression
#   4. Extra Trees
#   5. XGBoost Direct Multi-Horizon
#
# Primary task:
# Predict next-day demand profile:
# 01:00, 02:00, ..., 23:00
#
# IMPORTANT METHODOLOGY:
#
# TRAIN       = through 2022
# VALIDATION  = 2023
# TEST-A      = 2024
# TEST-B      = 2025 to March 2026
#
# Hyperparameters are selected ONLY on VALIDATION.
#
# TEST sets are NEVER used for model selection.
#
# Run this entire code in ONE Google Colab cell.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import sys
import json
import math
import time
import warnings
import subprocess

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")

np.random.seed(42)


pd.set_option(
    "display.max_columns",
    300
)

pd.set_option(
    "display.max_rows",
    200
)

pd.set_option(
    "display.width",
    260
)


print("=" * 110)

print(
    "STEP 3: MACHINE-LEARNING BASELINES"
)

print("=" * 110)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount(
    "/content/drive"
)


print(
    "\nGoogle Drive mounted successfully."
)


# =====================================================================
# 2. IMPORT SCIKIT-LEARN COMPONENTS
# =====================================================================

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import Ridge

from sklearn.ensemble import ExtraTreesRegressor

import joblib


# =====================================================================
# 3. ENSURE XGBOOST IS AVAILABLE
# =====================================================================

try:

    import xgboost as xgb

    from xgboost import XGBRegressor


except ImportError:

    print(
        "\nXGBoost not found."
    )

    print(
        "Installing XGBoost..."
    )

    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "xgboost"
        ]
    )

    import xgboost as xgb

    from xgboost import XGBRegressor


print(
    "\nXGBoost version:",
    xgb.__version__
)


# =====================================================================
# 4. DEFINE PROJECT DIRECTORIES
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)


MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "ml_baselines"
)


RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "ml_baselines"
)


FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "ml_baselines"
)


for directory in [

    MODEL_DIR,
    RESULT_DIR,
    FIGURE_DIR

]:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# =====================================================================
# 5. INPUT FILES
# =====================================================================

TRAIN_FILE = (
    FORECAST_DIR
    / "train_2015_2022.csv"
)


VALIDATION_FILE = (
    FORECAST_DIR
    / "validation_2023.csv"
)


TEST_FILE = (
    FORECAST_DIR
    / "test_2024_2026.csv"
)


for file_path in [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE

]:

    if not file_path.exists():

        raise FileNotFoundError(
            f"""
Required Step 2 file not found:

{file_path}

Please confirm Step 2 completed successfully.
"""
        )


# =====================================================================
# 6. LOAD DATASETS
# =====================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)


validation_df = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)


test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for dataframe in [

    train_df,
    validation_df,
    test_df

]:

    dataframe[
        "target_date"
    ] = pd.to_datetime(
        dataframe[
            "target_date"
        ],
        errors="coerce"
    )


print("\n" + "=" * 110)

print(
    "DATASETS LOADED"
)

print("=" * 110)


print(
    "TRAIN:",
    len(train_df),
    train_df[
        "target_date"
    ].min(),
    "to",
    train_df[
        "target_date"
    ].max()
)


print(
    "VALIDATION:",
    len(validation_df),
    validation_df[
        "target_date"
    ].min(),
    "to",
    validation_df[
        "target_date"
    ].max()
)


print(
    "ORIGINAL TEST:",
    len(test_df),
    test_df[
        "target_date"
    ].min(),
    "to",
    test_df[
        "target_date"
    ].max()
)


# =====================================================================
# 7. DEFINE TARGET HOURS
# =====================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 8. DEFINE FEATURE GROUPS EXACTLY AS STEP 2
# =====================================================================

CALENDAR_FEATURES = [

    "year",
    "month",
    "day",

    "dayofweek",
    "dayofyear",
    "weekofyear",

    "quarter",
    "is_weekend",

    "month_sin",
    "month_cos",

    "dow_sin",
    "dow_cos",

    "doy_sin",
    "doy_cos"
]


LAG1_FEATURES = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


LAG7_FEATURES = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


ROLL7_FEATURES = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


ROLL28_FEATURES = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


ROLL7STD_FEATURES = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


ROLL28STD_FEATURES = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


SUMMARY_FEATURES = [

    "lag1_daily_mean",
    "lag1_daily_max",
    "lag1_daily_min",
    "lag1_daily_range",
    "lag1_peak_hour",

    "lag7_daily_mean",
    "lag7_daily_max",
    "lag7_daily_min",
    "lag7_daily_range",
    "lag7_peak_hour",

    "mean_change_vs_last_week",
    "max_change_vs_last_week",
    "mean_ratio_vs_last_week"
]


FEATURE_COLUMNS = (

    CALENDAR_FEATURES

    +

    LAG1_FEATURES

    +

    LAG7_FEATURES

    +

    ROLL7_FEATURES

    +

    ROLL28_FEATURES

    +

    ROLL7STD_FEATURES

    +

    ROLL28STD_FEATURES

    +

    SUMMARY_FEATURES
)


print(
    "\nFeature count:",
    len(FEATURE_COLUMNS)
)


print(
    "Target count:",
    len(TARGET_COLUMNS)
)


# =====================================================================
# 9. FEATURE EXISTENCE CHECK
# =====================================================================

for dataframe_name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION",
        validation_df
    ),

    (
        "TEST",
        test_df
    )

]:

    missing_features = [

        feature

        for feature in FEATURE_COLUMNS

        if feature
        not in
        dataframe.columns
    ]


    missing_targets = [

        target

        for target in TARGET_COLUMNS

        if target
        not in
        dataframe.columns
    ]


    if len(
        missing_features
    ) > 0:

        raise ValueError(
            f"""
Missing features in {dataframe_name}:

{missing_features}
"""
        )


    if len(
        missing_targets
    ) > 0:

        raise ValueError(
            f"""
Missing targets in {dataframe_name}:

{missing_targets}
"""
        )


print(
    "\nAll expected feature and target columns found."
)


# =====================================================================
# 10. FINAL NaN CHECK
# =====================================================================

for dataframe_name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION",
        validation_df
    ),

    (
        "TEST",
        test_df
    )

]:

    feature_nan_count = int(

        dataframe[
            FEATURE_COLUMNS
        ]
        .isna()
        .sum()
        .sum()
    )


    target_nan_count = int(

        dataframe[
            TARGET_COLUMNS
        ]
        .isna()
        .sum()
        .sum()
    )


    print(
        f"\n{dataframe_name}:"
    )

    print(
        " Feature NaNs:",
        feature_nan_count
    )

    print(
        " Target NaNs:",
        target_nan_count
    )


    if (
        feature_nan_count > 0

        or

        target_nan_count > 0
    ):

        raise ValueError(
            f"{dataframe_name} contains NaN values."
        )


# =====================================================================
# 11. CREATE TWO TEMPORAL TEST SETS
# =====================================================================

# TEST-A:
# Calendar year 2024
#
# TEST-B:
# 2025 through available March 2026 records
#
# We will ALSO report ALL_TEST = 2024-2026.


test_a_df = (

    test_df.loc[

        test_df[
            "target_date"
        ]
        .between(

            pd.Timestamp(
                "2024-01-01"
            ),

            pd.Timestamp(
                "2024-12-31"
            )
        )
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


test_b_df = (

    test_df.loc[

        test_df[
            "target_date"
        ]
        >=
        pd.Timestamp(
            "2025-01-01"
        )
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 110)

print(
    "FINAL TEMPORAL EVALUATION WINDOWS"
)

print("=" * 110)


print(
    "TEST-A 2024:",
    len(test_a_df),
    "samples"
)


if len(
    test_a_df
) > 0:

    print(
        test_a_df[
            "target_date"
        ].min(),
        "to",
        test_a_df[
            "target_date"
        ].max()
    )


print(
    "\nTEST-B 2025-2026:",
    len(test_b_df),
    "samples"
)


if len(
    test_b_df
) > 0:

    print(
        test_b_df[
            "target_date"
        ].min(),
        "to",
        test_b_df[
            "target_date"
        ].max()
    )


# =====================================================================
# 12. PREPARE NUMPY ARRAYS
# =====================================================================

X_train = (

    train_df[
        FEATURE_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


Y_train = (

    train_df[
        TARGET_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


X_val = (

    validation_df[
        FEATURE_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


Y_val = (

    validation_df[
        TARGET_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


X_test = (

    test_df[
        FEATURE_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


Y_test = (

    test_df[
        TARGET_COLUMNS
    ]
    .to_numpy(
        dtype=np.float32
    )
)


print(
    "\nTraining matrix:",
    X_train.shape
)


print(
    "Training target matrix:",
    Y_train.shape
)


# =====================================================================
# 13. METRIC FUNCTIONS
# =====================================================================

def mae(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.mean(

            np.abs(
                y_true
                -
                y_pred
            )
        )
    )


def rmse(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.sqrt(

            np.mean(

                (
                    y_true
                    -
                    y_pred
                )
                **
                2
            )
        )
    )


def mape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            y_true
        ),

        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    y_true
                    -
                    y_pred
                )

                /

                denominator
            )
        )
    )


def smape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = (

        np.abs(
            y_true
        )

        +

        np.abs(
            y_pred
        )
    )


    valid = (
        denominator > 0
    )


    if not np.any(
        valid
    ):

        return np.nan


    return float(

        100

        *

        np.mean(

            2

            *

            np.abs(

                y_true[
                    valid
                ]

                -

                y_pred[
                    valid
                ]
            )

            /

            denominator[
                valid
            ]
        )
    )


# =====================================================================
# 14. MASE SCALE
# =====================================================================

# We use the TRAINING seasonal-naive D1 error as the scaling
# denominator.
#
# MASE < 1 means model MAE is better than this in-sample
# seasonal persistence scale.


TRAIN_D1_PRED = (

    train_df[
        LAG1_FEATURES
    ]
    .to_numpy(
        dtype=float
    )
)


TRAIN_MASE_SCALE = mae(
    Y_train,
    TRAIN_D1_PRED
)


print(
    "\nTraining seasonal D1 MAE scale:",
    round(
        TRAIN_MASE_SCALE,
        4
    ),
    "MW"
)


# =====================================================================
# 15. GENERAL METRIC FUNCTION
# =====================================================================

def calculate_metrics(
    y_true,
    y_pred,
    mase_scale=TRAIN_MASE_SCALE
):

    return {

        "MAE_MW":
            mae(
                y_true,
                y_pred
            ),

        "RMSE_MW":
            rmse(
                y_true,
                y_pred
            ),

        "MAPE_percent":
            mape(
                y_true,
                y_pred
            ),

        "sMAPE_percent":
            smape(
                y_true,
                y_pred
            ),

        "MASE":
            (
                mae(
                    y_true,
                    y_pred
                )

                /
                mase_scale
            )
    }


# =====================================================================
# 16. BASELINE PREDICTIONS
# =====================================================================

def get_naive_predictions(
    dataframe,
    prefix
):

    columns = [

        f"{prefix}_h{hour:02d}"

        for hour in HOURS
    ]


    return (

        dataframe[
            columns
        ]

        .to_numpy(
            dtype=float
        )
    )


NAIVE_D1_TEST = (
    get_naive_predictions(
        test_df,
        "lag1"
    )
)


NAIVE_D7_TEST = (
    get_naive_predictions(
        test_df,
        "lag7"
    )
)


# =====================================================================
# 17. VALIDATION BASELINE METRICS
# =====================================================================

val_d1_prediction = (
    get_naive_predictions(
        validation_df,
        "lag1"
    )
)


val_d7_prediction = (
    get_naive_predictions(
        validation_df,
        "lag7"
    )
)


print("\n" + "=" * 110)

print(
    "VALIDATION NAIVE BASELINES"
)

print("=" * 110)


validation_baseline_table = pd.DataFrame(
    [

        {
            "model":
                "Seasonal_Naive_D1",

            **calculate_metrics(
                Y_val,
                val_d1_prediction
            )
        },

        {
            "model":
                "Seasonal_Naive_D7",

            **calculate_metrics(
                Y_val,
                val_d7_prediction
            )
        }
    ]
)


display(
    validation_baseline_table.round(4)
)


# =====================================================================
# 18. RIDGE REGRESSION HYPERPARAMETER SEARCH
# =====================================================================

print("\n" + "=" * 110)

print(
    "RIDGE REGRESSION VALIDATION SEARCH"
)

print("=" * 110)


ridge_alphas = [

    0.01,
    0.1,
    1.0,
    10.0,
    100.0,
    1000.0
]


ridge_scaler_train = (
    StandardScaler()
)


X_train_scaled = (

    ridge_scaler_train
    .fit_transform(
        X_train
    )
)


X_val_scaled = (

    ridge_scaler_train
    .transform(
        X_val
    )
)


ridge_search_rows = []


for alpha in ridge_alphas:

    start_time = time.time()


    ridge_model = Ridge(
        alpha=alpha
    )


    ridge_model.fit(
        X_train_scaled,
        Y_train
    )


    validation_prediction = (
        ridge_model.predict(
            X_val_scaled
        )
    )


    metrics = calculate_metrics(
        Y_val,
        validation_prediction
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    ridge_search_rows.append(
        {

            "alpha":
                alpha,

            **metrics,

            "training_seconds":
                elapsed
        }
    )


    print(
        f"alpha={alpha:<8} "
        f"MAE={metrics['MAE_MW']:.3f} MW "
        f"MAPE={metrics['MAPE_percent']:.3f}%"
    )


ridge_search = pd.DataFrame(
    ridge_search_rows
)


ridge_search = (

    ridge_search
    .sort_values(
        "MAE_MW"
    )
    .reset_index(
        drop=True
    )
)


BEST_RIDGE_ALPHA = float(
    ridge_search.iloc[0][
        "alpha"
    ]
)


print(
    "\nBest Ridge alpha:",
    BEST_RIDGE_ALPHA
)


display(
    ridge_search.round(4)
)


# =====================================================================
# 19. EXTRA TREES HYPERPARAMETER SEARCH
# =====================================================================

print("\n" + "=" * 110)

print(
    "EXTRA TREES VALIDATION SEARCH"
)

print("=" * 110)


extra_configs = [

    {
        "name":
            "ET_A",

        "n_estimators":
            400,

        "max_depth":
            None,

        "min_samples_leaf":
            1,

        "max_features":
            1.0
    },

    {
        "name":
            "ET_B",

        "n_estimators":
            500,

        "max_depth":
            None,

        "min_samples_leaf":
            1,

        "max_features":
            0.70
    },

    {
        "name":
            "ET_C",

        "n_estimators":
            500,

        "max_depth":
            None,

        "min_samples_leaf":
            2,

        "max_features":
            0.80
    },

    {
        "name":
            "ET_D",

        "n_estimators":
            500,

        "max_depth":
            20,

        "min_samples_leaf":
            1,

        "max_features":
            0.80
    }
]


extra_search_rows = []


for config in extra_configs:

    print(
        "\nTraining",
        config[
            "name"
        ]
    )


    start_time = time.time()


    model = ExtraTreesRegressor(

        n_estimators=
            config[
                "n_estimators"
            ],

        max_depth=
            config[
                "max_depth"
            ],

        min_samples_leaf=
            config[
                "min_samples_leaf"
            ],

        max_features=
            config[
                "max_features"
            ],

        random_state=42,

        n_jobs=-1
    )


    model.fit(
        X_train,
        Y_train
    )


    prediction = model.predict(
        X_val
    )


    metrics = calculate_metrics(
        Y_val,
        prediction
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    extra_search_rows.append(
        {

            **config,

            **metrics,

            "training_seconds":
                elapsed
        }
    )


    print(
        f"Validation MAE: "
        f"{metrics['MAE_MW']:.3f} MW"
    )

    print(
        f"Validation MAPE: "
        f"{metrics['MAPE_percent']:.3f}%"
    )


extra_search = pd.DataFrame(
    extra_search_rows
)


extra_search = (

    extra_search
    .sort_values(
        "MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


BEST_EXTRA_NAME = (
    extra_search.iloc[0][
        "name"
    ]
)


BEST_EXTRA_CONFIG = next(

    config

    for config in extra_configs

    if config[
        "name"
    ]
    ==
    BEST_EXTRA_NAME
)


print(
    "\nBest Extra Trees configuration:",
    BEST_EXTRA_NAME
)


display(
    extra_search.round(4)
)


# =====================================================================
# 20. XGBOOST CONFIGURATION SEARCH
# =====================================================================

# We train one DIRECT model for each target hour.
#
# Therefore:
#
# model 01 predicts 01:00
# model 02 predicts 02:00
# ...
# model 23 predicts 23:00
#
# This avoids recursive error accumulation.


print("\n" + "=" * 110)

print(
    "XGBOOST DIRECT MULTI-HORIZON VALIDATION SEARCH"
)

print("=" * 110)


xgb_configs = [

    {
        "name":
            "XGB_A",

        "n_estimators":
            500,

        "learning_rate":
            0.03,

        "max_depth":
            3,

        "min_child_weight":
            3,

        "subsample":
            0.90,

        "colsample_bytree":
            0.90,

        "reg_lambda":
            5.0,

        "reg_alpha":
            0.0
    },

    {
        "name":
            "XGB_B",

        "n_estimators":
            500,

        "learning_rate":
            0.03,

        "max_depth":
            4,

        "min_child_weight":
            3,

        "subsample":
            0.90,

        "colsample_bytree":
            0.90,

        "reg_lambda":
            5.0,

        "reg_alpha":
            0.0
    },

    {
        "name":
            "XGB_C",

        "n_estimators":
            500,

        "learning_rate":
            0.03,

        "max_depth":
            5,

        "min_child_weight":
            3,

        "subsample":
            0.90,

        "colsample_bytree":
            0.85,

        "reg_lambda":
            8.0,

        "reg_alpha":
            0.05
    },

    {
        "name":
            "XGB_D",

        "n_estimators":
            350,

        "learning_rate":
            0.05,

        "max_depth":
            4,

        "min_child_weight":
            3,

        "subsample":
            0.90,

        "colsample_bytree":
            0.90,

        "reg_lambda":
            5.0,

        "reg_alpha":
            0.0
    }
]


def create_xgb_model(
    config
):

    return XGBRegressor(

        objective=
            "reg:squarederror",

        n_estimators=
            config[
                "n_estimators"
            ],

        learning_rate=
            config[
                "learning_rate"
            ],

        max_depth=
            config[
                "max_depth"
            ],

        min_child_weight=
            config[
                "min_child_weight"
            ],

        subsample=
            config[
                "subsample"
            ],

        colsample_bytree=
            config[
                "colsample_bytree"
            ],

        reg_lambda=
            config[
                "reg_lambda"
            ],

        reg_alpha=
            config[
                "reg_alpha"
            ],

        tree_method=
            "hist",

        random_state=42,

        n_jobs=-1,

        verbosity=0
    )


xgb_search_rows = []


for config in xgb_configs:

    print(
        "\n" + "-" * 90
    )

    print(
        "Training configuration:",
        config[
            "name"
        ]
    )

    print(
        "-" * 90
    )


    start_time = time.time()


    validation_prediction = np.zeros_like(
        Y_val,
        dtype=float
    )


    for target_index, hour in enumerate(
        HOURS
    ):

        model = create_xgb_model(
            config
        )


        model.fit(

            X_train,

            Y_train[
                :,
                target_index
            ]
        )


        validation_prediction[
            :,
            target_index
        ] = model.predict(
            X_val
        )


        if (
            target_index
            in
            [
                0,
                5,
                11,
                17,
                22
            ]
        ):

            print(
                f"  Completed target hour "
                f"{hour:02d}:00"
            )


    metrics = calculate_metrics(

        Y_val,

        validation_prediction
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    xgb_search_rows.append(
        {

            **config,

            **metrics,

            "training_seconds":
                elapsed
        }
    )


    print(
        "\nValidation MAE:",
        round(
            metrics[
                "MAE_MW"
            ],
            3
        ),
        "MW"
    )


    print(
        "Validation RMSE:",
        round(
            metrics[
                "RMSE_MW"
            ],
            3
        ),
        "MW"
    )


    print(
        "Validation MAPE:",
        round(
            metrics[
                "MAPE_percent"
            ],
            3
        ),
        "%"
    )


xgb_search = pd.DataFrame(
    xgb_search_rows
)


xgb_search = (

    xgb_search
    .sort_values(
        "MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


BEST_XGB_NAME = (
    xgb_search.iloc[0][
        "name"
    ]
)


BEST_XGB_CONFIG = next(

    config

    for config in xgb_configs

    if config[
        "name"
    ]
    ==
    BEST_XGB_NAME
)


print("\n" + "=" * 110)

print(
    "BEST XGBOOST CONFIGURATION"
)

print("=" * 110)


print(
    BEST_XGB_CONFIG
)


display(
    xgb_search.round(4)
)


# =====================================================================
# 21. ALGORITHM-LEVEL VALIDATION COMPARISON
# =====================================================================

algorithm_validation = pd.DataFrame(
    [

        {

            "model":
                "Seasonal_Naive_D1",

            **calculate_metrics(
                Y_val,
                val_d1_prediction
            )
        },

        {

            "model":
                "Seasonal_Naive_D7",

            **calculate_metrics(
                Y_val,
                val_d7_prediction
            )
        },

        {

            "model":
                "Ridge",

            "MAE_MW":
                float(
                    ridge_search.iloc[0][
                        "MAE_MW"
                    ]
                ),

            "RMSE_MW":
                float(
                    ridge_search.iloc[0][
                        "RMSE_MW"
                    ]
                ),

            "MAPE_percent":
                float(
                    ridge_search.iloc[0][
                        "MAPE_percent"
                    ]
                ),

            "sMAPE_percent":
                float(
                    ridge_search.iloc[0][
                        "sMAPE_percent"
                    ]
                ),

            "MASE":
                float(
                    ridge_search.iloc[0][
                        "MASE"
                    ]
                )
        },

        {

            "model":
                "ExtraTrees",

            "MAE_MW":
                float(
                    extra_search.iloc[0][
                        "MAE_MW"
                    ]
                ),

            "RMSE_MW":
                float(
                    extra_search.iloc[0][
                        "RMSE_MW"
                    ]
                ),

            "MAPE_percent":
                float(
                    extra_search.iloc[0][
                        "MAPE_percent"
                    ]
                ),

            "sMAPE_percent":
                float(
                    extra_search.iloc[0][
                        "sMAPE_percent"
                    ]
                ),

            "MASE":
                float(
                    extra_search.iloc[0][
                        "MASE"
                    ]
                )
        },

        {

            "model":
                "XGBoost",

            "MAE_MW":
                float(
                    xgb_search.iloc[0][
                        "MAE_MW"
                    ]
                ),

            "RMSE_MW":
                float(
                    xgb_search.iloc[0][
                        "RMSE_MW"
                    ]
                ),

            "MAPE_percent":
                float(
                    xgb_search.iloc[0][
                        "MAPE_percent"
                    ]
                ),

            "sMAPE_percent":
                float(
                    xgb_search.iloc[0][
                        "sMAPE_percent"
                    ]
                ),

            "MASE":
                float(
                    xgb_search.iloc[0][
                        "MASE"
                    ]
                )
        }
    ]
)


algorithm_validation = (

    algorithm_validation
    .sort_values(
        "MAE_MW"
    )
    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 110)

print(
    "VALIDATION MODEL LEADERBOARD"
)

print("=" * 110)


display(
    algorithm_validation.round(4)
)


# =====================================================================
# 22. COMBINE TRAIN + VALIDATION FOR FINAL FIT
# =====================================================================

# Hyperparameters have now been selected.
#
# We are allowed to combine TRAIN + VALIDATION before evaluating
# on the untouched temporal test periods.


train_val_df = pd.concat(

    [
        train_df,
        validation_df
    ],

    axis=0,

    ignore_index=True
)


train_val_df = (

    train_val_df
    .sort_values(
        "target_date"
    )

    .reset_index(
        drop=True
    )
)


X_train_val = (

    train_val_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )
)


Y_train_val = (

    train_val_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )
)


print(
    "\nCombined TRAIN + VALIDATION samples:",
    len(
        train_val_df
    )
)


# =====================================================================
# 23. FINAL RIDGE MODEL
# =====================================================================

print("\n" + "=" * 110)

print(
    "FINAL RIDGE FIT"
)

print("=" * 110)


final_scaler = StandardScaler()


X_train_val_scaled = (

    final_scaler
    .fit_transform(
        X_train_val
    )
)


X_test_scaled = (

    final_scaler
    .transform(
        X_test
    )
)


final_ridge = Ridge(
    alpha=
        BEST_RIDGE_ALPHA
)


final_ridge.fit(
    X_train_val_scaled,
    Y_train_val
)


RIDGE_TEST_PRED = (
    final_ridge.predict(
        X_test_scaled
    )
)


print(
    "Final Ridge training completed."
)


# =====================================================================
# 24. FINAL EXTRA TREES MODEL
# =====================================================================

print("\n" + "=" * 110)

print(
    "FINAL EXTRA TREES FIT"
)

print("=" * 110)


final_extra = ExtraTreesRegressor(

    n_estimators=
        BEST_EXTRA_CONFIG[
            "n_estimators"
        ],

    max_depth=
        BEST_EXTRA_CONFIG[
            "max_depth"
        ],

    min_samples_leaf=
        BEST_EXTRA_CONFIG[
            "min_samples_leaf"
        ],

    max_features=
        BEST_EXTRA_CONFIG[
            "max_features"
        ],

    random_state=42,

    n_jobs=-1
)


final_extra.fit(
    X_train_val,
    Y_train_val
)


EXTRA_TEST_PRED = (

    final_extra.predict(
        X_test
    )
)


print(
    "Final Extra Trees training completed."
)


# =====================================================================
# 25. FINAL XGBOOST DIRECT MODELS
# =====================================================================

print("\n" + "=" * 110)

print(
    "FINAL XGBOOST DIRECT 23-HOUR FIT"
)

print("=" * 110)


XGB_TEST_PRED = np.zeros_like(
    Y_test,
    dtype=float
)


xgb_models = []


for target_index, hour in enumerate(
    HOURS
):

    print(
        f"Training XGBoost target "
        f"{hour:02d}:00 "
        f"({target_index + 1}/23)"
    )


    model = create_xgb_model(
        BEST_XGB_CONFIG
    )


    model.fit(

        X_train_val,

        Y_train_val[
            :,
            target_index
        ]
    )


    XGB_TEST_PRED[
        :,
        target_index
    ] = model.predict(
        X_test
    )


    xgb_models.append(
        model
    )


print(
    "\nAll 23 final XGBoost models completed."
)


# =====================================================================
# 26. SAVE TRAINED MODELS
# =====================================================================

joblib.dump(

    final_scaler,

    MODEL_DIR
    /
    "ridge_scaler.joblib"
)


joblib.dump(

    final_ridge,

    MODEL_DIR
    /
    "ridge_model.joblib"
)


joblib.dump(

    final_extra,

    MODEL_DIR
    /
    "extra_trees_model.joblib"
)


joblib.dump(

    xgb_models,

    MODEL_DIR
    /
    "xgboost_23_direct_models.joblib"
)


print(
    "\nModels saved to Google Drive."
)


# =====================================================================
# 27. BUILD PREDICTION DICTIONARY
# =====================================================================

all_test_predictions = {

    "Seasonal_Naive_D1":
        NAIVE_D1_TEST,

    "Seasonal_Naive_D7":
        NAIVE_D7_TEST,

    "Ridge":
        RIDGE_TEST_PRED,

    "ExtraTrees":
        EXTRA_TEST_PRED,

    "XGBoost":
        XGB_TEST_PRED
}


# =====================================================================
# 28. TEMPORAL SUBSET MASKS
# =====================================================================

mask_test_a = (

    test_df[
        "target_date"
    ]
    .dt.year
    ==
    2024
).to_numpy()


mask_test_b = (

    test_df[
        "target_date"
    ]
    .dt.year
    >=
    2025
).to_numpy()


evaluation_splits = {

    "TEST_A_2024":
        mask_test_a,

    "TEST_B_2025_2026":
        mask_test_b,

    "ALL_TEST_2024_2026":
        np.ones(
            len(test_df),
            dtype=bool
        )
}


# =====================================================================
# 29. FULL MODEL LEADERBOARD
# =====================================================================

leaderboard_rows = []


for split_name, mask in (
    evaluation_splits.items()
):

    y_true = (
        Y_test[
            mask
        ]
    )


    d1_prediction = (
        NAIVE_D1_TEST[
            mask
        ]
    )


    d1_mae = mae(
        y_true,
        d1_prediction
    )


    for model_name, prediction in (
        all_test_predictions.items()
    ):

        y_pred = (
            prediction[
                mask
            ]
        )


        metrics = calculate_metrics(
            y_true,
            y_pred
        )


        skill_vs_d1 = (

            100

            *
            (
                1

                -

                metrics[
                    "MAE_MW"
                ]
                /
                d1_mae
            )
        )


        leaderboard_rows.append(
            {

                "split":
                    split_name,

                "model":
                    model_name,

                "samples":
                    int(
                        mask.sum()
                    ),

                **metrics,

                "MAE_skill_vs_D1_percent":
                    skill_vs_d1
            }
        )


leaderboard = pd.DataFrame(
    leaderboard_rows
)


leaderboard = (

    leaderboard
    .sort_values(
        [
            "split",
            "MAE_MW"
        ]
    )
    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 110)

print(
    "FINAL TEST LEADERBOARD"
)

print("=" * 110)


display(
    leaderboard.round(4)
)


# =====================================================================
# 30. HORIZON-WISE METRICS
# =====================================================================

horizon_rows = []


for split_name, mask in (
    evaluation_splits.items()
):

    for model_name, prediction in (
        all_test_predictions.items()
    ):

        for j, hour in enumerate(
            HOURS
        ):

            true_h = (
                Y_test[
                    mask,
                    j
                ]
            )


            pred_h = (
                prediction[
                    mask,
                    j
                ]
            )


            metrics = calculate_metrics(
                true_h,
                pred_h
            )


            horizon_rows.append(
                {

                    "split":
                        split_name,

                    "model":
                        model_name,

                    "hour":
                        hour,

                    **metrics
                }
            )


horizon_metrics = pd.DataFrame(
    horizon_rows
)


# =====================================================================
# 31. DAILY PROFILE UTILITY METRICS
# =====================================================================

def profile_utility_metrics(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    true_peak_value = (
        np.max(
            y_true,
            axis=1
        )
    )


    pred_peak_value = (
        np.max(
            y_pred,
            axis=1
        )
    )


    true_peak_position = (
        np.argmax(
            y_true,
            axis=1
        )
    )


    pred_peak_position = (
        np.argmax(
            y_pred,
            axis=1
        )
    )


    true_peak_hour = np.array(
        [
            HOURS[position]

            for position
            in true_peak_position
        ]
    )


    pred_peak_hour = np.array(
        [
            HOURS[position]

            for position
            in pred_peak_position
        ]
    )


    peak_hour_error = np.abs(

        true_peak_hour

        -

        pred_peak_hour
    )


    true_profile_sum = (
        np.sum(
            y_true,
            axis=1
        )
    )


    pred_profile_sum = (
        np.sum(
            y_pred,
            axis=1
        )
    )


    return {

        "Peak_Magnitude_MAE_MW":
            mae(
                true_peak_value,
                pred_peak_value
            ),

        "Peak_Hour_Exact_Accuracy_percent":
            100
            *
            float(
                np.mean(
                    peak_hour_error
                    ==
                    0
                )
            ),

        "Peak_Hour_Within_1h_Accuracy_percent":
            100
            *
            float(
                np.mean(
                    peak_hour_error
                    <=
                    1
                )
            ),

        "Mean_Peak_Hour_Error_hours":
            float(
                np.mean(
                    peak_hour_error
                )
            ),

        "Daily_Profile_Sum_MAE_MWh_proxy":
            mae(
                true_profile_sum,
                pred_profile_sum
            )
    }


utility_rows = []


for split_name, mask in (
    evaluation_splits.items()
):

    for model_name, prediction in (
        all_test_predictions.items()
    ):

        utility_rows.append(
            {

                "split":
                    split_name,

                "model":
                    model_name,

                **profile_utility_metrics(

                    Y_test[
                        mask
                    ],

                    prediction[
                        mask
                    ]
                )
            }
        )


utility_metrics = pd.DataFrame(
    utility_rows
)


print("\n" + "=" * 110)

print(
    "PRACTICAL PROFILE / PEAK METRICS"
)

print("=" * 110)


display(
    utility_metrics.round(4)
)


# =====================================================================
# 32. XGBOOST FEATURE IMPORTANCE
# =====================================================================

importance_matrix = []


for model in xgb_models:

    importance_matrix.append(
        model.feature_importances_
    )


importance_matrix = np.asarray(
    importance_matrix
)


average_importance = (
    importance_matrix.mean(
        axis=0
    )
)


feature_importance = pd.DataFrame(
    {

        "feature":
            FEATURE_COLUMNS,

        "mean_importance":
            average_importance
    }
)


feature_importance = (

    feature_importance
    .sort_values(
        "mean_importance",
        ascending=False
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nTop 30 XGBoost features:"
)


display(
    feature_importance.head(30)
)


# =====================================================================
# 33. FEATURE GROUP IMPORTANCE
# =====================================================================

def feature_group(
    feature
):

    if feature in CALENDAR_FEATURES:

        return "Calendar"


    if feature in LAG1_FEATURES:

        return "Previous_Day_Profile"


    if feature in LAG7_FEATURES:

        return "Previous_Week_Profile"


    if feature in ROLL7_FEATURES:

        return "Rolling_7Day_Mean"


    if feature in ROLL28_FEATURES:

        return "Rolling_28Day_Mean"


    if feature in ROLL7STD_FEATURES:

        return "Rolling_7Day_STD"


    if feature in ROLL28STD_FEATURES:

        return "Rolling_28Day_STD"


    if feature in SUMMARY_FEATURES:

        return "Historical_Summary"


    return "Other"


feature_importance[
    "feature_group"
] = (

    feature_importance[
        "feature"
    ]
    .apply(
        feature_group
    )
)


group_importance = (

    feature_importance

    .groupby(
        "feature_group"
    )[
        "mean_importance"
    ]

    .sum()

    .sort_values(
        ascending=False
    )

    .reset_index()
)


print(
    "\nXGBoost feature-group importance:"
)


display(
    group_importance
)


# =====================================================================
# 34. CREATE PREDICTION OUTPUT FILE
# =====================================================================

prediction_output = pd.DataFrame(
    {

        "target_date":
            test_df[
                "target_date"
            ]
    }
)


for j, hour in enumerate(
    HOURS
):

    prediction_output[
        f"actual_h{hour:02d}"
    ] = (
        Y_test[
            :,
            j
        ]
    )


for model_name, prediction in (
    all_test_predictions.items()
):

    safe_name = (

        model_name
        .lower()
        .replace(
            " ",
            "_"
        )
    )


    for j, hour in enumerate(
        HOURS
    ):

        prediction_output[
            f"{safe_name}_pred_h{hour:02d}"
        ] = (

            prediction[
                :,
                j
            ]
        )


# =====================================================================
# 35. SAVE RESULT TABLES
# =====================================================================

RIDGE_SEARCH_FILE = (
    RESULT_DIR
    / "ridge_validation_search.csv"
)


EXTRA_SEARCH_FILE = (
    RESULT_DIR
    / "extra_trees_validation_search.csv"
)


XGB_SEARCH_FILE = (
    RESULT_DIR
    / "xgboost_validation_search.csv"
)


VALIDATION_LEADERBOARD_FILE = (
    RESULT_DIR
    / "validation_model_leaderboard.csv"
)


TEST_LEADERBOARD_FILE = (
    RESULT_DIR
    / "final_test_leaderboard.csv"
)


HORIZON_METRICS_FILE = (
    RESULT_DIR
    / "test_metrics_by_hour.csv"
)


UTILITY_METRICS_FILE = (
    RESULT_DIR
    / "profile_utility_metrics.csv"
)


FEATURE_IMPORTANCE_FILE = (
    RESULT_DIR
    / "xgboost_feature_importance.csv"
)


GROUP_IMPORTANCE_FILE = (
    RESULT_DIR
    / "xgboost_feature_group_importance.csv"
)


PREDICTION_FILE = (
    RESULT_DIR
    / "all_test_predictions.csv"
)


ridge_search.to_csv(
    RIDGE_SEARCH_FILE,
    index=False
)


extra_search.to_csv(
    EXTRA_SEARCH_FILE,
    index=False
)


xgb_search.to_csv(
    XGB_SEARCH_FILE,
    index=False
)


algorithm_validation.to_csv(
    VALIDATION_LEADERBOARD_FILE,
    index=False
)


leaderboard.to_csv(
    TEST_LEADERBOARD_FILE,
    index=False
)


horizon_metrics.to_csv(
    HORIZON_METRICS_FILE,
    index=False
)


utility_metrics.to_csv(
    UTILITY_METRICS_FILE,
    index=False
)


feature_importance.to_csv(
    FEATURE_IMPORTANCE_FILE,
    index=False
)


group_importance.to_csv(
    GROUP_IMPORTANCE_FILE,
    index=False
)


prediction_output.to_csv(
    PREDICTION_FILE,
    index=False
)


# =====================================================================
# 36. FIGURE 1
# VALIDATION MODEL COMPARISON
# =====================================================================

plot_validation = (

    algorithm_validation
    .sort_values(
        "MAE_MW"
    )
)


plt.figure(
    figsize=(10, 6)
)


plt.bar(
    plot_validation[
        "model"
    ],
    plot_validation[
        "MAE_MW"
    ]
)


plt.ylabel(
    "Validation MAE (MW)"
)


plt.xlabel(
    "Model"
)


plt.title(
    "Validation Performance of Forecasting Models"
)


plt.xticks(
    rotation=25
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG1 = (
    FIGURE_DIR
    / "01_validation_model_comparison.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 37. FIGURE 2
# ALL-TEST MODEL COMPARISON
# =====================================================================

all_test_leaderboard = (

    leaderboard.loc[
        leaderboard[
            "split"
        ]
        ==
        "ALL_TEST_2024_2026"
    ]

    .sort_values(
        "MAE_MW"
    )
)


plt.figure(
    figsize=(10, 6)
)


plt.bar(
    all_test_leaderboard[
        "model"
    ],
    all_test_leaderboard[
        "MAE_MW"
    ]
)


plt.ylabel(
    "Test MAE (MW)"
)


plt.xlabel(
    "Model"
)


plt.title(
    "Out-of-Sample Next-Day Load Forecasting Performance"
)


plt.xticks(
    rotation=25
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG2 = (
    FIGURE_DIR
    / "02_test_model_comparison.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 38. FIGURE 3
# HORIZON-WISE MAE
# =====================================================================

all_test_hourly = (

    horizon_metrics.loc[
        horizon_metrics[
            "split"
        ]
        ==
        "ALL_TEST_2024_2026"
    ]
)


plt.figure(
    figsize=(14, 7)
)


for model_name in [

    "Seasonal_Naive_D1",
    "Ridge",
    "ExtraTrees",
    "XGBoost"

]:

    temp = (

        all_test_hourly.loc[
            all_test_hourly[
                "model"
            ]
            ==
            model_name
        ]

        .sort_values(
            "hour"
        )
    )


    plt.plot(
        temp[
            "hour"
        ],
        temp[
            "MAE_MW"
        ],
        marker="o",
        linewidth=1.8,
        label=model_name
    )


plt.xticks(
    HOURS
)


plt.xlabel(
    "Forecast Target Hour"
)


plt.ylabel(
    "MAE (MW)"
)


plt.title(
    "Forecast Error Across the 23-Slot Next-Day Load Profile"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG3 = (
    FIGURE_DIR
    / "03_test_mae_by_hour.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 39. FIGURE 4
# XGBOOST FEATURE GROUP IMPORTANCE
# =====================================================================

plt.figure(
    figsize=(11, 6)
)


plt.bar(
    group_importance[
        "feature_group"
    ],
    group_importance[
        "mean_importance"
    ]
)


plt.ylabel(
    "Aggregated Mean Importance"
)


plt.xlabel(
    "Feature Group"
)


plt.title(
    "XGBoost Feature-Group Importance Across Forecast Horizons"
)


plt.xticks(
    rotation=35,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG4 = (
    FIGURE_DIR
    / "04_xgboost_feature_group_importance.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 40. FIGURE 5
# EXAMPLE TEST PROFILE
# =====================================================================

# Use a deterministic middle observation from TEST-A 2024.


if len(
    test_a_df
) > 0:

    test_a_positions = np.where(
        mask_test_a
    )[0]


    example_global_index = (

        test_a_positions[
            len(
                test_a_positions
            )
            //
            2
        ]
    )


    actual_example = (
        Y_test[
            example_global_index
        ]
    )


    naive_example = (
        NAIVE_D1_TEST[
            example_global_index
        ]
    )


    ridge_example = (
        RIDGE_TEST_PRED[
            example_global_index
        ]
    )


    extra_example = (
        EXTRA_TEST_PRED[
            example_global_index
        ]
    )


    xgb_example = (
        XGB_TEST_PRED[
            example_global_index
        ]
    )


    example_date = (
        test_df.loc[
            example_global_index,
            "target_date"
        ]
    )


    plt.figure(
        figsize=(13, 7)
    )


    plt.plot(
        HOURS,
        actual_example,
        marker="o",
        linewidth=3,
        label="Actual"
    )


    plt.plot(
        HOURS,
        naive_example,
        marker="s",
        linewidth=1.5,
        label="Naive D1"
    )


    plt.plot(
        HOURS,
        ridge_example,
        linewidth=1.7,
        label="Ridge"
    )


    plt.plot(
        HOURS,
        extra_example,
        linewidth=1.7,
        label="Extra Trees"
    )


    plt.plot(
        HOURS,
        xgb_example,
        linewidth=2,
        label="XGBoost"
    )


    plt.xticks(
        HOURS
    )


    plt.xlabel(
        "Hour"
    )


    plt.ylabel(
        "Demand (MW)"
    )


    plt.title(
        "Example Next-Day Forecast: "
        +
        str(
            example_date.date()
        )
    )


    plt.legend()


    plt.grid(
        alpha=0.25
    )


    plt.tight_layout()


    FIG5 = (
        FIGURE_DIR
        / "05_example_2024_profile_forecast.png"
    )


    plt.savefig(
        FIG5,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()


# =====================================================================
# 41. FIGURE 6
# XGBOOST TOP 25 FEATURES
# =====================================================================

top_features = (
    feature_importance.head(
        25
    )
    .sort_values(
        "mean_importance",
        ascending=True
    )
)


plt.figure(
    figsize=(10, 9)
)


plt.barh(
    top_features[
        "feature"
    ],
    top_features[
        "mean_importance"
    ]
)


plt.xlabel(
    "Mean Feature Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top 25 XGBoost Predictors Across the 23 Forecast Horizons"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


FIG6 = (
    FIGURE_DIR
    / "06_xgboost_top25_features.png"
)


plt.savefig(
    FIG6,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 42. FIND BEST ALL-TEST MODEL
# =====================================================================

best_all_test_row = (

    all_test_leaderboard
    .iloc[0]
)


BEST_TEST_MODEL = (
    best_all_test_row[
        "model"
    ]
)


print("\n" + "=" * 110)

print(
    "BEST ALL-TEST MODEL"
)

print("=" * 110)


print(
    "Model:",
    BEST_TEST_MODEL
)


print(
    "MAE:",
    round(
        float(
            best_all_test_row[
                "MAE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "RMSE:",
    round(
        float(
            best_all_test_row[
                "RMSE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "MAPE:",
    round(
        float(
            best_all_test_row[
                "MAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "sMAPE:",
    round(
        float(
            best_all_test_row[
                "sMAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "MASE:",
    round(
        float(
            best_all_test_row[
                "MASE"
            ]
        ),
        4
    )
)


print(
    "MAE skill vs Naive D1:",
    round(
        float(
            best_all_test_row[
                "MAE_skill_vs_D1_percent"
            ]
        ),
        3
    ),
    "%"
)


# =====================================================================
# 43. SUMMARY JSON
# =====================================================================

summary = {

    "train_samples":
        int(
            len(
                train_df
            )
        ),

    "validation_samples":
        int(
            len(
                validation_df
            )
        ),

    "test_a_2024_samples":
        int(
            len(
                test_a_df
            )
        ),

    "test_b_2025_2026_samples":
        int(
            len(
                test_b_df
            )
        ),

    "feature_count":
        int(
            len(
                FEATURE_COLUMNS
            )
        ),

    "target_count":
        int(
            len(
                TARGET_COLUMNS
            )
        ),

    "best_ridge_alpha":
        float(
            BEST_RIDGE_ALPHA
        ),

    "best_extra_trees_configuration":
        str(
            BEST_EXTRA_NAME
        ),

    "best_xgboost_configuration":
        str(
            BEST_XGB_NAME
        ),

    "best_test_model":
        str(
            BEST_TEST_MODEL
        ),

    "best_test_MAE_MW":
        float(
            best_all_test_row[
                "MAE_MW"
            ]
        ),

    "best_test_RMSE_MW":
        float(
            best_all_test_row[
                "RMSE_MW"
            ]
        ),

    "best_test_MAPE_percent":
        float(
            best_all_test_row[
                "MAPE_percent"
            ]
        ),

    "best_test_sMAPE_percent":
        float(
            best_all_test_row[
                "sMAPE_percent"
            ]
        ),

    "best_test_MASE":
        float(
            best_all_test_row[
                "MASE"
            ]
        ),

    "best_test_MAE_skill_vs_D1_percent":
        float(
            best_all_test_row[
                "MAE_skill_vs_D1_percent"
            ]
        )
}


SUMMARY_FILE = (
    RESULT_DIR
    / "step3_summary.json"
)


with open(
    SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# =====================================================================
# 44. FINAL SANITY CHECKS
# =====================================================================

assert (
    RIDGE_TEST_PRED.shape
    ==
    Y_test.shape
)


assert (
    EXTRA_TEST_PRED.shape
    ==
    Y_test.shape
)


assert (
    XGB_TEST_PRED.shape
    ==
    Y_test.shape
)


assert np.isfinite(
    RIDGE_TEST_PRED
).all()


assert np.isfinite(
    EXTRA_TEST_PRED
).all()


assert np.isfinite(
    XGB_TEST_PRED
).all()


assert (
    len(
        test_a_df
    )
    >
    0
)


assert (
    len(
        test_b_df
    )
    >
    0
)


print(
    "\nAll Step 3 validation checks PASSED."
)


# =====================================================================
# 45. FINAL OUTPUT
# =====================================================================

print("\n")

print("=" * 110)

print(
    "STEP 3 COMPLETED SUCCESSFULLY"
)

print("=" * 110)


print(
    "\nTRAIN:",
    len(
        train_df
    )
)


print(
    "VALIDATION:",
    len(
        validation_df
    )
)


print(
    "TEST-A 2024:",
    len(
        test_a_df
    )
)


print(
    "TEST-B 2025-2026:",
    len(
        test_b_df
    )
)


print(
    "\nBest Ridge alpha:",
    BEST_RIDGE_ALPHA
)


print(
    "Best Extra Trees config:",
    BEST_EXTRA_NAME
)


print(
    "Best XGBoost config:",
    BEST_XGB_NAME
)


print("\n" + "-" * 110)

print(
    "BEST MODEL ON ALL TEST DATA"
)

print("-" * 110)


print(
    "Model:",
    BEST_TEST_MODEL
)


print(
    "MAE:",
    round(
        float(
            best_all_test_row[
                "MAE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "RMSE:",
    round(
        float(
            best_all_test_row[
                "RMSE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "MAPE:",
    round(
        float(
            best_all_test_row[
                "MAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "sMAPE:",
    round(
        float(
            best_all_test_row[
                "sMAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "MASE:",
    round(
        float(
            best_all_test_row[
                "MASE"
            ]
        ),
        4
    )
)


print(
    "MAE skill vs Naive D1:",
    round(
        float(
            best_all_test_row[
                "MAE_skill_vs_D1_percent"
            ]
        ),
        3
    ),
    "%"
)


print("\n" + "-" * 110)

print(
    "RESULT FILES"
)

print("-" * 110)


print(
    "\nFinal leaderboard:"
)

print(
    TEST_LEADERBOARD_FILE
)


print(
    "\nHorizon-wise metrics:"
)

print(
    HORIZON_METRICS_FILE
)


print(
    "\nPeak/profile utility metrics:"
)

print(
    UTILITY_METRICS_FILE
)


print(
    "\nXGBoost feature importance:"
)

print(
    FEATURE_IMPORTANCE_FILE
)


print(
    "\nPredictions:"
)

print(
    PREDICTION_FILE
)


print(
    "\nFigures:"
)

print(
    FIGURE_DIR
)


print(
    "\nModels:"
)

print(
    MODEL_DIR
)


print("\n" + "=" * 110)

print(
    "SEND THE FINAL STEP 3 SUMMARY AND FINAL TEST LEADERBOARD TO CHATGPT."
)

print(
    "DO NOT START LSTM OR TRANSFORMER YET."
)

print("=" * 110)
```
