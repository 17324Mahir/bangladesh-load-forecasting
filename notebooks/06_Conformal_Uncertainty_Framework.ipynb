```python
# =====================================================================
# STEP 5
# REGIME-AWARE CONFORMAL UNCERTAINTY FRAMEWORK
#
# Primary point forecaster:
# Ridge Regression
#
# Data protocol:
#
# 2015-2022       -> proper training
# Jan-Jun 2023    -> hyperparameter tuning
# Jul-Dec 2023    -> conformal calibration
# 2024-Mar 2026   -> untouched final evaluation
#
# Prediction intervals:
#
# 1. Global Absolute Conformal
# 2. Hour-specific Absolute Conformal
# 3. Hour-specific Normalized Conformal
# 4. Regime + Hour Normalized Conformal
# 5. Profile-Max Normalized Conformal
# 6. Regime-Aware Profile-Max Normalized Conformal
#
# Target coverage:
# 80%, 90%, 95%
#
# IMPORTANT:
# No 2024-2026 test residual is used for tuning/calibration.
#
# Run this entire block in ONE Google Colab cell.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import json
import math
import time
import warnings

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

import joblib


warnings.filterwarnings("ignore")

np.random.seed(42)

pd.set_option(
    "display.max_columns",
    300
)

pd.set_option(
    "display.max_rows",
    250
)

pd.set_option(
    "display.width",
    280
)


print("=" * 115)
print("STEP 5: REGIME-AWARE CONFORMAL UNCERTAINTY FRAMEWORK")
print("=" * 115)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount(
    "/content/drive"
)

print(
    "\nGoogle Drive mounted successfully."
)


# =====================================================================
# 2. PROJECT PATHS
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)


STEP3_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "ml_baselines"
)


MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "conformal_framework"
)


RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "conformal_framework"
)


FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "conformal_framework"
)


for directory in [

    MODEL_DIR,
    RESULT_DIR,
    FIGURE_DIR

]:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# =====================================================================
# 3. INPUT FILES
# =====================================================================

TRAIN_FILE = (
    FORECAST_DIR
    / "train_2015_2022.csv"
)


VALIDATION_FILE = (
    FORECAST_DIR
    / "validation_2023.csv"
)


TEST_FILE = (
    FORECAST_DIR
    / "test_2024_2026.csv"
)


STEP3_LEADERBOARD_FILE = (
    STEP3_RESULT_DIR
    / "final_test_leaderboard.csv"
)


for required_file in [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE

]:

    if not required_file.exists():

        raise FileNotFoundError(
            f"""
Required file not found:

{required_file}

Please make sure Steps 1-4 were completed.
"""
        )


# =====================================================================
# 4. LOAD DATA
# =====================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)


validation_2023 = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)


test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for dataframe in [

    train_df,
    validation_2023,
    test_df

]:

    dataframe[
        "target_date"
    ] = pd.to_datetime(

        dataframe[
            "target_date"
        ],

        errors="coerce"
    )


    dataframe.sort_values(
        "target_date",
        inplace=True
    )


    dataframe.reset_index(
        drop=True,
        inplace=True
    )


print("\nDatasets loaded successfully.")


print(
    "\nTRAIN:",
    len(train_df),
    train_df[
        "target_date"
    ].min(),
    "to",
    train_df[
        "target_date"
    ].max()
)


print(
    "\n2023 VALIDATION DATA:",
    len(validation_2023),
    validation_2023[
        "target_date"
    ].min(),
    "to",
    validation_2023[
        "target_date"
    ].max()
)


print(
    "\nFINAL TEST:",
    len(test_df),
    test_df[
        "target_date"
    ].min(),
    "to",
    test_df[
        "target_date"
    ].max()
)


# =====================================================================
# 5. DEFINE 23 TARGET HOURS
# =====================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 6. DEFINE EXACT SAME FEATURE SET USED IN STEP 3
# =====================================================================

CALENDAR_FEATURES = [

    "year",
    "month",
    "day",

    "dayofweek",
    "dayofyear",
    "weekofyear",

    "quarter",
    "is_weekend",

    "month_sin",
    "month_cos",

    "dow_sin",
    "dow_cos",

    "doy_sin",
    "doy_cos"
]


LAG1_FEATURES = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


LAG7_FEATURES = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


ROLL7_FEATURES = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


ROLL28_FEATURES = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


ROLL7STD_FEATURES = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


ROLL28STD_FEATURES = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


SUMMARY_FEATURES = [

    "lag1_daily_mean",
    "lag1_daily_max",
    "lag1_daily_min",
    "lag1_daily_range",
    "lag1_peak_hour",

    "lag7_daily_mean",
    "lag7_daily_max",
    "lag7_daily_min",
    "lag7_daily_range",
    "lag7_peak_hour",

    "mean_change_vs_last_week",
    "max_change_vs_last_week",
    "mean_ratio_vs_last_week"
]


FEATURE_COLUMNS = (

    CALENDAR_FEATURES

    +

    LAG1_FEATURES

    +

    LAG7_FEATURES

    +

    ROLL7_FEATURES

    +

    ROLL28_FEATURES

    +

    ROLL7STD_FEATURES

    +

    ROLL28STD_FEATURES

    +

    SUMMARY_FEATURES
)


print(
    "\nInput feature count:",
    len(FEATURE_COLUMNS)
)


print(
    "Forecast output count:",
    len(TARGET_COLUMNS)
)


# =====================================================================
# 7. COLUMN + LEAKAGE VALIDATION
# =====================================================================

for dataframe_name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION_2023",
        validation_2023
    ),

    (
        "TEST",
        test_df
    )

]:

    missing_features = [

        feature

        for feature in FEATURE_COLUMNS

        if feature
        not in
        dataframe.columns
    ]


    missing_targets = [

        target

        for target in TARGET_COLUMNS

        if target
        not in
        dataframe.columns
    ]


    if missing_features:

        raise ValueError(
            f"""
Missing feature columns in {dataframe_name}:

{missing_features}
"""
        )


    if missing_targets:

        raise ValueError(
            f"""
Missing target columns in {dataframe_name}:

{missing_targets}
"""
        )


if not set(
    FEATURE_COLUMNS
).isdisjoint(
    set(
        TARGET_COLUMNS
    )
):

    raise RuntimeError(
        "TARGET LEAKAGE FOUND."
    )


print(
    "\nFeature/target leakage check PASSED."
)


# =====================================================================
# 8. CHECK NaN / INFINITE VALUES
# =====================================================================

for dataframe_name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION",
        validation_2023
    ),

    (
        "TEST",
        test_df
    )

]:

    feature_values = (

        dataframe[
            FEATURE_COLUMNS
        ]

        .to_numpy(
            dtype=float
        )
    )


    target_values = (

        dataframe[
            TARGET_COLUMNS
        ]

        .to_numpy(
            dtype=float
        )
    )


    if not np.isfinite(
        feature_values
    ).all():

        raise ValueError(
            f"Non-finite feature found in {dataframe_name}."
        )


    if not np.isfinite(
        target_values
    ).all():

        raise ValueError(
            f"Non-finite target found in {dataframe_name}."
        )


print(
    "\nFinite-value validation PASSED."
)


# =====================================================================
# 9. SPLIT 2023 INTO TUNING AND CALIBRATION
# =====================================================================

TUNE_END = pd.Timestamp(
    "2023-06-30"
)


tune_df = (

    validation_2023.loc[

        validation_2023[
            "target_date"
        ]
        <=
        TUNE_END
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


calibration_df = (

    validation_2023.loc[

        validation_2023[
            "target_date"
        ]
        >
        TUNE_END
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 115)

print(
    "NESTED 2023 SPLIT"
)

print("=" * 115)


print(
    "TUNING:",
    len(tune_df),
    "samples"
)


print(
    tune_df[
        "target_date"
    ].min(),
    "to",
    tune_df[
        "target_date"
    ].max()
)


print(
    "\nCALIBRATION:",
    len(calibration_df),
    "samples"
)


print(
    calibration_df[
        "target_date"
    ].min(),
    "to",
    calibration_df[
        "target_date"
    ].max()
)


if (
    len(tune_df) < 50

    or

    len(calibration_df) < 50
):

    raise RuntimeError(
        "Tuning/calibration partition is unexpectedly too small."
    )


# =====================================================================
# 10. BASIC POINT FORECAST METRICS
# =====================================================================

def mae(
    y_true,
    y_pred
):

    return float(

        np.mean(

            np.abs(

                np.asarray(
                    y_true,
                    dtype=float
                )

                -

                np.asarray(
                    y_pred,
                    dtype=float
                )
            )
        )
    )


def rmse(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.sqrt(

            np.mean(

                (
                    y_true
                    -
                    y_pred
                )

                ** 2
            )
        )
    )


def mape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            y_true
        ),

        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    y_true
                    -
                    y_pred
                )

                /

                denominator
            )
        )
    )


def smape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = (

        np.abs(
            y_true
        )

        +

        np.abs(
            y_pred
        )
    )


    valid = (
        denominator > 0
    )


    return float(

        100

        *

        np.mean(

            2

            *

            np.abs(

                y_true[
                    valid
                ]

                -

                y_pred[
                    valid
                ]
            )

            /

            denominator[
                valid
            ]
        )
    )


# =====================================================================
# 11. PREPARE TRAIN/TUNE ARRAYS
# =====================================================================

X_train = (

    train_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


Y_train = (

    train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


X_tune = (

    tune_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


Y_tune = (

    tune_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


# =====================================================================
# 12. RIDGE ALPHA TUNING USING ONLY JAN-JUN 2023
# =====================================================================

RIDGE_ALPHAS = [

    0.01,
    0.1,
    1.0,
    10.0,
    100.0,
    1000.0
]


tuning_scaler = StandardScaler()


X_train_scaled = (

    tuning_scaler
    .fit_transform(
        X_train
    )
)


X_tune_scaled = (

    tuning_scaler
    .transform(
        X_tune
    )
)


ridge_tuning_rows = []


print("\n" + "=" * 115)

print(
    "RIDGE TUNING — JAN-JUN 2023 ONLY"
)

print("=" * 115)


for alpha_value in RIDGE_ALPHAS:

    model = Ridge(
        alpha=
            alpha_value
    )


    start_time = time.time()


    model.fit(
        X_train_scaled,
        Y_train
    )


    prediction = model.predict(
        X_tune_scaled
    )


    elapsed = (
        time.time()
        -
        start_time
    )


    result = {

        "alpha":
            alpha_value,

        "MAE_MW":
            mae(
                Y_tune,
                prediction
            ),

        "RMSE_MW":
            rmse(
                Y_tune,
                prediction
            ),

        "MAPE_percent":
            mape(
                Y_tune,
                prediction
            ),

        "sMAPE_percent":
            smape(
                Y_tune,
                prediction
            ),

        "training_seconds":
            elapsed
    }


    ridge_tuning_rows.append(
        result
    )


    print(

        f"alpha={alpha_value:<8} "

        f"MAE="
        f"{result['MAE_MW']:.3f} MW  "

        f"MAPE="
        f"{result['MAPE_percent']:.3f}%"
    )


ridge_tuning = pd.DataFrame(
    ridge_tuning_rows
)


ridge_tuning = (

    ridge_tuning

    .sort_values(
        "MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


BEST_ALPHA = float(

    ridge_tuning
    .iloc[0][
        "alpha"
    ]
)


print(
    "\nBest nested Ridge alpha:",
    BEST_ALPHA
)


display(
    ridge_tuning.round(4)
)


# =====================================================================
# 13. FINAL PROPER-TRAINING DATA
# =====================================================================

# After hyperparameter selection:
#
# Proper model training =
# 2015-2022 + Jan-Jun 2023
#
# Jul-Dec 2023 remains completely untouched for conformal calibration.


proper_train_df = pd.concat(

    [
        train_df,
        tune_df
    ],

    axis=0,

    ignore_index=True
)


proper_train_df = (

    proper_train_df

    .sort_values(
        "target_date"
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nProper training samples:",
    len(
        proper_train_df
    )
)


print(
    "Proper training period:",
    proper_train_df[
        "target_date"
    ].min(),
    "to",
    proper_train_df[
        "target_date"
    ].max()
)


# =====================================================================
# 14. FIT FINAL RIDGE FOR CONFORMAL FRAMEWORK
# =====================================================================

X_proper = (

    proper_train_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


Y_proper = (

    proper_train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


X_cal = (

    calibration_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


Y_cal = (

    calibration_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


X_test = (

    test_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


Y_test = (

    test_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float64
    )
)


final_scaler = StandardScaler()


X_proper_scaled = (

    final_scaler
    .fit_transform(
        X_proper
    )
)


X_cal_scaled = (

    final_scaler
    .transform(
        X_cal
    )
)


X_test_scaled = (

    final_scaler
    .transform(
        X_test
    )
)


conformal_ridge = Ridge(
    alpha=
        BEST_ALPHA
)


conformal_ridge.fit(
    X_proper_scaled,
    Y_proper
)


CAL_PRED = conformal_ridge.predict(
    X_cal_scaled
)


TEST_PRED = conformal_ridge.predict(
    X_test_scaled
)


print(
    "\nNested-split Ridge fitted successfully."
)


# =====================================================================
# 15. SAVE RIDGE + SCALER
# =====================================================================

SCALER_FILE = (

    MODEL_DIR
    /
    "conformal_ridge_scaler.joblib"
)


RIDGE_FILE = (

    MODEL_DIR
    /
    "conformal_ridge_model.joblib"
)


joblib.dump(
    final_scaler,
    SCALER_FILE
)


joblib.dump(
    conformal_ridge,
    RIDGE_FILE
)


# =====================================================================
# 16. POINT PERFORMANCE OF CONFORMAL RIDGE
# =====================================================================

def point_metrics(
    y_true,
    y_pred
):

    return {

        "MAE_MW":
            mae(
                y_true,
                y_pred
            ),

        "RMSE_MW":
            rmse(
                y_true,
                y_pred
            ),

        "MAPE_percent":
            mape(
                y_true,
                y_pred
            ),

        "sMAPE_percent":
            smape(
                y_true,
                y_pred
            )
    }


point_rows = []


test_dates = (
    test_df[
        "target_date"
    ]
)


POINT_SPLITS = {

    "TEST_A_2024":
        (
            test_dates.dt.year
            ==
            2024
        )
        .to_numpy(),

    "TEST_B_2025_2026":
        (
            test_dates.dt.year
            >=
            2025
        )
        .to_numpy(),

    "ALL_TEST_2024_2026":
        np.ones(
            len(
                test_df
            ),
            dtype=bool
        )
}


for split_name, split_mask in (
    POINT_SPLITS.items()
):

    metrics = point_metrics(

        Y_test[
            split_mask
        ],

        TEST_PRED[
            split_mask
        ]
    )


    point_rows.append(
        {

            "split":
                split_name,

            "samples":
                int(
                    split_mask.sum()
                ),

            **metrics
        }
    )


point_results = pd.DataFrame(
    point_rows
)


print("\n" + "=" * 115)

print(
    "NESTED-SPLIT RIDGE POINT FORECAST PERFORMANCE"
)

print("=" * 115)


display(
    point_results.round(4)
)


# =====================================================================
# 17. OPTIONAL STEP 3 RIDGE REFERENCE
# =====================================================================

STEP3_RIDGE_MAE = np.nan


if STEP3_LEADERBOARD_FILE.exists():

    step3_leaderboard = pd.read_csv(
        STEP3_LEADERBOARD_FILE
    )


    matching_rows = (

        step3_leaderboard.loc[

            (
                step3_leaderboard[
                    "split"
                ]
                ==
                "ALL_TEST_2024_2026"
            )

            &

            (
                step3_leaderboard[
                    "model"
                ]
                ==
                "Ridge"
            )
        ]
    )


    if len(
        matching_rows
    ) > 0:

        STEP3_RIDGE_MAE = float(

            matching_rows
            .iloc[0][
                "MAE_MW"
            ]
        )


        print(
            "\nStep 3 fully-refitted Ridge reference MAE:",
            round(
                STEP3_RIDGE_MAE,
                3
            ),
            "MW"
        )


        print(
            "NOTE: this is a point-forecast reference only."
        )


        print(
            "The conformal model intentionally uses less pre-test data "
            "because Jul-Dec 2023 is reserved for calibration."
        )


# =====================================================================
# 18. CALIBRATION RESIDUALS
# =====================================================================

CAL_RESIDUAL = (

    Y_cal

    -

    CAL_PRED
)


CAL_ABS_RESIDUAL = (

    np.abs(
        CAL_RESIDUAL
    )
)


# =====================================================================
# 19. NORMALIZED NONCONFORMITY SCORE
# =====================================================================

# Scale-normalized residual:
#
# |y - y_hat| / max(|y_hat|, 1000 MW)
#
# This allows interval width to adapt to the predicted load magnitude.


SCALE_FLOOR_MW = 1000.0


CAL_SCALE = np.maximum(

    np.abs(
        CAL_PRED
    ),

    SCALE_FLOOR_MW
)


CAL_NORMALIZED_SCORE = (

    CAL_ABS_RESIDUAL

    /

    CAL_SCALE
)


TEST_SCALE = np.maximum(

    np.abs(
        TEST_PRED
    ),

    SCALE_FLOOR_MW
)


# =====================================================================
# 20. CALIBRATION RESIDUAL DIAGNOSTICS BY HOUR
# =====================================================================

calibration_hour_rows = []


for hour_index, hour in enumerate(
    HOURS
):

    scores = (

        CAL_ABS_RESIDUAL[
            :,
            hour_index
        ]
    )


    calibration_hour_rows.append(
        {

            "hour":
                hour,

            "mean_abs_error_mw":
                float(
                    np.mean(
                        scores
                    )
                ),

            "median_abs_error_mw":
                float(
                    np.median(
                        scores
                    )
                ),

            "q80_abs_error_mw":
                float(
                    np.quantile(
                        scores,
                        0.80
                    )
                ),

            "q90_abs_error_mw":
                float(
                    np.quantile(
                        scores,
                        0.90
                    )
                ),

            "q95_abs_error_mw":
                float(
                    np.quantile(
                        scores,
                        0.95
                    )
                )
        }
    )


calibration_by_hour = pd.DataFrame(
    calibration_hour_rows
)


print(
    "\nCalibration absolute residuals by hour:"
)


display(
    calibration_by_hour.round(2)
)


# =====================================================================
# 21. FORECASTABLE LOAD REGIME DEFINITION
# =====================================================================

# Regime is based ONLY on previous-day mean load:
#
# lag1_daily_mean
#
# This is already known before target-day forecasting.
#
# Thresholds come ONLY from proper training data.


REGIME_FEATURE = (
    "lag1_daily_mean"
)


LOW_THRESHOLD = float(

    proper_train_df[
        REGIME_FEATURE
    ]
    .quantile(
        1 / 3
    )
)


HIGH_THRESHOLD = float(

    proper_train_df[
        REGIME_FEATURE
    ]
    .quantile(
        2 / 3
    )
)


print("\n" + "=" * 115)

print(
    "FORECASTABLE LOAD-REGIME THRESHOLDS"
)

print("=" * 115)


print(
    "LOW <",
    round(
        LOW_THRESHOLD,
        2
    ),
    "MW"
)


print(
    "NORMAL =",
    round(
        LOW_THRESHOLD,
        2
    ),
    "to",
    round(
        HIGH_THRESHOLD,
        2
    ),
    "MW"
)


print(
    "HIGH >",
    round(
        HIGH_THRESHOLD,
        2
    ),
    "MW"
)


# =====================================================================
# 22. REGIME CLASSIFICATION FUNCTION
# =====================================================================

def assign_regimes(
    values
):

    values = np.asarray(
        values,
        dtype=float
    )


    result = np.full(

        len(
            values
        ),

        "NORMAL",

        dtype=object
    )


    result[
        values
        <
        LOW_THRESHOLD
    ] = "LOW"


    result[
        values
        >
        HIGH_THRESHOLD
    ] = "HIGH"


    return result


CAL_REGIME = assign_regimes(

    calibration_df[
        REGIME_FEATURE
    ]
    .to_numpy(
        dtype=float
    )
)


TEST_REGIME = assign_regimes(

    test_df[
        REGIME_FEATURE
    ]
    .to_numpy(
        dtype=float
    )
)


REGIMES = [

    "LOW",
    "NORMAL",
    "HIGH"
]


# =====================================================================
# 23. REGIME COUNTS
# =====================================================================

regime_count_rows = []


for dataset_name, labels in [

    (
        "CALIBRATION",
        CAL_REGIME
    ),

    (
        "TEST",
        TEST_REGIME
    )

]:

    for regime in REGIMES:

        regime_count_rows.append(
            {

                "dataset":
                    dataset_name,

                "regime":
                    regime,

                "count":
                    int(
                        np.sum(
                            labels
                            ==
                            regime
                        )
                    )
            }
        )


regime_counts = pd.DataFrame(
    regime_count_rows
)


print(
    "\nRegime sample counts:"
)


display(
    regime_counts
)


# =====================================================================
# 24. CALIBRATION ERROR BY REGIME
# =====================================================================

calibration_regime_rows = []


for regime in REGIMES:

    mask = (
        CAL_REGIME
        ==
        regime
    )


    residuals = (

        CAL_ABS_RESIDUAL[
            mask
        ]
    )


    calibration_regime_rows.append(
        {

            "regime":
                regime,

            "days":
                int(
                    mask.sum()
                ),

            "MAE_MW":
                float(
                    np.mean(
                        residuals
                    )
                ),

            "median_absolute_error_MW":
                float(
                    np.median(
                        residuals
                    )
                ),

            "q90_absolute_error_MW":
                float(
                    np.quantile(
                        residuals,
                        0.90
                    )
                )
        }
    )


calibration_by_regime = pd.DataFrame(
    calibration_regime_rows
)


print(
    "\nCalibration error by forecastable regime:"
)


display(
    calibration_by_regime.round(3)
)


# =====================================================================
# 25. FINITE-SAMPLE SPLIT-CONFORMAL QUANTILE
# =====================================================================

def conformal_quantile(
    scores,
    alpha
):

    scores = np.asarray(
        scores,
        dtype=float
    )


    scores = scores[
        np.isfinite(
            scores
        )
    ]


    n = len(
        scores
    )


    if n == 0:

        return np.nan


    # Finite-sample conformal order statistic:
    #
    # ceil((n + 1) * (1 - alpha))


    rank = int(

        np.ceil(

            (
                n
                +
                1
            )

            *
            (
                1
                -
                alpha
            )
        )
    )


    rank = min(
        max(
            rank,
            1
        ),
        n
    )


    sorted_scores = np.sort(
        scores
    )


    return float(

        sorted_scores[
            rank - 1
        ]
    )


# =====================================================================
# 26. NOMINAL COVERAGES
# =====================================================================

COVERAGE_LEVELS = [

    0.80,
    0.90,
    0.95
]


MIN_REGIME_CALIBRATION_DAYS = 25


# =====================================================================
# 27. BUILD ALL CONFORMAL QUANTILES
# =====================================================================

quantile_store = {}

quantile_records = []


for coverage in COVERAGE_LEVELS:

    alpha = (
        1
        -
        coverage
    )


    coverage_key = int(
        round(
            coverage
            *
            100
        )
    )


    quantile_store[
        coverage_key
    ] = {}


    # -------------------------------------------------------------
    # A. GLOBAL ABSOLUTE
    # -------------------------------------------------------------

    q_global_abs = conformal_quantile(

        CAL_ABS_RESIDUAL.flatten(),

        alpha
    )


    quantile_store[
        coverage_key
    ][
        "GlobalAbsolute"
    ] = (
        q_global_abs
    )


    quantile_records.append(
        {

            "coverage_percent":
                coverage_key,

            "method":
                "GlobalAbsolute",

            "regime":
                "ALL",

            "hour":
                "ALL",

            "calibration_n":
                CAL_ABS_RESIDUAL.size,

            "quantile":
                q_global_abs,

            "fallback_used":
                False
        }
    )


    # -------------------------------------------------------------
    # B. HOUR-SPECIFIC ABSOLUTE
    # -------------------------------------------------------------

    q_hour_abs = np.zeros(
        23,
        dtype=float
    )


    for h_index, hour in enumerate(
        HOURS
    ):

        q_value = conformal_quantile(

            CAL_ABS_RESIDUAL[
                :,
                h_index
            ],

            alpha
        )


        q_hour_abs[
            h_index
        ] = (
            q_value
        )


        quantile_records.append(
            {

                "coverage_percent":
                    coverage_key,

                "method":
                    "HourAbsolute",

                "regime":
                    "ALL",

                "hour":
                    hour,

                "calibration_n":
                    len(
                        calibration_df
                    ),

                "quantile":
                    q_value,

                "fallback_used":
                    False
            }
        )


    quantile_store[
        coverage_key
    ][
        "HourAbsolute"
    ] = (
        q_hour_abs
    )


    # -------------------------------------------------------------
    # C. HOUR-SPECIFIC NORMALIZED
    # -------------------------------------------------------------

    q_hour_norm = np.zeros(
        23,
        dtype=float
    )


    for h_index, hour in enumerate(
        HOURS
    ):

        q_value = conformal_quantile(

            CAL_NORMALIZED_SCORE[
                :,
                h_index
            ],

            alpha
        )


        q_hour_norm[
            h_index
        ] = (
            q_value
        )


        quantile_records.append(
            {

                "coverage_percent":
                    coverage_key,

                "method":
                    "HourNormalized",

                "regime":
                    "ALL",

                "hour":
                    hour,

                "calibration_n":
                    len(
                        calibration_df
                    ),

                "quantile":
                    q_value,

                "fallback_used":
                    False
            }
        )


    quantile_store[
        coverage_key
    ][
        "HourNormalized"
    ] = (
        q_hour_norm
    )


    # -------------------------------------------------------------
    # D. REGIME + HOUR NORMALIZED
    #
    # Fall back to hour-normalized if a regime group is too small.
    # -------------------------------------------------------------

    regime_hour_q = {}


    for regime in REGIMES:

        regime_mask = (

            CAL_REGIME
            ==
            regime
        )


        regime_n = int(
            regime_mask.sum()
        )


        q_vector = np.zeros(
            23,
            dtype=float
        )


        for h_index, hour in enumerate(
            HOURS
        ):

            use_fallback = (

                regime_n
                <
                MIN_REGIME_CALIBRATION_DAYS
            )


            if use_fallback:

                q_value = (

                    q_hour_norm[
                        h_index
                    ]
                )


            else:

                q_value = conformal_quantile(

                    CAL_NORMALIZED_SCORE[
                        regime_mask,
                        h_index
                    ],

                    alpha
                )


            q_vector[
                h_index
            ] = (
                q_value
            )


            quantile_records.append(
                {

                    "coverage_percent":
                        coverage_key,

                    "method":
                        "RegimeHourNormalized",

                    "regime":
                        regime,

                    "hour":
                        hour,

                    "calibration_n":
                        regime_n,

                    "quantile":
                        q_value,

                    "fallback_used":
                        use_fallback
                }
            )


        regime_hour_q[
            regime
        ] = (
            q_vector
        )


    quantile_store[
        coverage_key
    ][
        "RegimeHourNormalized"
    ] = (
        regime_hour_q
    )


    # -------------------------------------------------------------
    # E. PROFILE MAX NORMALIZED
    #
    # score(day) =
    # maximum normalized residual across all 23 horizons.
    #
    # This targets simultaneous whole-profile coverage.
    # -------------------------------------------------------------

    profile_max_scores = (

        np.max(

            CAL_NORMALIZED_SCORE,

            axis=1
        )
    )


    q_profile = conformal_quantile(

        profile_max_scores,

        alpha
    )


    quantile_store[
        coverage_key
    ][
        "ProfileMaxNormalized"
    ] = (
        q_profile
    )


    quantile_records.append(
        {

            "coverage_percent":
                coverage_key,

            "method":
                "ProfileMaxNormalized",

            "regime":
                "ALL",

            "hour":
                "PROFILE",

            "calibration_n":
                len(
                    profile_max_scores
                ),

            "quantile":
                q_profile,

            "fallback_used":
                False
        }
    )


    # -------------------------------------------------------------
    # F. REGIME-AWARE PROFILE MAX NORMALIZED
    # -------------------------------------------------------------

    regime_profile_q = {}


    for regime in REGIMES:

        regime_mask = (

            CAL_REGIME
            ==
            regime
        )


        regime_n = int(
            regime_mask.sum()
        )


        if (
            regime_n
            <
            MIN_REGIME_CALIBRATION_DAYS
        ):

            q_value = (
                q_profile
            )

            use_fallback = True


        else:

            regime_profile_scores = (

                np.max(

                    CAL_NORMALIZED_SCORE[
                        regime_mask
                    ],

                    axis=1
                )
            )


            q_value = conformal_quantile(

                regime_profile_scores,

                alpha
            )


            use_fallback = False


        regime_profile_q[
            regime
        ] = (
            q_value
        )


        quantile_records.append(
            {

                "coverage_percent":
                    coverage_key,

                "method":
                    "RegimeProfileMaxNormalized",

                "regime":
                    regime,

                "hour":
                    "PROFILE",

                "calibration_n":
                    regime_n,

                "quantile":
                    q_value,

                "fallback_used":
                    use_fallback
            }
        )


    quantile_store[
        coverage_key
    ][
        "RegimeProfileMaxNormalized"
    ] = (
        regime_profile_q
    )


conformal_quantile_table = pd.DataFrame(
    quantile_records
)


# =====================================================================
# 28. INTERVAL CONSTRUCTION
# =====================================================================

def build_intervals(
    method,
    coverage_percent,
    predictions,
    prediction_scale,
    regimes
):

    stored = (

        quantile_store[
            coverage_percent
        ]
    )


    n_samples = (
        predictions.shape[0]
    )


    if method == "GlobalAbsolute":

        half_width = np.full_like(

            predictions,

            stored[
                "GlobalAbsolute"
            ],

            dtype=float
        )


    elif method == "HourAbsolute":

        half_width = np.tile(

            stored[
                "HourAbsolute"
            ][
                None,
                :
            ],

            (
                n_samples,
                1
            )
        )


    elif method == "HourNormalized":

        q_values = np.tile(

            stored[
                "HourNormalized"
            ][
                None,
                :
            ],

            (
                n_samples,
                1
            )
        )


        half_width = (

            q_values

            *
            prediction_scale
        )


    elif method == "RegimeHourNormalized":

        half_width = np.zeros_like(

            predictions,

            dtype=float
        )


        q_store = (

            stored[
                "RegimeHourNormalized"
            ]
        )


        for i in range(
            n_samples
        ):

            q_vector = (

                q_store[
                    regimes[
                        i
                    ]
                ]
            )


            half_width[
                i,
                :
            ] = (

                q_vector

                *
                prediction_scale[
                    i,
                    :
                ]
            )


    elif method == "ProfileMaxNormalized":

        half_width = (

            stored[
                "ProfileMaxNormalized"
            ]

            *
            prediction_scale
        )


    elif method == "RegimeProfileMaxNormalized":

        half_width = np.zeros_like(

            predictions,

            dtype=float
        )


        q_store = (

            stored[
                "RegimeProfileMaxNormalized"
            ]
        )


        for i in range(
            n_samples
        ):

            q_value = (

                q_store[
                    regimes[
                        i
                    ]
                ]
            )


            half_width[
                i,
                :
            ] = (

                q_value

                *
                prediction_scale[
                    i,
                    :
                ]
            )


    else:

        raise ValueError(
            f"Unknown method: {method}"
        )


    lower = (

        predictions

        -
        half_width
    )


    upper = (

        predictions

        +
        half_width
    )


    # Demand cannot be negative.
    #
    # Clipping only affects impossible negative lower bounds.

    lower = np.maximum(
        lower,
        0.0
    )


    return (
        lower,
        upper,
        half_width
    )


# =====================================================================
# 29. INTERVAL METRICS
# =====================================================================

def winkler_score(
    y_true,
    lower,
    upper,
    alpha
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    lower = np.asarray(
        lower,
        dtype=float
    )


    upper = np.asarray(
        upper,
        dtype=float
    )


    width = (
        upper
        -
        lower
    )


    score = (
        width.copy()
    )


    below = (
        y_true
        <
        lower
    )


    above = (
        y_true
        >
        upper
    )


    score[
        below
    ] += (

        2
        /
        alpha

        *

        (
            lower[
                below
            ]

            -
            y_true[
                below
            ]
        )
    )


    score[
        above
    ] += (

        2
        /
        alpha

        *

        (
            y_true[
                above
            ]

            -
            upper[
                above
            ]
        )
    )


    return float(
        np.mean(
            score
        )
    )


def interval_metrics(
    y_true,
    lower,
    upper,
    nominal_coverage,
    coverage_type
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    covered = (

        (
            y_true
            >=
            lower
        )

        &

        (
            y_true
            <=
            upper
        )
    )


    marginal_coverage = (

        100

        *
        float(
            np.mean(
                covered
            )
        )
    )


    profile_covered = (

        covered
        .all(
            axis=1
        )
    )


    profile_coverage = (

        100

        *
        float(
            np.mean(
                profile_covered
            )
        )
    )


    width = (

        upper

        -
        lower
    )


    mpiw = float(

        np.mean(
            width
        )
    )


    mean_actual = float(

        np.mean(
            np.abs(
                y_true
            )
        )
    )


    normalized_width = (

        100

        *
        mpiw

        /
        max(
            mean_actual,
            1e-6
        )
    )


    lower_miss = (

        100

        *
        float(
            np.mean(
                y_true
                <
                lower
            )
        )
    )


    upper_miss = (

        100

        *
        float(
            np.mean(
                y_true
                >
                upper
            )
        )
    )


    alpha = (
        1
        -
        nominal_coverage
    )


    interval_score = winkler_score(

        y_true,
        lower,
        upper,
        alpha
    )


    if (
        coverage_type
        ==
        "MARGINAL"
    ):

        primary_coverage = (
            marginal_coverage
        )


    else:

        primary_coverage = (
            profile_coverage
        )


    coverage_gap = (

        primary_coverage

        -
        (
            nominal_coverage
            *
            100
        )
    )


    return {

        "Marginal_Coverage_percent":
            marginal_coverage,

        "Profile_Coverage_percent":
            profile_coverage,

        "Primary_Coverage_percent":
            primary_coverage,

        "Coverage_Gap_percentage_points":
            coverage_gap,

        "MPIW_MW":
            mpiw,

        "Normalized_MPIW_percent":
            normalized_width,

        "Winkler_Score":
            interval_score,

        "Lower_Miss_percent":
            lower_miss,

        "Upper_Miss_percent":
            upper_miss
    }


# =====================================================================
# 30. CONFORMAL METHODS
# =====================================================================

METHODS = {

    "GlobalAbsolute":
        "MARGINAL",

    "HourAbsolute":
        "MARGINAL",

    "HourNormalized":
        "MARGINAL",

    "RegimeHourNormalized":
        "MARGINAL",

    "ProfileMaxNormalized":
        "SIMULTANEOUS",

    "RegimeProfileMaxNormalized":
        "SIMULTANEOUS"
}


# =====================================================================
# 31. FINAL INTERVAL EVALUATION
# =====================================================================

evaluation_rows = []


for split_name, split_mask in (
    POINT_SPLITS.items()
):

    split_true = (

        Y_test[
            split_mask
        ]
    )


    split_pred = (

        TEST_PRED[
            split_mask
        ]
    )


    split_scale = (

        TEST_SCALE[
            split_mask
        ]
    )


    split_regime = (

        TEST_REGIME[
            split_mask
        ]
    )


    for coverage in COVERAGE_LEVELS:

        coverage_percent = int(

            round(
                coverage
                *
                100
            )
        )


        for method, coverage_type in (
            METHODS.items()
        ):

            lower, upper, _ = (
                build_intervals(

                    method=
                        method,

                    coverage_percent=
                        coverage_percent,

                    predictions=
                        split_pred,

                    prediction_scale=
                        split_scale,

                    regimes=
                        split_regime
                )
            )


            metrics = interval_metrics(

                split_true,

                lower,

                upper,

                nominal_coverage=
                    coverage,

                coverage_type=
                    coverage_type
            )


            evaluation_rows.append(
                {

                    "split":
                        split_name,

                    "method":
                        method,

                    "coverage_type":
                        coverage_type,

                    "nominal_coverage_percent":
                        coverage_percent,

                    "days":
                        int(
                            split_mask.sum()
                        ),

                    **metrics
                }
            )


interval_evaluation = pd.DataFrame(
    evaluation_rows
)


print("\n" + "=" * 115)

print(
    "CONFORMAL INTERVAL EVALUATION"
)

print("=" * 115)


all_test_interval_results = (

    interval_evaluation.loc[

        interval_evaluation[
            "split"
        ]
        ==
        "ALL_TEST_2024_2026"
    ]

    .copy()

    .sort_values(
        [
            "nominal_coverage_percent",
            "coverage_type",
            "method"
        ]
    )
)


display(
    all_test_interval_results.round(4)
)


# =====================================================================
# 32. HOUR-WISE COVERAGE
# =====================================================================

hourly_interval_rows = []


for split_name, split_mask in (
    POINT_SPLITS.items()
):

    true_split = (

        Y_test[
            split_mask
        ]
    )


    pred_split = (

        TEST_PRED[
            split_mask
        ]
    )


    scale_split = (

        TEST_SCALE[
            split_mask
        ]
    )


    regime_split = (

        TEST_REGIME[
            split_mask
        ]
    )


    for coverage in COVERAGE_LEVELS:

        coverage_percent = int(

            round(
                coverage
                *
                100
            )
        )


        for method, coverage_type in (
            METHODS.items()
        ):

            lower, upper, _ = (
                build_intervals(

                    method,
                    coverage_percent,
                    pred_split,
                    scale_split,
                    regime_split
                )
            )


            for h_index, hour in enumerate(
                HOURS
            ):

                y_h = (

                    true_split[
                        :,
                        h_index
                    ]
                )


                low_h = (

                    lower[
                        :,
                        h_index
                    ]
                )


                high_h = (

                    upper[
                        :,
                        h_index
                    ]
                )


                coverage_h = (

                    100

                    *
                    np.mean(

                        (
                            y_h
                            >=
                            low_h
                        )

                        &

                        (
                            y_h
                            <=
                            high_h
                        )
                    )
                )


                width_h = (

                    np.mean(

                        high_h

                        -
                        low_h
                    )
                )


                hourly_interval_rows.append(
                    {

                        "split":
                            split_name,

                        "method":
                            method,

                        "coverage_type":
                            coverage_type,

                        "nominal_coverage_percent":
                            coverage_percent,

                        "hour":
                            hour,

                        "empirical_coverage_percent":
                            coverage_h,

                        "MPIW_MW":
                            width_h
                    }
                )


hourly_interval_results = pd.DataFrame(
    hourly_interval_rows
)


# =====================================================================
# 33. REGIME-WISE TEST INTERVAL PERFORMANCE
# =====================================================================

regime_interval_rows = []


for split_name, split_mask in (
    POINT_SPLITS.items()
):

    original_positions = np.where(
        split_mask
    )[0]


    for regime in REGIMES:

        regime_positions = (

            original_positions[

                TEST_REGIME[
                    original_positions
                ]
                ==
                regime
            ]
        )


        if len(
            regime_positions
        ) == 0:

            continue


        y_regime = (

            Y_test[
                regime_positions
            ]
        )


        pred_regime = (

            TEST_PRED[
                regime_positions
            ]
        )


        scale_regime = (

            TEST_SCALE[
                regime_positions
            ]
        )


        regime_labels = (

            TEST_REGIME[
                regime_positions
            ]
        )


        for coverage in COVERAGE_LEVELS:

            coverage_percent = int(

                round(
                    coverage
                    *
                    100
                )
            )


            for method, coverage_type in (
                METHODS.items()
            ):

                lower, upper, _ = (
                    build_intervals(

                        method,
                        coverage_percent,
                        pred_regime,
                        scale_regime,
                        regime_labels
                    )
                )


                metrics = interval_metrics(

                    y_regime,

                    lower,

                    upper,

                    nominal_coverage=
                        coverage,

                    coverage_type=
                        coverage_type
                )


                regime_interval_rows.append(
                    {

                        "split":
                            split_name,

                        "regime":
                            regime,

                        "days":
                            len(
                                regime_positions
                            ),

                        "method":
                            method,

                        "coverage_type":
                            coverage_type,

                        "nominal_coverage_percent":
                            coverage_percent,

                        **metrics
                    }
                )


regime_interval_results = pd.DataFrame(
    regime_interval_rows
)


# =====================================================================
# 34. PRIMARY 90% RESULTS
# =====================================================================

primary_90 = (

    interval_evaluation.loc[

        (
            interval_evaluation[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            interval_evaluation[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .copy()
)


print("\n" + "=" * 115)

print(
    "PRIMARY 90% INTERVAL COMPARISON — ALL TEST DATA"
)

print("=" * 115)


display(
    primary_90[
        [
            "method",
            "coverage_type",
            "Primary_Coverage_percent",
            "Coverage_Gap_percentage_points",
            "Marginal_Coverage_percent",
            "Profile_Coverage_percent",
            "MPIW_MW",
            "Normalized_MPIW_percent",
            "Winkler_Score"
        ]
    ]
    .round(4)
)


# =====================================================================
# 35. PROPOSED METHODS
# =====================================================================

PROPOSED_MARGINAL_METHOD = (
    "RegimeHourNormalized"
)


PROPOSED_PROFILE_METHOD = (
    "RegimeProfileMaxNormalized"
)


proposed_marginal_90 = (

    primary_90.loc[
        primary_90[
            "method"
        ]
        ==
        PROPOSED_MARGINAL_METHOD
    ]
    .iloc[0]
)


proposed_profile_90 = (

    primary_90.loc[
        primary_90[
            "method"
        ]
        ==
        PROPOSED_PROFILE_METHOD
    ]
    .iloc[0]
)


# =====================================================================
# 36. SAVE FINAL 90% PROPOSED MARGINAL INTERVALS
# =====================================================================

PROPOSED_90_LOWER, \
PROPOSED_90_UPPER, \
PROPOSED_90_HALF_WIDTH = (

    build_intervals(

        method=
            PROPOSED_MARGINAL_METHOD,

        coverage_percent=
            90,

        predictions=
            TEST_PRED,

        prediction_scale=
            TEST_SCALE,

        regimes=
            TEST_REGIME
    )
)


proposed_interval_output = pd.DataFrame(
    {

        "target_date":
            test_df[
                "target_date"
            ],

        "regime":
            TEST_REGIME
    }
)


for h_index, hour in enumerate(
    HOURS
):

    proposed_interval_output[
        f"actual_h{hour:02d}"
    ] = (

        Y_test[
            :,
            h_index
        ]
    )


    proposed_interval_output[
        f"point_pred_h{hour:02d}"
    ] = (

        TEST_PRED[
            :,
            h_index
        ]
    )


    proposed_interval_output[
        f"lower90_h{hour:02d}"
    ] = (

        PROPOSED_90_LOWER[
            :,
            h_index
        ]
    )


    proposed_interval_output[
        f"upper90_h{hour:02d}"
    ] = (

        PROPOSED_90_UPPER[
            :,
            h_index
        ]
    )


# =====================================================================
# 37. SAVE FINAL 90% SIMULTANEOUS PROFILE BAND
# =====================================================================

PROFILE90_LOWER, \
PROFILE90_UPPER, \
PROFILE90_HALF_WIDTH = (

    build_intervals(

        method=
            PROPOSED_PROFILE_METHOD,

        coverage_percent=
            90,

        predictions=
            TEST_PRED,

        prediction_scale=
            TEST_SCALE,

        regimes=
            TEST_REGIME
    )
)


for h_index, hour in enumerate(
    HOURS
):

    proposed_interval_output[
        f"profile_lower90_h{hour:02d}"
    ] = (

        PROFILE90_LOWER[
            :,
            h_index
        ]
    )


    proposed_interval_output[
        f"profile_upper90_h{hour:02d}"
    ] = (

        PROFILE90_UPPER[
            :,
            h_index
        ]
    )


# =====================================================================
# 38. SAVE ALL RESULT TABLES
# =====================================================================

RIDGE_TUNING_FILE = (

    RESULT_DIR
    /
    "nested_ridge_tuning.csv"
)


POINT_RESULT_FILE = (

    RESULT_DIR
    /
    "conformal_ridge_point_performance.csv"
)


CAL_HOUR_FILE = (

    RESULT_DIR
    /
    "calibration_residuals_by_hour.csv"
)


CAL_REGIME_FILE = (

    RESULT_DIR
    /
    "calibration_residuals_by_regime.csv"
)


REGIME_COUNT_FILE = (

    RESULT_DIR
    /
    "regime_counts.csv"
)


QUANTILE_FILE = (

    RESULT_DIR
    /
    "conformal_quantiles.csv"
)


INTERVAL_RESULT_FILE = (

    RESULT_DIR
    /
    "conformal_interval_evaluation.csv"
)


HOURLY_RESULT_FILE = (

    RESULT_DIR
    /
    "conformal_coverage_by_hour.csv"
)


REGIME_RESULT_FILE = (

    RESULT_DIR
    /
    "conformal_coverage_by_regime.csv"
)


PROPOSED_INTERVAL_FILE = (

    RESULT_DIR
    /
    "proposed_90percent_test_intervals.csv"
)


ridge_tuning.to_csv(
    RIDGE_TUNING_FILE,
    index=False
)


point_results.to_csv(
    POINT_RESULT_FILE,
    index=False
)


calibration_by_hour.to_csv(
    CAL_HOUR_FILE,
    index=False
)


calibration_by_regime.to_csv(
    CAL_REGIME_FILE,
    index=False
)


regime_counts.to_csv(
    REGIME_COUNT_FILE,
    index=False
)


conformal_quantile_table.to_csv(
    QUANTILE_FILE,
    index=False
)


interval_evaluation.to_csv(
    INTERVAL_RESULT_FILE,
    index=False
)


hourly_interval_results.to_csv(
    HOURLY_RESULT_FILE,
    index=False
)


regime_interval_results.to_csv(
    REGIME_RESULT_FILE,
    index=False
)


proposed_interval_output.to_csv(
    PROPOSED_INTERVAL_FILE,
    index=False
)


# =====================================================================
# 39. FIGURE 1
# RIDGE TUNING
# =====================================================================

ridge_plot = (

    ridge_tuning
    .sort_values(
        "alpha"
    )
)


plt.figure(
    figsize=(9, 6)
)


plt.semilogx(

    ridge_plot[
        "alpha"
    ],

    ridge_plot[
        "MAE_MW"
    ],

    marker="o"
)


plt.xlabel(
    "Ridge Alpha"
)


plt.ylabel(
    "Jan-Jun 2023 Tuning MAE (MW)"
)


plt.title(
    "Nested Ridge Hyperparameter Selection"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG1 = (

    FIGURE_DIR
    /
    "01_nested_ridge_tuning.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 40. FIGURE 2
# CALIBRATION RESIDUAL BY HOUR
# =====================================================================

plt.figure(
    figsize=(12, 6)
)


plt.plot(

    calibration_by_hour[
        "hour"
    ],

    calibration_by_hour[
        "mean_abs_error_mw"
    ],

    marker="o",

    label="Mean absolute residual"
)


plt.plot(

    calibration_by_hour[
        "hour"
    ],

    calibration_by_hour[
        "q90_abs_error_mw"
    ],

    marker="s",

    label="90th percentile residual"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Forecast Target Hour"
)


plt.ylabel(
    "Absolute Calibration Residual (MW)"
)


plt.title(
    "Horizon-Dependent Forecast Uncertainty"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG2 = (

    FIGURE_DIR
    /
    "02_calibration_residual_by_hour.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 41. FIGURE 3
# MARGINAL COVERAGE VS NOMINAL
# =====================================================================

marginal_plot = (

    interval_evaluation.loc[

        (
            interval_evaluation[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            interval_evaluation[
                "coverage_type"
            ]
            ==
            "MARGINAL"
        )
    ]
)


plt.figure(
    figsize=(11, 7)
)


for method in marginal_plot[
    "method"
].unique():

    temp = (

        marginal_plot.loc[
            marginal_plot[
                "method"
            ]
            ==
            method
        ]

        .sort_values(
            "nominal_coverage_percent"
        )
    )


    plt.plot(

        temp[
            "nominal_coverage_percent"
        ],

        temp[
            "Marginal_Coverage_percent"
        ],

        marker="o",

        linewidth=2,

        label=method
    )


plt.plot(

    [
        80,
        95
    ],

    [
        80,
        95
    ],

    linestyle="--",

    label="Ideal calibration"
)


plt.xlabel(
    "Nominal Coverage (%)"
)


plt.ylabel(
    "Empirical Marginal Coverage (%)"
)


plt.title(
    "Marginal Prediction-Interval Calibration"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG3 = (

    FIGURE_DIR
    /
    "03_marginal_coverage_calibration.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 42. FIGURE 4
# SIMULTANEOUS PROFILE COVERAGE
# =====================================================================

profile_plot = (

    interval_evaluation.loc[

        (
            interval_evaluation[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            interval_evaluation[
                "coverage_type"
            ]
            ==
            "SIMULTANEOUS"
        )
    ]
)


plt.figure(
    figsize=(10, 7)
)


for method in profile_plot[
    "method"
].unique():

    temp = (

        profile_plot.loc[
            profile_plot[
                "method"
            ]
            ==
            method
        ]

        .sort_values(
            "nominal_coverage_percent"
        )
    )


    plt.plot(

        temp[
            "nominal_coverage_percent"
        ],

        temp[
            "Profile_Coverage_percent"
        ],

        marker="o",

        linewidth=2,

        label=method
    )


plt.plot(

    [
        80,
        95
    ],

    [
        80,
        95
    ],

    linestyle="--",

    label="Ideal calibration"
)


plt.xlabel(
    "Nominal Whole-Profile Coverage (%)"
)


plt.ylabel(
    "Empirical Whole-Profile Coverage (%)"
)


plt.title(
    "Simultaneous 23-Slot Profile-Band Calibration"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG4 = (

    FIGURE_DIR
    /
    "04_simultaneous_profile_coverage.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 43. FIGURE 5
# 90% COVERAGE VS WIDTH
# =====================================================================

plot_90 = (

    interval_evaluation.loc[

        (
            interval_evaluation[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            interval_evaluation[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .copy()
)


plt.figure(
    figsize=(11, 7)
)


for _, row in plot_90.iterrows():

    plt.scatter(

        row[
            "MPIW_MW"
        ],

        row[
            "Primary_Coverage_percent"
        ],

        s=80
    )


    plt.annotate(

        row[
            "method"
        ],

        (
            row[
                "MPIW_MW"
            ],

            row[
                "Primary_Coverage_percent"
            ]
        ),

        xytext=(
            5,
            5
        ),

        textcoords="offset points"
    )


plt.axhline(

    90,

    linestyle="--"
)


plt.xlabel(
    "Mean Prediction Interval Width (MW)"
)


plt.ylabel(
    "Primary Empirical Coverage (%)"
)


plt.title(
    "Coverage–Sharpness Trade-Off at 90% Nominal Coverage"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG5 = (

    FIGURE_DIR
    /
    "05_coverage_width_tradeoff_90.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 44. FIGURE 6
# 90% PROPOSED MARGINAL COVERAGE BY HOUR
# =====================================================================

proposed_hourly_90 = (

    hourly_interval_results.loc[

        (
            hourly_interval_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            hourly_interval_results[
                "method"
            ]
            ==
            PROPOSED_MARGINAL_METHOD
        )

        &

        (
            hourly_interval_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .sort_values(
        "hour"
    )
)


plt.figure(
    figsize=(13, 6)
)


plt.plot(

    proposed_hourly_90[
        "hour"
    ],

    proposed_hourly_90[
        "empirical_coverage_percent"
    ],

    marker="o",

    linewidth=2
)


plt.axhline(

    90,

    linestyle="--",

    label="Nominal 90%"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Target Hour"
)


plt.ylabel(
    "Empirical Coverage (%)"
)


plt.title(
    "Proposed Regime-Hour Conformal Coverage Across Forecast Horizons"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG6 = (

    FIGURE_DIR
    /
    "06_proposed_90coverage_by_hour.png"
)


plt.savefig(
    FIG6,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 45. FIGURE 7
# EXAMPLE 90% PROPOSED INTERVAL
# =====================================================================

test_2024_positions = np.where(

    test_df[
        "target_date"
    ]
    .dt.year
    .eq(
        2024
    )
    .to_numpy()

)[0]


example_index = (

    test_2024_positions[

        len(
            test_2024_positions
        )

        //
        2
    ]
)


example_date = (

    test_df.loc[
        example_index,
        "target_date"
    ]
)


plt.figure(
    figsize=(13, 7)
)


plt.fill_between(

    HOURS,

    PROPOSED_90_LOWER[
        example_index
    ],

    PROPOSED_90_UPPER[
        example_index
    ],

    alpha=0.25,

    label="90% regime-hour conformal interval"
)


plt.plot(

    HOURS,

    Y_test[
        example_index
    ],

    marker="o",

    linewidth=2.5,

    label="Actual"
)


plt.plot(

    HOURS,

    TEST_PRED[
        example_index
    ],

    marker="s",

    linewidth=2,

    label="Ridge point forecast"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Hour"
)


plt.ylabel(
    "Demand (MW)"
)


plt.title(

    "Example Reliable Next-Day Load Forecast: "

    +
    str(
        example_date.date()
    )

    +
    f" | Regime={TEST_REGIME[example_index]}"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG7 = (

    FIGURE_DIR
    /
    "07_example_90percent_interval.png"
)


plt.savefig(
    FIG7,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 46. FIGURE 8
# PROPOSED MARGINAL 90% PERFORMANCE BY REGIME
# =====================================================================

proposed_regime_90 = (

    regime_interval_results.loc[

        (
            regime_interval_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            regime_interval_results[
                "method"
            ]
            ==
            PROPOSED_MARGINAL_METHOD
        )

        &

        (
            regime_interval_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .copy()
)


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    proposed_regime_90[
        "regime"
    ],

    proposed_regime_90[
        "Marginal_Coverage_percent"
    ]
)


plt.axhline(

    90,

    linestyle="--",

    label="Nominal 90%"
)


plt.ylabel(
    "Marginal Coverage (%)"
)


plt.xlabel(
    "Forecastable Load Regime"
)


plt.title(
    "Regime-Specific Reliability of Proposed 90% Intervals"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG8 = (

    FIGURE_DIR
    /
    "08_proposed_coverage_by_regime.png"
)


plt.savefig(
    FIG8,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 47. FINAL VALIDATION
# =====================================================================

assert (
    CAL_PRED.shape
    ==
    Y_cal.shape
)


assert (
    TEST_PRED.shape
    ==
    Y_test.shape
)


assert np.isfinite(
    CAL_PRED
).all()


assert np.isfinite(
    TEST_PRED
).all()


assert (
    len(
        calibration_df
    )
    >
    0
)


assert (

    calibration_df[
        "target_date"
    ]
    .max()

    <
    test_df[
        "target_date"
    ]
    .min()
)


assert np.all(

    PROPOSED_90_UPPER

    >=

    PROPOSED_90_LOWER
)


print(
    "\nAll Step 5 validation checks PASSED."
)


# =====================================================================
# 48. CREATE SUMMARY JSON
# =====================================================================

nested_all_test = (

    point_results.loc[
        point_results[
            "split"
        ]
        ==
        "ALL_TEST_2024_2026"
    ]

    .iloc[0]
)


summary = {

    "proper_training_samples":
        int(
            len(
                proper_train_df
            )
        ),

    "proper_training_end":
        str(
            proper_train_df[
                "target_date"
            ]
            .max()
        ),

    "calibration_samples":
        int(
            len(
                calibration_df
            )
        ),

    "calibration_start":
        str(
            calibration_df[
                "target_date"
            ]
            .min()
        ),

    "calibration_end":
        str(
            calibration_df[
                "target_date"
            ]
            .max()
        ),

    "test_samples":
        int(
            len(
                test_df
            )
        ),

    "nested_best_ridge_alpha":
        float(
            BEST_ALPHA
        ),

    "nested_ridge_test_MAE_MW":
        float(
            nested_all_test[
                "MAE_MW"
            ]
        ),

    "nested_ridge_test_RMSE_MW":
        float(
            nested_all_test[
                "RMSE_MW"
            ]
        ),

    "nested_ridge_test_MAPE_percent":
        float(
            nested_all_test[
                "MAPE_percent"
            ]
        ),

    "step3_fully_refitted_ridge_reference_MAE_MW":
        (
            None

            if np.isnan(
                STEP3_RIDGE_MAE
            )

            else float(
                STEP3_RIDGE_MAE
            )
        ),

    "regime_feature":
        REGIME_FEATURE,

    "low_regime_threshold_MW":
        float(
            LOW_THRESHOLD
        ),

    "high_regime_threshold_MW":
        float(
            HIGH_THRESHOLD
        ),

    "proposed_marginal_method":
        PROPOSED_MARGINAL_METHOD,

    "proposed_marginal_90_coverage_percent":
        float(
            proposed_marginal_90[
                "Primary_Coverage_percent"
            ]
        ),

    "proposed_marginal_90_MPIW_MW":
        float(
            proposed_marginal_90[
                "MPIW_MW"
            ]
        ),

    "proposed_marginal_90_Winkler":
        float(
            proposed_marginal_90[
                "Winkler_Score"
            ]
        ),

    "proposed_simultaneous_method":
        PROPOSED_PROFILE_METHOD,

    "proposed_simultaneous_90_profile_coverage_percent":
        float(
            proposed_profile_90[
                "Profile_Coverage_percent"
            ]
        ),

    "proposed_simultaneous_90_MPIW_MW":
        float(
            proposed_profile_90[
                "MPIW_MW"
            ]
        )
}


SUMMARY_FILE = (

    RESULT_DIR
    /
    "step5_summary.json"
)


with open(
    SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# =====================================================================
# 49. FINAL OUTPUT
# =====================================================================

print("\n")
print("=" * 115)

print(
    "STEP 5 COMPLETED SUCCESSFULLY"
)

print("=" * 115)


print(
    "\nProper model-training samples:",
    len(
        proper_train_df
    )
)


print(
    "Calibration samples:",
    len(
        calibration_df
    )
)


print(
    "Final test samples:",
    len(
        test_df
    )
)


print(
    "\nBest nested Ridge alpha:",
    BEST_ALPHA
)


print("\n" + "-" * 115)

print(
    "NESTED RIDGE — ALL TEST DATA"
)

print("-" * 115)


print(
    "MAE:",
    round(
        float(
            nested_all_test[
                "MAE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "RMSE:",
    round(
        float(
            nested_all_test[
                "RMSE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "MAPE:",
    round(
        float(
            nested_all_test[
                "MAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print("\n" + "-" * 115)

print(
    "PROPOSED MARGINAL 90% INTERVAL"
)

print("-" * 115)


print(
    "Method:",
    PROPOSED_MARGINAL_METHOD
)


print(
    "Empirical marginal coverage:",
    round(
        float(
            proposed_marginal_90[
                "Marginal_Coverage_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "Coverage gap:",
    round(
        float(
            proposed_marginal_90[
                "Coverage_Gap_percentage_points"
            ]
        ),
        3
    ),
    "percentage points"
)


print(
    "MPIW:",
    round(
        float(
            proposed_marginal_90[
                "MPIW_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "Normalized MPIW:",
    round(
        float(
            proposed_marginal_90[
                "Normalized_MPIW_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "Winkler score:",
    round(
        float(
            proposed_marginal_90[
                "Winkler_Score"
            ]
        ),
        3
    )
)


print("\n" + "-" * 115)

print(
    "PROPOSED SIMULTANEOUS 90% PROFILE BAND"
)

print("-" * 115)


print(
    "Method:",
    PROPOSED_PROFILE_METHOD
)


print(
    "Whole-profile coverage:",
    round(
        float(
            proposed_profile_90[
                "Profile_Coverage_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "MPIW:",
    round(
        float(
            proposed_profile_90[
                "MPIW_MW"
            ]
        ),
        3
    ),
    "MW"
)


print("\n" + "-" * 115)

print(
    "IMPORTANT OUTPUT FILES"
)

print("-" * 115)


print(
    "\nNested Ridge tuning:"
)

print(
    RIDGE_TUNING_FILE
)


print(
    "\nPoint performance:"
)

print(
    POINT_RESULT_FILE
)


print(
    "\nConformal interval evaluation:"
)

print(
    INTERVAL_RESULT_FILE
)


print(
    "\nHourly coverage:"
)

print(
    HOURLY_RESULT_FILE
)


print(
    "\nRegime coverage:"
)

print(
    REGIME_RESULT_FILE
)


print(
    "\nFinal proposed intervals:"
)

print(
    PROPOSED_INTERVAL_FILE
)


print(
    "\nFigures:"
)

print(
    FIGURE_DIR
)


print(
    "\nSummary:"
)

print(
    SUMMARY_FILE
)


print("\n" + "=" * 115)

print(
    "SEND THE FINAL STEP 5 SUMMARY AND PRIMARY 90% INTERVAL TABLE TO CHATGPT."
)

print(
    "DO NOT START EXTERNAL BPDB VALIDATION YET."
)

print("=" * 115)
```
