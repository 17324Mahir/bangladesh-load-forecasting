```python
# =====================================================================
# STEP 2 - FINAL CORRECTED VERSION
# FORECASTING-READY DATASET + EDA + NAIVE BASELINES
#
# Primary task:
# Predict the next day's observed national load profile
# from 01:00 to 23:00 = 23 outputs
#
# Secondary task:
# Predict the special PGCB 19:30 evening peak
#
# IMPORTANT:
# Step 1C must already be complete.
#
# This version fixes:
# ValueError:
# 'target_date' is both an index level and a column label
#
# Run this entire code in ONE Google Colab cell.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import json
import math
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 300)
pd.set_option("display.max_rows", 200)
pd.set_option("display.width", 260)


print("=" * 105)
print("STEP 2: FORECASTING DATASET + EDA + SEASONAL NAIVE BASELINES")
print("=" * 105)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount("/content/drive")

print("\nGoogle Drive mounted successfully.")


# =====================================================================
# 2. DEFINE PROJECT DIRECTORIES
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)

RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "forecasting_dataset"
)

FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "forecasting_dataset"
)


for directory in [
    PROCESSED_DIR,
    FORECAST_DIR,
    RESULT_DIR,
    FIGURE_DIR
]:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


print("\nProject root:")
print(PROJECT_ROOT)


# =====================================================================
# 3. INPUT FILES CREATED IN STEP 1C
# =====================================================================

HOURLY_FILE = (
    PROCESSED_DIR
    / "pgcb_final_observed_hourly_target.csv"
)

PEAK_FILE = (
    PROCESSED_DIR
    / "pgcb_final_1930_peak_target.csv"
)


if not HOURLY_FILE.exists():

    raise FileNotFoundError(
        f"""
Step 1C final hourly file was not found:

{HOURLY_FILE}

Please run Step 1C first.
"""
    )


print("\nMain hourly input:")
print(HOURLY_FILE)


print("\n19:30 peak input:")
print(PEAK_FILE)


# =====================================================================
# 4. LOAD FINAL CLEAN HOURLY DATASET
# =====================================================================

hourly = pd.read_csv(
    HOURLY_FILE,
    low_memory=False
)


hourly["datetime"] = pd.to_datetime(
    hourly["datetime"],
    errors="coerce"
)


hourly = (

    hourly
    .dropna(
        subset=["datetime"]
    )

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


# =====================================================================
# 5. DETERMINE FINAL TARGET COLUMN
# =====================================================================

if "demand_final" in hourly.columns:

    TARGET_COLUMN = "demand_final"


elif "demand_mw_clean" in hourly.columns:

    TARGET_COLUMN = "demand_mw_clean"


elif "demand_mw" in hourly.columns:

    TARGET_COLUMN = "demand_mw"


else:

    raise ValueError(
        "No valid final demand target column was found."
    )


hourly[TARGET_COLUMN] = pd.to_numeric(
    hourly[TARGET_COLUMN],
    errors="coerce"
)


print("\n" + "=" * 105)
print("FINAL STEP 1C DATASET LOADED")
print("=" * 105)


print(
    "Rows:",
    f"{len(hourly):,}"
)

print(
    "Period:",
    hourly["datetime"].min(),
    "to",
    hourly["datetime"].max()
)

print(
    "Target column:",
    TARGET_COLUMN
)


# =====================================================================
# 6. TIME COMPONENTS
# =====================================================================

hourly["date"] = (
    hourly["datetime"]
    .dt.normalize()
)


hourly["hour"] = (
    hourly["datetime"]
    .dt.hour
)


hourly["minute"] = (
    hourly["datetime"]
    .dt.minute
)


# =====================================================================
# 7. STRUCTURAL VALIDATION
# =====================================================================

if not hourly[
    "minute"
].eq(0).all():

    bad = hourly.loc[
        ~hourly[
            "minute"
        ].eq(0),
        [
            "datetime",
            TARGET_COLUMN
        ]
    ]

    print(bad.head())

    raise ValueError(
        "Non-integer-hour rows exist in the main forecasting dataset."
    )


if hourly[
    TARGET_COLUMN
].isna().any():

    raise ValueError(
        "Missing final demand values remain in the Step 1C target dataset."
    )


if hourly[
    "datetime"
].duplicated().any():

    raise ValueError(
        "Duplicate timestamps remain in the Step 1C dataset."
    )


print(
    "\nBasic structural checks passed."
)


# =====================================================================
# 8. DEFINE PRIMARY FORECASTING HOURS
# =====================================================================

# PGCB historical data does not provide a reliable conventional
# 00:00 observation.
#
# Therefore we use:
#
# 01:00
# 02:00
# ...
# 23:00
#
# = 23 observed target slots per day.


HOURS = list(
    range(
        1,
        24
    )
)


print(
    "\nPrimary target hours:"
)

print(HOURS)


print(
    "Number of target outputs:",
    len(HOURS)
)


# =====================================================================
# 9. FILTER PRIMARY PROFILE DATA
# =====================================================================

profile_source = (

    hourly.loc[
        hourly[
            "hour"
        ]
        .isin(HOURS)
    ]

    .copy()
)


print(
    "\n01:00-23:00 observations:",
    f"{len(profile_source):,}"
)


# =====================================================================
# 10. DATE-HOUR DUPLICATE CHECK
# =====================================================================

duplicate_date_hour = (

    profile_source
    .duplicated(
        subset=[
            "date",
            "hour"
        ],
        keep=False
    )
)


duplicate_count = int(
    duplicate_date_hour.sum()
)


print(
    "Duplicate date-hour observations:",
    duplicate_count
)


if duplicate_count > 0:

    display(

        profile_source.loc[
            duplicate_date_hour,
            [
                "datetime",
                "date",
                "hour",
                TARGET_COLUMN
            ]
        ]

        .sort_values(
            [
                "date",
                "hour"
            ]
        )

        .head(50)
    )

    raise ValueError(
        "Duplicate date-hour observations detected."
    )


# =====================================================================
# 11. CREATE DAILY 23-SLOT LOAD MATRIX
# =====================================================================

profile = (

    profile_source

    .pivot(
        index="date",
        columns="hour",
        values=TARGET_COLUMN
    )

    .reindex(
        columns=HOURS
    )
)


# Ensure DatetimeIndex

profile.index = pd.to_datetime(
    profile.index
)


profile.index.name = "target_date"


print(
    "\nRaw daily profile matrix created."
)

print(
    "Available dates:",
    len(profile)
)


# =====================================================================
# 12. REINDEX TO COMPLETE CALENDAR
# =====================================================================

# Critical:
#
# We reindex to every calendar date before creating lag features.
#
# Without this step:
#
# shift(1) = previous AVAILABLE row
#
# instead of:
#
# shift(1) = yesterday
#
# That would be wrong when complete days are missing.


calendar_index = pd.date_range(

    start=profile.index.min(),

    end=profile.index.max(),

    freq="D"
)


profile = profile.reindex(
    calendar_index
)


profile.index.name = "target_date"


print(
    "Full calendar length:",
    f"{len(profile):,}",
    "days"
)


# =====================================================================
# 13. DAILY TARGET COMPLETENESS
# =====================================================================

profile_completeness = pd.DataFrame(
    index=profile.index
)


profile_completeness[
    "observed_slots"
] = (

    profile
    .notna()
    .sum(
        axis=1
    )
)


profile_completeness[
    "missing_slots"
] = (

    23

    -

    profile_completeness[
        "observed_slots"
    ]
)


profile_completeness[
    "complete_23_slot_day"
] = (

    profile_completeness[
        "observed_slots"
    ]
    ==
    23
)


completeness_distribution = (

    profile_completeness[
        "observed_slots"
    ]

    .value_counts()

    .sort_index()

    .rename_axis(
        "observed_slots"
    )

    .reset_index(
        name="number_of_days"
    )
)


print("\n" + "=" * 105)
print("DAILY PROFILE COMPLETENESS")
print("=" * 105)


display(
    completeness_distribution
)


COMPLETE_TARGET_DAYS = int(

    profile_completeness[
        "complete_23_slot_day"
    ]
    .sum()
)


print(
    "\nCompletely usable 23-slot target days:",
    COMPLETE_TARGET_DAYS
)


# =====================================================================
# 14. SAVE RAW DAILY PROFILE MATRIX
# =====================================================================

profile_save = (
    profile.copy()
)


profile_save.columns = [

    f"demand_h{hour:02d}"

    for hour in HOURS
]


profile_save = (

    profile_save
    .reset_index()
)


RAW_PROFILE_FILE = (

    FORECAST_DIR
    /
    "daily_23slot_load_profile_raw.csv"
)


profile_save.to_csv(
    RAW_PROFILE_FILE,
    index=False
)


# =====================================================================
# 15. BUILD TARGET MATRIX
# =====================================================================

target_matrix = (
    profile.copy()
)


target_matrix.columns = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 16. PREVIOUS-DAY PROFILE FEATURES
# =====================================================================

# These are available at prediction time.


lag1 = (
    profile.shift(1)
)


lag1.columns = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 17. PREVIOUS-WEEK PROFILE FEATURES
# =====================================================================

lag7 = (
    profile.shift(7)
)


lag7.columns = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 18. CAUSAL 7-DAY ROLLING MEAN
# =====================================================================

# shift(1) is mandatory.
#
# It ensures today's target does not enter its own feature.


roll7 = (

    profile
    .shift(1)

    .rolling(
        window=7,
        min_periods=4
    )

    .mean()
)


roll7.columns = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 19. CAUSAL 28-DAY ROLLING MEAN
# =====================================================================

roll28 = (

    profile
    .shift(1)

    .rolling(
        window=28,
        min_periods=14
    )

    .mean()
)


roll28.columns = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 20. CAUSAL 7-DAY ROLLING STD
# =====================================================================

roll7_std = (

    profile
    .shift(1)

    .rolling(
        window=7,
        min_periods=4
    )

    .std()
)


roll7_std.columns = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 21. CAUSAL 28-DAY ROLLING STD
# =====================================================================

roll28_std = (

    profile
    .shift(1)

    .rolling(
        window=28,
        min_periods=14
    )

    .std()
)


roll28_std.columns = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 22. CALENDAR FEATURES
# =====================================================================

calendar = pd.DataFrame(
    index=profile.index
)


calendar["year"] = (
    calendar.index.year
)


calendar["month"] = (
    calendar.index.month
)


calendar["day"] = (
    calendar.index.day
)


calendar["dayofweek"] = (
    calendar.index.dayofweek
)


calendar["dayofyear"] = (
    calendar.index.dayofyear
)


calendar["weekofyear"] = (

    calendar.index
    .isocalendar()
    .week
    .astype(int)
    .to_numpy()
)


calendar["quarter"] = (
    calendar.index.quarter
)


# Friday = 4
# Saturday = 5

calendar["is_weekend"] = (

    calendar[
        "dayofweek"
    ]

    .isin(
        [
            4,
            5
        ]
    )

    .astype(int)
)


# =====================================================================
# 23. CYCLICAL CALENDAR FEATURES
# =====================================================================

calendar["month_sin"] = np.sin(

    2
    *
    np.pi
    *
    calendar[
        "month"
    ]
    /
    12
)


calendar["month_cos"] = np.cos(

    2
    *
    np.pi
    *
    calendar[
        "month"
    ]
    /
    12
)


calendar["dow_sin"] = np.sin(

    2
    *
    np.pi
    *
    calendar[
        "dayofweek"
    ]
    /
    7
)


calendar["dow_cos"] = np.cos(

    2
    *
    np.pi
    *
    calendar[
        "dayofweek"
    ]
    /
    7
)


calendar["doy_sin"] = np.sin(

    2
    *
    np.pi
    *
    calendar[
        "dayofyear"
    ]
    /
    365.25
)


calendar["doy_cos"] = np.cos(

    2
    *
    np.pi
    *
    calendar[
        "dayofyear"
    ]
    /
    365.25
)


# =====================================================================
# 24. PREVIOUS-DAY SUMMARY FEATURES
# =====================================================================

previous_day = (
    profile.shift(1)
)


summary_features = pd.DataFrame(
    index=profile.index
)


summary_features[
    "lag1_daily_mean"
] = (

    previous_day
    .mean(
        axis=1
    )
)


summary_features[
    "lag1_daily_max"
] = (

    previous_day
    .max(
        axis=1
    )
)


summary_features[
    "lag1_daily_min"
] = (

    previous_day
    .min(
        axis=1
    )
)


summary_features[
    "lag1_daily_range"
] = (

    summary_features[
        "lag1_daily_max"
    ]

    -

    summary_features[
        "lag1_daily_min"
    ]
)


summary_features[
    "lag1_peak_hour"
] = (

    previous_day
    .idxmax(
        axis=1
    )
)


# =====================================================================
# 25. PREVIOUS-WEEK SUMMARY FEATURES
# =====================================================================

previous_week = (
    profile.shift(7)
)


summary_features[
    "lag7_daily_mean"
] = (

    previous_week
    .mean(
        axis=1
    )
)


summary_features[
    "lag7_daily_max"
] = (

    previous_week
    .max(
        axis=1
    )
)


summary_features[
    "lag7_daily_min"
] = (

    previous_week
    .min(
        axis=1
    )
)


summary_features[
    "lag7_daily_range"
] = (

    summary_features[
        "lag7_daily_max"
    ]

    -

    summary_features[
        "lag7_daily_min"
    ]
)


summary_features[
    "lag7_peak_hour"
] = (

    previous_week
    .idxmax(
        axis=1
    )
)


# =====================================================================
# 26. TREND FEATURES
# =====================================================================

summary_features[
    "mean_change_vs_last_week"
] = (

    summary_features[
        "lag1_daily_mean"
    ]

    -

    summary_features[
        "lag7_daily_mean"
    ]
)


summary_features[
    "max_change_vs_last_week"
] = (

    summary_features[
        "lag1_daily_max"
    ]

    -

    summary_features[
        "lag7_daily_max"
    ]
)


summary_features[
    "mean_ratio_vs_last_week"
] = (

    summary_features[
        "lag1_daily_mean"
    ]

    /

    summary_features[
        "lag7_daily_mean"
    ]
    .replace(
        0,
        np.nan
    )
)


# =====================================================================
# 27. EXTREME-EVENT DAY LABEL
# =====================================================================

# Evaluation-only label.
#
# This MUST NOT be used as a forecasting feature.


evaluation_labels = pd.DataFrame(
    index=profile.index
)


if (
    "advanced_quality_status"
    in profile_source.columns
):

    temp_extreme = (
        profile_source.copy()
    )


    temp_extreme[
        "is_preserved_extreme"
    ] = (

        temp_extreme[
            "advanced_quality_status"
        ]
        .eq(
            "KEEP_EXTREME_EVENT"
        )

        .astype(int)
    )


    extreme_by_date = (

        temp_extreme
        .groupby(
            "date"
        )[
            "is_preserved_extreme"
        ]

        .max()

        .reindex(
            profile.index
        )

        .fillna(0)

        .astype(int)
    )


else:

    extreme_by_date = pd.Series(

        0,

        index=profile.index,

        dtype=int
    )


evaluation_labels[
    "target_extreme_event_day"
] = (
    extreme_by_date
)


# =====================================================================
# 28. TARGET-DAY ANALYSIS VARIABLES
# =====================================================================

# These are NOT model features.


evaluation_labels[
    "target_daily_mean"
] = (
    profile.mean(
        axis=1
    )
)


evaluation_labels[
    "target_daily_max"
] = (
    profile.max(
        axis=1
    )
)


evaluation_labels[
    "target_daily_min"
] = (
    profile.min(
        axis=1
    )
)


evaluation_labels[
    "target_peak_hour"
] = (
    profile.idxmax(
        axis=1
    )
)


# =====================================================================
# 29. LOAD STRICT 19:30 PEAK SERIES
# =====================================================================

peak_features = pd.DataFrame(
    index=profile.index
)


if PEAK_FILE.exists():

    peak = pd.read_csv(
        PEAK_FILE,
        low_memory=False
    )


    peak["datetime"] = pd.to_datetime(
        peak["datetime"],
        errors="coerce"
    )


    peak = (

        peak
        .dropna(
            subset=["datetime"]
        )

        .sort_values(
            "datetime"
        )

        .reset_index(
            drop=True
        )
    )


    peak["date"] = (
        peak["datetime"]
        .dt.normalize()
    )


    if "demand_original" in peak.columns:

        PEAK_TARGET_COLUMN = (
            "demand_original"
        )


    elif "demand_mw" in peak.columns:

        PEAK_TARGET_COLUMN = (
            "demand_mw"
        )


    elif "demand_final" in peak.columns:

        PEAK_TARGET_COLUMN = (
            "demand_final"
        )


    else:

        PEAK_TARGET_COLUMN = None


    if PEAK_TARGET_COLUMN is not None:

        peak[
            PEAK_TARGET_COLUMN
        ] = pd.to_numeric(
            peak[
                PEAK_TARGET_COLUMN
            ],
            errors="coerce"
        )


        peak_daily = (

            peak
            .dropna(
                subset=[
                    PEAK_TARGET_COLUMN
                ]
            )

            .drop_duplicates(
                subset=["date"],
                keep="first"
            )

            .set_index(
                "date"
            )[
                PEAK_TARGET_COLUMN
            ]

            .reindex(
                profile.index
            )
        )


        peak_features[
            "target_peak_1930"
        ] = (
            peak_daily
        )


        peak_features[
            "lag1_peak_1930"
        ] = (
            peak_daily.shift(1)
        )


        peak_features[
            "lag7_peak_1930"
        ] = (
            peak_daily.shift(7)
        )


        peak_features[
            "roll7mean_peak_1930"
        ] = (

            peak_daily
            .shift(1)

            .rolling(
                window=7,
                min_periods=4
            )

            .mean()
        )


        print(
            "\nStrict 19:30 series integrated."
        )


        print(
            "Available peak targets:",
            int(
                peak_features[
                    "target_peak_1930"
                ]
                .notna()
                .sum()
            )
        )


else:

    print(
        "\n19:30 peak file not found."
    )


# =====================================================================
# 30. COMBINE ALL DATAFRAMES
# =====================================================================

# IMPORTANT FIX:
#
# We keep target_date ONLY as index during concatenation.
#
# Then we reset the index exactly once.
#
# This prevents:
#
# ValueError:
# 'target_date' is both an index level and a column label


dataset = pd.concat(
    [
        calendar,

        lag1,

        lag7,

        roll7,

        roll28,

        roll7_std,

        roll28_std,

        summary_features,

        peak_features,

        evaluation_labels,

        target_matrix
    ],
    axis=1
)


# Ensure only one index name

dataset.index.name = (
    "target_date"
)


# Convert target_date from index to normal column ONCE.

dataset = (
    dataset
    .reset_index()
)


# Now index is a normal RangeIndex.

dataset.index.name = None


dataset[
    "target_date"
] = pd.to_datetime(
    dataset[
        "target_date"
    ]
)


print("\nCombined dataset created.")

print(
    "Rows:",
    f"{len(dataset):,}"
)

print(
    "Columns:",
    len(
        dataset.columns
    )
)


# =====================================================================
# 31. VERIFY TARGET_DATE AMBIGUITY IS FIXED
# =====================================================================

if (
    dataset.index.name
    ==
    "target_date"
):

    raise RuntimeError(
        "target_date still exists as index name."
    )


if (
    "target_date"
    not in
    dataset.columns
):

    raise RuntimeError(
        "target_date column was not created."
    )


print(
    "\ntarget_date ambiguity check PASSED."
)

print(
    "Index name:",
    dataset.index.name
)

print(
    "target_date exists as normal column:",
    "target_date" in dataset.columns
)


# =====================================================================
# 32. DEFINE TARGET COLUMNS
# =====================================================================

target_columns = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 33. DEFINE FEATURE GROUPS
# =====================================================================

calendar_feature_columns = [

    "year",

    "month",

    "day",

    "dayofweek",

    "dayofyear",

    "weekofyear",

    "quarter",

    "is_weekend",

    "month_sin",

    "month_cos",

    "dow_sin",

    "dow_cos",

    "doy_sin",

    "doy_cos"
]


lag1_columns = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


lag7_columns = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


roll7_columns = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


roll28_columns = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


roll7std_columns = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


roll28std_columns = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


summary_feature_columns = [

    "lag1_daily_mean",

    "lag1_daily_max",

    "lag1_daily_min",

    "lag1_daily_range",

    "lag1_peak_hour",

    "lag7_daily_mean",

    "lag7_daily_max",

    "lag7_daily_min",

    "lag7_daily_range",

    "lag7_peak_hour",

    "mean_change_vs_last_week",

    "max_change_vs_last_week",

    "mean_ratio_vs_last_week"
]


# =====================================================================
# 34. PRIMARY MODEL FEATURE LIST
# =====================================================================

# For the PRIMARY 23-slot forecasting model,
# we intentionally do NOT include 19:30 historical peak features yet.
#
# Reason:
# We first establish a clean demand-history + calendar baseline.
#
# The 19:30 historical features will be tested later as an ablation.


feature_columns = (

    calendar_feature_columns

    +

    lag1_columns

    +

    lag7_columns

    +

    roll7_columns

    +

    roll28_columns

    +

    roll7std_columns

    +

    roll28std_columns

    +

    summary_feature_columns
)


print(
    "\nPrimary feature count:",
    len(
        feature_columns
    )
)


print(
    "Target output count:",
    len(
        target_columns
    )
)


# =====================================================================
# 35. FEATURE LEAKAGE CHECK
# =====================================================================

leaking_columns = [

    column

    for column in feature_columns

    if column.startswith(
        "target_"
    )
]


if len(
    leaking_columns
) > 0:

    raise RuntimeError(
        f"""
TARGET LEAKAGE DETECTED:

{leaking_columns}
"""
    )


print(
    "\nTarget leakage check PASSED."
)


# =====================================================================
# 36. COMPLETE TARGET-DAY MASK
# =====================================================================

complete_target_mask = (

    dataset[
        target_columns
    ]

    .notna()

    .all(
        axis=1
    )
)


print(
    "\nComplete target days:",
    int(
        complete_target_mask.sum()
    )
)


# =====================================================================
# 37. REQUIRED HISTORICAL FEATURE MASK
# =====================================================================

# For a strict forecasting experiment we require:
#
# full previous-day profile
# full previous-week profile
# complete rolling mean features
# complete rolling std features


essential_history_columns = (

    lag1_columns

    +

    lag7_columns

    +

    roll7_columns

    +

    roll28_columns

    +

    roll7std_columns

    +

    roll28std_columns
)


complete_history_mask = (

    dataset[
        essential_history_columns
    ]

    .notna()

    .all(
        axis=1
    )
)


print(
    "Days with complete historical inputs:",
    int(
        complete_history_mask.sum()
    )
)


# =====================================================================
# 38. SUMMARY FEATURE AVAILABILITY
# =====================================================================

summary_feature_mask = (

    dataset[
        summary_feature_columns
    ]

    .notna()

    .all(
        axis=1
    )
)


print(
    "Days with complete historical summary features:",
    int(
        summary_feature_mask.sum()
    )
)


# =====================================================================
# 39. FINAL SAMPLE MASK
# =====================================================================

final_sample_mask = (

    complete_target_mask

    &

    complete_history_mask

    &

    summary_feature_mask
)


# =====================================================================
# 40. CREATE FINAL FORECAST SAMPLE DATASET
# =====================================================================

forecast_samples = (

    dataset.loc[
        final_sample_mask
    ]

    .copy()
)


# IMPORTANT:
# target_date is now ONLY a normal column,
# so sort_values is unambiguous.

forecast_samples = (

    forecast_samples

    .sort_values(
        by="target_date"
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 105)
print("FORECAST SAMPLE CONSTRUCTION")
print("=" * 105)


print(
    "Full calendar days:",
    f"{len(dataset):,}"
)


print(
    "Complete target days:",
    f"{complete_target_mask.sum():,}"
)


print(
    "Complete history days:",
    f"{complete_history_mask.sum():,}"
)


print(
    "Final forecasting samples:",
    f"{len(forecast_samples):,}"
)


# =====================================================================
# 41. FINAL DATASET LEAKAGE VALIDATION
# =====================================================================

assert (
    "target_date"
    in
    forecast_samples.columns
)


assert (
    forecast_samples.index.name
    is None
)


assert (
    set(
        target_columns
    )
    .isdisjoint(
        set(
            feature_columns
        )
    )
)


assert (

    forecast_samples[
        target_columns
    ]
    .notna()
    .all()
    .all()

)


assert (

    forecast_samples[
        feature_columns
    ]
    .notna()
    .all()
    .all()

)


print(
    "\nFinal forecasting-data validation PASSED."
)


# =====================================================================
# 42. CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# =====================================================================

# NO RANDOM SPLIT.
#
# Training:
# beginning -> 2022-12-31
#
# Validation:
# 2023
#
# Test:
# 2024 -> March 2026


forecast_samples[
    "split"
] = "UNASSIGNED"


train_mask = (

    forecast_samples[
        "target_date"
    ]
    <=
    pd.Timestamp(
        "2022-12-31"
    )
)


validation_mask = (

    forecast_samples[
        "target_date"
    ]
    .between(
        pd.Timestamp(
            "2023-01-01"
        ),
        pd.Timestamp(
            "2023-12-31"
        )
    )
)


test_mask = (

    forecast_samples[
        "target_date"
    ]
    >=
    pd.Timestamp(
        "2024-01-01"
    )
)


forecast_samples.loc[
    train_mask,
    "split"
] = "TRAIN"


forecast_samples.loc[
    validation_mask,
    "split"
] = "VALIDATION"


forecast_samples.loc[
    test_mask,
    "split"
] = "TEST"


# =====================================================================
# 43. CHECK UNASSIGNED SAMPLES
# =====================================================================

unassigned_count = int(

    forecast_samples[
        "split"
    ]
    .eq(
        "UNASSIGNED"
    )
    .sum()
)


print(
    "\nUnassigned samples:",
    unassigned_count
)


if unassigned_count > 0:

    display(

        forecast_samples.loc[
            forecast_samples[
                "split"
            ]
            ==
            "UNASSIGNED",
            [
                "target_date",
                "split"
            ]
        ]
    )

    raise RuntimeError(
        "Some forecasting samples were not assigned to a split."
    )


# =====================================================================
# 44. SPLIT SUMMARY
# =====================================================================

split_order = [
    "TRAIN",
    "VALIDATION",
    "TEST"
]


split_summary = (

    forecast_samples[
        "split"
    ]

    .value_counts()

    .reindex(
        split_order
    )

    .rename_axis(
        "split"
    )

    .reset_index(
        name="samples"
    )
)


split_summary[
    "percent"
] = (

    100

    *

    split_summary[
        "samples"
    ]

    /

    len(
        forecast_samples
    )
)


print("\n" + "=" * 105)
print("CHRONOLOGICAL DATA SPLIT")
print("=" * 105)


display(
    split_summary
)


# =====================================================================
# 45. CREATE SPLIT DATAFRAMES
# =====================================================================

train_df = (

    forecast_samples.loc[
        forecast_samples[
            "split"
        ]
        ==
        "TRAIN"
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


validation_df = (

    forecast_samples.loc[
        forecast_samples[
            "split"
        ]
        ==
        "VALIDATION"
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


test_df = (

    forecast_samples.loc[
        forecast_samples[
            "split"
        ]
        ==
        "TEST"
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


print(
    "\nTRAIN samples:",
    f"{len(train_df):,}"
)


print(
    "VALIDATION samples:",
    f"{len(validation_df):,}"
)


print(
    "TEST samples:",
    f"{len(test_df):,}"
)


# =====================================================================
# 46. FEATURE MANIFEST
# =====================================================================

feature_manifest = pd.DataFrame(
    {
        "feature":
            feature_columns
    }
)


def determine_group(
    feature
):

    if feature in calendar_feature_columns:
        return "calendar"

    if feature in lag1_columns:
        return "previous_day_profile"

    if feature in lag7_columns:
        return "previous_week_profile"

    if feature in roll7_columns:
        return "rolling_7day_mean"

    if feature in roll28_columns:
        return "rolling_28day_mean"

    if feature in roll7std_columns:
        return "rolling_7day_std"

    if feature in roll28std_columns:
        return "rolling_28day_std"

    if feature in summary_feature_columns:
        return "historical_summary"

    return "other"


feature_manifest[
    "feature_group"
] = (

    feature_manifest[
        "feature"
    ]

    .apply(
        determine_group
    )
)


feature_group_summary = (

    feature_manifest[
        "feature_group"
    ]

    .value_counts()

    .rename_axis(
        "feature_group"
    )

    .reset_index(
        name="feature_count"
    )
)


print(
    "\nFeature groups:"
)

display(
    feature_group_summary
)


# =====================================================================
# 47. METRIC FUNCTIONS
# =====================================================================

def mae(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    return float(

        np.mean(

            np.abs(
                y_true
                -
                y_pred
            )
        )
    )


def rmse(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    return float(

        np.sqrt(

            np.mean(

                (
                    y_true
                    -
                    y_pred
                )
                **
                2
            )
        )
    )


def mape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = np.maximum(
        np.abs(
            y_true
        ),
        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    y_true
                    -
                    y_pred
                )

                /

                denominator
            )
        )
    )


def smape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = (

        np.abs(
            y_true
        )

        +

        np.abs(
            y_pred
        )
    )


    valid = (
        denominator
        >
        0
    )


    if not np.any(
        valid
    ):

        return np.nan


    return float(

        100

        *

        np.mean(

            2

            *

            np.abs(

                y_true[
                    valid
                ]

                -

                y_pred[
                    valid
                ]
            )

            /

            denominator[
                valid
            ]
        )
    )


# =====================================================================
# 48. PROFILE BASELINE EVALUATION FUNCTION
# =====================================================================

def evaluate_profile_baseline(
    dataframe,
    prediction_prefix,
    baseline_name,
    split_name
):

    if len(
        dataframe
    ) == 0:

        return None, []


    y_true = (

        dataframe[
            target_columns
        ]

        .to_numpy(
            dtype=float
        )
    )


    pred_columns = [

        f"{prediction_prefix}_h{hour:02d}"

        for hour in HOURS
    ]


    y_pred = (

        dataframe[
            pred_columns
        ]

        .to_numpy(
            dtype=float
        )
    )


    overall_result = {

        "split":
            split_name,

        "model":
            baseline_name,

        "samples":
            len(
                dataframe
            ),

        "MAE_MW":
            mae(
                y_true,
                y_pred
            ),

        "RMSE_MW":
            rmse(
                y_true,
                y_pred
            ),

        "MAPE_percent":
            mape(
                y_true,
                y_pred
            ),

        "sMAPE_percent":
            smape(
                y_true,
                y_pred
            )
    }


    hourly_results = []


    for j, hour in enumerate(
        HOURS
    ):

        true_hour = (
            y_true[
                :,
                j
            ]
        )

        pred_hour = (
            y_pred[
                :,
                j
            ]
        )


        hourly_results.append(
            {

                "split":
                    split_name,

                "model":
                    baseline_name,

                "hour":
                    hour,

                "MAE_MW":
                    mae(
                        true_hour,
                        pred_hour
                    ),

                "RMSE_MW":
                    rmse(
                        true_hour,
                        pred_hour
                    ),

                "MAPE_percent":
                    mape(
                        true_hour,
                        pred_hour
                    ),

                "sMAPE_percent":
                    smape(
                        true_hour,
                        pred_hour
                    )
            }
        )


    return (
        overall_result,
        hourly_results
    )


# =====================================================================
# 49. EVALUATE MANDATORY NAIVE BASELINES
# =====================================================================

# Model 1:
# Previous day's same-hour profile
#
# Model 2:
# Same weekday one week earlier


overall_results = []

hour_results = []


split_frames = {

    "TRAIN":
        train_df,

    "VALIDATION":
        validation_df,

    "TEST":
        test_df
}


for split_name, split_frame in (
    split_frames.items()
):

    # -------------------------------------------------------------
    # Previous-day persistence
    # -------------------------------------------------------------

    overall, hourly_metrics = (
        evaluate_profile_baseline(

            dataframe=
                split_frame,

            prediction_prefix=
                "lag1",

            baseline_name=
                "Seasonal_Naive_D1",

            split_name=
                split_name
        )
    )


    if overall is not None:

        overall_results.append(
            overall
        )


        hour_results.extend(
            hourly_metrics
        )


    # -------------------------------------------------------------
    # Previous-week seasonal persistence
    # -------------------------------------------------------------

    overall, hourly_metrics = (
        evaluate_profile_baseline(

            dataframe=
                split_frame,

            prediction_prefix=
                "lag7",

            baseline_name=
                "Seasonal_Naive_D7",

            split_name=
                split_name
        )
    )


    if overall is not None:

        overall_results.append(
            overall
        )


        hour_results.extend(
            hourly_metrics
        )


baseline_overall = pd.DataFrame(
    overall_results
)


baseline_by_hour = pd.DataFrame(
    hour_results
)


print("\n" + "=" * 105)
print("MANDATORY SEASONAL NAIVE BASELINES")
print("=" * 105)


display(
    baseline_overall.round(4)
)


# =====================================================================
# 50. TEST-SET BASELINES
# =====================================================================

test_baselines = (

    baseline_overall.loc[
        baseline_overall[
            "split"
        ]
        ==
        "TEST"
    ]

    .sort_values(
        by="MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nTEST-SET BASELINE RESULTS:"
)

display(
    test_baselines.round(4)
)


if len(
    test_baselines
) == 0:

    raise RuntimeError(
        "No test baseline results were generated."
    )


best_test_naive = (

    test_baselines
    .iloc[0]
    .to_dict()
)


# =====================================================================
# 51. REGIME-WISE BASELINE ANALYSIS
# =====================================================================

regime_results = []


normal_test = (

    test_df.loc[
        test_df[
            "target_extreme_event_day"
        ]
        ==
        0
    ]

    .copy()
)


extreme_test = (

    test_df.loc[
        test_df[
            "target_extreme_event_day"
        ]
        ==
        1
    ]

    .copy()
)


for regime_name, regime_frame in [

    (
        "NORMAL_TEST_DAY",
        normal_test
    ),

    (
        "PRESERVED_EXTREME_TEST_DAY",
        extreme_test
    )
]:

    if len(
        regime_frame
    ) == 0:

        continue


    y_true = (

        regime_frame[
            target_columns
        ]

        .to_numpy(
            dtype=float
        )
    )


    for prefix, model_name in [

        (
            "lag1",
            "Seasonal_Naive_D1"
        ),

        (
            "lag7",
            "Seasonal_Naive_D7"
        )
    ]:

        pred_columns = [

            f"{prefix}_h{hour:02d}"

            for hour in HOURS
        ]


        y_pred = (

            regime_frame[
                pred_columns
            ]

            .to_numpy(
                dtype=float
            )
        )


        regime_results.append(
            {

                "regime":
                    regime_name,

                "model":
                    model_name,

                "days":
                    len(
                        regime_frame
                    ),

                "MAE_MW":
                    mae(
                        y_true,
                        y_pred
                    ),

                "RMSE_MW":
                    rmse(
                        y_true,
                        y_pred
                    ),

                "MAPE_percent":
                    mape(
                        y_true,
                        y_pred
                    ),

                "sMAPE_percent":
                    smape(
                        y_true,
                        y_pred
                    )
            }
        )


regime_baseline_results = pd.DataFrame(
    regime_results
)


print(
    "\nBaseline performance by target regime:"
)

display(
    regime_baseline_results.round(4)
)


# =====================================================================
# 52. 19:30 PEAK BASELINES
# =====================================================================

peak_baseline_rows = []


if (
    "target_peak_1930"
    in
    forecast_samples.columns
):

    for split_name, split_frame in (
        split_frames.items()
    ):

        for pred_column, model_name in [

            (
                "lag1_peak_1930",
                "Peak_Naive_D1"
            ),

            (
                "lag7_peak_1930",
                "Peak_Naive_D7"
            )
        ]:

            if (
                pred_column
                not in
                split_frame.columns
            ):

                continue


            temp = (

                split_frame[
                    [
                        "target_peak_1930",
                        pred_column
                    ]
                ]

                .dropna()
            )


            if len(
                temp
            ) == 0:

                continue


            peak_baseline_rows.append(
                {

                    "split":
                        split_name,

                    "model":
                        model_name,

                    "samples":
                        len(
                            temp
                        ),

                    "MAE_MW":
                        mae(
                            temp[
                                "target_peak_1930"
                            ],
                            temp[
                                pred_column
                            ]
                        ),

                    "RMSE_MW":
                        rmse(
                            temp[
                                "target_peak_1930"
                            ],
                            temp[
                                pred_column
                            ]
                        ),

                    "MAPE_percent":
                        mape(
                            temp[
                                "target_peak_1930"
                            ],
                            temp[
                                pred_column
                            ]
                        ),

                    "sMAPE_percent":
                        smape(
                            temp[
                                "target_peak_1930"
                            ],
                            temp[
                                pred_column
                            ]
                        )
                }
            )


peak_baseline_results = pd.DataFrame(
    peak_baseline_rows
)


if len(
    peak_baseline_results
) > 0:

    print(
        "\nStrict 19:30 peak baselines:"
    )

    display(
        peak_baseline_results.round(4)
    )


# =====================================================================
# 53. TRAINING-PERIOD EDA PROFILE
# =====================================================================

train_dates = pd.DatetimeIndex(

    train_df[
        "target_date"
    ]
)


train_profile = (

    profile.loc[
        profile.index.isin(
            train_dates
        )
    ]
)


average_profile = pd.DataFrame(
    {

        "hour":
            HOURS,

        "mean_demand_mw":
            train_profile
            .mean(
                axis=0
            )
            .to_numpy(),

        "median_demand_mw":
            train_profile
            .median(
                axis=0
            )
            .to_numpy(),

        "std_demand_mw":
            train_profile
            .std(
                axis=0
            )
            .to_numpy()
    }
)


print(
    "\nAverage TRAINING-period hourly profile:"
)

display(
    average_profile.round(2)
)


# =====================================================================
# 54. DAILY EDA DATAFRAME
# =====================================================================

eda_daily = pd.DataFrame(
    index=profile.index
)


eda_daily[
    "daily_mean"
] = (
    profile.mean(
        axis=1
    )
)


eda_daily[
    "daily_max"
] = (
    profile.max(
        axis=1
    )
)


eda_daily[
    "daily_min"
] = (
    profile.min(
        axis=1
    )
)


eda_daily[
    "observed_slots"
] = (
    profile.notna().sum(
        axis=1
    )
)


eda_daily[
    "complete_day"
] = (

    eda_daily[
        "observed_slots"
    ]
    ==
    23
)


eda_daily["year"] = (
    eda_daily.index.year
)


eda_daily["month"] = (
    eda_daily.index.month
)


eda_daily["weekday"] = (
    eda_daily.index.dayofweek
)


eda_complete = (

    eda_daily.loc[
        eda_daily[
            "complete_day"
        ]
    ]

    .copy()
)


# =====================================================================
# 55. MONTHLY CHARACTERISTICS
# =====================================================================

monthly_profile = (

    eda_complete

    .groupby(
        "month"
    )

    .agg(

        mean_daily_load_mw=(
            "daily_mean",
            "mean"
        ),

        mean_daily_peak_mw=(
            "daily_max",
            "mean"
        ),

        number_of_days=(
            "daily_mean",
            "size"
        )
    )

    .reset_index()
)


print(
    "\nMonthly demand characteristics:"
)

display(
    monthly_profile.round(2)
)


# =====================================================================
# 56. WEEKDAY CHARACTERISTICS
# =====================================================================

weekday_profile = (

    eda_complete

    .groupby(
        "weekday"
    )

    .agg(

        mean_daily_load_mw=(
            "daily_mean",
            "mean"
        ),

        mean_daily_peak_mw=(
            "daily_max",
            "mean"
        ),

        number_of_days=(
            "daily_mean",
            "size"
        )
    )

    .reset_index()
)


WEEKDAY_NAMES = {

    0:
        "Monday",

    1:
        "Tuesday",

    2:
        "Wednesday",

    3:
        "Thursday",

    4:
        "Friday",

    5:
        "Saturday",

    6:
        "Sunday"
}


weekday_profile[
    "weekday_name"
] = (

    weekday_profile[
        "weekday"
    ]

    .map(
        WEEKDAY_NAMES
    )
)


print(
    "\nWeekday demand characteristics:"
)

display(
    weekday_profile.round(2)
)


# =====================================================================
# 57. YEARLY CHARACTERISTICS
# =====================================================================

yearly_summary = (

    eda_complete

    .groupby(
        "year"
    )

    .agg(

        mean_demand_mw=(
            "daily_mean",
            "mean"
        ),

        mean_peak_mw=(
            "daily_max",
            "mean"
        ),

        maximum_observed_peak_mw=(
            "daily_max",
            "max"
        ),

        complete_days=(
            "daily_mean",
            "size"
        )
    )

    .reset_index()
)


print(
    "\nYearly demand characteristics:"
)

display(
    yearly_summary.round(2)
)


# =====================================================================
# 58. SAVE DATASETS
# =====================================================================

ALL_SAMPLES_FILE = (

    FORECAST_DIR
    /
    "forecasting_samples_23slot.csv"
)


TRAIN_FILE = (

    FORECAST_DIR
    /
    "train_2015_2022.csv"
)


VALIDATION_FILE = (

    FORECAST_DIR
    /
    "validation_2023.csv"
)


TEST_FILE = (

    FORECAST_DIR
    /
    "test_2024_2026.csv"
)


forecast_samples.to_csv(
    ALL_SAMPLES_FILE,
    index=False
)


train_df.to_csv(
    TRAIN_FILE,
    index=False
)


validation_df.to_csv(
    VALIDATION_FILE,
    index=False
)


test_df.to_csv(
    TEST_FILE,
    index=False
)


# =====================================================================
# 59. SAVE RESULT TABLES
# =====================================================================

FEATURE_MANIFEST_FILE = (

    RESULT_DIR
    /
    "feature_manifest.csv"
)


FEATURE_GROUP_FILE = (

    RESULT_DIR
    /
    "feature_group_summary.csv"
)


SPLIT_SUMMARY_FILE = (

    RESULT_DIR
    /
    "split_summary.csv"
)


COMPLETENESS_FILE = (

    RESULT_DIR
    /
    "daily_profile_completeness.csv"
)


BASELINE_OVERALL_FILE = (

    RESULT_DIR
    /
    "naive_baseline_overall.csv"
)


BASELINE_HOUR_FILE = (

    RESULT_DIR
    /
    "naive_baseline_by_hour.csv"
)


REGIME_BASELINE_FILE = (

    RESULT_DIR
    /
    "naive_baseline_by_regime.csv"
)


PEAK_BASELINE_FILE = (

    RESULT_DIR
    /
    "naive_peak_baseline.csv"
)


AVERAGE_PROFILE_FILE = (

    RESULT_DIR
    /
    "average_hourly_profile.csv"
)


MONTHLY_PROFILE_FILE = (

    RESULT_DIR
    /
    "monthly_load_characteristics.csv"
)


WEEKDAY_PROFILE_FILE = (

    RESULT_DIR
    /
    "weekday_load_characteristics.csv"
)


YEARLY_SUMMARY_FILE = (

    RESULT_DIR
    /
    "yearly_load_summary.csv"
)


feature_manifest.to_csv(
    FEATURE_MANIFEST_FILE,
    index=False
)


feature_group_summary.to_csv(
    FEATURE_GROUP_FILE,
    index=False
)


split_summary.to_csv(
    SPLIT_SUMMARY_FILE,
    index=False
)


profile_completeness.reset_index().to_csv(
    COMPLETENESS_FILE,
    index=False
)


baseline_overall.to_csv(
    BASELINE_OVERALL_FILE,
    index=False
)


baseline_by_hour.to_csv(
    BASELINE_HOUR_FILE,
    index=False
)


regime_baseline_results.to_csv(
    REGIME_BASELINE_FILE,
    index=False
)


peak_baseline_results.to_csv(
    PEAK_BASELINE_FILE,
    index=False
)


average_profile.to_csv(
    AVERAGE_PROFILE_FILE,
    index=False
)


monthly_profile.to_csv(
    MONTHLY_PROFILE_FILE,
    index=False
)


weekday_profile.to_csv(
    WEEKDAY_PROFILE_FILE,
    index=False
)


yearly_summary.to_csv(
    YEARLY_SUMMARY_FILE,
    index=False
)


print(
    "\nForecasting datasets and result tables saved."
)


# =====================================================================
# 60. FIGURE 1
# AVERAGE NATIONAL DAILY LOAD PROFILE
# =====================================================================

plt.figure(
    figsize=(12, 6)
)


plt.plot(
    average_profile[
        "hour"
    ],

    average_profile[
        "mean_demand_mw"
    ],

    marker="o",

    linewidth=2,

    label="Mean demand"
)


plt.plot(
    average_profile[
        "hour"
    ],

    average_profile[
        "median_demand_mw"
    ],

    marker="s",

    linewidth=1.5,

    label="Median demand"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Hour of Day"
)


plt.ylabel(
    "Electricity Demand (MW)"
)


plt.title(
    "Average National Electricity Load Profile"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG1 = (

    FIGURE_DIR
    /
    "01_average_daily_load_profile.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 61. FIGURE 2
# DAILY MEAN AND DAILY PEAK
# =====================================================================

plt.figure(
    figsize=(17, 6)
)


plt.plot(
    eda_complete.index,

    eda_complete[
        "daily_mean"
    ],

    linewidth=0.8,

    alpha=0.80,

    label="Daily mean load"
)


plt.plot(
    eda_complete.index,

    eda_complete[
        "daily_max"
    ],

    linewidth=0.7,

    alpha=0.65,

    label="Daily peak load"
)


plt.xlabel(
    "Date"
)


plt.ylabel(
    "Electricity Demand (MW)"
)


plt.title(
    "Long-Term Evolution of Bangladesh National Electricity Demand"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG2 = (

    FIGURE_DIR
    /
    "02_daily_mean_and_peak_timeseries.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 62. FIGURE 3
# MONTHLY PATTERN
# =====================================================================

plt.figure(
    figsize=(11, 6)
)


plt.plot(
    monthly_profile[
        "month"
    ],

    monthly_profile[
        "mean_daily_load_mw"
    ],

    marker="o",

    linewidth=2,

    label="Mean daily load"
)


plt.plot(
    monthly_profile[
        "month"
    ],

    monthly_profile[
        "mean_daily_peak_mw"
    ],

    marker="s",

    linewidth=2,

    label="Mean daily peak"
)


plt.xticks(
    range(
        1,
        13
    )
)


plt.xlabel(
    "Month"
)


plt.ylabel(
    "Electricity Demand (MW)"
)


plt.title(
    "Seasonal Electricity Demand Pattern"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG3 = (

    FIGURE_DIR
    /
    "03_monthly_demand_pattern.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 63. FIGURE 4
# WEEKDAY PATTERN
# =====================================================================

plt.figure(
    figsize=(11, 6)
)


plt.plot(
    weekday_profile[
        "weekday"
    ],

    weekday_profile[
        "mean_daily_load_mw"
    ],

    marker="o",

    linewidth=2,

    label="Mean daily load"
)


plt.plot(
    weekday_profile[
        "weekday"
    ],

    weekday_profile[
        "mean_daily_peak_mw"
    ],

    marker="s",

    linewidth=2,

    label="Mean daily peak"
)


plt.xticks(
    weekday_profile[
        "weekday"
    ],

    weekday_profile[
        "weekday_name"
    ],

    rotation=30
)


plt.ylabel(
    "Electricity Demand (MW)"
)


plt.title(
    "Electricity Demand Pattern by Day of Week"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG4 = (

    FIGURE_DIR
    /
    "04_weekday_demand_pattern.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 64. FIGURE 5
# TEST BASELINE MAE BY TARGET HOUR
# =====================================================================

test_hour_metrics = (

    baseline_by_hour.loc[
        baseline_by_hour[
            "split"
        ]
        ==
        "TEST"
    ]

    .copy()
)


plt.figure(
    figsize=(12, 6)
)


for model_name in (

    test_hour_metrics[
        "model"
    ]
    .unique()
):

    temp = (

        test_hour_metrics.loc[
            test_hour_metrics[
                "model"
            ]
            ==
            model_name
        ]

        .sort_values(
            "hour"
        )
    )


    plt.plot(
        temp[
            "hour"
        ],

        temp[
            "MAE_MW"
        ],

        marker="o",

        linewidth=2,

        label=model_name
    )


plt.xticks(
    HOURS
)


plt.xlabel(
    "Forecast Target Hour"
)


plt.ylabel(
    "MAE (MW)"
)


plt.title(
    "Seasonal Naive Forecast Error Across the Next-Day Load Profile"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG5 = (

    FIGURE_DIR
    /
    "05_test_naive_error_by_hour.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 65. FIGURE 6
# EXAMPLE TEST-DAY FORECAST
# =====================================================================

if len(
    test_df
) > 0:

    example_position = (
        len(
            test_df
        )
        //
        2
    )


    example = (
        test_df.iloc[
            example_position
        ]
    )


    actual_values = np.array(
        [

            example[
                f"target_h{hour:02d}"
            ]

            for hour in HOURS
        ],
        dtype=float
    )


    d1_values = np.array(
        [

            example[
                f"lag1_h{hour:02d}"
            ]

            for hour in HOURS
        ],
        dtype=float
    )


    d7_values = np.array(
        [

            example[
                f"lag7_h{hour:02d}"
            ]

            for hour in HOURS
        ],
        dtype=float
    )


    plt.figure(
        figsize=(12, 6)
    )


    plt.plot(
        HOURS,
        actual_values,
        marker="o",
        linewidth=2.5,
        label="Actual target day"
    )


    plt.plot(
        HOURS,
        d1_values,
        marker="s",
        linewidth=1.6,
        label="Previous day"
    )


    plt.plot(
        HOURS,
        d7_values,
        marker="^",
        linewidth=1.6,
        label="Previous week"
    )


    plt.xticks(
        HOURS
    )


    plt.xlabel(
        "Hour of Day"
    )


    plt.ylabel(
        "Electricity Demand (MW)"
    )


    example_date = (

        pd.Timestamp(
            example[
                "target_date"
            ]
        )

        .date()
    )


    plt.title(
        f"Example Next-Day Load Profile: {example_date}"
    )


    plt.legend()


    plt.grid(
        alpha=0.25
    )


    plt.tight_layout()


    FIG6 = (

        FIGURE_DIR
        /
        "06_example_test_day_profile.png"
    )


    plt.savefig(
        FIG6,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()


# =====================================================================
# 66. FIGURE 7
# YEARLY MEAN AND PEAK LOAD
# =====================================================================

plt.figure(
    figsize=(12, 6)
)


plt.plot(
    yearly_summary[
        "year"
    ],

    yearly_summary[
        "mean_demand_mw"
    ],

    marker="o",

    linewidth=2,

    label="Mean demand"
)


plt.plot(
    yearly_summary[
        "year"
    ],

    yearly_summary[
        "mean_peak_mw"
    ],

    marker="s",

    linewidth=2,

    label="Mean daily peak"
)


plt.xlabel(
    "Year"
)


plt.ylabel(
    "Electricity Demand (MW)"
)


plt.title(
    "Long-Term Change in National Electricity Demand"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG7 = (

    FIGURE_DIR
    /
    "07_yearly_demand_trend.png"
)


plt.savefig(
    FIG7,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 67. CREATE STEP 2 SUMMARY JSON
# =====================================================================

step2_summary = {

    "study_start":
        str(
            profile.index.min()
        ),

    "study_end":
        str(
            profile.index.max()
        ),

    "complete_23slot_target_days":
        int(
            COMPLETE_TARGET_DAYS
        ),

    "final_forecasting_samples":
        int(
            len(
                forecast_samples
            )
        ),

    "train_samples":
        int(
            len(
                train_df
            )
        ),

    "validation_samples":
        int(
            len(
                validation_df
            )
        ),

    "test_samples":
        int(
            len(
                test_df
            )
        ),

    "input_feature_count":
        int(
            len(
                feature_columns
            )
        ),

    "output_target_count":
        int(
            len(
                target_columns
            )
        ),

    "target_hours":
        HOURS,

    "extreme_test_days":
        int(
            test_df[
                "target_extreme_event_day"
            ]
            .sum()
        ),

    "best_test_naive_model":
        str(
            best_test_naive[
                "model"
            ]
        ),

    "best_test_naive_MAE_MW":
        float(
            best_test_naive[
                "MAE_MW"
            ]
        ),

    "best_test_naive_RMSE_MW":
        float(
            best_test_naive[
                "RMSE_MW"
            ]
        ),

    "best_test_naive_MAPE_percent":
        float(
            best_test_naive[
                "MAPE_percent"
            ]
        ),

    "best_test_naive_sMAPE_percent":
        float(
            best_test_naive[
                "sMAPE_percent"
            ]
        )
}


STEP2_JSON = (

    RESULT_DIR
    /
    "step2_summary.json"
)


with open(
    STEP2_JSON,
    "w"
) as file:

    json.dump(
        step2_summary,
        file,
        indent=4
    )


# =====================================================================
# 68. FINAL VALIDATION
# =====================================================================

assert (
    len(
        forecast_samples
    )
    >
    0
)


assert (
    len(
        train_df
    )
    >
    0
)


assert (
    len(
        validation_df
    )
    >
    0
)


assert (
    len(
        test_df
    )
    >
    0
)


assert (

    forecast_samples[
        "target_date"
    ]
    .is_monotonic_increasing

)


assert (

    forecast_samples[
        target_columns
    ]
    .notna()
    .all()
    .all()

)


assert (

    forecast_samples[
        feature_columns
    ]
    .notna()
    .all()
    .all()

)


assert (

    set(
        feature_columns
    )
    .isdisjoint(
        set(
            target_columns
        )
    )

)


print(
    "\nAll Step 2 final validation checks PASSED."
)


# =====================================================================
# 69. FINAL OUTPUT SUMMARY
# =====================================================================

print("\n")
print("=" * 105)
print("STEP 2 COMPLETED SUCCESSFULLY")
print("=" * 105)


print(
    "\nComplete 23-slot target days:",
    f"{COMPLETE_TARGET_DAYS:,}"
)


print(
    "Final forecasting samples:",
    f"{len(forecast_samples):,}"
)


print(
    "\nTRAIN samples:",
    f"{len(train_df):,}"
)


print(
    "VALIDATION samples:",
    f"{len(validation_df):,}"
)


print(
    "TEST samples:",
    f"{len(test_df):,}"
)


print(
    "\nInput features:",
    len(
        feature_columns
    )
)


print(
    "Forecast outputs:",
    len(
        target_columns
    )
)


print(
    "Extreme-event TEST days:",
    int(
        test_df[
            "target_extreme_event_day"
        ]
        .sum()
    )
)


print("\n" + "-" * 105)
print("BEST TEST-SET NAIVE BASELINE")
print("-" * 105)


print(
    "Model:",
    best_test_naive[
        "model"
    ]
)


print(
    "MAE:",
    round(
        float(
            best_test_naive[
                "MAE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "RMSE:",
    round(
        float(
            best_test_naive[
                "RMSE_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "MAPE:",
    round(
        float(
            best_test_naive[
                "MAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "sMAPE:",
    round(
        float(
            best_test_naive[
                "sMAPE_percent"
            ]
        ),
        3
    ),
    "%"
)


print("\n" + "-" * 105)
print("FILES SAVED")
print("-" * 105)


print(
    "\n1. All forecasting samples:"
)

print(
    ALL_SAMPLES_FILE
)


print(
    "\n2. Training dataset:"
)

print(
    TRAIN_FILE
)


print(
    "\n3. Validation dataset:"
)

print(
    VALIDATION_FILE
)


print(
    "\n4. Test dataset:"
)

print(
    TEST_FILE
)


print(
    "\n5. Naive baseline results:"
)

print(
    BASELINE_OVERALL_FILE
)


print(
    "\n6. Hour-wise naive results:"
)

print(
    BASELINE_HOUR_FILE
)


print(
    "\n7. Figures:"
)

print(
    FIGURE_DIR
)


print(
    "\n8. Step 2 summary:"
)

print(
    STEP2_JSON
)


print("\n" + "=" * 105)

print(
    "SEND THE FINAL STEP 2 SUMMARY TO CHATGPT."
)

print(
    "ALSO SEND FIGURES 01, 02, AND 05."
)

print(
    "DO NOT START ML/DL MODEL TRAINING YET."
)

print("=" * 105)
```
