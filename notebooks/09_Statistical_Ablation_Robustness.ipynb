```python
# =====================================================================
# STEP 7
# STATISTICAL SIGNIFICANCE + FEATURE ABLATION + TEMPORAL ROBUSTNESS
# + BPDB EVENING_PEAK SENSITIVITY
#
# IMPORTANT:
# ----------
# Steps 1-6 are now FROZEN.
#
# This notebook DOES NOT:
#   - tune Ridge again
#   - select a new model
#   - change the test period
#   - change conformal window
#   - use BPDB for training
#
# Primary frozen point model:
#       Ridge alpha = 10
#
# Primary final point-model training:
#       2015-2022 + 2023
#
# Locked test:
#       2024-01-01 to 2026-03-08
#
# Main analyses:
#
# A. Reproduce frozen Ridge result
# B. Paired statistical significance
# C. 7-day + 14-day moving-block bootstrap
# D. HAC / Diebold-Mariano-style loss-difference tests
# E. Holm correction for multiple model comparisons
# F. Leave-one-feature-group-out Ridge ablation
# G. Temporal robustness
# H. BPDB Evening_Peak timing sensitivity
#
# Run the entire block in ONE Colab cell.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import re
import json
import math
import warnings

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

from scipy.stats import norm

from IPython.display import display


warnings.filterwarnings("ignore")

np.random.seed(42)


pd.set_option(
    "display.max_columns",
    300
)

pd.set_option(
    "display.max_rows",
    400
)

pd.set_option(
    "display.width",
    300
)


print("=" * 125)

print(
    "STEP 7: STATISTICAL SIGNIFICANCE + ABLATION + ROBUSTNESS"
)

print("=" * 125)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount(
    "/content/drive"
)


print(
    "\nGoogle Drive mounted successfully."
)


# =====================================================================
# 2. PROJECT DIRECTORIES
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)


PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)


STEP3_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "ml_baselines"
)


DEEP_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "deep_sequence"
)


EXTERNAL_RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "external_bpdb_validation"
)


RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "statistical_ablation_robustness"
)


FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "statistical_ablation_robustness"
)


RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =====================================================================
# 3. INPUT FILES
# =====================================================================

TRAIN_FILE = (
    FORECAST_DIR
    / "train_2015_2022.csv"
)


VALIDATION_FILE = (
    FORECAST_DIR
    / "validation_2023.csv"
)


TEST_FILE = (
    FORECAST_DIR
    / "test_2024_2026.csv"
)


ML_PREDICTION_FILE = (
    STEP3_RESULT_DIR
    / "all_test_predictions.csv"
)


DEEP_PREDICTION_FILE = (
    DEEP_RESULT_DIR
    / "deep_test_predictions.csv"
)


PGCB_AUDIT_FILE = (
    PROCESSED_DIR
    / "pgcb_all_rows_quality_flagged.csv"
)


BPDB_FILE = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "bpdb"
    / "BPDB_version_2.xlsx"
)


required_files = [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE,
    PGCB_AUDIT_FILE,
    BPDB_FILE
]


print("\n" + "=" * 125)

print(
    "REQUIRED FILE CHECK"
)

print("=" * 125)


for path in required_files:

    print(
        path,
        "->",
        path.exists()
    )


    if not path.exists():

        raise FileNotFoundError(
            f"""
Required file does not exist:

{path}
"""
        )


print(
    "\nAll core files found."
)


# =====================================================================
# 4. LOAD TRAIN / VALIDATION / TEST
# =====================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)


validation_df = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)


test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for dataframe in [

    train_df,
    validation_df,
    test_df
]:

    dataframe[
        "target_date"
    ] = pd.to_datetime(

        dataframe[
            "target_date"
        ],

        errors="coerce"
    )


    dataframe.sort_values(
        "target_date",
        inplace=True
    )


    dataframe.reset_index(
        drop=True,
        inplace=True
    )


print(
    "\nTRAIN:",
    len(train_df),
    train_df["target_date"].min(),
    "to",
    train_df["target_date"].max()
)


print(
    "VALIDATION:",
    len(validation_df),
    validation_df["target_date"].min(),
    "to",
    validation_df["target_date"].max()
)


print(
    "TEST:",
    len(test_df),
    test_df["target_date"].min(),
    "to",
    test_df["target_date"].max()
)


# =====================================================================
# 5. FORECAST HOURS AND TARGETS
# =====================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


assert len(
    TARGET_COLUMNS
) == 23


# =====================================================================
# 6. EXACT FROZEN FEATURE GROUPS
# =====================================================================

CALENDAR_FEATURES = [

    "year",
    "month",
    "day",
    "dayofweek",
    "dayofyear",
    "weekofyear",
    "quarter",
    "is_weekend",
    "month_sin",
    "month_cos",
    "dow_sin",
    "dow_cos",
    "doy_sin",
    "doy_cos"
]


PREVIOUS_DAY_PROFILE = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


PREVIOUS_WEEK_PROFILE = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


ROLLING7_MEAN = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


ROLLING28_MEAN = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


ROLLING7_STD = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


ROLLING28_STD = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


HISTORICAL_SUMMARY = [

    "lag1_daily_mean",
    "lag1_daily_max",
    "lag1_daily_min",
    "lag1_daily_range",
    "lag1_peak_hour",

    "lag7_daily_mean",
    "lag7_daily_max",
    "lag7_daily_min",
    "lag7_daily_range",
    "lag7_peak_hour",

    "mean_change_vs_last_week",
    "max_change_vs_last_week",
    "mean_ratio_vs_last_week"
]


FEATURE_GROUPS = {

    "Previous_Day_Profile":
        PREVIOUS_DAY_PROFILE,

    "Previous_Week_Profile":
        PREVIOUS_WEEK_PROFILE,

    "Rolling7_Mean":
        ROLLING7_MEAN,

    "Rolling28_Mean":
        ROLLING28_MEAN,

    "Rolling7_STD":
        ROLLING7_STD,

    "Rolling28_STD":
        ROLLING28_STD,

    "Calendar":
        CALENDAR_FEATURES,

    "Historical_Summary":
        HISTORICAL_SUMMARY
}


FEATURE_COLUMNS = []


for group_columns in (
    FEATURE_GROUPS.values()
):

    FEATURE_COLUMNS.extend(
        group_columns
    )


assert (
    len(
        FEATURE_COLUMNS
    )
    ==
    165
), (
    f"Expected 165 features; found "
    f"{len(FEATURE_COLUMNS)}"
)


assert (
    len(
        set(
            FEATURE_COLUMNS
        )
    )
    ==
    165
)


assert set(
    FEATURE_COLUMNS
).isdisjoint(
    set(
        TARGET_COLUMNS
    )
)


print(
    "\nFrozen feature count:",
    len(
        FEATURE_COLUMNS
    )
)


print(
    "Target count:",
    len(
        TARGET_COLUMNS
    )
)


print(
    "\nFeature groups:"
)


for group_name, columns in (
    FEATURE_GROUPS.items()
):

    print(
        f"{group_name:<25}: {len(columns)}"
    )


# =====================================================================
# 7. VALIDATE COLUMNS AND FINITE VALUES
# =====================================================================

for dataset_name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION",
        validation_df
    ),

    (
        "TEST",
        test_df
    )
]:

    missing_features = [

        column

        for column in FEATURE_COLUMNS

        if column
        not in
        dataframe.columns
    ]


    missing_targets = [

        column

        for column in TARGET_COLUMNS

        if column
        not in
        dataframe.columns
    ]


    if missing_features:

        raise ValueError(
            f"""
{dataset_name} missing features:

{missing_features}
"""
        )


    if missing_targets:

        raise ValueError(
            f"""
{dataset_name} missing targets:

{missing_targets}
"""
        )


    values = (

        dataframe[
            FEATURE_COLUMNS
            +
            TARGET_COLUMNS
        ]

        .to_numpy(
            dtype=float
        )
    )


    if not np.isfinite(
        values
    ).all():

        raise ValueError(
            f"""
Non-finite values detected in {dataset_name}.
"""
        )


print(
    "\nFeature/target integrity checks PASSED."
)


# =====================================================================
# 8. METRIC FUNCTIONS
# =====================================================================

def mae(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.mean(

            np.abs(

                y_true

                -

                y_pred
            )
        )
    )


def rmse(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.sqrt(

            np.mean(

                (
                    y_true

                    -

                    y_pred
                )

                ** 2
            )
        )
    )


def mape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            y_true
        ),

        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    y_true

                    -

                    y_pred
                )

                /

                denominator
            )
        )
    )


def smape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = (

        np.abs(
            y_true
        )

        +

        np.abs(
            y_pred
        )
    )


    valid = (
        denominator > 0
    )


    return float(

        100

        *

        np.mean(

            2

            *

            np.abs(

                y_true[
                    valid
                ]

                -

                y_pred[
                    valid
                ]
            )

            /

            denominator[
                valid
            ]
        )
    )


def metric_dict(
    y_true,
    y_pred
):

    return {

        "MAE_MW":
            mae(
                y_true,
                y_pred
            ),

        "RMSE_MW":
            rmse(
                y_true,
                y_pred
            ),

        "MAPE_percent":
            mape(
                y_true,
                y_pred
            ),

        "sMAPE_percent":
            smape(
                y_true,
                y_pred
            )
    }


# =====================================================================
# 9. PREPARE MATRICES
# =====================================================================

X_train = (

    train_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_train = (

    train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


X_val = (

    validation_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_val = (

    validation_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


X_test = (

    test_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_test = (

    test_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


PRETEST_DF = pd.concat(

    [
        train_df,
        validation_df
    ],

    ignore_index=True
)


PRETEST_DF = (

    PRETEST_DF

    .sort_values(
        "target_date"
    )

    .reset_index(
        drop=True
    )
)


# =====================================================================
# 10. REPRODUCE FROZEN RIDGE MODEL
# =====================================================================

FROZEN_ALPHA = 10.0


X_pretest = (

    PRETEST_DF[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_pretest = (

    PRETEST_DF[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


frozen_scaler = StandardScaler()


X_pretest_scaled = (

    frozen_scaler
    .fit_transform(
        X_pretest
    )
)


X_test_scaled = (

    frozen_scaler
    .transform(
        X_test
    )
)


frozen_ridge = Ridge(
    alpha=
        FROZEN_ALPHA
)


frozen_ridge.fit(
    X_pretest_scaled,
    Y_pretest
)


RIDGE_PRED = frozen_ridge.predict(
    X_test_scaled
)


ridge_metrics = metric_dict(
    Y_test,
    RIDGE_PRED
)


print("\n" + "=" * 125)

print(
    "FROZEN RIDGE REPRODUCTION"
)

print("=" * 125)


print(
    "Alpha:",
    FROZEN_ALPHA
)


print(
    "MAE:",
    round(
        ridge_metrics[
            "MAE_MW"
        ],
        4
    ),
    "MW"
)


print(
    "RMSE:",
    round(
        ridge_metrics[
            "RMSE_MW"
        ],
        4
    ),
    "MW"
)


print(
    "MAPE:",
    round(
        ridge_metrics[
            "MAPE_percent"
        ],
        4
    ),
    "%"
)


EXPECTED_STEP3_RIDGE_MAE = (
    437.1586
)


reproduction_difference = (

    ridge_metrics[
        "MAE_MW"
    ]

    -

    EXPECTED_STEP3_RIDGE_MAE
)


print(
    "\nDifference from Step 3 reported Ridge MAE:",
    round(
        reproduction_difference,
        6
    ),
    "MW"
)


if abs(
    reproduction_difference
) > 1.0:

    print(
        "\nWARNING:"
    )

    print(
        "Recomputed Ridge differs from Step 3 by >1 MW."
    )

    print(
        "Continue only as a robustness reproduction, "
        "not as a replacement of Step 3."
    )


else:

    print(
        "\nFrozen Ridge reproduction PASSED."
    )


# =====================================================================
# 11. RECOMPUTE D1 AND D7 BASELINES DIRECTLY FROM FORECAST FEATURES
# =====================================================================

D1_COLUMNS = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


D7_COLUMNS = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


D1_PRED = (

    test_df[
        D1_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


D7_PRED = (

    test_df[
        D7_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


print(
    "\nD1 test MAE:",
    round(
        mae(
            Y_test,
            D1_PRED
        ),
        4
    ),
    "MW"
)


print(
    "D7 test MAE:",
    round(
        mae(
            Y_test,
            D7_PRED
        ),
        4
    ),
    "MW"
)


# =====================================================================
# 12. MODEL-PREDICTION STORE
# =====================================================================

MODEL_PREDICTIONS = {

    "Ridge":
        RIDGE_PRED,

    "D1":
        D1_PRED,

    "D7":
        D7_PRED
}


# =====================================================================
# 13. OPTIONAL PREDICTION EXTRACTION FROM EXISTING RESULT FILES
#
# This does NOT train or tune anything.
#
# If ET/XGB/LSTM/GRU/etc. predictions are present, they are added only
# for paired statistical comparison.
# =====================================================================

def prepare_prediction_source(
    file_path
):

    if not file_path.exists():

        return None


    dataframe = pd.read_csv(
        file_path,
        low_memory=False
    )


    date_column = None


    for candidate in [

        "target_date",
        "date"
    ]:

        if candidate in dataframe.columns:

            date_column = candidate

            break


    if (
        date_column
        is None
    ):

        return None


    dataframe[
        date_column
    ] = pd.to_datetime(

        dataframe[
            date_column
        ],

        errors="coerce"
    )


    dataframe = (

        dataframe
        .dropna(
            subset=[
                date_column
            ]
        )

        .drop_duplicates(
            subset=[
                date_column
            ]
        )

        .set_index(
            date_column
        )
    )


    test_dates = pd.DatetimeIndex(

        test_df[
            "target_date"
        ]
    )


    missing_dates = (

        test_dates
        .difference(
            dataframe.index
        )
    )


    if len(
        missing_dates
    ) > 0:

        return None


    dataframe = (

        dataframe
        .reindex(
            test_dates
        )
    )


    return dataframe


def extract_keyword_prediction_matrix(
    dataframe,
    keywords
):

    if dataframe is None:

        return None


    selected_columns = []


    for hour in HOURS:

        hour_pattern = re.compile(

            rf"h0?{hour}$",

            flags=re.IGNORECASE
        )


        candidates = []


        for column in dataframe.columns:

            lower = str(
                column
            ).lower()


            if (
                any(
                    keyword.lower()
                    in
                    lower

                    for keyword
                    in
                    keywords
                )

                and

                hour_pattern.search(
                    lower
                )

                and

                "actual"
                not in
                lower

                and

                "target"
                not in
                lower
            ):

                candidates.append(
                    column
                )


        if len(
            candidates
        ) == 0:

            return None


        candidates = sorted(

            candidates,

            key=lambda column: (

                0
                if "pred"
                in
                str(column).lower()
                else
                1,

                len(
                    str(
                        column
                    )
                )
            )
        )


        selected_columns.append(
            candidates[0]
        )


    if len(
        selected_columns
    ) != 23:

        return None


    matrix = (

        dataframe[
            selected_columns
        ]

        .to_numpy(
            dtype=float
        )
    )


    if matrix.shape != (
        len(
            test_df
        ),
        23
    ):

        return None


    if not np.isfinite(
        matrix
    ).all():

        return None


    return matrix


ml_prediction_source = (
    prepare_prediction_source(
        ML_PREDICTION_FILE
    )
)


deep_prediction_source = (
    prepare_prediction_source(
        DEEP_PREDICTION_FILE
    )
)


OPTIONAL_MODELS = [

    (
        "ExtraTrees",
        ml_prediction_source,
        [
            "extra",
            "extratrees",
            "extra_trees"
        ]
    ),

    (
        "XGBoost",
        ml_prediction_source,
        [
            "xgb",
            "xgboost"
        ]
    ),

    (
        "LSTM",
        deep_prediction_source,
        [
            "lstm"
        ]
    ),

    (
        "GRU",
        deep_prediction_source,
        [
            "gru"
        ]
    ),

    (
        "Transformer",
        deep_prediction_source,
        [
            "transformer"
        ]
    ),

    (
        "TemporalCNN",
        deep_prediction_source,
        [
            "cnn",
            "temporalcnn",
            "temporal_cnn"
        ]
    )
]


print("\n" + "=" * 125)

print(
    "OPTIONAL SAVED MODEL PREDICTION DETECTION"
)

print("=" * 125)


for (
    model_name,
    source,
    keywords
) in OPTIONAL_MODELS:

    matrix = (

        extract_keyword_prediction_matrix(
            source,
            keywords
        )
    )


    if matrix is not None:

        MODEL_PREDICTIONS[
            model_name
        ] = matrix


        print(
            model_name,
            "-> detected"
        )


    else:

        print(
            model_name,
            "-> not detected; skipped"
        )


print(
    "\nModels available for statistical comparison:"
)

print(
    list(
        MODEL_PREDICTIONS.keys()
    )
)


# =====================================================================
# 14. DAILY LOSS SERIES
#
# Each day contributes ONE loss observation:
#
# mean absolute error across the 23 target slots.
#
# This avoids pretending that the 23 within-day horizons are
# statistically independent observations.
# =====================================================================

DAILY_LOSSES = {}


for model_name, prediction in (
    MODEL_PREDICTIONS.items()
):

    DAILY_LOSSES[
        model_name
    ] = (

        np.mean(

            np.abs(

                Y_test

                -

                prediction
            ),

            axis=1
        )
    )


# =====================================================================
# 15. CIRCULAR MOVING-BLOCK BOOTSTRAP
# =====================================================================

def block_bootstrap_mean_ci(
    values,
    block_length=7,
    n_bootstrap=5000,
    seed=42
):

    values = np.asarray(
        values,
        dtype=float
    )


    values = values[
        np.isfinite(
            values
        )
    ]


    n = len(
        values
    )


    if n == 0:

        return (
            np.nan,
            np.nan,
            np.nan
        )


    rng = np.random.default_rng(
        seed
    )


    number_of_blocks = int(

        np.ceil(

            n

            /

            block_length
        )
    )


    bootstrap_means = np.empty(
        n_bootstrap,
        dtype=float
    )


    offsets = np.arange(
        block_length
    )


    for iteration in range(
        n_bootstrap
    ):

        starts = rng.integers(

            0,

            n,

            size=
                number_of_blocks
        )


        indices = np.concatenate(
            [

                (
                    start

                    +

                    offsets
                )

                %
                n

                for start in starts
            ]
        )[
            :n
        ]


        bootstrap_means[
            iteration
        ] = float(

            np.mean(

                values[
                    indices
                ]
            )
        )


    lower = float(

        np.quantile(
            bootstrap_means,
            0.025
        )
    )


    upper = float(

        np.quantile(
            bootstrap_means,
            0.975
        )
    )


    observed = float(
        np.mean(
            values
        )
    )


    return (
        observed,
        lower,
        upper
    )


# =====================================================================
# 16. HAC / DM-STYLE MEAN LOSS-DIFFERENCE TEST
#
# Positive differential:
#
# comparator loss - Ridge loss > 0
#
# means Ridge is better.
#
# Bartlett/Newey-West long-run variance estimator.
# =====================================================================

def hac_loss_difference_test(
    difference,
    max_lag=7
):

    d = np.asarray(
        difference,
        dtype=float
    )


    d = d[
        np.isfinite(
            d
        )
    ]


    n = len(
        d
    )


    if n < 20:

        return {

            "mean_difference":
                np.nan,

            "hac_se":
                np.nan,

            "z_stat":
                np.nan,

            "p_value":
                np.nan,

            "max_lag":
                max_lag
        }


    mean_d = float(
        np.mean(
            d
        )
    )


    centered = (

        d

        -

        mean_d
    )


    gamma0 = float(

        np.dot(

            centered,

            centered
        )

        /
        n
    )


    long_run_variance = (
        gamma0
    )


    for lag in range(
        1,
        max_lag + 1
    ):

        covariance = float(

            np.dot(

                centered[
                    lag:
                ],

                centered[
                    :-lag
                ]
            )

            /
            n
        )


        weight = (

            1

            -

            lag

            /
            (
                max_lag
                +
                1
            )
        )


        long_run_variance += (

            2

            *
            weight

            *
            covariance
        )


    long_run_variance = max(

        long_run_variance,

        1e-12
    )


    hac_se = float(

        np.sqrt(

            long_run_variance

            /
            n
        )
    )


    z_stat = (

        mean_d

        /
        hac_se
    )


    p_value = float(

        2

        *
        (
            1

            -
            norm.cdf(

                abs(
                    z_stat
                )
            )
        )
    )


    return {

        "mean_difference":
            mean_d,

        "hac_se":
            hac_se,

        "z_stat":
            float(
                z_stat
            ),

        "p_value":
            p_value,

        "max_lag":
            max_lag
    }


# =====================================================================
# 17. HOLM MULTIPLE-COMPARISON CORRECTION
# =====================================================================

def holm_adjust(
    p_values
):

    p_values = np.asarray(
        p_values,
        dtype=float
    )


    m = len(
        p_values
    )


    order = np.argsort(
        p_values
    )


    adjusted = np.empty(
        m,
        dtype=float
    )


    running_max = 0.0


    for rank, index in enumerate(
        order
    ):

        multiplier = (

            m

            -

            rank
        )


        value = min(

            1.0,

            multiplier

            *
            p_values[
                index
            ]
        )


        running_max = max(

            running_max,

            value
        )


        adjusted[
            index
        ] = running_max


    return adjusted


# =====================================================================
# 18. STATISTICAL COMPARISON: RIDGE VS EACH COMPARATOR
# =====================================================================

RIDGE_DAILY = (
    DAILY_LOSSES[
        "Ridge"
    ]
)


statistical_rows = []


for model_name in (
    MODEL_PREDICTIONS.keys()
):

    if model_name == "Ridge":

        continue


    comparator_daily = (

        DAILY_LOSSES[
            model_name
        ]
    )


    difference = (

        comparator_daily

        -

        RIDGE_DAILY
    )


    comparator_mae = float(

        np.mean(
            comparator_daily
        )
    )


    ridge_daily_mae = float(

        np.mean(
            RIDGE_DAILY
        )
    )


    skill = (

        100

        *
        (
            comparator_mae

            -

            ridge_daily_mae
        )

        /
        comparator_mae
    )


    observed7, lower7, upper7 = (

        block_bootstrap_mean_ci(

            difference,

            block_length=
                7,

            n_bootstrap=
                5000,

            seed=
                42
        )
    )


    observed14, lower14, upper14 = (

        block_bootstrap_mean_ci(

            difference,

            block_length=
                14,

            n_bootstrap=
                5000,

            seed=
                142
        )
    )


    hac7 = (

        hac_loss_difference_test(

            difference,

            max_lag=
                7
        )
    )


    hac14 = (

        hac_loss_difference_test(

            difference,

            max_lag=
                14
        )
    )


    ridge_wins = (

        100

        *
        float(

            np.mean(

                RIDGE_DAILY

                <

                comparator_daily
            )
        )
    )


    ties = (

        100

        *
        float(

            np.mean(

                np.isclose(

                    RIDGE_DAILY,

                    comparator_daily,

                    atol=
                        1e-9
                )
            )
        )
    )


    statistical_rows.append(
        {

            "Comparator":
                model_name,

            "Ridge_MAE_MW":
                ridge_daily_mae,

            "Comparator_MAE_MW":
                comparator_mae,

            "Ridge_Skill_percent":
                skill,

            "Mean_Daily_MAE_Improvement_MW":
                observed7,

            "Bootstrap7_CI_Lower_MW":
                lower7,

            "Bootstrap7_CI_Upper_MW":
                upper7,

            "Bootstrap14_CI_Lower_MW":
                lower14,

            "Bootstrap14_CI_Upper_MW":
                upper14,

            "HAC7_z":
                hac7[
                    "z_stat"
                ],

            "HAC7_p_raw":
                hac7[
                    "p_value"
                ],

            "HAC14_z":
                hac14[
                    "z_stat"
                ],

            "HAC14_p_raw":
                hac14[
                    "p_value"
                ],

            "Ridge_Wins_Days_percent":
                ridge_wins,

            "Tie_Days_percent":
                ties
        }
    )


statistical_results = pd.DataFrame(
    statistical_rows
)


if len(
    statistical_results
) > 0:

    statistical_results[
        "HAC7_p_Holm"
    ] = holm_adjust(

        statistical_results[
            "HAC7_p_raw"
        ].to_numpy()
    )


    statistical_results[
        "HAC14_p_Holm"
    ] = holm_adjust(

        statistical_results[
            "HAC14_p_raw"
        ].to_numpy()
    )


    statistical_results[
        "Bootstrap7_Excludes_Zero"
    ] = (

        (
            statistical_results[
                "Bootstrap7_CI_Lower_MW"
            ]
            >
            0
        )

        |

        (
            statistical_results[
                "Bootstrap7_CI_Upper_MW"
            ]
            <
            0
        )
    )


    statistical_results[
        "Bootstrap14_Excludes_Zero"
    ] = (

        (
            statistical_results[
                "Bootstrap14_CI_Lower_MW"
            ]
            >
            0
        )

        |

        (
            statistical_results[
                "Bootstrap14_CI_Upper_MW"
            ]
            <
            0
        )
    )


    statistical_results = (

        statistical_results

        .sort_values(
            "Comparator_MAE_MW"
        )

        .reset_index(
            drop=True
        )
    )


print("\n" + "=" * 125)

print(
    "PAIRED STATISTICAL SIGNIFICANCE RESULTS"
)

print(
    "Positive improvement = Ridge has lower daily MAE."
)

print("=" * 125)


display(
    statistical_results.round(
        5
    )
)


# =====================================================================
# 19. PRIMARY RIDGE VS D1 RESULT
# =====================================================================

ridge_vs_d1 = (

    statistical_results.loc[

        statistical_results[
            "Comparator"
        ]
        ==
        "D1"
    ]
)


if len(
    ridge_vs_d1
) != 1:

    raise RuntimeError(
        """
Primary Ridge-vs-D1 statistical comparison missing.
"""
    )


ridge_vs_d1 = (
    ridge_vs_d1.iloc[0]
)


print("\n" + "=" * 125)

print(
    "PRIMARY SIGNIFICANCE RESULT: RIDGE VS D1"
)

print("=" * 125)


print(
    "Ridge MAE:",
    round(
        ridge_vs_d1[
            "Ridge_MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "D1 MAE:",
    round(
        ridge_vs_d1[
            "Comparator_MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "MAE skill:",
    round(
        ridge_vs_d1[
            "Ridge_Skill_percent"
        ],
        3
    ),
    "%"
)


print(
    "Mean daily improvement:",
    round(
        ridge_vs_d1[
            "Mean_Daily_MAE_Improvement_MW"
        ],
        3
    ),
    "MW"
)


print(
    "7-day block-bootstrap 95% CI:",
    round(
        ridge_vs_d1[
            "Bootstrap7_CI_Lower_MW"
        ],
        3
    ),
    "to",
    round(
        ridge_vs_d1[
            "Bootstrap7_CI_Upper_MW"
        ],
        3
    ),
    "MW"
)


print(
    "14-day block-bootstrap 95% CI:",
    round(
        ridge_vs_d1[
            "Bootstrap14_CI_Lower_MW"
        ],
        3
    ),
    "to",
    round(
        ridge_vs_d1[
            "Bootstrap14_CI_Upper_MW"
        ],
        3
    ),
    "MW"
)


print(
    "HAC-7 Holm-adjusted p:",
    ridge_vs_d1[
        "HAC7_p_Holm"
    ]
)


# =====================================================================
# 20. FEATURE-GROUP ABLATION
#
# IMPORTANT:
#
# We keep alpha = 10 for every ablation.
#
# We DO NOT retune the hyperparameter for each ablated model.
#
# This makes the ablation an explanatory robustness analysis rather
# than a new model-selection exercise.
# =====================================================================

def fit_ridge_subset(
    train_dataframe,
    eval_dataframe,
    selected_features,
    alpha=10.0
):

    X_fit = (

        train_dataframe[
            selected_features
        ]

        .to_numpy(
            dtype=float
        )
    )


    Y_fit = (

        train_dataframe[
            TARGET_COLUMNS
        ]

        .to_numpy(
            dtype=float
        )
    )


    X_eval = (

        eval_dataframe[
            selected_features
        ]

        .to_numpy(
            dtype=float
        )
    )


    scaler = StandardScaler()


    X_fit_scaled = (

        scaler
        .fit_transform(
            X_fit
        )
    )


    X_eval_scaled = (

        scaler
        .transform(
            X_eval
        )
    )


    model = Ridge(
        alpha=
            alpha
    )


    model.fit(
        X_fit_scaled,
        Y_fit
    )


    prediction = model.predict(
        X_eval_scaled
    )


    return prediction


# =====================================================================
# 21. FULL-FEATURE VALIDATION REFERENCE
# =====================================================================

FULL_VALIDATION_PRED = (

    fit_ridge_subset(

        train_dataframe=
            train_df,

        eval_dataframe=
            validation_df,

        selected_features=
            FEATURE_COLUMNS,

        alpha=
            FROZEN_ALPHA
    )
)


FULL_VALIDATION_METRICS = (

    metric_dict(

        Y_val,

        FULL_VALIDATION_PRED
    )
)


FULL_TEST_METRICS = (

    metric_dict(

        Y_test,

        RIDGE_PRED
    )
)


ablation_rows = []


ablation_rows.append(
    {

        "Ablation":
            "FULL_MODEL",

        "Removed_Group":
            "NONE",

        "Features_Remaining":
            165,

        "Validation_MAE_MW":
            FULL_VALIDATION_METRICS[
                "MAE_MW"
            ],

        "Validation_RMSE_MW":
            FULL_VALIDATION_METRICS[
                "RMSE_MW"
            ],

        "Test_MAE_MW":
            FULL_TEST_METRICS[
                "MAE_MW"
            ],

        "Test_RMSE_MW":
            FULL_TEST_METRICS[
                "RMSE_MW"
            ],

        "Test_MAPE_percent":
            FULL_TEST_METRICS[
                "MAPE_percent"
            ],

        "Delta_Test_MAE_vs_Full_MW":
            0.0,

        "Relative_Test_MAE_Change_percent":
            0.0
    }
)


# =====================================================================
# 22. LEAVE-ONE-GROUP-OUT ABLATION LOOP
# =====================================================================

print("\n" + "=" * 125)

print(
    "RUNNING LEAVE-ONE-FEATURE-GROUP-OUT ABLATION"
)

print("=" * 125)


for group_name, group_columns in (
    FEATURE_GROUPS.items()
):

    selected_features = [

        feature

        for feature in FEATURE_COLUMNS

        if feature
        not in
        group_columns
    ]


    # -------------------------------------------------------------
    # Validation:
    # train = 2015-2022
    # eval  = 2023
    # -------------------------------------------------------------

    validation_prediction = (

        fit_ridge_subset(

            train_dataframe=
                train_df,

            eval_dataframe=
                validation_df,

            selected_features=
                selected_features,

            alpha=
                FROZEN_ALPHA
        )
    )


    validation_metrics = (

        metric_dict(

            Y_val,

            validation_prediction
        )
    )


    # -------------------------------------------------------------
    # Locked-test diagnostic:
    # train = 2015-2023
    # eval  = 2024-Mar2026
    #
    # This is an explanatory post-hoc ablation.
    # It cannot replace the frozen full model.
    # -------------------------------------------------------------

    test_prediction = (

        fit_ridge_subset(

            train_dataframe=
                PRETEST_DF,

            eval_dataframe=
                test_df,

            selected_features=
                selected_features,

            alpha=
                FROZEN_ALPHA
        )
    )


    test_metrics = (

        metric_dict(

            Y_test,

            test_prediction
        )
    )


    delta = (

        test_metrics[
            "MAE_MW"
        ]

        -

        FULL_TEST_METRICS[
            "MAE_MW"
        ]
    )


    relative_delta = (

        100

        *
        delta

        /
        FULL_TEST_METRICS[
            "MAE_MW"
        ]
    )


    ablation_rows.append(
        {

            "Ablation":
                f"Remove_{group_name}",

            "Removed_Group":
                group_name,

            "Features_Remaining":
                len(
                    selected_features
                ),

            "Validation_MAE_MW":
                validation_metrics[
                    "MAE_MW"
                ],

            "Validation_RMSE_MW":
                validation_metrics[
                    "RMSE_MW"
                ],

            "Test_MAE_MW":
                test_metrics[
                    "MAE_MW"
                ],

            "Test_RMSE_MW":
                test_metrics[
                    "RMSE_MW"
                ],

            "Test_MAPE_percent":
                test_metrics[
                    "MAPE_percent"
                ],

            "Delta_Test_MAE_vs_Full_MW":
                delta,

            "Relative_Test_MAE_Change_percent":
                relative_delta
        }
    )


    print(
        f"{group_name:<25} "
        f"Val MAE={validation_metrics['MAE_MW']:.2f} "
        f"Test MAE={test_metrics['MAE_MW']:.2f} "
        f"Delta={delta:+.2f} MW"
    )


ablation_results = pd.DataFrame(
    ablation_rows
)


ablation_results = (

    ablation_results

    .sort_values(
        "Delta_Test_MAE_vs_Full_MW",

        ascending=False
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 125)

print(
    "FEATURE-GROUP ABLATION RESULTS"
)

print("=" * 125)


display(
    ablation_results.round(
        4
    )
)


# =====================================================================
# 23. TEMPORAL ROBUSTNESS
# =====================================================================

TEST_DATES = (

    test_df[
        "target_date"
    ]
)


TEMPORAL_SPLITS = {

    "ALL_TEST_2024_2026":
        np.ones(
            len(
                test_df
            ),
            dtype=bool
        ),

    "2024":
        (
            TEST_DATES.dt.year
            ==
            2024
        )
        .to_numpy(),

    "2025":
        (
            TEST_DATES.dt.year
            ==
            2025
        )
        .to_numpy(),

    "2026_PARTIAL_to_Mar08":
        (
            TEST_DATES.dt.year
            ==
            2026
        )
        .to_numpy()
}


temporal_rows = []


for split_name, mask in (
    TEMPORAL_SPLITS.items()
):

    if (
        mask.sum()
        ==
        0
    ):

        continue


    for model_name, prediction in (
        MODEL_PREDICTIONS.items()
    ):

        metrics = (

            metric_dict(

                Y_test[
                    mask
                ],

                prediction[
                    mask
                ]
            )
        )


        temporal_rows.append(
            {

                "Period":
                    split_name,

                "Model":
                    model_name,

                "Days":
                    int(
                        mask.sum()
                    ),

                **metrics
            }
        )


temporal_results = pd.DataFrame(
    temporal_rows
)


# =====================================================================
# 24. RIDGE TEMPORAL SKILL VS D1
# =====================================================================

temporal_skill_rows = []


for split_name, mask in (
    TEMPORAL_SPLITS.items()
):

    if (
        mask.sum()
        ==
        0
    ):

        continue


    ridge_mae_split = (

        mae(

            Y_test[
                mask
            ],

            RIDGE_PRED[
                mask
            ]
        )
    )


    d1_mae_split = (

        mae(

            Y_test[
                mask
            ],

            D1_PRED[
                mask
            ]
        )
    )


    skill = (

        100

        *
        (
            d1_mae_split

            -

            ridge_mae_split
        )

        /
        d1_mae_split
    )


    temporal_skill_rows.append(
        {

            "Period":
                split_name,

            "Days":
                int(
                    mask.sum()
                ),

            "Ridge_MAE_MW":
                ridge_mae_split,

            "D1_MAE_MW":
                d1_mae_split,

            "Ridge_Skill_vs_D1_percent":
                skill
        }
    )


temporal_skill = pd.DataFrame(
    temporal_skill_rows
)


print("\n" + "=" * 125)

print(
    "TEMPORAL ROBUSTNESS: RIDGE VS D1"
)

print("=" * 125)


display(
    temporal_skill.round(
        4
    )
)


# =====================================================================
# 25. MONTH-OF-YEAR ROBUSTNESS
#
# Descriptive only.
# =====================================================================

monthly_rows = []


for month in range(
    1,
    13
):

    mask = (

        TEST_DATES.dt.month
        ==
        month
    ).to_numpy()


    if (
        mask.sum()
        <
        10
    ):

        continue


    ridge_month_mae = (

        mae(

            Y_test[
                mask
            ],

            RIDGE_PRED[
                mask
            ]
        )
    )


    d1_month_mae = (

        mae(

            Y_test[
                mask
            ],

            D1_PRED[
                mask
            ]
        )
    )


    monthly_rows.append(
        {

            "Month":
                month,

            "Days":
                int(
                    mask.sum()
                ),

            "Ridge_MAE_MW":
                ridge_month_mae,

            "D1_MAE_MW":
                d1_month_mae,

            "Ridge_Skill_vs_D1_percent":
                100
                *
                (
                    d1_month_mae
                    -
                    ridge_month_mae
                )
                /
                d1_month_mae
        }
    )


monthly_robustness = pd.DataFrame(
    monthly_rows
)


# =====================================================================
# 26. BPDB EVENING_PEAK TIMING SENSITIVITY
# =====================================================================

print("\n" + "=" * 125)

print(
    "BPDB EVENING_PEAK TIMING SENSITIVITY"
)

print("=" * 125)


# =====================================================================
# 27. LOAD BPDB
# =====================================================================

bpdb = pd.read_excel(
    BPDB_FILE
)


BPDB_DATE_COLUMN = "Date"


if (
    BPDB_DATE_COLUMN
    not in
    bpdb.columns
):

    raise ValueError(
        "BPDB Date column missing."
    )


bpdb[
    BPDB_DATE_COLUMN
] = pd.to_datetime(

    bpdb[
        BPDB_DATE_COLUMN
    ],

    errors="coerce",

    dayfirst=True
)


bpdb = (

    bpdb

    .dropna(
        subset=[
            BPDB_DATE_COLUMN
        ]
    )

    .sort_values(
        BPDB_DATE_COLUMN
    )

    .reset_index(
        drop=True
    )
)


bpdb[
    "date"
] = (

    bpdb[
        BPDB_DATE_COLUMN
    ]
    .dt.normalize()
)


assert (
    bpdb[
        "date"
    ].min()
    ==
    pd.Timestamp(
        "2019-11-21"
    )
)


assert (
    bpdb[
        "date"
    ].max()
    ==
    pd.Timestamp(
        "2024-12-30"
    )
)


BPDB_TARGET_COLUMN = (
    "Max. Demand at eve. peak (Generation end)"
)


if (
    BPDB_TARGET_COLUMN
    not in
    bpdb.columns
):

    raise ValueError(
        """
BPDB evening-demand column missing.
"""
    )


bpdb[
    "bpdb_evening_generation_mw"
] = pd.to_numeric(

    bpdb[
        BPDB_TARGET_COLUMN
    ],

    errors="coerce"
)


bpdb.loc[

    ~bpdb[
        "bpdb_evening_generation_mw"
    ].between(
        1000,
        30000
    ),

    "bpdb_evening_generation_mw"

] = np.nan


# =====================================================================
# 28. LOAD PGCB ALL-ROWS AUDIT
# =====================================================================

pgcb = pd.read_csv(
    PGCB_AUDIT_FILE,
    low_memory=False
)


pgcb[
    "datetime"
] = pd.to_datetime(

    pgcb[
        "datetime"
    ],

    errors="coerce"
)


pgcb = (

    pgcb
    .dropna(
        subset=[
            "datetime"
        ]
    )

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


pgcb[
    "date"
] = (

    pgcb[
        "datetime"
    ]
    .dt.normalize()
)


pgcb[
    "hour"
] = (

    pgcb[
        "datetime"
    ]
    .dt.hour
)


pgcb[
    "minute"
] = (

    pgcb[
        "datetime"
    ]
    .dt.minute
)


pgcb[
    "remarks_clean"
] = (

    pgcb[
        "remarks"
    ]

    .fillna(
        ""
    )

    .astype(str)

    .str.strip()

    .str.lower()
)


# =====================================================================
# 29. PGCB DEMAND COLUMN
# =====================================================================

if (
    "demand_mw"
    in
    pgcb.columns
):

    PGCB_DEMAND = (
        "demand_mw"
    )


elif (
    "demand_original"
    in
    pgcb.columns
):

    PGCB_DEMAND = (
        "demand_original"
    )


else:

    raise ValueError(
        "PGCB demand column not found."
    )


pgcb[
    PGCB_DEMAND
] = pd.to_numeric(

    pgcb[
        PGCB_DEMAND
    ],

    errors="coerce"
)


# =====================================================================
# 30. PGCB QUALITY FLAG
# =====================================================================

if (
    "demand_corrupt"
    in
    pgcb.columns
):

    corrupt_text = (

        pgcb[
            "demand_corrupt"
        ]

        .astype(str)

        .str.strip()

        .str.lower()
    )


    demand_corrupt = (

        corrupt_text

        .isin(
            [
                "true",
                "1",
                "yes"
            ]
        )
    )


else:

    demand_corrupt = (

        pd.Series(

            False,

            index=
                pgcb.index
        )
    )


# =====================================================================
# 31. ALL VALID EXPLICIT EVENING_PEAK ROWS
# =====================================================================

explicit_mask = (

    pgcb[
        "remarks_clean"
    ]
    .eq(
        "evening_peak"
    )

    &

    (
        ~demand_corrupt
    )

    &

    pgcb[
        PGCB_DEMAND
    ]
    .between(
        1000,
        30000
    )
)


explicit_all = (

    pgcb.loc[
        explicit_mask
    ]

    .copy()
)


# =====================================================================
# 32. TIMING SENSITIVITY DEFINITIONS
# =====================================================================

explicit_17_22 = (

    explicit_all.loc[

        explicit_all[
            "hour"
        ]
        .between(
            17,
            22
        )
    ]

    .copy()
)


explicit_18_22 = (

    explicit_all.loc[

        explicit_all[
            "hour"
        ]
        .between(
            18,
            22
        )
    ]

    .copy()
)


explicit_exclude_01_12 = (

    explicit_all.loc[

        ~explicit_all[
            "hour"
        ]
        .isin(
            [
                1,
                12
            ]
        )
    ]

    .copy()
)


print(
    "All explicit Evening_Peak rows:",
    len(
        explicit_all
    )
)


print(
    "17:00-22:59 rows:",
    len(
        explicit_17_22
    )
)


print(
    "18:00-22:59 rows:",
    len(
        explicit_18_22
    )
)


print(
    "After excluding hour 01 and 12:",
    len(
        explicit_exclude_01_12
    )
)


# =====================================================================
# 33. SOURCE AGREEMENT METRIC FUNCTION
# =====================================================================

def source_agreement_metrics(
    explicit_dataframe,
    definition_name
):

    daily = (

        explicit_dataframe

        .groupby(
            "date",
            as_index=False
        )

        .agg(

            pgcb_peak_mw=(
                PGCB_DEMAND,
                "max"
            )
        )
    )


    merged = (

        bpdb[
            [
                "date",
                "bpdb_evening_generation_mw"
            ]
        ]

        .merge(

            daily,

            on="date",

            how="inner"
        )

        .dropna()
    )


    actual = (

        merged[
            "bpdb_evening_generation_mw"
        ]

        .to_numpy(
            dtype=float
        )
    )


    comparison = (

        merged[
            "pgcb_peak_mw"
        ]

        .to_numpy(
            dtype=float
        )
    )


    if len(
        merged
    ) < 2:

        return {

            "Definition":
                definition_name,

            "Rows":
                len(
                    explicit_dataframe
                ),

            "Matched_Days":
                len(
                    merged
                ),

            "Pearson_r":
                np.nan,

            "MAE_MW":
                np.nan,

            "RMSE_MW":
                np.nan,

            "MAPE_percent":
                np.nan,

            "Bias_MW":
                np.nan
        }


    return {

        "Definition":
            definition_name,

        "Rows":
            len(
                explicit_dataframe
            ),

        "Matched_Days":
            len(
                merged
            ),

        "Pearson_r":
            float(

                np.corrcoef(

                    actual,

                    comparison
                )[
                    0,
                    1
                ]
            ),

        "MAE_MW":
            mae(
                actual,
                comparison
            ),

        "RMSE_MW":
            rmse(
                actual,
                comparison
            ),

        "MAPE_percent":
            mape(
                actual,
                comparison
            ),

        "Bias_MW":
            float(

                np.mean(

                    comparison

                    -

                    actual
                )
            )
    }


# =====================================================================
# 34. RUN EVENING-PEAK SENSITIVITY
# =====================================================================

evening_sensitivity = pd.DataFrame(
    [

        source_agreement_metrics(

            explicit_all,

            "All_Explicit_EveningPeak"
        ),

        source_agreement_metrics(

            explicit_17_22,

            "Explicit_17to22"
        ),

        source_agreement_metrics(

            explicit_18_22,

            "Explicit_18to22"
        ),

        source_agreement_metrics(

            explicit_exclude_01_12,

            "Explicit_Exclude_01_and_12"
        )
    ]
)


print("\n" + "=" * 125)

print(
    "EVENING_PEAK SOURCE-AGREEMENT SENSITIVITY"
)

print("=" * 125)


display(
    evening_sensitivity.round(
        5
    )
)


# =====================================================================
# 35. QUANTIFY EFFECT OF THE UNUSUAL 01:00 / 12:00 ROWS
# =====================================================================

unusual_evening_rows = (

    explicit_all.loc[

        explicit_all[
            "hour"
        ]
        .isin(
            [
                1,
                12
            ]
        )
    ]

    [
        [
            "datetime",
            "date",
            PGCB_DEMAND,
            "remarks"
        ]
    ]

    .copy()
)


print(
    "\nUnusual explicit Evening_Peak rows at 01:00 or 12:00:"
)


display(
    unusual_evening_rows
)


# =====================================================================
# 36. SAVE STATISTICAL RESULTS
# =====================================================================

STATISTICAL_FILE = (

    RESULT_DIR
    /
    "paired_statistical_significance.csv"
)


ABLATION_FILE = (

    RESULT_DIR
    /
    "ridge_feature_group_ablation.csv"
)


TEMPORAL_FILE = (

    RESULT_DIR
    /
    "temporal_robustness_all_models.csv"
)


TEMPORAL_SKILL_FILE = (

    RESULT_DIR
    /
    "ridge_temporal_skill_vs_d1.csv"
)


MONTHLY_FILE = (

    RESULT_DIR
    /
    "monthly_robustness.csv"
)


EVENING_SENSITIVITY_FILE = (

    RESULT_DIR
    /
    "bpdb_evening_peak_timing_sensitivity.csv"
)


UNUSUAL_EVENING_FILE = (

    RESULT_DIR
    /
    "unusual_evening_peak_rows.csv"
)


statistical_results.to_csv(
    STATISTICAL_FILE,
    index=False
)


ablation_results.to_csv(
    ABLATION_FILE,
    index=False
)


temporal_results.to_csv(
    TEMPORAL_FILE,
    index=False
)


temporal_skill.to_csv(
    TEMPORAL_SKILL_FILE,
    index=False
)


monthly_robustness.to_csv(
    MONTHLY_FILE,
    index=False
)


evening_sensitivity.to_csv(
    EVENING_SENSITIVITY_FILE,
    index=False
)


unusual_evening_rows.to_csv(
    UNUSUAL_EVENING_FILE,
    index=False
)


# =====================================================================
# 37. FIGURE 1
# FEATURE ABLATION
# =====================================================================

ablation_plot = (

    ablation_results.loc[

        ablation_results[
            "Removed_Group"
        ]
        !=
        "NONE"
    ]

    .sort_values(
        "Delta_Test_MAE_vs_Full_MW"
    )

    .copy()
)


plt.figure(
    figsize=(
        11,
        7
    )
)


plt.barh(

    ablation_plot[
        "Removed_Group"
    ],

    ablation_plot[
        "Delta_Test_MAE_vs_Full_MW"
    ]
)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Change in Test MAE After Removing Feature Group (MW)"
)


plt.ylabel(
    "Removed Feature Group"
)


plt.title(
    "Leave-One-Feature-Group-Out Ridge Ablation"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


FIG1 = (

    FIGURE_DIR
    /
    "01_feature_group_ablation.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 38. FIGURE 2
# BLOCK-BOOTSTRAP CONFIDENCE INTERVALS
# =====================================================================

if len(
    statistical_results
) > 0:

    bootstrap_plot = (

        statistical_results

        .sort_values(
            "Mean_Daily_MAE_Improvement_MW"
        )

        .copy()
    )


    centers = (

        bootstrap_plot[
            "Mean_Daily_MAE_Improvement_MW"
        ]
        .to_numpy()
    )


    lower_errors = (

        centers

        -

        bootstrap_plot[
            "Bootstrap7_CI_Lower_MW"
        ]
        .to_numpy()
    )


    upper_errors = (

        bootstrap_plot[
            "Bootstrap7_CI_Upper_MW"
        ]
        .to_numpy()

        -

        centers
    )


    positions = np.arange(
        len(
            bootstrap_plot
        )
    )


    plt.figure(
        figsize=(
            11,
            7
        )
    )


    plt.errorbar(

        centers,

        positions,

        xerr=
            np.vstack(
                [
                    lower_errors,
                    upper_errors
                ]
            ),

        fmt="o",

        capsize=4
    )


    plt.yticks(

        positions,

        bootstrap_plot[
            "Comparator"
        ]
    )


    plt.axvline(
        0,
        linestyle="--"
    )


    plt.xlabel(
        "Daily MAE Improvement of Ridge (MW)"
    )


    plt.ylabel(
        "Comparator"
    )


    plt.title(
        "7-Day Block-Bootstrap 95% Confidence Intervals"
    )


    plt.grid(
        axis="x",
        alpha=0.25
    )


    plt.tight_layout()


    FIG2 = (

        FIGURE_DIR
        /
        "02_block_bootstrap_significance.png"
    )


    plt.savefig(
        FIG2,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()


# =====================================================================
# 39. FIGURE 3
# TEMPORAL ROBUSTNESS RIDGE VS D1
# =====================================================================

temporal_plot = (

    temporal_skill.loc[

        temporal_skill[
            "Period"
        ]
        !=
        "ALL_TEST_2024_2026"
    ]

    .copy()
)


if len(
    temporal_plot
) > 0:

    x = np.arange(
        len(
            temporal_plot
        )
    )


    width = 0.35


    plt.figure(
        figsize=(
            11,
            6
        )
    )


    plt.bar(

        x
        -
        width
        /
        2,

        temporal_plot[
            "Ridge_MAE_MW"
        ],

        width=

            width,

        label=
            "Ridge"
    )


    plt.bar(

        x
        +
        width
        /
        2,

        temporal_plot[
            "D1_MAE_MW"
        ],

        width=
            width,

        label=
            "D1"
    )


    plt.xticks(

        x,

        temporal_plot[
            "Period"
        ],

        rotation=20
    )


    plt.ylabel(
        "MAE (MW)"
    )


    plt.xlabel(
        "Evaluation Period"
    )


    plt.title(
        "Temporal Robustness of Ridge vs Previous-Day Baseline"
    )


    plt.legend()


    plt.grid(
        axis="y",
        alpha=0.25
    )


    plt.tight_layout()


    FIG3 = (

        FIGURE_DIR
        /
        "03_temporal_robustness_ridge_vs_d1.png"
    )


    plt.savefig(
        FIG3,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()


# =====================================================================
# 40. FIGURE 4
# MONTHLY RIDGE SKILL
# =====================================================================

if len(
    monthly_robustness
) > 0:

    plt.figure(
        figsize=(
            12,
            6
        )
    )


    plt.plot(

        monthly_robustness[
            "Month"
        ],

        monthly_robustness[
            "Ridge_Skill_vs_D1_percent"
        ],

        marker="o",

        linewidth=2
    )


    plt.axhline(
        0,
        linestyle="--"
    )


    plt.xticks(
        range(
            1,
            13
        )
    )


    plt.xlabel(
        "Calendar Month"
    )


    plt.ylabel(
        "Ridge MAE Skill vs D1 (%)"
    )


    plt.title(
        "Month-of-Year Robustness of Ridge Forecasting Skill"
    )


    plt.grid(
        alpha=0.25
    )


    plt.tight_layout()


    FIG4 = (

        FIGURE_DIR
        /
        "04_monthly_ridge_skill.png"
    )


    plt.savefig(
        FIG4,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()


# =====================================================================
# 41. FIGURE 5
# BPDB EVENING_PEAK TIMING SENSITIVITY
# =====================================================================

plt.figure(
    figsize=(
        11,
        6
    )
)


plt.bar(

    evening_sensitivity[
        "Definition"
    ],

    evening_sensitivity[
        "MAE_MW"
    ]
)


plt.ylabel(
    "PGCB-BPDB Source Agreement MAE (MW)"
)


plt.xlabel(
    "Evening_Peak Definition"
)


plt.title(
    "Sensitivity of BPDB-PGCB Agreement to Evening_Peak Timing"
)


plt.xticks(
    rotation=25,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG5 = (

    FIGURE_DIR
    /
    "05_evening_peak_timing_sensitivity.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 42. FIGURE 6
# DAILY RIDGE VS D1 ERROR DIFFERENCE THROUGH TIME
# =====================================================================

daily_difference = (

    DAILY_LOSSES[
        "D1"
    ]

    -

    DAILY_LOSSES[
        "Ridge"
    ]
)


daily_difference_series = pd.Series(

    daily_difference,

    index=
        test_df[
            "target_date"
        ]
)


rolling30_difference = (

    daily_difference_series

    .rolling(
        window=30,
        min_periods=10
    )

    .mean()
)


plt.figure(
    figsize=(
        15,
        6
    )
)


plt.plot(

    daily_difference_series.index,

    rolling30_difference.values,

    linewidth=1.8
)


plt.axhline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Date"
)


plt.ylabel(
    "30-Day Mean D1 MAE − Ridge MAE (MW)"
)


plt.title(
    "Temporal Stability of Ridge Improvement Over D1"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG6 = (

    FIGURE_DIR
    /
    "06_rolling_ridge_improvement_vs_d1.png"
)


plt.savefig(
    FIG6,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 43. IDENTIFY MOST IMPORTANT ABLATION
#
# EXPLANATORY ONLY.
#
# Do NOT use this to alter the frozen primary model.
# =====================================================================

non_full_ablation = (

    ablation_results.loc[

        ablation_results[
            "Removed_Group"
        ]
        !=
        "NONE"
    ]

    .copy()
)


largest_damage_row = (

    non_full_ablation

    .sort_values(
        "Delta_Test_MAE_vs_Full_MW",

        ascending=False
    )

    .iloc[0]
)


# =====================================================================
# 44. PRIMARY EVENING-PEAK SENSITIVITY ROWS
# =====================================================================

all_evening_row = (

    evening_sensitivity.loc[

        evening_sensitivity[
            "Definition"
        ]
        ==
        "All_Explicit_EveningPeak"
    ]

    .iloc[0]
)


restricted_evening_row = (

    evening_sensitivity.loc[

        evening_sensitivity[
            "Definition"
        ]
        ==
        "Explicit_Exclude_01_and_12"
    ]

    .iloc[0]
)


# =====================================================================
# 45. SUMMARY JSON
# =====================================================================

summary = {

    "frozen_ridge_alpha":
        float(
            FROZEN_ALPHA
        ),

    "frozen_ridge_test_MAE_MW":
        float(
            ridge_metrics[
                "MAE_MW"
            ]
        ),

    "frozen_ridge_test_RMSE_MW":
        float(
            ridge_metrics[
                "RMSE_MW"
            ]
        ),

    "frozen_ridge_test_MAPE_percent":
        float(
            ridge_metrics[
                "MAPE_percent"
            ]
        ),

    "d1_test_MAE_MW":
        float(
            mae(
                Y_test,
                D1_PRED
            )
        ),

    "ridge_skill_vs_d1_percent":
        float(
            ridge_vs_d1[
                "Ridge_Skill_percent"
            ]
        ),

    "ridge_vs_d1_mean_daily_improvement_MW":
        float(
            ridge_vs_d1[
                "Mean_Daily_MAE_Improvement_MW"
            ]
        ),

    "ridge_vs_d1_block7_CI_lower_MW":
        float(
            ridge_vs_d1[
                "Bootstrap7_CI_Lower_MW"
            ]
        ),

    "ridge_vs_d1_block7_CI_upper_MW":
        float(
            ridge_vs_d1[
                "Bootstrap7_CI_Upper_MW"
            ]
        ),

    "ridge_vs_d1_block14_CI_lower_MW":
        float(
            ridge_vs_d1[
                "Bootstrap14_CI_Lower_MW"
            ]
        ),

    "ridge_vs_d1_block14_CI_upper_MW":
        float(
            ridge_vs_d1[
                "Bootstrap14_CI_Upper_MW"
            ]
        ),

    "ridge_vs_d1_HAC7_p_Holm":
        float(
            ridge_vs_d1[
                "HAC7_p_Holm"
            ]
        ),

    "largest_ablation_damage_group":
        str(
            largest_damage_row[
                "Removed_Group"
            ]
        ),

    "largest_ablation_test_MAE_increase_MW":
        float(
            largest_damage_row[
                "Delta_Test_MAE_vs_Full_MW"
            ]
        ),

    "bpdb_all_explicit_MAE_MW":
        float(
            all_evening_row[
                "MAE_MW"
            ]
        ),

    "bpdb_restricted_explicit_MAE_MW":
        float(
            restricted_evening_row[
                "MAE_MW"
            ]
        ),

    "bpdb_all_explicit_Pearson_r":
        float(
            all_evening_row[
                "Pearson_r"
            ]
        ),

    "bpdb_restricted_explicit_Pearson_r":
        float(
            restricted_evening_row[
                "Pearson_r"
            ]
        ),

    "unusual_01_12_evening_peak_rows":
        int(
            len(
                unusual_evening_rows
            )
        ),

    "test_days":
        int(
            len(
                test_df
            )
        ),

    "analysis_note":
        (
            "Ablation and sensitivity analyses are explanatory "
            "post-hoc robustness analyses and do not modify the "
            "frozen primary forecasting model."
        )
}


SUMMARY_FILE = (

    RESULT_DIR
    /
    "step7_summary.json"
)


with open(
    SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# =====================================================================
# 46. FINAL VALIDATION CHECKS
# =====================================================================

assert (
    RIDGE_PRED.shape
    ==
    Y_test.shape
)


assert (
    D1_PRED.shape
    ==
    Y_test.shape
)


assert (
    len(
        DAILY_LOSSES[
            "Ridge"
        ]
    )
    ==
    len(
        test_df
    )
)


assert np.isfinite(
    RIDGE_PRED
).all()


assert np.isfinite(
    D1_PRED
).all()


assert (
    len(
        ablation_results
    )
    ==
    9
)


assert (
    len(
        evening_sensitivity
    )
    ==
    4
)


assert (
    bpdb[
        "date"
    ].min()
    ==
    pd.Timestamp(
        "2019-11-21"
    )
)


assert (
    bpdb[
        "date"
    ].max()
    ==
    pd.Timestamp(
        "2024-12-30"
    )
)


print(
    "\nAll Step 7 validation checks PASSED."
)


# =====================================================================
# 47. FINAL OUTPUT
# =====================================================================

print("\n")

print("=" * 125)

print(
    "STEP 7 COMPLETED SUCCESSFULLY"
)

print("=" * 125)


# =====================================================================
# 48. PRIMARY STATISTICAL RESULT
# =====================================================================

print("\n" + "-" * 125)

print(
    "A. RIDGE VS D1 STATISTICAL SIGNIFICANCE"
)

print("-" * 125)


print(
    "Ridge MAE:",
    round(
        ridge_vs_d1[
            "Ridge_MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "D1 MAE:",
    round(
        ridge_vs_d1[
            "Comparator_MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "Ridge skill vs D1:",
    round(
        ridge_vs_d1[
            "Ridge_Skill_percent"
        ],
        3
    ),
    "%"
)


print(
    "Mean daily improvement:",
    round(
        ridge_vs_d1[
            "Mean_Daily_MAE_Improvement_MW"
        ],
        3
    ),
    "MW"
)


print(
    "7-day block-bootstrap 95% CI:",
    round(
        ridge_vs_d1[
            "Bootstrap7_CI_Lower_MW"
        ],
        3
    ),
    "to",
    round(
        ridge_vs_d1[
            "Bootstrap7_CI_Upper_MW"
        ],
        3
    ),
    "MW"
)


print(
    "14-day block-bootstrap 95% CI:",
    round(
        ridge_vs_d1[
            "Bootstrap14_CI_Lower_MW"
        ],
        3
    ),
    "to",
    round(
        ridge_vs_d1[
            "Bootstrap14_CI_Upper_MW"
        ],
        3
    ),
    "MW"
)


print(
    "HAC-7 Holm p:",
    ridge_vs_d1[
        "HAC7_p_Holm"
    ]
)


print(
    "Ridge wins on:",
    round(
        ridge_vs_d1[
            "Ridge_Wins_Days_percent"
        ],
        2
    ),
    "% of test days"
)


# =====================================================================
# 49. PRIMARY ABLATION RESULT
# =====================================================================

print("\n" + "-" * 125)

print(
    "B. FEATURE ABLATION"
)

print("-" * 125)


print(
    "Largest increase in test MAE occurred when removing:"
)

print(
    largest_damage_row[
        "Removed_Group"
    ]
)


print(
    "MAE increase:",
    round(
        largest_damage_row[
            "Delta_Test_MAE_vs_Full_MW"
        ],
        3
    ),
    "MW"
)


print(
    "Relative MAE change:",
    round(
        largest_damage_row[
            "Relative_Test_MAE_Change_percent"
        ],
        3
    ),
    "%"
)


# =====================================================================
# 50. PRIMARY EVENING-PEAK SENSITIVITY
# =====================================================================

print("\n" + "-" * 125)

print(
    "C. EVENING_PEAK TIMING SENSITIVITY"
)

print("-" * 125)


print(
    "Unusual explicit rows at 01:00 or 12:00:",
    len(
        unusual_evening_rows
    )
)


print(
    "\nAll explicit Evening_Peak:"
)


print(
    "n =",
    int(
        all_evening_row[
            "Matched_Days"
        ]
    )
)


print(
    "r =",
    round(
        all_evening_row[
            "Pearson_r"
        ],
        6
    )
)


print(
    "MAE =",
    round(
        all_evening_row[
            "MAE_MW"
        ],
        3
    ),
    "MW"
)


print(
    "\nAfter excluding 01:00 and 12:00:"
)


print(
    "n =",
    int(
        restricted_evening_row[
            "Matched_Days"
        ]
    )
)


print(
    "r =",
    round(
        restricted_evening_row[
            "Pearson_r"
        ],
        6
    )
)


print(
    "MAE =",
    round(
        restricted_evening_row[
            "MAE_MW"
        ],
        3
    ),
    "MW"
)


# =====================================================================
# 51. OUTPUT FILES
# =====================================================================

print("\n" + "-" * 125)

print(
    "IMPORTANT OUTPUT FILES"
)

print("-" * 125)


print(
    "\nStatistical significance:"
)

print(
    STATISTICAL_FILE
)


print(
    "\nFeature ablation:"
)

print(
    ABLATION_FILE
)


print(
    "\nTemporal robustness:"
)

print(
    TEMPORAL_SKILL_FILE
)


print(
    "\nMonthly robustness:"
)

print(
    MONTHLY_FILE
)


print(
    "\nEvening_Peak sensitivity:"
)

print(
    EVENING_SENSITIVITY_FILE
)


print(
    "\nUnusual Evening_Peak rows:"
)

print(
    UNUSUAL_EVENING_FILE
)


print(
    "\nFigures:"
)

print(
    FIGURE_DIR
)


print(
    "\nStep 7 summary:"
)

print(
    SUMMARY_FILE
)


print("\n" + "=" * 125)

print(
    "SEND THESE OUTPUTS TO CHATGPT:"
)

print(
    "1. STEP 7 COMPLETED SUCCESSFULLY"
)

print(
    "2. PAIRED STATISTICAL SIGNIFICANCE RESULTS"
)

print(
    "3. FEATURE-GROUP ABLATION RESULTS"
)

print(
    "4. TEMPORAL ROBUSTNESS: RIDGE VS D1"
)

print(
    "5. EVENING_PEAK SOURCE-AGREEMENT SENSITIVITY"
)

print("=" * 125)
```
