```python
# =====================================================================
# STEP 1C - FINAL CORRECTED VERSION
# ADVANCED TEMPORAL + DIGIT-SHIFT ANOMALY SCREENING
# AND FINAL PGCB FORECASTING TARGET CREATION
#
# Research:
# Reliable Multi-Horizon Electricity Load Forecasting in Bangladesh
#
# Run this entire code in ONE Google Colab cell.
#
# IMPORTANT:
# Step 1B must already have been completed.
# This code does NOT train any forecasting model.
# =====================================================================


# =====================================================================
# 0. IMPORT LIBRARIES
# =====================================================================

import os
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 150)
pd.set_option("display.max_rows", 150)
pd.set_option("display.width", 220)

print("=" * 95)
print("STEP 1C: ADVANCED TEMPORAL + DIGIT-SHIFT DATA QUALITY SCREENING")
print("=" * 95)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount("/content/drive")

print("\nGoogle Drive mounted successfully.")


# =====================================================================
# 2. PROJECT PATHS
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

QUALITY_DIR = (
    PROJECT_ROOT
    / "data"
    / "quality_logs"
)

RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "advanced_cleaning"
)

FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "advanced_cleaning"
)

for directory in [
    PROCESSED_DIR,
    QUALITY_DIR,
    RESULT_DIR,
    FIGURE_DIR
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# =====================================================================
# 3. LOAD STEP 1B OUTPUT
# =====================================================================

INPUT_FILE = (
    PROCESSED_DIR
    / "pgcb_all_rows_quality_flagged.csv"
)

if not INPUT_FILE.exists():

    raise FileNotFoundError(
        f"""
Step 1B output was not found.

Expected file:

{INPUT_FILE}

Please confirm that Step 1B completed successfully.
"""
    )


df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)

df["datetime"] = pd.to_datetime(
    df["datetime"],
    errors="coerce"
)

df = (
    df
    .sort_values("datetime")
    .reset_index(drop=True)
)

print("\nStep 1B dataset loaded successfully.")

print(
    "Rows:",
    f"{len(df):,}"
)

print(
    "Period:",
    df["datetime"].min(),
    "to",
    df["datetime"].max()
)


# =====================================================================
# 4. HELPER FOR BOOLEAN COLUMNS
# =====================================================================

def safe_bool(series):

    if pd.api.types.is_bool_dtype(series):
        return series.fillna(False)

    return (
        series
        .astype(str)
        .str.strip()
        .str.lower()
        .map(
            {
                "true": True,
                "false": False,
                "1": True,
                "0": False,
                "yes": True,
                "no": False
            }
        )
        .fillna(False)
        .astype(bool)
    )


# =====================================================================
# 5. PRESERVE ORIGINAL VALUES
# =====================================================================

df["demand_original"] = (
    pd.to_numeric(
        df["demand_mw"],
        errors="coerce"
    )
)

df["generation_original"] = (
    pd.to_numeric(
        df["generation_mw"],
        errors="coerce"
    )
)

df["load_shedding_original"] = (
    pd.to_numeric(
        df["load_shedding"],
        errors="coerce"
    )
)


# =====================================================================
# 6. RESTORE STEP 1B BOOLEAN FLAGS SAFELY
# =====================================================================

boolean_columns = [

    "demand_corrupt",
    "generation_corrupt",
    "load_shedding_corrupt",

    "severe_physical_inconsistency",
    "physical_review_required",

    "duplicate_timestamp"
]


for col in boolean_columns:

    if col not in df.columns:
        df[col] = False

    df[col] = safe_bool(
        df[col]
    )


# =====================================================================
# 7. STRICT TIMESTAMP CLASSIFICATION
# =====================================================================

# We now distinguish:
#
# 1. Normal integer-hour observations
# 2. Exact 19:30 special PGCB observations
# 3. Other non-hour timestamps
#
# We DO NOT use the remarks column to determine temporal resolution.


df["hour_exact"] = (
    df["datetime"]
    .dt.hour
)

df["minute_exact"] = (
    df["datetime"]
    .dt.minute
)


df["strict_regular_hour"] = (

    df["datetime"].notna()

    &

    (df["minute_exact"] == 0)
)


df["strict_1930_peak"] = (

    df["datetime"].notna()

    &

    (df["hour_exact"] == 19)

    &

    (df["minute_exact"] == 30)
)


df["strict_other_time"] = (

    df["datetime"].notna()

    &

    (~df["strict_regular_hour"])

    &

    (~df["strict_1930_peak"])
)


print("\n" + "=" * 95)
print("STRICT TIMESTAMP STRUCTURE")
print("=" * 95)

print(
    "Regular integer-hour rows:",
    f"{df['strict_regular_hour'].sum():,}"
)

print(
    "Exact 19:30 rows:",
    f"{df['strict_1930_peak'].sum():,}"
)

print(
    "Other non-hour rows:",
    f"{df['strict_other_time'].sum():,}"
)


# =====================================================================
# 8. CREATE REGULAR-HOUR WORKING DATASET
# =====================================================================

regular = (

    df.loc[
        df["strict_regular_hour"]
    ]

    .copy()

    .sort_values("datetime")

    .reset_index()
    .rename(
        columns={
            "index":
            "master_index"
        }
    )
)


print(
    "\nRegular-hour observations entering Step 1C:",
    f"{len(regular):,}"
)


# =====================================================================
# 9. CALENDAR VARIABLES
# =====================================================================

regular["date"] = (
    regular["datetime"]
    .dt.normalize()
)

regular["year"] = (
    regular["datetime"]
    .dt.year
)

regular["month"] = (
    regular["datetime"]
    .dt.month
)

regular["day"] = (
    regular["datetime"]
    .dt.day
)

regular["hour"] = (
    regular["datetime"]
    .dt.hour
)

regular["weekday"] = (
    regular["datetime"]
    .dt.dayofweek
)


# =====================================================================
# 10. SAME-HOUR LOCAL MEDIAN AND MAD
# =====================================================================

# We compare each point against the same hour on nearby dates.
#
# Example:
#
# 03:00 is compared with nearby 03:00 values.
#
# This prevents the normal daily load curve itself from appearing
# anomalous.


def robust_mad(values):

    values = np.asarray(
        values,
        dtype=float
    )

    values = values[
        np.isfinite(values)
    ]

    if len(values) == 0:
        return np.nan

    median = np.median(
        values
    )

    return np.median(
        np.abs(
            values - median
        )
    )


regular["same_hour_median_29d"] = np.nan
regular["same_hour_mad_29d"] = np.nan


print(
    "\nCalculating robust same-hour temporal references..."
)


for hr in sorted(
    regular["hour"]
    .dropna()
    .unique()
):

    mask = (
        regular["hour"]
        == hr
    )

    temp = (
        regular.loc[
            mask,
            "demand_original"
        ]
    )

    rolling_median = (

        temp
        .rolling(
            window=29,
            center=True,
            min_periods=9
        )
        .median()
    )

    rolling_mad = (

        temp
        .rolling(
            window=29,
            center=True,
            min_periods=9
        )
        .apply(
            robust_mad,
            raw=True
        )
    )

    regular.loc[
        mask,
        "same_hour_median_29d"
    ] = rolling_median.values

    regular.loc[
        mask,
        "same_hour_mad_29d"
    ] = rolling_mad.values


print(
    "Same-hour references calculated."
)


# =====================================================================
# 11. SAME-HOUR DEVIATION
# =====================================================================

regular[
    "same_hour_absolute_deviation_mw"
] = (

    regular["demand_original"]

    -

    regular["same_hour_median_29d"]

).abs()


regular[
    "same_hour_relative_deviation"
] = (

    regular[
        "same_hour_absolute_deviation_mw"
    ]

    /

    regular[
        "same_hour_median_29d"
    ]
    .abs()
    .clip(lower=1)
)


# =====================================================================
# 12. ROBUST Z-SCORE
# =====================================================================

regular["robust_z"] = (

    0.6745

    *

    regular[
        "same_hour_absolute_deviation_mw"
    ]

    /

    regular[
        "same_hour_mad_29d"
    ]
    .replace(
        0,
        np.nan
    )
)


# =====================================================================
# 13. IMMEDIATE PREVIOUS AND NEXT HOUR REFERENCES
# =====================================================================

regular["previous_datetime"] = (
    regular["datetime"]
    .shift(1)
)

regular["next_datetime"] = (
    regular["datetime"]
    .shift(-1)
)

regular["previous_demand"] = (
    regular["demand_original"]
    .shift(1)
)

regular["next_demand"] = (
    regular["demand_original"]
    .shift(-1)
)


regular[
    "minutes_from_previous"
] = (

    regular["datetime"]

    -

    regular["previous_datetime"]

).dt.total_seconds() / 60


regular[
    "minutes_to_next"
] = (

    regular["next_datetime"]

    -

    regular["datetime"]

).dt.total_seconds() / 60


# Only accept an actual previous/next hour.

regular["previous_hour_valid"] = (

    regular[
        "minutes_from_previous"
    ]
    .between(
        55,
        65
    )
)


regular["next_hour_valid"] = (

    regular[
        "minutes_to_next"
    ]
    .between(
        55,
        65
    )
)


regular["valid_previous_demand"] = np.where(

    regular[
        "previous_hour_valid"
    ],

    regular[
        "previous_demand"
    ],

    np.nan
)


regular["valid_next_demand"] = np.where(

    regular[
        "next_hour_valid"
    ],

    regular[
        "next_demand"
    ],

    np.nan
)


# =====================================================================
# 14. NEIGHBOR REFERENCE
# =====================================================================

regular["neighbor_median"] = (

    regular[
        [
            "valid_previous_demand",
            "valid_next_demand"
        ]
    ]
    .median(
        axis=1,
        skipna=True
    )
)


regular[
    "neighbor_absolute_deviation_mw"
] = (

    regular["demand_original"]

    -

    regular["neighbor_median"]

).abs()


regular[
    "neighbor_relative_deviation"
] = (

    regular[
        "neighbor_absolute_deviation_mw"
    ]

    /

    regular[
        "neighbor_median"
    ]
    .abs()
    .clip(lower=1)
)


# =====================================================================
# 15. EXACT 24-HOUR AND 168-HOUR HISTORICAL REFERENCES
# =====================================================================

reference_lookup = (

    regular[
        [
            "datetime",
            "demand_original"
        ]
    ]

    .drop_duplicates(
        subset="datetime"
    )

    .set_index(
        "datetime"
    )[
        "demand_original"
    ]
)


regular["demand_24h_ago"] = [

    reference_lookup.get(
        dt - pd.Timedelta(hours=24),
        np.nan
    )

    for dt in regular["datetime"]
]


regular["demand_168h_ago"] = [

    reference_lookup.get(
        dt - pd.Timedelta(hours=168),
        np.nan
    )

    for dt in regular["datetime"]
]


regular[
    "relative_to_24h"
] = (

    (
        regular["demand_original"]

        -

        regular["demand_24h_ago"]
    )
    .abs()

    /

    regular[
        "demand_24h_ago"
    ]
    .abs()
    .clip(lower=1)
)


regular[
    "relative_to_168h"
] = (

    (
        regular["demand_original"]

        -

        regular["demand_168h_ago"]
    )
    .abs()

    /

    regular[
        "demand_168h_ago"
    ]
    .abs()
    .clip(lower=1)
)


# =====================================================================
# 16. TEMPORAL EXTREME CANDIDATE
# =====================================================================

# A candidate must be:
#
# > 50% away from its same-hour local baseline
#
# AND
#
# > 1500 MW away in absolute magnitude.


regular[
    "temporal_extreme_candidate"
] = (

    (
        regular[
            "same_hour_relative_deviation"
        ]
        > 0.50
    )

    &

    (
        regular[
            "same_hour_absolute_deviation_mw"
        ]
        > 1500
    )
)


print(
    "\nInitial temporal extreme candidates:",
    int(
        regular[
            "temporal_extreme_candidate"
        ]
        .sum()
    )
)


# =====================================================================
# 17. DETECT CONSECUTIVE EXTREME EPISODES
# =====================================================================

# Genuine grid disturbances may persist for several hours.
#
# A single corrupted record often appears as an isolated point.


candidate = (
    regular[
        "temporal_extreme_candidate"
    ]
    .fillna(False)
    .astype(bool)
)


episode_ids = []

episode_counter = 0


for i in range(
    len(regular)
):

    if not candidate.iloc[i]:

        episode_ids.append(
            np.nan
        )

        continue


    start_new_episode = False


    if i == 0:

        start_new_episode = True

    else:

        previous_was_candidate = (
            candidate.iloc[i - 1]
        )

        gap = (
            regular.loc[
                i,
                "minutes_from_previous"
            ]
        )

        if (
            not previous_was_candidate
            or
            pd.isna(gap)
            or
            gap > 65
        ):

            start_new_episode = True


    if start_new_episode:

        episode_counter += 1


    episode_ids.append(
        float(
            episode_counter
        )
    )


regular[
    "extreme_episode_id"
] = episode_ids


episode_sizes = (

    regular.loc[
        regular[
            "extreme_episode_id"
        ]
        .notna()
    ]

    .groupby(
        "extreme_episode_id"
    )

    .size()
)


regular[
    "extreme_episode_length"
] = (

    regular[
        "extreme_episode_id"
    ]

    .map(
        episode_sizes
    )

    .fillna(0)

    .astype(int)
)


regular[
    "multi_hour_extreme_event"
] = (

    regular[
        "temporal_extreme_candidate"
    ]

    &

    (
        regular[
            "extreme_episode_length"
        ]
        >= 3
    )
)


# =====================================================================
# 18. DIGIT / DECIMAL SHIFT DETECTOR
# =====================================================================

# Examples:
#
# 74223 -> perhaps 7422.3
#
# 80 -> perhaps 8000
#
# 611 -> perhaps 6110
#
# This detector NEVER automatically changes the value.
# It only records a candidate correction.


SCALE_FACTORS = [
    0.01,
    0.1,
    10.0,
    100.0
]


def detect_digit_shift(
    observed,
    reference
):

    if (
        pd.isna(observed)
        or
        pd.isna(reference)
        or
        observed <= 0
        or
        reference <= 0
    ):

        return (
            False,
            np.nan,
            np.nan,
            np.nan,
            np.nan
        )


    raw_error = (

        abs(
            observed
            -
            reference
        )

        /
        reference
    )


    best_factor = np.nan
    best_value = np.nan
    best_error = raw_error


    for factor in SCALE_FACTORS:

        candidate_value = (
            observed
            *
            factor
        )

        candidate_error = (

            abs(
                candidate_value
                -
                reference
            )

            /
            reference
        )


        if candidate_error < best_error:

            best_error = (
                candidate_error
            )

            best_factor = (
                factor
            )

            best_value = (
                candidate_value
            )


    improvement_ratio = (

        raw_error
        /
        max(
            best_error,
            1e-9
        )
    )


    is_digit_shift = (

        raw_error > 0.50

        and

        best_error <= 0.20

        and

        improvement_ratio >= 3
    )


    return (

        is_digit_shift,

        best_factor,

        best_value,

        best_error,

        improvement_ratio
    )


digit_results = [

    detect_digit_shift(
        observed,
        reference
    )

    for observed, reference
    in zip(
        regular[
            "demand_original"
        ],
        regular[
            "same_hour_median_29d"
        ]
    )
]


regular[
    "digit_shift_candidate"
] = [
    result[0]
    for result in digit_results
]

regular[
    "digit_shift_best_factor"
] = [
    result[1]
    for result in digit_results
]

regular[
    "digit_shift_candidate_value"
] = [
    result[2]
    for result in digit_results
]

regular[
    "digit_shift_candidate_relative_error"
] = [
    result[3]
    for result in digit_results
]

regular[
    "digit_shift_improvement_ratio"
] = [
    result[4]
    for result in digit_results
]


print(
    "Digit/decimal-shift candidates:",
    int(
        regular[
            "digit_shift_candidate"
        ]
        .sum()
    )
)


# =====================================================================
# 19. ISOLATED TEMPORAL ANOMALY
# =====================================================================

regular[
    "isolated_temporal_anomaly"
] = (

    regular[
        "temporal_extreme_candidate"
    ]

    &

    (
        regular[
            "neighbor_relative_deviation"
        ]
        > 0.50
    )

    &

    (
        regular[
            "neighbor_absolute_deviation_mw"
        ]
        > 1500
    )

    &

    (
        regular[
            "extreme_episode_length"
        ]
        <= 2
    )
)


# =====================================================================
# 20. HISTORICAL DISAGREEMENT
# =====================================================================

regular[
    "history_disagreement"
] = (

    regular[
        "demand_24h_ago"
    ]
    .notna()

    &

    regular[
        "demand_168h_ago"
    ]
    .notna()

    &

    (
        regular[
            "relative_to_24h"
        ]
        > 0.50
    )

    &

    (
        regular[
            "relative_to_168h"
        ]
        > 0.50
    )
)


# =====================================================================
# 21. VERY LARGE ABSOLUTE VALUE SIGNAL
# =====================================================================

# This is NOT an automatic deletion rule.
#
# It is only extra evidence.
#
# The threshold is deliberately generous for Bangladesh national load.


regular[
    "extreme_scale_signal"
] = (

    regular[
        "demand_original"
    ]
    > 30000
)


# =====================================================================
# 22. EXTREMELY SMALL ISOLATED VALUE SIGNAL
# =====================================================================

regular[
    "tiny_isolated_signal"
] = (

    (
        regular[
            "demand_original"
        ]
        < 1000
    )

    &

    (
        regular[
            "same_hour_median_29d"
        ]
        > 3000
    )

    &

    (
        regular[
            "extreme_episode_length"
        ]
        <= 2
    )
)


# =====================================================================
# 23. ANOMALY EVIDENCE SCORE
# =====================================================================

# Transparent rule-based evidence score.
#
# This is NOT machine learning.


regular[
    "anomaly_evidence_score"
] = 0


# Strong previous Step 1B target corruption signal

regular.loc[
    regular[
        "demand_corrupt"
    ],
    "anomaly_evidence_score"
] += 5


# Possible decimal/digit error

regular.loc[
    regular[
        "digit_shift_candidate"
    ],
    "anomaly_evidence_score"
] += 3


# Isolated spike/drop

regular.loc[
    regular[
        "isolated_temporal_anomaly"
    ],
    "anomaly_evidence_score"
] += 2


# Strong disagreement with historical demand

regular.loc[
    regular[
        "history_disagreement"
    ],
    "anomaly_evidence_score"
] += 1


# Extreme robust deviation

regular.loc[
    regular[
        "robust_z"
    ]
    > 10,
    "anomaly_evidence_score"
] += 1


# Physical inconsistency

regular.loc[
    regular[
        "severe_physical_inconsistency"
    ],
    "anomaly_evidence_score"
] += 2


# Extremely large value

regular.loc[
    regular[
        "extreme_scale_signal"
    ],
    "anomaly_evidence_score"
] += 2


# Tiny isolated point

regular.loc[
    regular[
        "tiny_isolated_signal"
    ],
    "anomaly_evidence_score"
] += 2


# =====================================================================
# 24. EVENT PRESERVATION SIGNAL
# =====================================================================

# Preserve prolonged unusual episodes when:
#
# - at least 3 consecutive hours are extreme
# - no Step 1B demand corruption
# - no digit-shift signature
# - no severe physical inconsistency
#
# This helps preserve real grid disturbances.


regular[
    "event_preservation_signal"
] = (

    regular[
        "multi_hour_extreme_event"
    ]

    &

    (~regular[
        "demand_corrupt"
    ])

    &

    (~regular[
        "digit_shift_candidate"
    ])

    &

    (~regular[
        "severe_physical_inconsistency"
    ])
)


# =====================================================================
# 25. FINAL ADVANCED QUALITY CLASSIFICATION
# =====================================================================

regular[
    "advanced_quality_status"
] = "KEEP_NORMAL"


# Preserve real-looking multi-hour events

regular.loc[

    regular[
        "event_preservation_signal"
    ],

    "advanced_quality_status"

] = "KEEP_EXTREME_EVENT"


# Moderate anomaly evidence

review_condition = (

    (
        regular[
            "anomaly_evidence_score"
        ]
        >= 2
    )

    &

    (
        regular[
            "anomaly_evidence_score"
        ]
        < 4
    )

    &

    (~regular[
        "event_preservation_signal"
    ])
)


regular.loc[
    review_condition,
    "advanced_quality_status"
] = "REVIEW"


# Strong anomaly evidence

exclude_condition = (

    (
        regular[
            "anomaly_evidence_score"
        ]
        >= 4
    )

    &

    (~regular[
        "event_preservation_signal"
    ])
)


regular.loc[
    exclude_condition,
    "advanced_quality_status"
] = "EXCLUDE_TARGET"


# Step 1B demand corruption always excluded

regular.loc[
    regular[
        "demand_corrupt"
    ],
    "advanced_quality_status"
] = "EXCLUDE_TARGET"


# Strong isolated digit-shift case

strong_digit_error = (

    regular[
        "digit_shift_candidate"
    ]

    &

    regular[
        "isolated_temporal_anomaly"
    ]
)


regular.loc[
    strong_digit_error,
    "advanced_quality_status"
] = "EXCLUDE_TARGET"


# Extremely large digit-shift-like values

regular.loc[

    (
        regular[
            "extreme_scale_signal"
        ]

        &

        regular[
            "digit_shift_candidate"
        ]
    ),

    "advanced_quality_status"

] = "EXCLUDE_TARGET"


# =====================================================================
# 26. FINAL TARGET
# =====================================================================

# Original demand is preserved separately.
#
# We DO NOT automatically repair an anomalous value.
#
# REVIEW and EXCLUDE_TARGET become missing targets.
#
# This is conservative and scientifically safer.


regular[
    "demand_final"
] = (
    regular[
        "demand_original"
    ]
)


regular.loc[

    regular[
        "advanced_quality_status"
    ]
    .isin(
        [
            "REVIEW",
            "EXCLUDE_TARGET"
        ]
    ),

    "demand_final"

] = np.nan


regular[
    "final_target_usable"
] = (
    regular[
        "demand_final"
    ]
    .notna()
)


# =====================================================================
# 27. HUMAN-READABLE QUALITY REASONS
# =====================================================================

def build_reason(row):

    reasons = []


    if row[
        "demand_corrupt"
    ]:

        reasons.append(
            "step1b_demand_corruption"
        )


    if row[
        "digit_shift_candidate"
    ]:

        reasons.append(
            "possible_decimal_or_digit_shift"
        )


    if row[
        "isolated_temporal_anomaly"
    ]:

        reasons.append(
            "isolated_temporal_spike_or_drop"
        )


    if row[
        "history_disagreement"
    ]:

        reasons.append(
            "disagrees_with_24h_and_168h_history"
        )


    if row[
        "severe_physical_inconsistency"
    ]:

        reasons.append(
            "physical_balance_violation"
        )


    if row[
        "extreme_scale_signal"
    ]:

        reasons.append(
            "extremely_large_demand_value"
        )


    if row[
        "tiny_isolated_signal"
    ]:

        reasons.append(
            "extremely_small_isolated_value"
        )


    if row[
        "multi_hour_extreme_event"
    ]:

        reasons.append(
            "multi_hour_extreme_episode"
        )


    if row[
        "event_preservation_signal"
    ]:

        reasons.append(
            "preserved_possible_real_grid_event"
        )


    if len(reasons) == 0:

        return "normal"


    return ";".join(
        reasons
    )


regular[
    "advanced_quality_reason"
] = regular.apply(
    build_reason,
    axis=1
)


# =====================================================================
# 28. ADVANCED QUALITY SUMMARY
# =====================================================================

advanced_summary = (

    regular[
        "advanced_quality_status"
    ]

    .value_counts(
        dropna=False
    )

    .rename_axis(
        "advanced_quality_status"
    )

    .reset_index(
        name="count"
    )
)


advanced_summary[
    "percent"
] = (

    100

    *

    advanced_summary[
        "count"
    ]

    /

    len(
        regular
    )
)


print("\n" + "=" * 95)
print("ADVANCED QUALITY SUMMARY")
print("=" * 95)

display(
    advanced_summary
)


# =====================================================================
# 29. DETECTOR SUMMARY
# =====================================================================

detector_summary = pd.DataFrame({

    "detector": [

        "Step 1B demand corruption",

        "Temporal extreme candidate",

        "Digit/decimal shift candidate",

        "Isolated temporal anomaly",

        "Multi-hour extreme-event rows",

        "History disagreement",

        "Extreme > 30,000 MW signal",

        "Tiny isolated value signal",

        "Event-preservation signal",

        "Final excluded/review targets",

        "Final usable targets"
    ],

    "count": [

        int(
            regular[
                "demand_corrupt"
            ]
            .sum()
        ),

        int(
            regular[
                "temporal_extreme_candidate"
            ]
            .sum()
        ),

        int(
            regular[
                "digit_shift_candidate"
            ]
            .sum()
        ),

        int(
            regular[
                "isolated_temporal_anomaly"
            ]
            .sum()
        ),

        int(
            regular[
                "multi_hour_extreme_event"
            ]
            .sum()
        ),

        int(
            regular[
                "history_disagreement"
            ]
            .sum()
        ),

        int(
            regular[
                "extreme_scale_signal"
            ]
            .sum()
        ),

        int(
            regular[
                "tiny_isolated_signal"
            ]
            .sum()
        ),

        int(
            regular[
                "event_preservation_signal"
            ]
            .sum()
        ),

        int(
            (
                ~regular[
                    "final_target_usable"
                ]
            )
            .sum()
        ),

        int(
            regular[
                "final_target_usable"
            ]
            .sum()
        )
    ]
})


detector_summary[
    "percent"
] = (

    100

    *

    detector_summary[
        "count"
    ]

    /

    len(
        regular
    )
)


print("\nDetector summary:")

display(
    detector_summary
)


# =====================================================================
# 30. DIGIT SHIFT EXAMPLES
# =====================================================================

# IMPORTANT:
# This fixes the KeyError from the previous version.
#
# same_hour_relative_deviation is now explicitly included.


digit_shift_columns = [

    "datetime",

    "generation_original",
    "demand_original",
    "load_shedding_original",

    "same_hour_median_29d",
    "same_hour_relative_deviation",
    "same_hour_absolute_deviation_mw",

    "neighbor_median",
    "neighbor_relative_deviation",

    "demand_24h_ago",
    "demand_168h_ago",

    "digit_shift_best_factor",
    "digit_shift_candidate_value",
    "digit_shift_candidate_relative_error",
    "digit_shift_improvement_ratio",

    "robust_z",

    "extreme_episode_id",
    "extreme_episode_length",

    "anomaly_evidence_score",

    "advanced_quality_status",
    "advanced_quality_reason"
]


digit_shift_examples = (

    regular.loc[

        regular[
            "digit_shift_candidate"
        ],

        digit_shift_columns
    ]

    .sort_values(
        by="same_hour_relative_deviation",
        ascending=False
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 95)
print("POSSIBLE DECIMAL / DIGIT-SHIFT CASES")
print("=" * 95)

print(
    "Total digit/decimal-shift candidates:",
    len(
        digit_shift_examples
    )
)


if len(
    digit_shift_examples
) > 0:

    display(
        digit_shift_examples.head(60)
    )

else:

    print(
        "No digit/decimal-shift candidates detected."
    )


# =====================================================================
# 31. EXCLUDED / REVIEW OBSERVATIONS
# =====================================================================

excluded_columns = [

    "datetime",

    "generation_original",
    "demand_original",
    "load_shedding_original",

    "same_hour_median_29d",
    "same_hour_relative_deviation",

    "neighbor_median",
    "neighbor_relative_deviation",

    "demand_24h_ago",
    "demand_168h_ago",

    "robust_z",

    "digit_shift_candidate",
    "digit_shift_best_factor",
    "digit_shift_candidate_value",

    "extreme_episode_id",
    "extreme_episode_length",

    "anomaly_evidence_score",

    "advanced_quality_status",
    "advanced_quality_reason"
]


excluded_examples = (

    regular.loc[

        ~regular[
            "final_target_usable"
        ],

        excluded_columns
    ]

    .sort_values(
        by=[
            "anomaly_evidence_score",
            "datetime"
        ],
        ascending=[
            False,
            True
        ]
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 95)
print("EXCLUDED / REVIEW TARGET OBSERVATIONS")
print("=" * 95)

print(
    "Total excluded/review:",
    len(
        excluded_examples
    )
)

display(
    excluded_examples.head(100)
)


# =====================================================================
# 32. PRESERVED EXTREME EVENTS
# =====================================================================

preserved_columns = [

    "datetime",

    "generation_original",
    "demand_original",
    "load_shedding_original",

    "same_hour_median_29d",
    "same_hour_relative_deviation",

    "neighbor_median",

    "extreme_episode_id",
    "extreme_episode_length",

    "advanced_quality_status",
    "advanced_quality_reason"
]


preserved_events = (

    regular.loc[

        regular[
            "advanced_quality_status"
        ]
        ==
        "KEEP_EXTREME_EVENT",

        preserved_columns
    ]

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 95)
print("PRESERVED POSSIBLE REAL EXTREME EVENTS")
print("=" * 95)

print(
    "Preserved extreme-event rows:",
    len(
        preserved_events
    )
)

display(
    preserved_events.head(100)
)


# =====================================================================
# 33. CHECK KNOWN SUSPICIOUS CASES FROM STEP 1B
# =====================================================================

check_dates = pd.to_datetime([

    "2016-09-16 07:00:00",

    "2016-12-09 21:00:00",

    "2017-05-17 13:00:00",

    "2017-08-26 16:00:00",

    "2017-10-16 10:00:00",

    "2018-01-04 21:00:00",

    "2018-12-18 09:00:00",

    "2020-10-20 08:00:00",

    "2020-12-24 03:00:00",

    "2023-04-18 19:00:00",

    "2024-04-16 08:00:00"
])


sanity_columns = [

    "datetime",

    "generation_original",
    "demand_original",
    "load_shedding_original",

    "same_hour_median_29d",
    "same_hour_relative_deviation",

    "neighbor_median",

    "digit_shift_candidate",
    "digit_shift_best_factor",
    "digit_shift_candidate_value",

    "extreme_episode_length",

    "anomaly_evidence_score",

    "advanced_quality_status",
    "advanced_quality_reason"
]


historical_sanity_check = (

    regular.loc[

        regular[
            "datetime"
        ]
        .isin(
            check_dates
        ),

        sanity_columns
    ]

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 95)
print("KNOWN SUSPICIOUS-CASE SANITY CHECK")
print("=" * 95)

display(
    historical_sanity_check
)


# =====================================================================
# 34. CREATE FINAL OBSERVED-HOUR TARGET DATASET
# =====================================================================

final_hourly = (

    regular.loc[
        regular[
            "final_target_usable"
        ]
    ]

    .copy()

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nFinal usable observed-hour targets:",
    f"{len(final_hourly):,}"
)


# =====================================================================
# 35. OBSERVATION COUNT BY HOUR
# =====================================================================

hour_distribution = (

    final_hourly[
        "hour"
    ]

    .value_counts()

    .sort_index()

    .rename_axis(
        "hour"
    )

    .reset_index(
        name="observation_count"
    )
)


print("\nObserved integer-hour distribution:")

display(
    hour_distribution
)


# =====================================================================
# 36. DAILY COMPLETENESS
# =====================================================================

daily_hour_counts = (

    final_hourly

    .groupby(
        final_hourly[
            "datetime"
        ]
        .dt.date
    )

    .agg(

        observed_integer_hours=(
            "hour",
            "nunique"
        ),

        first_hour=(
            "hour",
            "min"
        ),

        last_hour=(
            "hour",
            "max"
        )
    )

    .reset_index()
)


daily_hour_counts.columns = [

    "date",

    "observed_integer_hours",

    "first_hour",

    "last_hour"
]


daily_count_distribution = (

    daily_hour_counts[
        "observed_integer_hours"
    ]

    .value_counts()

    .sort_index()

    .rename_axis(
        "observed_integer_hours"
    )

    .reset_index(
        name="number_of_days"
    )
)


print(
    "\nFinal daily integer-hour completeness:"
)

display(
    daily_count_distribution
)


# =====================================================================
# 37. CREATE STRICT 19:30 DATASET
# =====================================================================

peak1930 = (

    df.loc[
        df[
            "strict_1930_peak"
        ]
    ]

    .copy()

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


if "demand_corrupt" in peak1930.columns:

    peak1930[
        "demand_corrupt"
    ] = safe_bool(
        peak1930[
            "demand_corrupt"
        ]
    )

else:

    peak1930[
        "demand_corrupt"
    ] = False


peak1930[
    "peak_target_usable"
] = (

    peak1930[
        "demand_original"
    ]
    .notna()

    &

    (~peak1930[
        "demand_corrupt"
    ])
)


peak1930_final = (

    peak1930.loc[
        peak1930[
            "peak_target_usable"
        ]
    ]

    .copy()

    .sort_values(
        "datetime"
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nStrict 19:30 observations:",
    f"{len(peak1930):,}"
)

print(
    "Usable strict 19:30 targets:",
    f"{len(peak1930_final):,}"
)


# =====================================================================
# 38. SAVE ALL OUTPUT FILES
# =====================================================================

ADVANCED_AUDIT_FILE = (

    PROCESSED_DIR
    /
    "pgcb_regular_hours_advanced_audit.csv"
)


FINAL_HOURLY_FILE = (

    PROCESSED_DIR
    /
    "pgcb_final_observed_hourly_target.csv"
)


FINAL_PEAK_FILE = (

    PROCESSED_DIR
    /
    "pgcb_final_1930_peak_target.csv"
)


ADVANCED_LOG_FILE = (

    QUALITY_DIR
    /
    "pgcb_advanced_quality_log.csv"
)


ADVANCED_SUMMARY_FILE = (

    RESULT_DIR
    /
    "advanced_quality_summary.csv"
)


DETECTOR_SUMMARY_FILE = (

    RESULT_DIR
    /
    "advanced_detector_summary.csv"
)


DIGIT_SHIFT_FILE = (

    RESULT_DIR
    /
    "digit_shift_candidates.csv"
)


PRESERVED_EVENTS_FILE = (

    RESULT_DIR
    /
    "preserved_extreme_events.csv"
)


HOUR_DISTRIBUTION_FILE = (

    RESULT_DIR
    /
    "hour_distribution.csv"
)


DAILY_COMPLETENESS_FILE = (

    RESULT_DIR
    /
    "daily_integer_hour_completeness.csv"
)


SANITY_FILE = (

    RESULT_DIR
    /
    "historical_suspicious_case_check.csv"
)


regular.to_csv(
    ADVANCED_AUDIT_FILE,
    index=False
)


final_hourly.to_csv(
    FINAL_HOURLY_FILE,
    index=False
)


peak1930_final.to_csv(
    FINAL_PEAK_FILE,
    index=False
)


excluded_examples.to_csv(
    ADVANCED_LOG_FILE,
    index=False
)


advanced_summary.to_csv(
    ADVANCED_SUMMARY_FILE,
    index=False
)


detector_summary.to_csv(
    DETECTOR_SUMMARY_FILE,
    index=False
)


digit_shift_examples.to_csv(
    DIGIT_SHIFT_FILE,
    index=False
)


preserved_events.to_csv(
    PRESERVED_EVENTS_FILE,
    index=False
)


hour_distribution.to_csv(
    HOUR_DISTRIBUTION_FILE,
    index=False
)


daily_hour_counts.to_csv(
    DAILY_COMPLETENESS_FILE,
    index=False
)


historical_sanity_check.to_csv(
    SANITY_FILE,
    index=False
)


print(
    "\nCSV outputs saved successfully."
)


# =====================================================================
# 39. FIGURE 1
# RAW VS FINAL DEMAND
# =====================================================================

plt.figure(
    figsize=(17, 6)
)


plt.plot(
    regular[
        "datetime"
    ],
    regular[
        "demand_original"
    ],
    linewidth=0.40,
    alpha=0.30,
    label="Raw observed demand"
)


plt.plot(
    final_hourly[
        "datetime"
    ],
    final_hourly[
        "demand_final"
    ],
    linewidth=0.55,
    alpha=0.85,
    label="Final usable demand"
)


# Use a realistic plotting range so a handful of corrupt values
# do not destroy readability.

reasonable_upper = (

    final_hourly[
        "demand_final"
    ]
    .quantile(
        0.9995
    )
    *
    1.25
)


reasonable_upper = max(
    reasonable_upper,
    18000
)


plt.ylim(
    0,
    reasonable_upper
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "Electricity Demand (MW)"
)

plt.title(
    "PGCB National Electricity Demand After Advanced Quality Screening"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


FIG1 = (

    FIGURE_DIR
    /
    "01_raw_vs_final_demand.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =====================================================================
# 40. FIGURE 2
# EXCLUDED VS PRESERVED
# =====================================================================

plt.figure(
    figsize=(17, 6)
)


plt.plot(
    regular[
        "datetime"
    ],
    regular[
        "demand_original"
    ],
    linewidth=0.35,
    alpha=0.30,
    label="Observed demand"
)


excluded_plot = (

    regular.loc[
        ~regular[
            "final_target_usable"
        ]
    ]
)


plt.scatter(
    excluded_plot[
        "datetime"
    ],
    excluded_plot[
        "demand_original"
    ],
    s=30,
    marker="x",
    label="Excluded / review"
)


preserved_plot = (

    regular.loc[
        regular[
            "advanced_quality_status"
        ]
        ==
        "KEEP_EXTREME_EVENT"
    ]
)


plt.scatter(
    preserved_plot[
        "datetime"
    ],
    preserved_plot[
        "demand_original"
    ],
    s=18,
    alpha=0.75,
    label="Preserved extreme event"
)


plt.ylim(
    0,
    reasonable_upper
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "Demand (MW)"
)

plt.title(
    "Advanced Screening: Data Errors vs Preserved Extreme Grid Events"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


FIG2 = (

    FIGURE_DIR
    /
    "02_excluded_vs_preserved_events.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =====================================================================
# 41. FIGURE 3
# LOCAL TEMPORAL REFERENCE SCATTER
# =====================================================================

plot_df = (

    regular[
        [
            "same_hour_median_29d",
            "demand_original",
            "final_target_usable"
        ]
    ]

    .dropna()
)


usable_plot = (

    plot_df.loc[
        plot_df[
            "final_target_usable"
        ]
    ]
)


bad_plot = (

    plot_df.loc[
        ~plot_df[
            "final_target_usable"
        ]
    ]
)


if len(
    usable_plot
) > 20000:

    usable_sample = (
        usable_plot.sample(
            n=20000,
            random_state=42
        )
    )

else:

    usable_sample = (
        usable_plot.copy()
    )


local_limit = (

    usable_plot[
        "same_hour_median_29d"
    ]
    .quantile(
        0.999
    )
    *
    1.25
)


local_limit = max(
    local_limit,
    18000
)


plt.figure(
    figsize=(8, 8)
)


plt.scatter(
    usable_sample[
        "same_hour_median_29d"
    ],
    usable_sample[
        "demand_original"
    ],
    s=8,
    alpha=0.25,
    label="Usable observations"
)


plt.scatter(
    bad_plot[
        "same_hour_median_29d"
    ],
    bad_plot[
        "demand_original"
    ],
    s=30,
    marker="x",
    label="Excluded / review"
)


plt.plot(
    [
        0,
        local_limit
    ],
    [
        0,
        local_limit
    ],
    linestyle="--",
    linewidth=1.2,
    label="Observed = local reference"
)


plt.xlim(
    0,
    local_limit
)

plt.ylim(
    0,
    local_limit
)


plt.xlabel(
    "Same-Hour Local Median Demand (MW)"
)

plt.ylabel(
    "Observed Demand (MW)"
)

plt.title(
    "Temporal Plausibility Screening of PGCB Demand"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


FIG3 = (

    FIGURE_DIR
    /
    "03_temporal_plausibility_scatter.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =====================================================================
# 42. FIGURE 4
# OBSERVATION AVAILABILITY BY HOUR
# =====================================================================

plt.figure(
    figsize=(12, 6)
)


plt.bar(
    hour_distribution[
        "hour"
    ],
    hour_distribution[
        "observation_count"
    ]
)


plt.xticks(
    range(
        0,
        24
    )
)


plt.xlabel(
    "Hour of Day"
)

plt.ylabel(
    "Number of Observations"
)

plt.title(
    "Availability of PGCB Integer-Hour Electricity Demand Observations"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()


FIG4 = (

    FIGURE_DIR
    /
    "04_hourly_observation_availability.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =====================================================================
# 43. FIGURE 5
# ANOMALY SCORE DISTRIBUTION
# =====================================================================

score_counts = (

    regular[
        "anomaly_evidence_score"
    ]

    .value_counts()

    .sort_index()
)


plt.figure(
    figsize=(10, 6)
)


plt.bar(
    score_counts.index.astype(str),
    score_counts.values
)


plt.xlabel(
    "Anomaly Evidence Score"
)

plt.ylabel(
    "Number of Observations"
)

plt.title(
    "Distribution of Rule-Based Data Quality Evidence Scores"
)

plt.grid(
    axis="y",
    alpha=0.25
)

plt.tight_layout()


FIG5 = (

    FIGURE_DIR
    /
    "05_anomaly_evidence_score_distribution.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =====================================================================
# 44. CREATE STEP 1C JSON SUMMARY
# =====================================================================

summary_json = {

    "input_regular_hour_rows":
        int(
            len(
                regular
            )
        ),

    "final_usable_regular_targets":
        int(
            regular[
                "final_target_usable"
            ]
            .sum()
        ),

    "final_excluded_or_review_targets":
        int(
            (
                ~regular[
                    "final_target_usable"
                ]
            )
            .sum()
        ),

    "temporal_extreme_candidates":
        int(
            regular[
                "temporal_extreme_candidate"
            ]
            .sum()
        ),

    "digit_shift_candidates":
        int(
            regular[
                "digit_shift_candidate"
            ]
            .sum()
        ),

    "isolated_temporal_anomalies":
        int(
            regular[
                "isolated_temporal_anomaly"
            ]
            .sum()
        ),

    "multi_hour_extreme_event_rows":
        int(
            regular[
                "multi_hour_extreme_event"
            ]
            .sum()
        ),

    "preserved_extreme_event_rows":
        int(
            regular[
                "event_preservation_signal"
            ]
            .sum()
        ),

    "strict_1930_peak_rows":
        int(
            len(
                peak1930
            )
        ),

    "usable_strict_1930_peak_rows":
        int(
            len(
                peak1930_final
            )
        ),

    "first_regular_timestamp":
        str(
            regular[
                "datetime"
            ]
            .min()
        ),

    "last_regular_timestamp":
        str(
            regular[
                "datetime"
            ]
            .max()
        )
}


SUMMARY_JSON_FILE = (

    RESULT_DIR
    /
    "step_1c_summary.json"
)


with open(
    SUMMARY_JSON_FILE,
    "w"
) as file:

    json.dump(
        summary_json,
        file,
        indent=4
    )


# =====================================================================
# 45. FINAL VALIDATION
# =====================================================================

assert (
    final_hourly[
        "datetime"
    ]
    .is_monotonic_increasing
), "Final dataset is not chronological."


assert (
    final_hourly[
        "demand_final"
    ]
    .notna()
    .all()
), "Final target contains NaN values."


assert (
    final_hourly[
        "minute_exact"
    ]
    .eq(0)
    .all()
), "Non-integer-hour observations entered the main target dataset."


if len(
    peak1930_final
) > 0:

    assert (
        peak1930_final[
            "datetime"
        ]
        .dt.hour
        .eq(19)
        .all()
    )

    assert (
        peak1930_final[
            "datetime"
        ]
        .dt.minute
        .eq(30)
        .all()
    )


print(
    "\nAll final validation checks passed."
)


# =====================================================================
# 46. FINAL OUTPUT SUMMARY
# =====================================================================

print("\n")
print("=" * 95)
print("STEP 1C COMPLETED SUCCESSFULLY")
print("=" * 95)


print(
    "\nInput regular integer-hour observations:",
    f"{len(regular):,}"
)


print(
    "Final usable demand targets:",
    f"{regular['final_target_usable'].sum():,}"
)


print(
    "Excluded/review targets:",
    f"{(~regular['final_target_usable']).sum():,}"
)


print(
    "\nTemporal extreme candidates:",
    f"{regular['temporal_extreme_candidate'].sum():,}"
)


print(
    "Digit/decimal shift candidates:",
    f"{regular['digit_shift_candidate'].sum():,}"
)


print(
    "Isolated temporal anomalies:",
    f"{regular['isolated_temporal_anomaly'].sum():,}"
)


print(
    "Multi-hour extreme-event observations:",
    f"{regular['multi_hour_extreme_event'].sum():,}"
)


print(
    "Preserved possible real extreme-event observations:",
    f"{regular['event_preservation_signal'].sum():,}"
)


print(
    "\nStrict 19:30 observations:",
    f"{len(peak1930):,}"
)


print(
    "Usable strict 19:30 observations:",
    f"{len(peak1930_final):,}"
)


print("\n" + "-" * 95)
print("IMPORTANT OUTPUT FILES")
print("-" * 95)


print(
    "\n1. FINAL observed-hour forecasting target:"
)

print(
    FINAL_HOURLY_FILE
)


print(
    "\n2. Complete advanced audit dataset:"
)

print(
    ADVANCED_AUDIT_FILE
)


print(
    "\n3. Strict 19:30 evening-peak target:"
)

print(
    FINAL_PEAK_FILE
)


print(
    "\n4. Excluded/review quality log:"
)

print(
    ADVANCED_LOG_FILE
)


print(
    "\n5. Digit-shift candidates:"
)

print(
    DIGIT_SHIFT_FILE
)


print(
    "\n6. Preserved extreme-event records:"
)

print(
    PRESERVED_EVENTS_FILE
)


print(
    "\n7. Figures:"
)

print(
    FIGURE_DIR
)


print(
    "\n8. Advanced-cleaning results:"
)

print(
    RESULT_DIR
)


print("\n" + "=" * 95)

print(
    "DO NOT START MODEL TRAINING YET."
)

print(
    "SEND THE FINAL STEP 1C SUMMARY AND FIGURES TO CHATGPT."
)

print("=" * 95)
```
