```python
# ======================================================================
# STEP W4
# WEATHER EFFECT SIGNIFICANCE + POST-HOC WEATHER FEATURE ABLATION
#
# PURPOSE
# -------
# A. Test the paired daily-loss difference:
#
#       Frozen Ridge
#            vs
#       Ridge + Historical Weather
#
# B. Use the SAME dependence-aware statistical protocol as Step 7:
#       - daily MAE as statistical unit
#       - 5,000 moving-block bootstrap replications
#       - 7-day blocks
#       - 14-day robustness blocks
#       - Newey-West HAC tests with lags 7 and 14
#
# C. Diagnose weather feature groups using leave-one-group-out ablation.
#
# IMPORTANT
# ---------
# The locked test has already been observed in W3.
#
# Therefore:
#   W4 IS POST-HOC / DIAGNOSTIC.
#
# DO NOT:
#   - select a new final forecasting model from W4 test results
#   - replace the frozen Ridge based on these ablations
#   - claim a test-discovered weather subset is confirmatory
#
# Frozen point model:
#       Ridge alpha = 10
#
# Full historical-weather model:
#       Ridge alpha = 10
#
# ======================================================================


# ======================================================================
# 0. IMPORTS
# ======================================================================

import json
import math
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

from scipy.stats import norm

from google.colab import drive
from IPython.display import display


warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)

N_BOOTSTRAP = 5000
RIDGE_ALPHA = 10.0


pd.set_option(
    "display.max_columns",
    350
)

pd.set_option(
    "display.max_rows",
    300
)

pd.set_option(
    "display.width",
    320
)


print("=" * 125)
print("STEP W4: WEATHER SIGNIFICANCE + FEATURE-GROUP ABLATION")
print("=" * 125)


# ======================================================================
# 1. MOUNT DRIVE
# ======================================================================

drive.mount(
    "/content/drive"
)


# ======================================================================
# 2. PROJECT PATHS
# ======================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


WEATHER_FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting_weather"
)


RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "weather_extension"
)


FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "weather_extension"
)


RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


FIGURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ======================================================================
# 3. INPUT FILES
# ======================================================================

TRAIN_FILE = (
    WEATHER_FORECAST_DIR
    / "train_2015_2022_weather.csv"
)

VALIDATION_FILE = (
    WEATHER_FORECAST_DIR
    / "validation_2023_weather.csv"
)

TEST_FILE = (
    WEATHER_FORECAST_DIR
    / "test_2024_2026_weather.csv"
)

MANIFEST_FILE = (
    WEATHER_FORECAST_DIR
    / "weather_feature_manifest.csv"
)

DAILY_PAIR_FILE = (
    RESULT_DIR
    / "weather_daily_paired_errors.csv"
)

W3_LEADERBOARD_FILE = (
    RESULT_DIR
    / "weather_test_leaderboard.csv"
)


REQUIRED_FILES = [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE,
    MANIFEST_FILE,
    DAILY_PAIR_FILE,
    W3_LEADERBOARD_FILE

]


print("\nRequired files:")


for path in REQUIRED_FILES:

    print(
        path,
        "->",
        path.exists()
    )

    if not path.exists():

        raise FileNotFoundError(
            f"Required W3/W2 file missing: {path}"
        )


# ======================================================================
# 4. LOAD DATASETS
# ======================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)

validation_df = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)

test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for df in [

    train_df,
    validation_df,
    test_df

]:

    df["target_date"] = pd.to_datetime(
        df["target_date"],
        errors="coerce"
    )

    df.sort_values(
        "target_date",
        inplace=True
    )

    df.reset_index(
        drop=True,
        inplace=True
    )


manifest = pd.read_csv(
    MANIFEST_FILE
)


daily_pair = pd.read_csv(
    DAILY_PAIR_FILE
)


daily_pair["target_date"] = pd.to_datetime(
    daily_pair["target_date"],
    errors="coerce"
)


w3_leaderboard = pd.read_csv(
    W3_LEADERBOARD_FILE
)


print("\n" + "=" * 125)
print("DATA LOADED")
print("=" * 125)


print(
    "Train:",
    len(train_df)
)

print(
    "Validation:",
    len(validation_df)
)

print(
    "Test:",
    len(test_df)
)

print(
    "Paired test days:",
    len(daily_pair)
)


assert len(train_df) == 2236
assert len(validation_df) == 273
assert len(test_df) == 709
assert len(daily_pair) == 709


# ======================================================================
# 5. FORECAST TARGETS
# ======================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS

]


assert len(
    TARGET_COLUMNS
) == 23


# ======================================================================
# 6. FROZEN 165 BASE FEATURES
# ======================================================================

CALENDAR_FEATURES = [

    "year",
    "month",
    "day",
    "dayofweek",
    "dayofyear",
    "weekofyear",
    "quarter",
    "is_weekend",
    "month_sin",
    "month_cos",
    "dow_sin",
    "dow_cos",
    "doy_sin",
    "doy_cos"

]


PREVIOUS_DAY_PROFILE = [

    f"lag1_h{hour:02d}"

    for hour in HOURS

]


PREVIOUS_WEEK_PROFILE = [

    f"lag7_h{hour:02d}"

    for hour in HOURS

]


ROLLING7_MEAN = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS

]


ROLLING28_MEAN = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS

]


ROLLING7_STD = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS

]


ROLLING28_STD = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS

]


HISTORICAL_SUMMARY = [

    "lag1_daily_mean",
    "lag1_daily_max",
    "lag1_daily_min",
    "lag1_daily_range",
    "lag1_peak_hour",

    "lag7_daily_mean",
    "lag7_daily_max",
    "lag7_daily_min",
    "lag7_daily_range",
    "lag7_peak_hour",

    "mean_change_vs_last_week",
    "max_change_vs_last_week",
    "mean_ratio_vs_last_week"

]


BASE_FEATURE_COLUMNS = (

    PREVIOUS_DAY_PROFILE
    +
    PREVIOUS_WEEK_PROFILE
    +
    ROLLING7_MEAN
    +
    ROLLING28_MEAN
    +
    ROLLING7_STD
    +
    ROLLING28_STD
    +
    CALENDAR_FEATURES
    +
    HISTORICAL_SUMMARY

)


assert len(
    BASE_FEATURE_COLUMNS
) == 165


# ======================================================================
# 7. READ WEATHER FEATURE GROUPS FROM W2 MANIFEST
# ======================================================================

required_manifest_columns = [

    "feature_group",
    "feature"

]


for column in required_manifest_columns:

    if column not in manifest.columns:

        raise ValueError(
            f"Manifest missing column: {column}"
        )


WEATHER_GROUPS = {}


for group_name, group_df in manifest.groupby(
    "feature_group",
    sort=False
):

    WEATHER_GROUPS[
        group_name
    ] = (
        group_df[
            "feature"
        ]
        .astype(str)
        .tolist()
    )


WEATHER_FEATURE_COLUMNS = sorted(

    set(

        feature

        for features
        in WEATHER_GROUPS.values()

        for feature
        in features

    )

)


COMBINED_FEATURE_COLUMNS = (

    BASE_FEATURE_COLUMNS
    +
    WEATHER_FEATURE_COLUMNS

)


print("\n" + "=" * 125)
print("WEATHER FEATURE GROUPS")
print("=" * 125)


for group_name, features in (
    WEATHER_GROUPS.items()
):

    print(
        f"{group_name:40s}",
        len(features)
    )


print(
    "\nBase features:",
    len(
        BASE_FEATURE_COLUMNS
    )
)

print(
    "Weather features:",
    len(
        WEATHER_FEATURE_COLUMNS
    )
)

print(
    "Combined:",
    len(
        COMBINED_FEATURE_COLUMNS
    )
)


assert len(
    WEATHER_FEATURE_COLUMNS
) == 89

assert len(
    COMBINED_FEATURE_COLUMNS
) == 254


# ======================================================================
# 8. DATA INTEGRITY
# ======================================================================

for split_name, df in [

    ("TRAIN", train_df),
    ("VALIDATION", validation_df),
    ("TEST", test_df)

]:

    required = (

        COMBINED_FEATURE_COLUMNS
        +
        TARGET_COLUMNS

    )


    missing = [

        feature

        for feature in required

        if feature not in df.columns

    ]


    if missing:

        raise ValueError(
            f"{split_name}: missing columns {missing}"
        )


    nan_count = int(

        df[
            required
        ]
        .isna()
        .sum()
        .sum()

    )


    print(
        split_name,
        "missing cells:",
        nan_count
    )


    assert nan_count == 0


print(
    "\nDataset integrity checks PASSED."
)


# ======================================================================
# 9. W3 REFERENCE VALUES
# ======================================================================

baseline_row = (

    w3_leaderboard.loc[
        w3_leaderboard[
            "model"
        ]
        ==
        "Frozen_Ridge_165"
    ]
    .iloc[0]
)


weather_row = (

    w3_leaderboard.loc[
        w3_leaderboard[
            "model"
        ]
        ==
        "Ridge_Plus_Historical_Weather"
    ]
    .iloc[0]
)


BASELINE_TEST_MAE = float(
    baseline_row[
        "MAE_MW"
    ]
)


FULL_WEATHER_TEST_MAE = float(
    weather_row[
        "MAE_MW"
    ]
)


print("\n" + "=" * 125)
print("W3 FROZEN REFERENCE")
print("=" * 125)


print(
    "Baseline Ridge MAE:",
    round(
        BASELINE_TEST_MAE,
        4
    )
)


print(
    "Full weather Ridge MAE:",
    round(
        FULL_WEATHER_TEST_MAE,
        4
    )
)


print(
    "Weather improvement:",
    round(
        BASELINE_TEST_MAE
        -
        FULL_WEATHER_TEST_MAE,
        4
    ),
    "MW"
)


# ======================================================================
# 10. PAIRED DAILY LOSS DIFFERENCE
# ======================================================================
#
# Positive:
#       Weather model is better.
#
# Negative:
#       Weather model is worse.
# ======================================================================

required_pair_columns = [

    "baseline_daily_MAE_MW",
    "weather_daily_MAE_MW"

]


for column in required_pair_columns:

    if column not in daily_pair.columns:

        raise ValueError(
            f"Missing W3 daily-pair column: {column}"
        )


loss_difference = (

    daily_pair[
        "baseline_daily_MAE_MW"
    ].to_numpy(
        dtype=float
    )

    -

    daily_pair[
        "weather_daily_MAE_MW"
    ].to_numpy(
        dtype=float
    )

)


MEAN_DIFFERENCE = float(
    np.mean(
        loss_difference
    )
)


MEDIAN_DIFFERENCE = float(
    np.median(
        loss_difference
    )
)


WEATHER_WIN_RATE = float(

    100
    *
    np.mean(
        loss_difference
        >
        0
    )

)


print("\n" + "=" * 125)
print("PAIRED DAILY WEATHER EFFECT")
print("=" * 125)


print(
    "Mean improvement:",
    round(
        MEAN_DIFFERENCE,
        4
    ),
    "MW"
)


print(
    "Median improvement:",
    round(
        MEDIAN_DIFFERENCE,
        4
    ),
    "MW"
)


print(
    "Weather model wins:",
    round(
        WEATHER_WIN_RATE,
        3
    ),
    "%"
)


# ======================================================================
# 11. MOVING-BLOCK BOOTSTRAP
# ======================================================================

def moving_block_bootstrap_mean_ci(
    values,
    block_length,
    n_bootstrap=5000,
    seed=42
):

    values = np.asarray(
        values,
        dtype=float
    )


    n = len(
        values
    )


    if block_length <= 0:

        raise ValueError(
            "block_length must be positive."
        )


    if block_length > n:

        raise ValueError(
            "block_length exceeds sample length."
        )


    rng = np.random.default_rng(
        seed
    )


    max_start = (
        n
        -
        block_length
        +
        1
    )


    blocks_needed = int(

        math.ceil(
            n
            /
            block_length
        )

    )


    bootstrap_means = np.empty(
        n_bootstrap,
        dtype=float
    )


    for bootstrap_index in range(
        n_bootstrap
    ):

        starts = rng.integers(

            low=0,

            high=max_start,

            size=blocks_needed

        )


        sample_parts = [

            values[
                start:
                start
                +
                block_length
            ]

            for start in starts

        ]


        bootstrap_sample = np.concatenate(
            sample_parts
        )[:n]


        bootstrap_means[
            bootstrap_index
        ] = np.mean(
            bootstrap_sample
        )


    lower = float(

        np.percentile(
            bootstrap_means,
            2.5
        )

    )


    upper = float(

        np.percentile(
            bootstrap_means,
            97.5
        )

    )


    return {

        "block_length":
            block_length,

        "bootstrap_repetitions":
            n_bootstrap,

        "mean_difference_MW":
            float(
                np.mean(
                    values
                )
            ),

        "CI_lower_MW":
            lower,

        "CI_upper_MW":
            upper

    }


bootstrap7 = (
    moving_block_bootstrap_mean_ci(

        loss_difference,

        block_length=7,

        n_bootstrap=
            N_BOOTSTRAP,

        seed=
            SEED

    )
)


bootstrap14 = (
    moving_block_bootstrap_mean_ci(

        loss_difference,

        block_length=14,

        n_bootstrap=
            N_BOOTSTRAP,

        seed=
            SEED + 1

    )
)


bootstrap_results = pd.DataFrame([

    bootstrap7,
    bootstrap14

])


print("\n" + "=" * 125)
print("MOVING-BLOCK BOOTSTRAP")
print("=" * 125)


display(
    bootstrap_results.round(5)
)


# ======================================================================
# 12. NEWey-WEST HAC MEAN-DIFFERENCE TEST
# ======================================================================

def newey_west_mean_test(
    values,
    max_lag
):

    values = np.asarray(
        values,
        dtype=float
    )


    n = len(
        values
    )


    mean_value = float(
        np.mean(
            values
        )
    )


    centered = (

        values
        -
        mean_value

    )


    gamma0 = float(

        np.dot(
            centered,
            centered
        )

        /
        n

    )


    long_run_variance = gamma0


    for lag in range(
        1,
        max_lag + 1
    ):

        covariance = float(

            np.dot(

                centered[
                    lag:
                ],

                centered[
                    :-lag
                ]

            )

            /
            n

        )


        bartlett_weight = (

            1

            -

            lag
            /
            (
                max_lag
                +
                1
            )

        )


        long_run_variance += (

            2

            *
            bartlett_weight

            *
            covariance

        )


    long_run_variance = max(
        long_run_variance,
        0.0
    )


    standard_error = float(

        np.sqrt(

            long_run_variance
            /
            n

        )

    )


    if standard_error == 0:

        z_value = np.nan
        p_value = np.nan

    else:

        z_value = (

            mean_value
            /
            standard_error

        )


        p_value = float(

            2

            *
            norm.sf(
                abs(
                    z_value
                )
            )

        )


    return {

        "HAC_lag":
            max_lag,

        "mean_difference_MW":
            mean_value,

        "standard_error_MW":
            standard_error,

        "z":
            z_value,

        "p_value":
            p_value

    }


hac7 = (
    newey_west_mean_test(
        loss_difference,
        7
    )
)


hac14 = (
    newey_west_mean_test(
        loss_difference,
        14
    )
)


hac_results = pd.DataFrame([

    hac7,
    hac14

])


print("\n" + "=" * 125)
print("NEWEY-WEST HAC TEST")
print("=" * 125)


display(
    hac_results.round(6)
)


# ======================================================================
# 13. SIGNIFICANCE INTERPRETATION
# ======================================================================

CI7_LOWER = bootstrap7[
    "CI_lower_MW"
]

CI7_UPPER = bootstrap7[
    "CI_upper_MW"
]


CI14_LOWER = bootstrap14[
    "CI_lower_MW"
]

CI14_UPPER = bootstrap14[
    "CI_upper_MW"
]


if (
    CI7_LOWER > 0
    and
    CI14_LOWER > 0
):

    SIGNIFICANCE_INTERPRETATION = (
        "Historical weather improved forecasting under both "
        "block-bootstrap configurations."
    )


elif (
    CI7_UPPER < 0
    and
    CI14_UPPER < 0
):

    SIGNIFICANCE_INTERPRETATION = (
        "Historical weather significantly worsened forecasting "
        "under both block-bootstrap configurations."
    )


else:

    SIGNIFICANCE_INTERPRETATION = (
        "The weather effect is not consistently distinguishable "
        "from zero under both block-bootstrap configurations."
    )


print("\nInterpretation:")

print(
    SIGNIFICANCE_INTERPRETATION
)


# ======================================================================
# 14. HELPER METRICS
# ======================================================================

def mae(
    actual,
    predicted
):

    return float(

        np.mean(

            np.abs(
                actual
                -
                predicted
            )

        )

    )


def rmse(
    actual,
    predicted
):

    return float(

        np.sqrt(

            np.mean(

                (
                    actual
                    -
                    predicted
                )
                ** 2

            )

        )

    )


def mape(
    actual,
    predicted
):

    actual = np.asarray(
        actual,
        dtype=float
    )

    predicted = np.asarray(
        predicted,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            actual
        ),

        1e-6

    )


    return float(

        100
        *
        np.mean(

            np.abs(
                actual
                -
                predicted
            )

            /
            denominator

        )

    )


# ======================================================================
# 15. FIXED-ALPHA RIDGE EVALUATION FUNCTION
# ======================================================================
#
# For diagnostic ablation only.
#
# alpha remains fixed at 10.
#
# Validation and test results are reported.
# Test results MUST NOT be used to redefine the final model.
# ======================================================================

train_val_df = pd.concat(

    [
        train_df,
        validation_df
    ],

    axis=0,

    ignore_index=True

)


train_val_df = (

    train_val_df

    .sort_values(
        "target_date"
    )

    .reset_index(
        drop=True
    )

)


Y_train = (

    train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )

)


Y_validation = (

    validation_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )

)


Y_train_val = (

    train_val_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )

)


Y_test = (

    test_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )

)


def evaluate_fixed_ridge(
    feature_columns
):

    # --------------------------------------------------------------
    # Validation model
    # --------------------------------------------------------------

    X_train = (

        train_df[
            feature_columns
        ]

        .to_numpy(
            dtype=np.float32
        )

    )


    X_validation = (

        validation_df[
            feature_columns
        ]

        .to_numpy(
            dtype=np.float32
        )

    )


    validation_scaler = StandardScaler()


    X_train_scaled = (

        validation_scaler

        .fit_transform(
            X_train
        )

    )


    X_validation_scaled = (

        validation_scaler

        .transform(
            X_validation
        )

    )


    validation_model = Ridge(
        alpha=
            RIDGE_ALPHA
    )


    validation_model.fit(

        X_train_scaled,

        Y_train

    )


    validation_prediction = (

        validation_model

        .predict(
            X_validation_scaled
        )

    )


    # --------------------------------------------------------------
    # Final model
    # --------------------------------------------------------------

    X_train_val = (

        train_val_df[
            feature_columns
        ]

        .to_numpy(
            dtype=np.float32
        )

    )


    X_test = (

        test_df[
            feature_columns
        ]

        .to_numpy(
            dtype=np.float32
        )

    )


    final_scaler = StandardScaler()


    X_train_val_scaled = (

        final_scaler

        .fit_transform(
            X_train_val
        )

    )


    X_test_scaled = (

        final_scaler

        .transform(
            X_test
        )

    )


    final_model = Ridge(
        alpha=
            RIDGE_ALPHA
    )


    final_model.fit(

        X_train_val_scaled,

        Y_train_val

    )


    test_prediction = (

        final_model

        .predict(
            X_test_scaled
        )

    )


    return {

        "Validation_MAE_MW":
            mae(
                Y_validation,
                validation_prediction
            ),

        "Validation_RMSE_MW":
            rmse(
                Y_validation,
                validation_prediction
            ),

        "Test_MAE_MW":
            mae(
                Y_test,
                test_prediction
            ),

        "Test_RMSE_MW":
            rmse(
                Y_test,
                test_prediction
            ),

        "Test_MAPE_percent":
            mape(
                Y_test,
                test_prediction
            )

    }


# ======================================================================
# 16. REPRODUCE FULL WEATHER MODEL
# ======================================================================

print("\n" + "=" * 125)
print("REPRODUCING FULL WEATHER MODEL")
print("=" * 125)


full_weather_metrics = (
    evaluate_fixed_ridge(
        COMBINED_FEATURE_COLUMNS
    )
)


print(
    "Validation MAE:",
    round(
        full_weather_metrics[
            "Validation_MAE_MW"
        ],
        4
    )
)


print(
    "Test MAE:",
    round(
        full_weather_metrics[
            "Test_MAE_MW"
        ],
        4
    )
)


print(
    "Difference from W3:",
    round(

        full_weather_metrics[
            "Test_MAE_MW"
        ]

        -

        FULL_WEATHER_TEST_MAE,

        6
    ),
    "MW"
)


assert (

    abs(

        full_weather_metrics[
            "Test_MAE_MW"
        ]

        -

        FULL_WEATHER_TEST_MAE

    )

    <
    0.1

)


print(
    "Full weather reproduction PASSED."
)


# ======================================================================
# 17. LEAVE-ONE-WEATHER-GROUP-OUT ABLATION
# ======================================================================

print("\n" + "=" * 125)
print("POST-HOC WEATHER FEATURE-GROUP ABLATION")
print("=" * 125)


ablation_rows = []


# Full model reference

ablation_rows.append({

    "Ablation":
        "Full_Weather_Model",

    "Removed_Group":
        "None",

    "Features_Remaining":
        len(
            COMBINED_FEATURE_COLUMNS
        ),

    **full_weather_metrics,

    "Delta_Validation_MAE_vs_FullWeather_MW":
        0.0,

    "Delta_Test_MAE_vs_FullWeather_MW":
        0.0

})


# --------------------------------------------------------------
# Remove one weather group at a time
# --------------------------------------------------------------

for group_name, group_features in (
    WEATHER_GROUPS.items()
):

    feature_set = set(
        group_features
    )


    remaining_features = [

        feature

        for feature in (
            COMBINED_FEATURE_COLUMNS
        )

        if feature
        not in feature_set

    ]


    metrics = evaluate_fixed_ridge(
        remaining_features
    )


    validation_delta = (

        metrics[
            "Validation_MAE_MW"
        ]

        -

        full_weather_metrics[
            "Validation_MAE_MW"
        ]

    )


    test_delta = (

        metrics[
            "Test_MAE_MW"
        ]

        -

        full_weather_metrics[
            "Test_MAE_MW"
        ]

    )


    ablation_rows.append({

        "Ablation":
            f"Remove_{group_name}",

        "Removed_Group":
            group_name,

        "Features_Remaining":
            len(
                remaining_features
            ),

        **metrics,

        "Delta_Validation_MAE_vs_FullWeather_MW":
            validation_delta,

        "Delta_Test_MAE_vs_FullWeather_MW":
            test_delta

    })


    print(

        f"{group_name:40s}",

        f"Val={metrics['Validation_MAE_MW']:.2f}",

        f"Test={metrics['Test_MAE_MW']:.2f}",

        f"Test Δ={test_delta:+.2f}"

    )


# ======================================================================
# 18. REMOVE ALL WEATHER
# ======================================================================

baseline_metrics = (
    evaluate_fixed_ridge(
        BASE_FEATURE_COLUMNS
    )
)


ablation_rows.append({

    "Ablation":
        "Remove_All_Weather",

    "Removed_Group":
        "All_Weather",

    "Features_Remaining":
        len(
            BASE_FEATURE_COLUMNS
        ),

    **baseline_metrics,

    "Delta_Validation_MAE_vs_FullWeather_MW":

        baseline_metrics[
            "Validation_MAE_MW"
        ]

        -

        full_weather_metrics[
            "Validation_MAE_MW"
        ],

    "Delta_Test_MAE_vs_FullWeather_MW":

        baseline_metrics[
            "Test_MAE_MW"
        ]

        -

        full_weather_metrics[
            "Test_MAE_MW"
        ]

})


ablation_results = pd.DataFrame(
    ablation_rows
)


print("\n" + "=" * 125)
print("WEATHER ABLATION RESULTS")
print("=" * 125)


display(
    ablation_results.round(4)
)


# ======================================================================
# 19. ADD INTERPRETATION COLUMN
# ======================================================================
#
# Delta:
#
# > 0:
#   removing the group worsened the full-weather model
#   -> group contributed positively within this specification.
#
# < 0:
#   removing the group improved the full-weather model
#   -> group may be redundant/harmful.
#
# POST-HOC ONLY.
# ======================================================================

def interpret_ablation_delta(
    value
):

    if value > 1e-6:

        return (
            "Removal worsened performance"
        )


    if value < -1e-6:

        return (
            "Removal improved performance"
        )


    return (
        "No material numerical change"
    )


ablation_results[
    "PostHoc_Test_Interpretation"
] = (

    ablation_results[
        "Delta_Test_MAE_vs_FullWeather_MW"
    ]

    .apply(
        interpret_ablation_delta
    )

)


# ======================================================================
# 20. FULL WEATHER RIDGE COEFFICIENT DIAGNOSTICS
# ======================================================================
#
# Features are standardized.
#
# Thus coefficient magnitude represents approximate MW change
# associated with a 1-SD feature change, conditional on the
# correlated feature set.
#
# DO NOT interpret coefficients causally.
# ======================================================================

X_train_val_full = (

    train_val_df[
        COMBINED_FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=np.float32
    )

)


coef_scaler = StandardScaler()


X_train_val_full_scaled = (

    coef_scaler

    .fit_transform(
        X_train_val_full
    )

)


coef_model = Ridge(
    alpha=
        RIDGE_ALPHA
)


coef_model.fit(

    X_train_val_full_scaled,

    Y_train_val

)


coefficient_matrix = np.asarray(
    coef_model.coef_,
    dtype=float
)


# Expected:
# 23 targets × 254 predictors

assert (
    coefficient_matrix.shape[0]
    ==
    23
)


coefficient_rows = []


for group_name, group_features in (
    WEATHER_GROUPS.items()
):

    indices = [

        COMBINED_FEATURE_COLUMNS.index(
            feature
        )

        for feature in group_features

    ]


    group_coefficients = (

        coefficient_matrix[
            :,
            indices
        ]

    )


    coefficient_rows.append({

        "Weather_Group":
            group_name,

        "Feature_Count":
            len(
                group_features
            ),

        "Mean_Absolute_Standardized_Coefficient_MW":
            float(

                np.mean(

                    np.abs(
                        group_coefficients
                    )

                )

            ),

        "Median_Absolute_Standardized_Coefficient_MW":
            float(

                np.median(

                    np.abs(
                        group_coefficients
                    )

                )

            ),

        "Maximum_Absolute_Standardized_Coefficient_MW":
            float(

                np.max(

                    np.abs(
                        group_coefficients
                    )

                )

            )

    })


coefficient_summary = pd.DataFrame(
    coefficient_rows
)


coefficient_summary = (

    coefficient_summary

    .sort_values(

        "Mean_Absolute_Standardized_Coefficient_MW",

        ascending=False

    )

    .reset_index(
        drop=True
    )

)


print("\n" + "=" * 125)
print("WEATHER RIDGE COEFFICIENT DIAGNOSTICS")
print("=" * 125)


display(
    coefficient_summary.round(4)
)


# ======================================================================
# 21. MONTHLY PAIRED DIAGNOSTIC
# ======================================================================

daily_pair[
    "year"
] = (
    daily_pair[
        "target_date"
    ].dt.year
)


daily_pair[
    "month"
] = (
    daily_pair[
        "target_date"
    ].dt.month
)


daily_pair[
    "weather_improvement_MW"
] = (
    loss_difference
)


monthly_summary = (

    daily_pair

    .groupby(
        "month"
    )

    .agg(

        test_days=(
            "target_date",
            "size"
        ),

        mean_weather_improvement_MW=(
            "weather_improvement_MW",
            "mean"
        ),

        median_weather_improvement_MW=(
            "weather_improvement_MW",
            "median"
        ),

        weather_win_rate_percent=(
            "weather_improvement_MW",
            lambda x:
                100
                *
                np.mean(
                    x > 0
                )
        )

    )

    .reset_index()

)


print("\n" + "=" * 125)
print("MONTH-OF-YEAR WEATHER EFFECT")
print("=" * 125)


display(
    monthly_summary.round(4)
)


# ======================================================================
# 22. SAVE RESULTS
# ======================================================================

BOOTSTRAP_FILE = (
    RESULT_DIR
    / "weather_paired_block_bootstrap.csv"
)


HAC_FILE = (
    RESULT_DIR
    / "weather_paired_hac_tests.csv"
)


ABLATION_FILE = (
    RESULT_DIR
    / "weather_feature_group_ablation.csv"
)


COEFFICIENT_FILE = (
    RESULT_DIR
    / "weather_ridge_coefficient_groups.csv"
)


MONTHLY_FILE = (
    RESULT_DIR
    / "weather_monthly_effect.csv"
)


bootstrap_results.to_csv(
    BOOTSTRAP_FILE,
    index=False
)


hac_results.to_csv(
    HAC_FILE,
    index=False
)


ablation_results.to_csv(
    ABLATION_FILE,
    index=False
)


coefficient_summary.to_csv(
    COEFFICIENT_FILE,
    index=False
)


monthly_summary.to_csv(
    MONTHLY_FILE,
    index=False
)


# ======================================================================
# 23. FIGURE — ABLATION
# ======================================================================

ablation_plot = (

    ablation_results.loc[

        ~ablation_results[
            "Removed_Group"
        ]
        .isin(
            [
                "None",
                "All_Weather"
            ]
        )

    ]

    .copy()

)


ablation_plot = (

    ablation_plot

    .sort_values(
        "Delta_Test_MAE_vs_FullWeather_MW"
    )

)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.barh(

    ablation_plot[
        "Removed_Group"
    ],

    ablation_plot[
        "Delta_Test_MAE_vs_FullWeather_MW"
    ]

)


plt.axvline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Change in Test MAE After Removing Weather Group (MW)"
)


plt.ylabel(
    "Removed Weather Feature Group"
)


plt.title(
    "Post-Hoc Historical-Weather Feature-Group Ablation"
)


plt.grid(
    axis="x",
    alpha=0.25
)


plt.tight_layout()


ABLATION_FIGURE = (

    FIGURE_DIR
    / "weather_feature_group_ablation.png"

)


plt.savefig(

    ABLATION_FIGURE,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# 24. FIGURE — MONTHLY WEATHER EFFECT
# ======================================================================

plt.figure(
    figsize=(
        10,
        5
    )
)


plt.bar(

    monthly_summary[
        "month"
    ],

    monthly_summary[
        "mean_weather_improvement_MW"
    ]

)


plt.axhline(
    0,
    linestyle="--"
)


plt.xlabel(
    "Month"
)


plt.ylabel(
    "Mean Daily MAE Improvement from Weather (MW)"
)


plt.title(
    "Month-of-Year Historical-Weather Forecast Effect"
)


plt.xticks(
    range(
        1,
        13
    )
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


MONTHLY_FIGURE = (

    FIGURE_DIR
    / "weather_monthly_effect.png"

)


plt.savefig(

    MONTHLY_FIGURE,

    dpi=300,

    bbox_inches="tight"

)


plt.show()


# ======================================================================
# 25. SUMMARY
# ======================================================================

summary = {

    "step":
        "W4",

    "analysis_status":
        "post_hoc_diagnostic_only",

    "test_days":
        int(
            len(
                daily_pair
            )
        ),

    "baseline_test_MAE_MW":
        float(
            BASELINE_TEST_MAE
        ),

    "full_weather_test_MAE_MW":
        float(
            FULL_WEATHER_TEST_MAE
        ),

    "mean_daily_weather_improvement_MW":
        float(
            MEAN_DIFFERENCE
        ),

    "weather_win_rate_percent":
        float(
            WEATHER_WIN_RATE
        ),

    "bootstrap7_CI_lower_MW":
        float(
            CI7_LOWER
        ),

    "bootstrap7_CI_upper_MW":
        float(
            CI7_UPPER
        ),

    "bootstrap14_CI_lower_MW":
        float(
            CI14_LOWER
        ),

    "bootstrap14_CI_upper_MW":
        float(
            CI14_UPPER
        ),

    "HAC7_p_value":
        float(
            hac7[
                "p_value"
            ]
        ),

    "HAC14_p_value":
        float(
            hac14[
                "p_value"
            ]
        ),

    "interpretation":
        SIGNIFICANCE_INTERPRETATION,

    "selection_warning":
        (
            "Weather subgroup results are post-hoc diagnostics. "
            "Do not redefine the frozen final model from test ablations."
        )

}


SUMMARY_FILE = (

    RESULT_DIR
    / "stepW4_summary.json"

)


with open(
    SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# ======================================================================
# 26. FINAL OUTPUT
# ======================================================================

print("\n")
print("=" * 125)
print("STEP W4 COMPLETED SUCCESSFULLY")
print("=" * 125)


print("\nA. FULL WEATHER EFFECT")

print(
    "Frozen Ridge MAE:",
    round(
        BASELINE_TEST_MAE,
        3
    ),
    "MW"
)


print(
    "Historical-weather Ridge MAE:",
    round(
        FULL_WEATHER_TEST_MAE,
        3
    ),
    "MW"
)


print(
    "Mean daily weather improvement:",
    round(
        MEAN_DIFFERENCE,
        3
    ),
    "MW"
)


print(
    "Weather wins on:",
    round(
        WEATHER_WIN_RATE,
        2
    ),
    "% of test days"
)


print("\nB. BLOCK-BOOTSTRAP 95% CI")


print(
    "7-day:",
    round(
        CI7_LOWER,
        3
    ),
    "to",
    round(
        CI7_UPPER,
        3
    ),
    "MW"
)


print(
    "14-day:",
    round(
        CI14_LOWER,
        3
    ),
    "to",
    round(
        CI14_UPPER,
        3
    ),
    "MW"
)


print("\nC. HAC TESTS")


def format_p(
    p
):

    if not np.isfinite(
        p
    ):

        return "NA"

    if p < 0.001:

        return "<0.001"

    return f"{p:.4f}"


print(
    "HAC-7 p:",
    format_p(
        hac7[
            "p_value"
        ]
    )
)


print(
    "HAC-14 p:",
    format_p(
        hac14[
            "p_value"
        ]
    )
)


print("\nD. INTERPRETATION")

print(
    SIGNIFICANCE_INTERPRETATION
)


print("\nIMPORTANT:")

print(
    "Weather ablations are POST-HOC diagnostics only."
)

print(
    "Do not use locked-test ablation results to replace "
    "the frozen primary Ridge model."
)


print("\nOutput files:")

print(
    BOOTSTRAP_FILE
)

print(
    HAC_FILE
)

print(
    ABLATION_FILE
)

print(
    COEFFICIENT_FILE
)

print(
    MONTHLY_FILE
)

print(
    SUMMARY_FILE
)


print("=" * 125)
```
