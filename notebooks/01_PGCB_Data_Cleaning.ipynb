```python
# ================================================================
# STEP 1B
# PHYSICS-AWARE DATA CLEANING AND QUALITY AUDIT
# PGCB HOURLY ELECTRICITY DATASET
#
# Research project:
# Reliable Multi-Horizon Electricity Load Forecasting in Bangladesh
#
# Run this entire block in ONE Google Colab cell.
# ================================================================


# ================================================================
# 0. IMPORT LIBRARIES
# ================================================================

import os
import re
import json
import math
import shutil
import zipfile
import warnings

from pathlib import Path
from datetime import timedelta

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 160)

print("=" * 80)
print("STEP 1B: PGCB PHYSICS-AWARE DATA CLEANING")
print("=" * 80)


# ================================================================
# 1. MOUNT GOOGLE DRIVE
# ================================================================

from google.colab import drive

drive.mount("/content/drive")

print("\nGoogle Drive mounted successfully.")


# ================================================================
# 2. CREATE PROJECT DIRECTORIES
# ================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
QUALITY_DIR = PROJECT_ROOT / "data" / "quality_logs"

FIGURE_DIR = PROJECT_ROOT / "figures" / "data_cleaning"
RESULT_DIR = PROJECT_ROOT / "results" / "data_cleaning"

for directory in [
    RAW_DIR,
    PROCESSED_DIR,
    QUALITY_DIR,
    FIGURE_DIR,
    RESULT_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

print("\nProject directories created:")
print(PROJECT_ROOT)


# ================================================================
# 3. UPLOAD THE HOURLY DATASET
# ================================================================

from google.colab import files

print("\n" + "=" * 80)
print("UPLOAD YOUR HOURLY PGCB ZIP OR CSV FILE NOW")
print("=" * 80)

uploaded = files.upload()

if len(uploaded) == 0:
    raise RuntimeError("No file was uploaded.")

uploaded_names = list(uploaded.keys())

print("\nUploaded:")
for name in uploaded_names:
    print(" -", name)


# ================================================================
# 4. COPY UPLOADED FILE TO GOOGLE DRIVE
# ================================================================

for filename in uploaded_names:

    source = Path("/content") / filename
    destination = RAW_DIR / filename

    shutil.copy2(source, destination)

print("\nRaw uploaded file copied to:")
print(RAW_DIR)


# ================================================================
# 5. FIND / EXTRACT THE PGCB CSV AUTOMATICALLY
# ================================================================

WORK_DIR = Path("/content/pgcb_work")

if WORK_DIR.exists():
    shutil.rmtree(WORK_DIR)

WORK_DIR.mkdir(parents=True, exist_ok=True)


csv_candidates = []


for filename in uploaded_names:

    path = Path("/content") / filename

    # ------------------------------------------------------------
    # ZIP INPUT
    # ------------------------------------------------------------

    if filename.lower().endswith(".zip"):

        print(f"\nExtracting ZIP: {filename}")

        with zipfile.ZipFile(path, "r") as z:
            z.extractall(WORK_DIR)

    # ------------------------------------------------------------
    # CSV INPUT
    # ------------------------------------------------------------

    elif filename.lower().endswith(".csv"):

        csv_candidates.append(path)


# Search recursively after extraction

csv_candidates.extend(
    list(WORK_DIR.rglob("*.csv"))
)


if len(csv_candidates) == 0:

    raise FileNotFoundError(
        "No CSV file could be found inside the uploaded file."
    )


print("\nCSV candidates found:")

for p in csv_candidates:
    print(" -", p)


# ================================================================
# 6. IDENTIFY THE CORRECT PGCB FILE
# ================================================================

pgcb_path = None

required_columns = {
    "datetime",
    "generation_mw",
    "demand_mw",
    "load_shedding"
}


for candidate in csv_candidates:

    try:

        temp = pd.read_csv(candidate, nrows=5)

        cols = set(
            temp.columns.astype(str).str.strip()
        )

        if required_columns.issubset(cols):

            pgcb_path = candidate
            break

    except Exception:
        pass


if pgcb_path is None:

    raise RuntimeError(
        "Could not identify the PGCB hourly CSV automatically."
    )


print("\nCorrect PGCB dataset detected:")
print(pgcb_path)


# Copy actual CSV to Google Drive

raw_csv_drive_path = RAW_DIR / "pgcb_V1_raw.csv"

shutil.copy2(
    pgcb_path,
    raw_csv_drive_path
)

print("\nOriginal CSV preserved at:")
print(raw_csv_drive_path)


# ================================================================
# 7. LOAD DATA
# ================================================================

df = pd.read_csv(
    pgcb_path,
    low_memory=False
)

print("\n" + "=" * 80)
print("RAW DATASET LOADED")
print("=" * 80)

print("Rows    :", len(df))
print("Columns :", len(df.columns))

print("\nColumn names:")

for col in df.columns:
    print(" -", col)


print("\nFirst 5 observations:")
display(df.head())


# ================================================================
# 8. CLEAN COLUMN NAMES
# ================================================================

df.columns = (
    df.columns
      .astype(str)
      .str.strip()
      .str.lower()
      .str.replace(" ", "_", regex=False)
)


# ================================================================
# 9. CHECK REQUIRED VARIABLES
# ================================================================

required_columns = [
    "datetime",
    "generation_mw",
    "demand_mw",
    "load_shedding"
]


missing_required = [
    c for c in required_columns
    if c not in df.columns
]


if missing_required:

    raise ValueError(
        f"Required columns missing: {missing_required}"
    )


# ================================================================
# 10. PARSE TIMESTAMP
# ================================================================

df["datetime"] = pd.to_datetime(
    df["datetime"],
    errors="coerce"
)


bad_timestamp_count = (
    df["datetime"].isna().sum()
)


print("\nInvalid timestamps:",
      bad_timestamp_count)


if bad_timestamp_count > 0:

    print(
        "Rows with invalid timestamps will be preserved in the "
        "quality log but cannot be used for forecasting."
    )


# ================================================================
# 11. IDENTIFY NUMERIC VARIABLES
# ================================================================

numeric_columns = [

    "generation_mw",
    "demand_mw",
    "load_shedding",

    "gas",
    "liquid_fuel",
    "coal",
    "hydro",
    "solar",
    "wind",

    "india_bheramara_hvdc",
    "india_tripura",
    "india_adani",
    "nepal"
]


numeric_columns = [
    c for c in numeric_columns
    if c in df.columns
]


# Convert to numerical values

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


print("\nNumeric conversion completed.")


# ================================================================
# 12. SORT DATA CHRONOLOGICALLY
# ================================================================

df = (
    df
    .sort_values("datetime")
    .reset_index(drop=True)
)


# Create permanent original row identifier

df["source_row_id"] = np.arange(
    1,
    len(df) + 1
)


# ================================================================
# 13. BASIC TIME VARIABLES
# ================================================================

df["date"] = df["datetime"].dt.date

df["year"] = df["datetime"].dt.year

df["month"] = df["datetime"].dt.month

df["day"] = df["datetime"].dt.day

df["hour"] = df["datetime"].dt.hour

df["minute"] = df["datetime"].dt.minute

df["weekday"] = df["datetime"].dt.dayofweek

df["time_of_day"] = (
    df["datetime"]
    .dt.strftime("%H:%M")
)


# ================================================================
# 14. IDENTIFY OBSERVATION TYPE
# ================================================================

# PGCB dataset contains ordinary hourly observations plus
# a special evening-peak observation around 19:30.

df["is_regular_hour"] = (

    df["datetime"].notna()

    &

    (df["minute"] == 0)

)


df["is_1930_peak"] = (

    df["datetime"].notna()

    &

    (df["hour"] == 19)

    &

    (df["minute"] == 30)

)


# Some datasets also explicitly mark Evening_Peak in remarks.

if "remarks" in df.columns:

    remark_norm = (
        df["remarks"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
    )

    df["remarks_evening_peak"] = (
        remark_norm.str.contains(
            "evening",
            regex=False
        )
    )

else:

    df["remarks_evening_peak"] = False


df["is_evening_peak"] = (

    df["is_1930_peak"]

    |

    df["remarks_evening_peak"]

)


df["observation_type"] = "other"

df.loc[
    df["is_regular_hour"],
    "observation_type"
] = "regular_hour"


df.loc[
    df["is_evening_peak"],
    "observation_type"
] = "evening_peak"


print("\nObservation-type counts:")

print(
    df["observation_type"]
    .value_counts(dropna=False)
)


# ================================================================
# 15. DUPLICATE TIMESTAMP ANALYSIS
# ================================================================

df["duplicate_timestamp"] = (

    df["datetime"]
    .duplicated(keep=False)

    &

    df["datetime"].notna()

)


duplicate_count = (
    df["duplicate_timestamp"].sum()
)


print("\nDuplicate timestamp rows:",
      duplicate_count)


# ================================================================
# 16. TIMESTAMP INTERVAL ANALYSIS
# ================================================================

df["interval_minutes"] = (

    df["datetime"]
    .diff()
    .dt.total_seconds()
    .div(60)

)


interval_summary = (
    df["interval_minutes"]
    .value_counts()
    .sort_index()
    .reset_index()
)

interval_summary.columns = [
    "interval_minutes",
    "count"
]


print("\nMost common timestamp intervals:")

display(
    interval_summary
    .sort_values(
        "count",
        ascending=False
    )
    .head(15)
)


# ================================================================
# 17. COMPLETE MISSING CALENDAR DAYS
# ================================================================

valid_dates = (
    pd.Series(
        df["datetime"]
        .dropna()
        .dt.normalize()
        .unique()
    )
    .sort_values()
)


date_min = valid_dates.min()

date_max = valid_dates.max()


full_dates = pd.date_range(
    start=date_min,
    end=date_max,
    freq="D"
)


missing_full_days = (

    full_dates.difference(
        pd.DatetimeIndex(valid_dates)
    )

)


print("\nDataset period:")

print(
    date_min.date(),
    "to",
    date_max.date()
)


print(
    "\nCompletely missing calendar days:",
    len(missing_full_days)
)


if len(missing_full_days) > 0:

    display(
        pd.DataFrame(
            {
                "missing_date":
                    missing_full_days
            }
        )
    )


# ================================================================
# 18. MISSING-VALUE REPORT
# ================================================================

missing_report = pd.DataFrame({

    "column": df.columns,

    "missing_count": [
        df[c].isna().sum()
        for c in df.columns
    ],

    "missing_percent": [
        100 * df[c].isna().mean()
        for c in df.columns
    ]

})


missing_report = (
    missing_report
    .sort_values(
        "missing_percent",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\nMissing-value report:")

display(
    missing_report.head(30)
)


# ================================================================
# 19. IMPORTANT:
# DO NOT INTERPRET ALL MISSING GENERATION SOURCES AS MISSING DATA
# ================================================================

# Solar, wind, Adani, Nepal, etc. may only exist in later
# periods of the historical record.
#
# Therefore we find their first non-missing observation.


availability_report = []


for col in numeric_columns:

    nonmissing = df.loc[
        df[col].notna(),
        ["datetime", col]
    ]

    if len(nonmissing) > 0:

        first_date = (
            nonmissing["datetime"].min()
        )

        last_date = (
            nonmissing["datetime"].max()
        )

    else:

        first_date = pd.NaT
        last_date = pd.NaT


    availability_report.append({

        "feature": col,

        "missing_count":
            int(df[col].isna().sum()),

        "missing_percent":
            100 * df[col].isna().mean(),

        "first_available":
            first_date,

        "last_available":
            last_date
    })


availability_report = pd.DataFrame(
    availability_report
)


print("\nFeature availability report:")

display(availability_report)


# ================================================================
# 20. PHYSICAL CONSISTENCY RELATION
# ================================================================

# Fundamental operational relationship:
#
# Demand ≈ Generation + Load Shedding
#
# We do NOT assume perfect equality due to reporting,
# rounding and operational differences.


df["supply_plus_shedding"] = (

    df["generation_mw"]

    +

    df["load_shedding"]
)


df["physical_residual_mw"] = (

    df["demand_mw"]

    -

    df["supply_plus_shedding"]
)


df["physical_abs_error_mw"] = (

    df["physical_residual_mw"]
    .abs()
)


# Relative physical discrepancy.
#
# We use max(demand, generation+LS, 1)
# so division remains numerically stable.

physical_scale = pd.concat(
    [
        df["demand_mw"].abs(),

        df["supply_plus_shedding"]
          .abs(),

        pd.Series(
            1,
            index=df.index
        )
    ],
    axis=1
).max(axis=1)


df["physical_relative_error"] = (

    df["physical_abs_error_mw"]

    /

    physical_scale
)


print("\nPhysical consistency error:")

print(
    df["physical_abs_error_mw"]
    .describe(
        percentiles=[
            .50,
            .90,
            .95,
            .99,
            .999
        ]
    )
)


# ================================================================
# 21. DEFINE SEVERE PHYSICAL INCONSISTENCY
# ================================================================

# Conservative criterion:
#
# 1. Absolute mismatch > 1000 MW
#
# AND
#
# 2. Relative mismatch > 15%
#
# This does NOT classify small rounding/reporting differences
# as corruption.


df["severe_physical_inconsistency"] = (

    (df["physical_abs_error_mw"] > 1000)

    &

    (df["physical_relative_error"] > 0.15)

)


print(
    "\nSevere physical inconsistencies:",
    df["severe_physical_inconsistency"].sum()
)


# ================================================================
# 22. BASIC PHYSICAL IMPOSSIBILITY FLAGS
# ================================================================

df["demand_nonpositive"] = (

    df["demand_mw"].isna()

    |

    (df["demand_mw"] <= 0)

)


df["generation_negative"] = (

    df["generation_mw"].notna()

    &

    (df["generation_mw"] < 0)

)


df["load_shedding_negative"] = (

    df["load_shedding"].notna()

    &

    (df["load_shedding"] < 0)

)


# ================================================================
# 23. TEMPORAL REFERENCE:
# SAME TIME-OF-DAY LOCAL MEDIAN
# ================================================================

# We now compare each observation with observations at the
# same time of day around nearby dates.
#
# Example:
#
# 19:00 is compared mostly with nearby 19:00 observations.
#
# This avoids comparing a 3 AM load directly against a
# normal 8 PM peak.


def centered_same_time_median(
    dataframe,
    column,
    window=29
):

    result = pd.Series(
        np.nan,
        index=dataframe.index,
        dtype=float
    )

    for time_value, idx in (
        dataframe
        .groupby("time_of_day")
        .groups
        .items()
    ):

        indices = list(idx)

        temp = dataframe.loc[
            indices,
            column
        ]

        med = (
            temp
            .rolling(
                window=window,
                center=True,
                min_periods=7
            )
            .median()
        )

        result.loc[indices] = med.values

    return result


print(
    "\nCalculating local temporal reference..."
)


df["demand_local_median"] = (
    centered_same_time_median(
        df,
        "demand_mw",
        window=29
    )
)


df["generation_local_median"] = (
    centered_same_time_median(
        df,
        "generation_mw",
        window=29
    )
)


# ================================================================
# 24. LOCAL TEMPORAL DEVIATION
# ================================================================

df["demand_local_relative_deviation"] = (

    (
        df["demand_mw"]

        -

        df["demand_local_median"]
    )
    .abs()

    /

    df["demand_local_median"]
      .abs()
      .clip(lower=1)
)


df["generation_local_relative_deviation"] = (

    (
        df["generation_mw"]

        -

        df["generation_local_median"]
    )
    .abs()

    /

    df["generation_local_median"]
      .abs()
      .clip(lower=1)
)


# ================================================================
# 25. TEMPORAL EXTREME FLAGS
# ================================================================

# IMPORTANT:
#
# These are NOT automatically called corrupted.
#
# Extreme values may represent genuine:
#
# - grid stress
# - supply shortage
# - holidays
# - special events
# - unusual operating regimes


df["demand_temporal_extreme"] = (

    (
        df["demand_local_relative_deviation"]
        > 0.60
    )

    &

    (
        (
            df["demand_mw"]

            -

            df["demand_local_median"]
        )
        .abs()
        > 2000
    )
)


df["generation_temporal_extreme"] = (

    (
        df["generation_local_relative_deviation"]
        > 0.60
    )

    &

    (
        (
            df["generation_mw"]

            -

            df["generation_local_median"]
        )
        .abs()
        > 2000
    )
)


# ================================================================
# 26. DETERMINE WHICH FIELD IS PROBABLY CORRUPTED
# ================================================================

# We only make a high-confidence field-level corruption decision
# when physical consistency is severely violated.
#
# If demand is enormously abnormal relative to its local history,
# while generation remains reasonable:
#
# -> demand probably corrupted.
#
# If generation is enormously abnormal but demand reasonable:
#
# -> generation probably corrupted.


df["demand_corrupt"] = False

df["generation_corrupt"] = False

df["load_shedding_corrupt"] = False


# ------------------------------------------------
# Demand likely corrupt
# ------------------------------------------------

condition_demand_corrupt = (

    df["severe_physical_inconsistency"]

    &

    (
        df["demand_local_relative_deviation"]
        >
        (
            2.0
            *
            df["generation_local_relative_deviation"]
        )
    )

    &

    (
        df["demand_local_relative_deviation"]
        > 0.60
    )
)


df.loc[
    condition_demand_corrupt,
    "demand_corrupt"
] = True


# ------------------------------------------------
# Generation likely corrupt
# ------------------------------------------------

condition_generation_corrupt = (

    df["severe_physical_inconsistency"]

    &

    (
        df["generation_local_relative_deviation"]
        >
        (
            2.0
            *
            df["demand_local_relative_deviation"]
        )
    )

    &

    (
        df["generation_local_relative_deviation"]
        > 0.60
    )
)


df.loc[
    condition_generation_corrupt,
    "generation_corrupt"
] = True


# ------------------------------------------------
# Load-shedding likely corrupt
# ------------------------------------------------

# When demand and generation are both near their local
# normal values, but the physical relation is severely
# broken, load-shedding is the likely incorrect field.

condition_ls_corrupt = (

    df["severe_physical_inconsistency"]

    &

    (~df["demand_corrupt"])

    &

    (~df["generation_corrupt"])

    &

    (
        df["demand_local_relative_deviation"]
        < 0.40
    )

    &

    (
        df["generation_local_relative_deviation"]
        < 0.40
    )

    &

    (
        df["load_shedding"]
        >
        1000
    )
)


df.loc[
    condition_ls_corrupt,
    "load_shedding_corrupt"
] = True


# ================================================================
# 27. NONPOSITIVE/NEGATIVE VALUES = DEFINITE FIELD PROBLEMS
# ================================================================

df.loc[
    df["demand_nonpositive"],
    "demand_corrupt"
] = True


df.loc[
    df["generation_negative"],
    "generation_corrupt"
] = True


df.loc[
    df["load_shedding_negative"],
    "load_shedding_corrupt"
] = True


# ================================================================
# 28. UNRESOLVED PHYSICAL INCONSISTENCY
# ================================================================

df["physical_review_required"] = (

    df["severe_physical_inconsistency"]

    &

    (~df["demand_corrupt"])

    &

    (~df["generation_corrupt"])

    &

    (~df["load_shedding_corrupt"])

)


# ================================================================
# 29. PRESERVE PHYSICALLY CONSISTENT EXTREME EVENTS
# ================================================================

df["contextual_extreme_keep"] = (

    (
        df["demand_temporal_extreme"]

        |

        df["generation_temporal_extreme"]
    )

    &

    (~df["severe_physical_inconsistency"])

)


# ================================================================
# 30. CREATE QUALITY LABEL
# ================================================================

df["quality_status"] = "CLEAN"


df.loc[
    df["contextual_extreme_keep"],
    "quality_status"
] = "CONTEXTUAL_EXTREME_KEEP"


df.loc[
    df["physical_review_required"],
    "quality_status"
] = "PHYSICAL_INCONSISTENCY_REVIEW"


high_confidence_corruption = (

    df["demand_corrupt"]

    |

    df["generation_corrupt"]

    |

    df["load_shedding_corrupt"]
)


df.loc[
    high_confidence_corruption,
    "quality_status"
] = "HIGH_CONFIDENCE_CORRUPTION"


df.loc[
    df["duplicate_timestamp"],
    "quality_status"
] = "DUPLICATE_TIMESTAMP"


df.loc[
    df["datetime"].isna(),
    "quality_status"
] = "INVALID_TIMESTAMP"


# ================================================================
# 31. CREATE HUMAN-READABLE REASONS
# ================================================================

def build_reason(row):

    reasons = []

    if row["demand_corrupt"]:
        reasons.append(
            "demand_field_likely_corrupted"
        )

    if row["generation_corrupt"]:
        reasons.append(
            "generation_field_likely_corrupted"
        )

    if row["load_shedding_corrupt"]:
        reasons.append(
            "load_shedding_field_likely_corrupted"
        )

    if row["severe_physical_inconsistency"]:
        reasons.append(
            "severe_physical_balance_violation"
        )

    if row["demand_temporal_extreme"]:
        reasons.append(
            "unusual_demand_vs_local_history"
        )

    if row["generation_temporal_extreme"]:
        reasons.append(
            "unusual_generation_vs_local_history"
        )

    if row["contextual_extreme_keep"]:
        reasons.append(
            "extreme_but_physically_consistent_keep"
        )

    if row["physical_review_required"]:
        reasons.append(
            "manual_review_required"
        )

    if row["duplicate_timestamp"]:
        reasons.append(
            "duplicate_timestamp"
        )

    if pd.isna(row["datetime"]):
        reasons.append(
            "invalid_timestamp"
        )

    if len(reasons) == 0:
        return "none"

    return ";".join(reasons)


df["quality_reason"] = df.apply(
    build_reason,
    axis=1
)


# ================================================================
# 32. CREATE CLEAN FIELD COPIES
# ================================================================

# We DO NOT overwrite the original measurements.
#
# Original columns remain untouched.
#
# Suspected corrupted field values become NaN in the
# *_clean variables.


df["demand_mw_clean"] = (
    df["demand_mw"].copy()
)


df.loc[
    df["demand_corrupt"],
    "demand_mw_clean"
] = np.nan


df["generation_mw_clean"] = (
    df["generation_mw"].copy()
)


df.loc[
    df["generation_corrupt"],
    "generation_mw_clean"
] = np.nan


df["load_shedding_clean"] = (
    df["load_shedding"].copy()
)


df.loc[
    df["load_shedding_corrupt"],
    "load_shedding_clean"
] = np.nan


# ================================================================
# 33. MODEL-TARGET USABILITY
# ================================================================

# Demand target can be used if:
#
# - timestamp valid
# - demand exists
# - demand itself is NOT identified as corrupted
# - timestamp is not duplicate
#
# Importantly:
#
# A corrupted generation field does NOT automatically
# force us to discard a valid demand target.


df["target_usable"] = (

    df["datetime"].notna()

    &

    df["demand_mw_clean"].notna()

    &

    (~df["duplicate_timestamp"])

)


# ================================================================
# 34. REGULAR-HOUR MODEL DATASET
# ================================================================

regular_clean = (

    df.loc[
        df["is_regular_hour"]
        &
        df["target_usable"]
    ]
    .copy()
)


# We preserve unusual physically consistent events.

regular_clean = (
    regular_clean
    .sort_values("datetime")
    .reset_index(drop=True)
)


# ================================================================
# 35. EVENING-PEAK DATASET
# ================================================================

evening_peak_clean = (

    df.loc[
        df["is_evening_peak"]
        &
        df["target_usable"]
    ]
    .copy()
)


evening_peak_clean = (
    evening_peak_clean
    .sort_values("datetime")
    .reset_index(drop=True)
)


# ================================================================
# 36. QUALITY LOG
# ================================================================

quality_log = (

    df.loc[
        df["quality_status"]
        != "CLEAN"
    ]
    .copy()
)


quality_log_columns = [

    "source_row_id",
    "datetime",

    "generation_mw",
    "demand_mw",
    "load_shedding",

    "physical_residual_mw",
    "physical_abs_error_mw",
    "physical_relative_error",

    "demand_local_median",
    "generation_local_median",

    "demand_local_relative_deviation",
    "generation_local_relative_deviation",

    "demand_corrupt",
    "generation_corrupt",
    "load_shedding_corrupt",

    "contextual_extreme_keep",

    "quality_status",
    "quality_reason"
]


quality_log_columns = [
    c for c in quality_log_columns
    if c in quality_log.columns
]


quality_log = quality_log[
    quality_log_columns
]


# ================================================================
# 37. QUALITY SUMMARY
# ================================================================

quality_summary = (

    df["quality_status"]
    .value_counts(dropna=False)
    .rename_axis("quality_status")
    .reset_index(name="row_count")
)


quality_summary["percent"] = (

    100
    *
    quality_summary["row_count"]
    /
    len(df)
)


print("\n" + "=" * 80)
print("QUALITY STATUS SUMMARY")
print("=" * 80)

display(quality_summary)


# ================================================================
# 38. FIELD-LEVEL CORRUPTION SUMMARY
# ================================================================

field_summary = pd.DataFrame({

    "issue": [

        "Demand field likely corrupted",

        "Generation field likely corrupted",

        "Load-shedding field likely corrupted",

        "Severe physical inconsistency",

        "Physical inconsistency requiring review",

        "Contextual extreme preserved",

        "Duplicate timestamp",

        "Usable demand target"
    ],

    "count": [

        int(df["demand_corrupt"].sum()),

        int(df["generation_corrupt"].sum()),

        int(df["load_shedding_corrupt"].sum()),

        int(
            df[
                "severe_physical_inconsistency"
            ].sum()
        ),

        int(
            df[
                "physical_review_required"
            ].sum()
        ),

        int(
            df[
                "contextual_extreme_keep"
            ].sum()
        ),

        int(
            df[
                "duplicate_timestamp"
            ].sum()
        ),

        int(
            df[
                "target_usable"
            ].sum()
        )
    ]
})


field_summary["percent_of_all_rows"] = (

    100
    *
    field_summary["count"]
    /
    len(df)
)


print("\nField-level summary:")

display(field_summary)


# ================================================================
# 39. SHOW LARGEST PHYSICAL ERRORS
# ================================================================

largest_errors = (

    df[
        [
            "datetime",
            "generation_mw",
            "demand_mw",
            "load_shedding",
            "physical_residual_mw",
            "physical_abs_error_mw",
            "quality_status",
            "quality_reason"
        ]
    ]

    .sort_values(
        "physical_abs_error_mw",
        ascending=False
    )

    .head(30)
)


print("\nLargest physical consistency violations:")

display(largest_errors)


# ================================================================
# 40. SHOW CONTEXTUAL EXTREMES THAT WE ARE KEEPING
# ================================================================

extreme_keep_examples = (

    df.loc[
        df["contextual_extreme_keep"],

        [
            "datetime",
            "generation_mw",
            "demand_mw",
            "load_shedding",

            "demand_local_median",

            "physical_abs_error_mw",

            "quality_status"
        ]
    ]

    .head(30)
)


print(
    "\nExamples of extreme observations PRESERVED rather than deleted:"
)

display(extreme_keep_examples)


# ================================================================
# 41. DAILY RECORD COUNT REPORT
# ================================================================

daily_counts = (

    df.dropna(subset=["datetime"])

    .groupby(
        df["datetime"].dt.date
    )

    .size()

    .rename("records_per_day")

    .reset_index()

)


daily_counts.columns = [
    "date",
    "records_per_day"
]


print("\nRecords per calendar day:")

display(
    daily_counts[
        "records_per_day"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "records_per_day"
    )
    .reset_index(
        name="number_of_days"
    )
)


# ================================================================
# 42. SAVE CSV FILES
# ================================================================

flagged_path = (
    PROCESSED_DIR
    /
    "pgcb_all_rows_quality_flagged.csv"
)


regular_path = (
    PROCESSED_DIR
    /
    "pgcb_regular_hourly_clean.csv"
)


peak_path = (
    PROCESSED_DIR
    /
    "pgcb_evening_peak_clean.csv"
)


quality_log_path = (
    QUALITY_DIR
    /
    "pgcb_quality_log.csv"
)


quality_summary_path = (
    RESULT_DIR
    /
    "quality_status_summary.csv"
)


field_summary_path = (
    RESULT_DIR
    /
    "field_corruption_summary.csv"
)


missing_report_path = (
    RESULT_DIR
    /
    "missing_value_report.csv"
)


availability_path = (
    RESULT_DIR
    /
    "feature_availability_report.csv"
)


interval_path = (
    RESULT_DIR
    /
    "timestamp_interval_report.csv"
)


missing_days_path = (
    RESULT_DIR
    /
    "missing_full_days.csv"
)


df.to_csv(
    flagged_path,
    index=False
)


regular_clean.to_csv(
    regular_path,
    index=False
)


evening_peak_clean.to_csv(
    peak_path,
    index=False
)


quality_log.to_csv(
    quality_log_path,
    index=False
)


quality_summary.to_csv(
    quality_summary_path,
    index=False
)


field_summary.to_csv(
    field_summary_path,
    index=False
)


missing_report.to_csv(
    missing_report_path,
    index=False
)


availability_report.to_csv(
    availability_path,
    index=False
)


interval_summary.to_csv(
    interval_path,
    index=False
)


pd.DataFrame(
    {
        "missing_date":
            missing_full_days
    }
).to_csv(
    missing_days_path,
    index=False
)


# ================================================================
# 43. FIGURE 1:
# FULL DEMAND TIME SERIES
# ================================================================

plt.figure(
    figsize=(16, 6)
)

plt.plot(
    df["datetime"],
    df["demand_mw"],
    linewidth=0.6,
    alpha=0.75,
    label="Raw demand"
)


plt.xlabel("Date")
plt.ylabel("Demand (MW)")

plt.title(
    "PGCB National Electricity Demand: Raw Historical Series"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()


figure1_path = (
    FIGURE_DIR
    /
    "01_raw_demand_time_series.png"
)


plt.savefig(
    figure1_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 44. FIGURE 2:
# CLEAN TARGET TIME SERIES
# ================================================================

plt.figure(
    figsize=(16, 6)
)


plt.plot(
    regular_clean["datetime"],
    regular_clean["demand_mw_clean"],
    linewidth=0.6,
    alpha=0.80
)


plt.xlabel("Date")
plt.ylabel("Demand (MW)")

plt.title(
    "PGCB National Electricity Demand After High-Confidence "
    "Target Corruption Filtering"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()


figure2_path = (
    FIGURE_DIR
    /
    "02_clean_demand_time_series.png"
)


plt.savefig(
    figure2_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 45. FIGURE 3:
# PHYSICAL CONSISTENCY SCATTER
# ================================================================

plot_sample = (

    df[
        [
            "demand_mw",
            "supply_plus_shedding",
            "severe_physical_inconsistency"
        ]
    ]

    .dropna()

    .sample(
        n=min(
            20000,
            len(
                df.dropna(
                    subset=[
                        "demand_mw",
                        "supply_plus_shedding"
                    ]
                )
            )
        ),
        random_state=42
    )
)


normal_points = (
    plot_sample[
        ~plot_sample[
            "severe_physical_inconsistency"
        ]
    ]
)


bad_points = (
    plot_sample[
        plot_sample[
            "severe_physical_inconsistency"
        ]
    ]
)


plt.figure(
    figsize=(8, 8)
)


plt.scatter(
    normal_points[
        "supply_plus_shedding"
    ],

    normal_points[
        "demand_mw"
    ],

    s=8,
    alpha=0.25,
    label="Operational observations"
)


if len(bad_points) > 0:

    plt.scatter(
        bad_points[
            "supply_plus_shedding"
        ],

        bad_points[
            "demand_mw"
        ],

        s=25,
        marker="x",
        label="Severe inconsistency"
    )


combined_values = pd.concat(
    [
        plot_sample[
            "demand_mw"
        ],

        plot_sample[
            "supply_plus_shedding"
        ]
    ]
)


# Prevent gigantic malformed observations from destroying
# figure readability.

plot_limit = combined_values.quantile(
    0.999
)


plot_limit = max(
    plot_limit * 1.10,
    1
)


plt.plot(
    [0, plot_limit],
    [0, plot_limit],
    linestyle="--",
    linewidth=1.2,
    label="Ideal: Demand = Generation + Load Shedding"
)


plt.xlim(
    0,
    plot_limit
)

plt.ylim(
    0,
    plot_limit
)


plt.xlabel(
    "Generation + Load Shedding (MW)"
)

plt.ylabel(
    "Reported Demand (MW)"
)


plt.title(
    "Physical Consistency of PGCB Operational Records"
)


plt.legend()

plt.grid(
    alpha=0.25
)

plt.tight_layout()


figure3_path = (
    FIGURE_DIR
    /
    "03_physical_consistency_scatter.png"
)


plt.savefig(
    figure3_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 46. FIGURE 4:
# PHYSICAL ERROR DISTRIBUTION
# ================================================================

valid_error = (

    df.loc[
        df["physical_abs_error_mw"].notna()
        &
        (
            df["physical_abs_error_mw"]
            <
            df["physical_abs_error_mw"]
              .quantile(0.999)
        ),

        "physical_abs_error_mw"
    ]
)


plt.figure(
    figsize=(10, 6)
)


plt.hist(
    valid_error,
    bins=70
)


plt.xlabel(
    "|Demand - (Generation + Load Shedding)| (MW)"
)

plt.ylabel(
    "Number of observations"
)


plt.title(
    "Distribution of Physical Balance Error"
)


plt.grid(
    alpha=0.25
)

plt.tight_layout()


figure4_path = (
    FIGURE_DIR
    /
    "04_physical_balance_error_distribution.png"
)


plt.savefig(
    figure4_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 47. CREATE A MACHINE-READABLE JSON SUMMARY
# ================================================================

summary_json = {

    "total_raw_rows":
        int(len(df)),

    "start_datetime":
        str(df["datetime"].min()),

    "end_datetime":
        str(df["datetime"].max()),

    "duplicate_timestamp_rows":
        int(
            df["duplicate_timestamp"].sum()
        ),

    "completely_missing_days":
        int(len(missing_full_days)),

    "regular_hour_rows":
        int(
            df["is_regular_hour"].sum()
        ),

    "evening_peak_rows":
        int(
            df["is_evening_peak"].sum()
        ),

    "demand_corrupt_count":
        int(
            df["demand_corrupt"].sum()
        ),

    "generation_corrupt_count":
        int(
            df["generation_corrupt"].sum()
        ),

    "load_shedding_corrupt_count":
        int(
            df[
                "load_shedding_corrupt"
            ].sum()
        ),

    "severe_physical_inconsistency_count":
        int(
            df[
                "severe_physical_inconsistency"
            ].sum()
        ),

    "physical_review_required_count":
        int(
            df[
                "physical_review_required"
            ].sum()
        ),

    "contextual_extreme_keep_count":
        int(
            df[
                "contextual_extreme_keep"
            ].sum()
        ),

    "usable_regular_hour_targets":
        int(
            len(regular_clean)
        ),

    "usable_evening_peak_targets":
        int(
            len(evening_peak_clean)
        )
}


summary_json_path = (
    RESULT_DIR
    /
    "step_1b_summary.json"
)


with open(
    summary_json_path,
    "w"
) as f:

    json.dump(
        summary_json,
        f,
        indent=4
    )


# ================================================================
# 48. FINAL VALIDATION
# ================================================================

assert (
    regular_clean["datetime"]
    .is_monotonic_increasing
), "Regular dataset is not chronological."


assert (
    evening_peak_clean["datetime"]
    .is_monotonic_increasing
), "Evening peak dataset is not chronological."


assert (
    regular_clean[
        "demand_mw_clean"
    ]
    .notna()
    .all()
), "Clean regular dataset still contains missing demand target."


print("\nValidation checks passed.")


# ================================================================
# 49. FINAL OUTPUT SUMMARY
# ================================================================

print("\n")
print("=" * 80)
print("STEP 1B COMPLETED SUCCESSFULLY")
print("=" * 80)

print(
    f"\nRaw observations: "
    f"{len(df):,}"
)

print(
    f"Regular hourly usable observations: "
    f"{len(regular_clean):,}"
)

print(
    f"Evening-peak usable observations: "
    f"{len(evening_peak_clean):,}"
)


print(
    "\nHigh-confidence field corruption:"
)

print(
    " Demand:",
    int(
        df["demand_corrupt"].sum()
    )
)

print(
    " Generation:",
    int(
        df["generation_corrupt"].sum()
    )
)

print(
    " Load shedding:",
    int(
        df[
            "load_shedding_corrupt"
        ].sum()
    )
)


print(
    "\nSevere physical inconsistencies:",
    int(
        df[
            "severe_physical_inconsistency"
        ].sum()
    )
)


print(
    "Unresolved physical inconsistencies:",
    int(
        df[
            "physical_review_required"
        ].sum()
    )
)


print(
    "Contextual extreme observations PRESERVED:",
    int(
        df[
            "contextual_extreme_keep"
        ].sum()
    )
)


print(
    "\nCompletely missing calendar days:",
    len(missing_full_days)
)


print("\n" + "-" * 80)

print("FILES SAVED TO GOOGLE DRIVE")

print("-" * 80)


print("\n1. Full quality-flagged dataset:")
print(flagged_path)


print("\n2. Clean regular-hour forecasting dataset:")
print(regular_path)


print("\n3. Clean evening-peak dataset:")
print(peak_path)


print("\n4. Quality log:")
print(quality_log_path)


print("\n5. Cleaning figures:")
print(FIGURE_DIR)


print("\n6. Audit results:")
print(RESULT_DIR)


print("\n" + "=" * 80)
print(
    "DO NOT START MODEL TRAINING YET."
)
print(
    "SEND THE FINAL PRINTED SUMMARY TO CHATGPT FIRST."
)
print("=" * 80)
```
