```python
# =====================================================================
# STEP 5B
# TEMPORALLY ADAPTIVE + TREND-NORMALIZED CONFORMAL FORECASTING
#
# PURPOSE
# -------
# Correct the undercoverage observed in Step 5.
#
# Point forecaster:
# Ridge Regression
#
# Static model protocol:
#
# 2015-2022       -> original training
# Jan-Jun 2023    -> Ridge hyperparameter tuning
# Jul-Dec 2023    -> initial conformal calibration
# 2024-Mar 2026   -> sequential/prequential evaluation
#
#
# IMPORTANT PREQUENTIAL RULE
# --------------------------
#
# For test day t:
#
# 1. Forecast t
# 2. Build interval using ONLY residuals observed before t
# 3. Evaluate forecast after actual t becomes available
# 4. Residual from t may then be used for t+1
#
# Therefore:
#
# NO SAME-DAY TARGET LEAKAGE
# NO FUTURE TARGET LEAKAGE
#
#
# METHODS
# -------
#
# STATIC:
#   1. Static Hour-Normalized
#   2. Static Trend-Regime + Hour-Normalized
#
# ADAPTIVE:
#   3. Adaptive Hour-Normalized
#   4. Adaptive Trend-Regime + Hour-Normalized
#   5. Adaptive Profile-Max
#   6. Adaptive Trend-Regime Profile-Max
#
#
# Sensitivity windows:
#
# 90 recent days
# 180 recent days <- PRIMARY operational setting
# 365 recent days
#
#
# Nominal coverages:
#
# 80%
# 90%
# 95%
#
# =====================================================================


# =====================================================================
# 0. IMPORTS
# =====================================================================

import os
import json
import math
import time
import warnings

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

import joblib


warnings.filterwarnings("ignore")

np.random.seed(42)

pd.set_option(
    "display.max_columns",
    300
)

pd.set_option(
    "display.max_rows",
    300
)

pd.set_option(
    "display.width",
    280
)


print("=" * 120)

print(
    "STEP 5B: ADAPTIVE + TREND-NORMALIZED CONFORMAL FRAMEWORK"
)

print("=" * 120)


# =====================================================================
# 1. MOUNT GOOGLE DRIVE
# =====================================================================

from google.colab import drive

drive.mount(
    "/content/drive"
)


print(
    "\nGoogle Drive mounted successfully."
)


# =====================================================================
# 2. PROJECT PATHS
# =====================================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/Load_Forecasting_Paper"
)


FORECAST_DIR = (
    PROJECT_ROOT
    / "data"
    / "forecasting"
)


MODEL_DIR = (
    PROJECT_ROOT
    / "models"
    / "adaptive_conformal"
)


RESULT_DIR = (
    PROJECT_ROOT
    / "results"
    / "adaptive_conformal"
)


FIGURE_DIR = (
    PROJECT_ROOT
    / "figures"
    / "adaptive_conformal"
)


for directory in [

    MODEL_DIR,
    RESULT_DIR,
    FIGURE_DIR

]:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# =====================================================================
# 3. INPUT FILES
# =====================================================================

TRAIN_FILE = (
    FORECAST_DIR
    / "train_2015_2022.csv"
)


VALIDATION_FILE = (
    FORECAST_DIR
    / "validation_2023.csv"
)


TEST_FILE = (
    FORECAST_DIR
    / "test_2024_2026.csv"
)


for path in [

    TRAIN_FILE,
    VALIDATION_FILE,
    TEST_FILE

]:

    if not path.exists():

        raise FileNotFoundError(
            f"""
Required forecasting dataset not found:

{path}

Please confirm Step 2 completed successfully.
"""
        )


# =====================================================================
# 4. LOAD DATA
# =====================================================================

train_df = pd.read_csv(
    TRAIN_FILE,
    low_memory=False
)


validation_df = pd.read_csv(
    VALIDATION_FILE,
    low_memory=False
)


test_df = pd.read_csv(
    TEST_FILE,
    low_memory=False
)


for dataframe in [

    train_df,
    validation_df,
    test_df

]:

    dataframe[
        "target_date"
    ] = pd.to_datetime(

        dataframe[
            "target_date"
        ],

        errors="coerce"
    )


    dataframe.sort_values(
        "target_date",
        inplace=True
    )


    dataframe.reset_index(
        drop=True,
        inplace=True
    )


print("\nDatasets loaded.")


print(
    "TRAIN:",
    len(train_df)
)


print(
    "2023:",
    len(validation_df)
)


print(
    "TEST:",
    len(test_df)
)


# =====================================================================
# 5. TARGET HOURS
# =====================================================================

HOURS = list(
    range(
        1,
        24
    )
)


TARGET_COLUMNS = [

    f"target_h{hour:02d}"

    for hour in HOURS
]


# =====================================================================
# 6. FEATURE DEFINITIONS
# =====================================================================

CALENDAR_FEATURES = [

    "year",

    "month",

    "day",

    "dayofweek",

    "dayofyear",

    "weekofyear",

    "quarter",

    "is_weekend",

    "month_sin",

    "month_cos",

    "dow_sin",

    "dow_cos",

    "doy_sin",

    "doy_cos"
]


LAG1_FEATURES = [

    f"lag1_h{hour:02d}"

    for hour in HOURS
]


LAG7_FEATURES = [

    f"lag7_h{hour:02d}"

    for hour in HOURS
]


ROLL7_FEATURES = [

    f"roll7mean_h{hour:02d}"

    for hour in HOURS
]


ROLL28_FEATURES = [

    f"roll28mean_h{hour:02d}"

    for hour in HOURS
]


ROLL7STD_FEATURES = [

    f"roll7std_h{hour:02d}"

    for hour in HOURS
]


ROLL28STD_FEATURES = [

    f"roll28std_h{hour:02d}"

    for hour in HOURS
]


SUMMARY_FEATURES = [

    "lag1_daily_mean",

    "lag1_daily_max",

    "lag1_daily_min",

    "lag1_daily_range",

    "lag1_peak_hour",

    "lag7_daily_mean",

    "lag7_daily_max",

    "lag7_daily_min",

    "lag7_daily_range",

    "lag7_peak_hour",

    "mean_change_vs_last_week",

    "max_change_vs_last_week",

    "mean_ratio_vs_last_week"
]


FEATURE_COLUMNS = (

    CALENDAR_FEATURES

    +

    LAG1_FEATURES

    +

    LAG7_FEATURES

    +

    ROLL7_FEATURES

    +

    ROLL28_FEATURES

    +

    ROLL7STD_FEATURES

    +

    ROLL28STD_FEATURES

    +

    SUMMARY_FEATURES
)


print(
    "\nFeature count:",
    len(FEATURE_COLUMNS)
)


print(
    "Target count:",
    len(TARGET_COLUMNS)
)


# =====================================================================
# 7. STRUCTURAL VALIDATION
# =====================================================================

for name, dataframe in [

    (
        "TRAIN",
        train_df
    ),

    (
        "VALIDATION",
        validation_df
    ),

    (
        "TEST",
        test_df
    )

]:

    missing_features = [

        column

        for column in FEATURE_COLUMNS

        if column not in dataframe.columns
    ]


    missing_targets = [

        column

        for column in TARGET_COLUMNS

        if column not in dataframe.columns
    ]


    if missing_features:

        raise ValueError(
            f"{name} missing features: {missing_features}"
        )


    if missing_targets:

        raise ValueError(
            f"{name} missing targets: {missing_targets}"
        )


    matrix = dataframe[
        FEATURE_COLUMNS
        +
        TARGET_COLUMNS
    ].to_numpy(
        dtype=float
    )


    if not np.isfinite(
        matrix
    ).all():

        raise ValueError(
            f"{name} contains non-finite values."
        )


print(
    "\nDataset validation PASSED."
)


# =====================================================================
# 8. NESTED 2023 SPLIT
# =====================================================================

TUNE_END = pd.Timestamp(
    "2023-06-30"
)


tune_df = (

    validation_df.loc[

        validation_df[
            "target_date"
        ]
        <=
        TUNE_END
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


initial_calibration_df = (

    validation_df.loc[

        validation_df[
            "target_date"
        ]
        >
        TUNE_END
    ]

    .copy()

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 120)

print(
    "NESTED TEMPORAL DESIGN"
)

print("=" * 120)


print(
    "Original training:",
    len(train_df)
)


print(
    "Tuning Jan-Jun 2023:",
    len(tune_df)
)


print(
    "Initial calibration Jul-Dec 2023:",
    len(initial_calibration_df)
)


print(
    "Sequential test:",
    len(test_df)
)


# =====================================================================
# 9. METRIC FUNCTIONS
# =====================================================================

def mae(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.mean(

            np.abs(
                y_true
                -
                y_pred
            )
        )
    )


def rmse(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    return float(

        np.sqrt(

            np.mean(

                (
                    y_true
                    -
                    y_pred
                )

                ** 2
            )
        )
    )


def mape(
    y_true,
    y_pred
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    y_pred = np.asarray(
        y_pred,
        dtype=float
    )


    denominator = np.maximum(

        np.abs(
            y_true
        ),

        1e-6
    )


    return float(

        100

        *

        np.mean(

            np.abs(

                (
                    y_true
                    -
                    y_pred
                )

                /

                denominator
            )
        )
    )


# =====================================================================
# 10. RIDGE TUNING
# =====================================================================

RIDGE_ALPHAS = [

    0.01,

    0.1,

    1.0,

    10.0,

    100.0,

    1000.0
]


X_train = (

    train_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_train = (

    train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


X_tune = (

    tune_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_tune = (

    tune_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


tune_scaler = StandardScaler()


X_train_scaled = (

    tune_scaler
    .fit_transform(
        X_train
    )
)


X_tune_scaled = (

    tune_scaler
    .transform(
        X_tune
    )
)


tune_rows = []


for alpha in RIDGE_ALPHAS:

    model = Ridge(
        alpha=alpha
    )


    model.fit(
        X_train_scaled,
        Y_train
    )


    prediction = model.predict(
        X_tune_scaled
    )


    tune_rows.append(
        {

            "alpha":
                alpha,

            "MAE_MW":
                mae(
                    Y_tune,
                    prediction
                ),

            "RMSE_MW":
                rmse(
                    Y_tune,
                    prediction
                ),

            "MAPE_percent":
                mape(
                    Y_tune,
                    prediction
                )
        }
    )


ridge_tuning = pd.DataFrame(
    tune_rows
)


ridge_tuning = (

    ridge_tuning
    .sort_values(
        "MAE_MW"
    )

    .reset_index(
        drop=True
    )
)


BEST_ALPHA = float(

    ridge_tuning
    .iloc[0][
        "alpha"
    ]
)


print(
    "\nBest Ridge alpha:",
    BEST_ALPHA
)


display(
    ridge_tuning.round(4)
)


# =====================================================================
# 11. PROPER TRAINING SET
# =====================================================================

proper_train_df = pd.concat(

    [
        train_df,
        tune_df
    ],

    ignore_index=True
)


proper_train_df = (

    proper_train_df
    .sort_values(
        "target_date"
    )

    .reset_index(
        drop=True
    )
)


# =====================================================================
# 12. FINAL STATIC POINT MODEL
# =====================================================================

X_proper = (

    proper_train_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_proper = (

    proper_train_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


X_cal = (

    initial_calibration_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_cal = (

    initial_calibration_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


X_test = (

    test_df[
        FEATURE_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


Y_test = (

    test_df[
        TARGET_COLUMNS
    ]

    .to_numpy(
        dtype=float
    )
)


final_scaler = StandardScaler()


X_proper_scaled = (

    final_scaler
    .fit_transform(
        X_proper
    )
)


X_cal_scaled = (

    final_scaler
    .transform(
        X_cal
    )
)


X_test_scaled = (

    final_scaler
    .transform(
        X_test
    )
)


ridge_model = Ridge(
    alpha=
        BEST_ALPHA
)


ridge_model.fit(
    X_proper_scaled,
    Y_proper
)


CAL_PRED = ridge_model.predict(
    X_cal_scaled
)


TEST_PRED = ridge_model.predict(
    X_test_scaled
)


print(
    "\nStatic Ridge point model fitted."
)


# =====================================================================
# 13. SAVE MODEL
# =====================================================================

joblib.dump(

    final_scaler,

    MODEL_DIR
    /
    "adaptive_conformal_scaler.joblib"
)


joblib.dump(

    ridge_model,

    MODEL_DIR
    /
    "adaptive_conformal_ridge.joblib"
)


# =====================================================================
# 14. POINT FORECAST PERFORMANCE
# =====================================================================

print("\n" + "=" * 120)

print(
    "POINT FORECAST PERFORMANCE"
)

print("=" * 120)


print(
    "All-test MAE:",
    round(
        mae(
            Y_test,
            TEST_PRED
        ),
        3
    ),
    "MW"
)


print(
    "All-test RMSE:",
    round(
        rmse(
            Y_test,
            TEST_PRED
        ),
        3
    ),
    "MW"
)


print(
    "All-test MAPE:",
    round(
        mape(
            Y_test,
            TEST_PRED
        ),
        3
    ),
    "%"
)


# =====================================================================
# 15. NORMALIZED NONCONFORMITY SCORES
# =====================================================================

SCALE_FLOOR_MW = 1000.0


CAL_ABS_ERROR = np.abs(

    Y_cal

    -

    CAL_PRED
)


CAL_SCALE = np.maximum(

    np.abs(
        CAL_PRED
    ),

    SCALE_FLOOR_MW
)


CAL_SCORE = (

    CAL_ABS_ERROR

    /

    CAL_SCALE
)


TEST_ABS_ERROR = np.abs(

    Y_test

    -

    TEST_PRED
)


TEST_SCALE = np.maximum(

    np.abs(
        TEST_PRED
    ),

    SCALE_FLOOR_MW
)


TEST_SCORE = (

    TEST_ABS_ERROR

    /

    TEST_SCALE
)


CAL_PROFILE_SCORE = np.max(

    CAL_SCORE,

    axis=1
)


TEST_PROFILE_SCORE = np.max(

    TEST_SCORE,

    axis=1
)


# =====================================================================
# 16. TREND-NORMALIZED FORECASTABLE REGIME
# =====================================================================

# Instead of absolute MW thresholds:
#
# previous-day mean
# -----------------
# recent 28-day expected mean
#
#
# A value near 1 means yesterday was near the recent expected level.
#
# This removes much of the long-term demand growth effect.


def create_regime_signal(
    dataframe
):

    recent_expected_profile_mean = (

        dataframe[
            ROLL28_FEATURES
        ]

        .mean(
            axis=1
        )

        .to_numpy(
            dtype=float
        )
    )


    lag1_mean = (

        dataframe[
            "lag1_daily_mean"
        ]

        .to_numpy(
            dtype=float
        )
    )


    signal = (

        lag1_mean

        /

        np.maximum(
            recent_expected_profile_mean,
            1.0
        )
    )


    return signal


PROPER_REGIME_SIGNAL = create_regime_signal(
    proper_train_df
)


CAL_REGIME_SIGNAL = create_regime_signal(
    initial_calibration_df
)


TEST_REGIME_SIGNAL = create_regime_signal(
    test_df
)


# =====================================================================
# 17. REGIME THRESHOLDS FROM PROPER TRAINING ONLY
# =====================================================================

LOW_THRESHOLD = float(

    np.quantile(

        PROPER_REGIME_SIGNAL,

        1 / 3
    )
)


HIGH_THRESHOLD = float(

    np.quantile(

        PROPER_REGIME_SIGNAL,

        2 / 3
    )
)


print("\n" + "=" * 120)

print(
    "TREND-NORMALIZED REGIME THRESHOLDS"
)

print("=" * 120)


print(
    "LOW <",
    round(
        LOW_THRESHOLD,
        5
    )
)


print(
    "NORMAL:",
    round(
        LOW_THRESHOLD,
        5
    ),
    "to",
    round(
        HIGH_THRESHOLD,
        5
    )
)


print(
    "HIGH >",
    round(
        HIGH_THRESHOLD,
        5
    )
)


# =====================================================================
# 18. REGIME FUNCTION
# =====================================================================

def assign_regime(
    signal
):

    signal = np.asarray(
        signal,
        dtype=float
    )


    labels = np.full(

        len(
            signal
        ),

        "NORMAL",

        dtype=object
    )


    labels[
        signal
        <
        LOW_THRESHOLD
    ] = "LOW"


    labels[
        signal
        >
        HIGH_THRESHOLD
    ] = "HIGH"


    return labels


CAL_REGIME = assign_regime(
    CAL_REGIME_SIGNAL
)


TEST_REGIME = assign_regime(
    TEST_REGIME_SIGNAL
)


REGIMES = [

    "LOW",

    "NORMAL",

    "HIGH"
]


# =====================================================================
# 19. REGIME DISTRIBUTION
# =====================================================================

regime_rows = []


for dataset_name, labels in [

    (
        "INITIAL_CALIBRATION",
        CAL_REGIME
    ),

    (
        "TEST",
        TEST_REGIME
    )
]:

    for regime in REGIMES:

        count = int(

            np.sum(
                labels
                ==
                regime
            )
        )


        regime_rows.append(
            {

                "dataset":
                    dataset_name,

                "regime":
                    regime,

                "count":
                    count,

                "percent":
                    100
                    *
                    count
                    /
                    len(
                        labels
                    )
            }
        )


regime_distribution = pd.DataFrame(
    regime_rows
)


print(
    "\nTrend-normalized regime distribution:"
)


display(
    regime_distribution.round(3)
)


# =====================================================================
# 20. FINITE-SAMPLE CONFORMAL QUANTILE
# =====================================================================

def conformal_quantile(
    scores,
    alpha
):

    scores = np.asarray(
        scores,
        dtype=float
    )


    scores = scores[
        np.isfinite(
            scores
        )
    ]


    n = len(
        scores
    )


    if n == 0:

        return np.nan


    rank = int(

        np.ceil(

            (
                n
                +
                1
            )

            *
            (
                1
                -
                alpha
            )
        )
    )


    rank = max(
        rank,
        1
    )


    rank = min(
        rank,
        n
    )


    ordered = np.sort(
        scores
    )


    return float(

        ordered[
            rank
            -
            1
        ]
    )


# =====================================================================
# 21. COVERAGE LEVELS
# =====================================================================

COVERAGES = [

    0.80,

    0.90,

    0.95
]


WINDOWS = [

    90,

    180,

    365
]


PRIMARY_WINDOW = 180


MIN_REGIME_SAMPLES = 25


# =====================================================================
# 22. STATIC CONFORMAL INTERVALS
# =====================================================================

def build_static_intervals(
    coverage,
    regime_aware=False
):

    alpha = (
        1
        -
        coverage
    )


    n_test = (
        len(
            test_df
        )
    )


    lower = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    upper = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    # -------------------------------------------------------------
    # Global hour-specific normalized fallback
    # -------------------------------------------------------------

    global_hour_q = np.array(
        [

            conformal_quantile(

                CAL_SCORE[
                    :,
                    hour_index
                ],

                alpha
            )

            for hour_index in range(
                23
            )
        ]
    )


    for i in range(
        n_test
    ):

        if not regime_aware:

            q_vector = (
                global_hour_q
            )


        else:

            regime = (

                TEST_REGIME[
                    i
                ]
            )


            regime_mask = (

                CAL_REGIME
                ==
                regime
            )


            regime_count = int(
                regime_mask.sum()
            )


            if (
                regime_count
                >=
                MIN_REGIME_SAMPLES
            ):

                q_vector = np.array(
                    [

                        conformal_quantile(

                            CAL_SCORE[
                                regime_mask,
                                hour_index
                            ],

                            alpha
                        )

                        for hour_index
                        in range(
                            23
                        )
                    ]
                )


            else:

                q_vector = (
                    global_hour_q
                )


        width = (

            q_vector

            *
            TEST_SCALE[
                i
            ]
        )


        lower[
            i
        ] = (

            TEST_PRED[
                i
            ]

            -
            width
        )


        upper[
            i
        ] = (

            TEST_PRED[
                i
            ]

            +
            width
        )


    lower = np.maximum(
        lower,
        0.0
    )


    return (
        lower,
        upper
    )


# =====================================================================
# 23. STATIC PROFILE-MAX INTERVAL
# =====================================================================

def build_static_profile_interval(
    coverage,
    regime_aware=False
):

    alpha = (
        1
        -
        coverage
    )


    global_q = conformal_quantile(

        CAL_PROFILE_SCORE,

        alpha
    )


    lower = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    upper = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    for i in range(
        len(
            test_df
        )
    ):

        if not regime_aware:

            q = (
                global_q
            )


        else:

            regime = (

                TEST_REGIME[
                    i
                ]
            )


            regime_mask = (

                CAL_REGIME
                ==
                regime
            )


            if (
                int(
                    regime_mask.sum()
                )
                >=
                MIN_REGIME_SAMPLES
            ):

                q = conformal_quantile(

                    CAL_PROFILE_SCORE[
                        regime_mask
                    ],

                    alpha
                )


            else:

                q = (
                    global_q
                )


        half_width = (

            q

            *
            TEST_SCALE[
                i
            ]
        )


        lower[
            i
        ] = (

            TEST_PRED[
                i
            ]

            -
            half_width
        )


        upper[
            i
        ] = (

            TEST_PRED[
                i
            ]

            +
            half_width
        )


    lower = np.maximum(
        lower,
        0.0
    )


    return (
        lower,
        upper
    )


# =====================================================================
# 24. PREQUENTIAL ADAPTIVE INTERVAL BUILDER
# =====================================================================

def build_adaptive_intervals(
    coverage,
    window,
    regime_aware=False,
    profile_max=False
):

    alpha = (
        1
        -
        coverage
    )


    n_test = (
        len(
            test_df
        )
    )


    lower = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    upper = np.zeros_like(
        TEST_PRED,
        dtype=float
    )


    # -------------------------------------------------------------
    # Initial history consists ONLY of Jul-Dec 2023 calibration.
    # -------------------------------------------------------------

    score_history = [

        CAL_SCORE[
            i
        ]
        .copy()

        for i in range(
            len(
                CAL_SCORE
            )
        )
    ]


    profile_history = [

        float(
            CAL_PROFILE_SCORE[
                i
            ]
        )

        for i in range(
            len(
                CAL_PROFILE_SCORE
            )
        )
    ]


    regime_history = [

        str(
            CAL_REGIME[
                i
            ]
        )

        for i in range(
            len(
                CAL_REGIME
            )
        )
    ]


    # -------------------------------------------------------------
    # STRICT PREQUENTIAL LOOP
    # -------------------------------------------------------------

    for test_index in range(
        n_test
    ):

        # ---------------------------------------------------------
        # Only the MOST RECENT `window` already-observed days.
        # Current test day is NOT yet in history.
        # ---------------------------------------------------------

        history_start = max(

            0,

            len(
                score_history
            )

            -
            window
        )


        recent_scores = np.asarray(

            score_history[
                history_start:
            ],

            dtype=float
        )


        recent_profiles = np.asarray(

            profile_history[
                history_start:
            ],

            dtype=float
        )


        recent_regimes = np.asarray(

            regime_history[
                history_start:
            ],

            dtype=object
        )


        current_regime = (

            TEST_REGIME[
                test_index
            ]
        )


        # ---------------------------------------------------------
        # PROFILE-MAX METHOD
        # ---------------------------------------------------------

        if profile_max:

            fallback_q = conformal_quantile(

                recent_profiles,

                alpha
            )


            if regime_aware:

                regime_mask = (

                    recent_regimes
                    ==
                    current_regime
                )


                regime_count = int(
                    regime_mask.sum()
                )


                if (
                    regime_count
                    >=
                    MIN_REGIME_SAMPLES
                ):

                    q = conformal_quantile(

                        recent_profiles[
                            regime_mask
                        ],

                        alpha
                    )


                else:

                    q = (
                        fallback_q
                    )


            else:

                q = (
                    fallback_q
                )


            half_width = (

                q

                *
                TEST_SCALE[
                    test_index
                ]
            )


        # ---------------------------------------------------------
        # HOUR-SPECIFIC METHOD
        # ---------------------------------------------------------

        else:

            fallback_q_vector = np.array(
                [

                    conformal_quantile(

                        recent_scores[
                            :,
                            hour_index
                        ],

                        alpha
                    )

                    for hour_index
                    in range(
                        23
                    )
                ]
            )


            if regime_aware:

                regime_mask = (

                    recent_regimes
                    ==
                    current_regime
                )


                regime_count = int(
                    regime_mask.sum()
                )


                if (
                    regime_count
                    >=
                    MIN_REGIME_SAMPLES
                ):

                    regime_scores = (

                        recent_scores[
                            regime_mask
                        ]
                    )


                    q_vector = np.array(
                        [

                            conformal_quantile(

                                regime_scores[
                                    :,
                                    hour_index
                                ],

                                alpha
                            )

                            for hour_index
                            in range(
                                23
                            )
                        ]
                    )


                else:

                    q_vector = (
                        fallback_q_vector
                    )


            else:

                q_vector = (
                    fallback_q_vector
                )


            half_width = (

                q_vector

                *
                TEST_SCALE[
                    test_index
                ]
            )


        # ---------------------------------------------------------
        # BUILD CURRENT DAY INTERVAL
        # ---------------------------------------------------------

        lower[
            test_index
        ] = (

            TEST_PRED[
                test_index
            ]

            -
            half_width
        )


        upper[
            test_index
        ] = (

            TEST_PRED[
                test_index
            ]

            +
            half_width
        )


        # ---------------------------------------------------------
        # ONLY NOW:
        #
        # Current day's actual outcome has hypothetically become known.
        #
        # Add its score for forecasting NEXT DAY.
        # ---------------------------------------------------------

        score_history.append(

            TEST_SCORE[
                test_index
            ]
            .copy()
        )


        profile_history.append(

            float(
                TEST_PROFILE_SCORE[
                    test_index
                ]
            )
        )


        regime_history.append(

            str(
                current_regime
            )
        )


    lower = np.maximum(
        lower,
        0.0
    )


    return (
        lower,
        upper
    )


# =====================================================================
# 25. WINKLER SCORE
# =====================================================================

def winkler_score(
    y_true,
    lower,
    upper,
    alpha
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    lower = np.asarray(
        lower,
        dtype=float
    )


    upper = np.asarray(
        upper,
        dtype=float
    )


    width = (

        upper

        -

        lower
    )


    score = (
        width.copy()
    )


    below = (

        y_true
        <
        lower
    )


    above = (

        y_true
        >
        upper
    )


    score[
        below
    ] += (

        2

        /
        alpha

        *

        (
            lower[
                below
            ]

            -
            y_true[
                below
            ]
        )
    )


    score[
        above
    ] += (

        2

        /
        alpha

        *

        (
            y_true[
                above
            ]

            -
            upper[
                above
            ]
        )
    )


    return float(

        np.mean(
            score
        )
    )


# =====================================================================
# 26. INTERVAL METRIC FUNCTION
# =====================================================================

def evaluate_interval(
    y_true,
    lower,
    upper,
    nominal_coverage,
    primary_type="MARGINAL"
):

    y_true = np.asarray(
        y_true,
        dtype=float
    )


    covered = (

        (
            y_true
            >=
            lower
        )

        &

        (
            y_true
            <=
            upper
        )
    )


    marginal_coverage = (

        100

        *
        float(
            covered.mean()
        )
    )


    profile_coverage = (

        100

        *
        float(

            covered
            .all(
                axis=1
            )
            .mean()
        )
    )


    width = (

        upper

        -

        lower
    )


    mpiw = float(

        np.mean(
            width
        )
    )


    mean_load = float(

        np.mean(

            np.abs(
                y_true
            )
        )
    )


    normalized_width = (

        100

        *
        mpiw

        /
        max(
            mean_load,
            1e-6
        )
    )


    if (
        primary_type
        ==
        "MARGINAL"
    ):

        primary_coverage = (

            marginal_coverage
        )


    else:

        primary_coverage = (

            profile_coverage
        )


    nominal_percent = (

        nominal_coverage

        *
        100
    )


    return {

        "Marginal_Coverage_percent":
            marginal_coverage,

        "Profile_Coverage_percent":
            profile_coverage,

        "Primary_Coverage_percent":
            primary_coverage,

        "Coverage_Gap_percentage_points":
            (
                primary_coverage
                -
                nominal_percent
            ),

        "MPIW_MW":
            mpiw,

        "Normalized_MPIW_percent":
            normalized_width,

        "Winkler_Score":
            winkler_score(

                y_true,

                lower,

                upper,

                1
                -
                nominal_coverage
            )
    }


# =====================================================================
# 27. TEST WINDOWS
# =====================================================================

TEST_DATES = (

    test_df[
        "target_date"
    ]
)


TEST_SPLITS = {

    "TEST_A_2024":

        (
            TEST_DATES
            .dt.year
            ==
            2024
        )
        .to_numpy(),

    "TEST_B_2025_2026":

        (
            TEST_DATES
            .dt.year
            >=
            2025
        )
        .to_numpy(),

    "ALL_TEST_2024_2026":

        np.ones(
            len(
                test_df
            ),
            dtype=bool
        )
}


# =====================================================================
# 28. GENERATE STATIC RESULTS
# =====================================================================

interval_predictions = {}


for coverage in COVERAGES:

    coverage_int = int(
        coverage
        *
        100
    )


    # -------------------------------------------------------------
    # Static Hour Normalized
    # -------------------------------------------------------------

    lower, upper = build_static_intervals(

        coverage=
            coverage,

        regime_aware=
            False
    )


    interval_predictions[
        (
            "StaticHourNormalized",
            coverage_int
        )
    ] = (

        lower,
        upper,
        "MARGINAL"
    )


    # -------------------------------------------------------------
    # Static Trend-Regime Hour Normalized
    # -------------------------------------------------------------

    lower, upper = build_static_intervals(

        coverage=
            coverage,

        regime_aware=
            True
    )


    interval_predictions[
        (
            "StaticTrendRegimeHourNormalized",
            coverage_int
        )
    ] = (

        lower,
        upper,
        "MARGINAL"
    )


    # -------------------------------------------------------------
    # Static Profile-Max
    # -------------------------------------------------------------

    lower, upper = build_static_profile_interval(

        coverage=
            coverage,

        regime_aware=
            False
    )


    interval_predictions[
        (
            "StaticProfileMax",
            coverage_int
        )
    ] = (

        lower,
        upper,
        "SIMULTANEOUS"
    )


    # -------------------------------------------------------------
    # Static Trend-Regime Profile-Max
    # -------------------------------------------------------------

    lower, upper = build_static_profile_interval(

        coverage=
            coverage,

        regime_aware=
            True
    )


    interval_predictions[
        (
            "StaticTrendRegimeProfileMax",
            coverage_int
        )
    ] = (

        lower,
        upper,
        "SIMULTANEOUS"
    )


# =====================================================================
# 29. GENERATE ADAPTIVE RESULTS FOR EACH WINDOW
# =====================================================================

print("\n" + "=" * 120)

print(
    "GENERATING PREQUENTIAL ADAPTIVE INTERVALS"
)

print("=" * 120)


for window in WINDOWS:

    print(
        f"\nWindow = {window} recent observed days"
    )


    for coverage in COVERAGES:

        coverage_int = int(
            coverage
            *
            100
        )


        # ---------------------------------------------------------
        # Adaptive hour normalized
        # ---------------------------------------------------------

        lower, upper = build_adaptive_intervals(

            coverage=
                coverage,

            window=
                window,

            regime_aware=
                False,

            profile_max=
                False
        )


        interval_predictions[
            (
                f"AdaptiveHourNormalized_W{window}",
                coverage_int
            )
        ] = (

            lower,
            upper,
            "MARGINAL"
        )


        # ---------------------------------------------------------
        # Adaptive trend-regime hour normalized
        # ---------------------------------------------------------

        lower, upper = build_adaptive_intervals(

            coverage=
                coverage,

            window=
                window,

            regime_aware=
                True,

            profile_max=
                False
        )


        interval_predictions[
            (
                f"AdaptiveTrendRegimeHourNormalized_W{window}",
                coverage_int
            )
        ] = (

            lower,
            upper,
            "MARGINAL"
        )


        # ---------------------------------------------------------
        # Adaptive profile max
        # ---------------------------------------------------------

        lower, upper = build_adaptive_intervals(

            coverage=
                coverage,

            window=
                window,

            regime_aware=
                False,

            profile_max=
                True
        )


        interval_predictions[
            (
                f"AdaptiveProfileMax_W{window}",
                coverage_int
            )
        ] = (

            lower,
            upper,
            "SIMULTANEOUS"
        )


        # ---------------------------------------------------------
        # Adaptive trend-regime profile max
        # ---------------------------------------------------------

        lower, upper = build_adaptive_intervals(

            coverage=
                coverage,

            window=
                window,

            regime_aware=
                True,

            profile_max=
                True
        )


        interval_predictions[
            (
                f"AdaptiveTrendRegimeProfileMax_W{window}",
                coverage_int
            )
        ] = (

            lower,
            upper,
            "SIMULTANEOUS"
        )


print(
    "\nAll adaptive intervals generated."
)


# =====================================================================
# 30. EVALUATE EVERY METHOD
# =====================================================================

result_rows = []


for (
    method,
    coverage_int
), (
    lower,
    upper,
    primary_type
) in interval_predictions.items():

    nominal_coverage = (

        coverage_int

        /
        100
    )


    for split_name, split_mask in (
        TEST_SPLITS.items()
    ):

        metrics = evaluate_interval(

            Y_test[
                split_mask
            ],

            lower[
                split_mask
            ],

            upper[
                split_mask
            ],

            nominal_coverage=
                nominal_coverage,

            primary_type=
                primary_type
        )


        result_rows.append(
            {

                "split":
                    split_name,

                "method":
                    method,

                "coverage_type":
                    primary_type,

                "nominal_coverage_percent":
                    coverage_int,

                "days":
                    int(
                        split_mask.sum()
                    ),

                **metrics
            }
        )


evaluation_results = pd.DataFrame(
    result_rows
)


# =====================================================================
# 31. PRIMARY 180-DAY METHODS
# =====================================================================

PRIMARY_MARGINAL_METHOD = (

    f"AdaptiveTrendRegimeHourNormalized_W"
    f"{PRIMARY_WINDOW}"
)


PRIMARY_PROFILE_METHOD = (

    f"AdaptiveTrendRegimeProfileMax_W"
    f"{PRIMARY_WINDOW}"
)


print("\n" + "=" * 120)

print(
    "PRIMARY 90% COMPARISON — ALL TEST DATA"
)

print("=" * 120)


primary_90_methods = [

    "StaticHourNormalized",

    "StaticTrendRegimeHourNormalized",

    f"AdaptiveHourNormalized_W{PRIMARY_WINDOW}",

    PRIMARY_MARGINAL_METHOD,

    "StaticProfileMax",

    "StaticTrendRegimeProfileMax",

    f"AdaptiveProfileMax_W{PRIMARY_WINDOW}",

    PRIMARY_PROFILE_METHOD
]


primary_90 = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )

        &

        (
            evaluation_results[
                "method"
            ]
            .isin(
                primary_90_methods
            )
        )
    ]

    .copy()
)


primary_90[
    "method_order"
] = (

    primary_90[
        "method"
    ]

    .map(
        {

            method:
                index

            for index, method
            in enumerate(
                primary_90_methods
            )
        }
    )
)


primary_90 = (

    primary_90

    .sort_values(
        "method_order"
    )

    .drop(
        columns=[
            "method_order"
        ]
    )

    .reset_index(
        drop=True
    )
)


display(

    primary_90[
        [
            "method",

            "coverage_type",

            "Primary_Coverage_percent",

            "Coverage_Gap_percentage_points",

            "Marginal_Coverage_percent",

            "Profile_Coverage_percent",

            "MPIW_MW",

            "Normalized_MPIW_percent",

            "Winkler_Score"
        ]
    ]

    .round(4)
)


# =====================================================================
# 32. WINDOW-SENSITIVITY TABLE
# =====================================================================

window_sensitivity_methods = []


for window in WINDOWS:

    window_sensitivity_methods.extend(
        [

            f"AdaptiveHourNormalized_W{window}",

            f"AdaptiveTrendRegimeHourNormalized_W{window}",

            f"AdaptiveProfileMax_W{window}",

            f"AdaptiveTrendRegimeProfileMax_W{window}"
        ]
    )


window_sensitivity = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )

        &

        (
            evaluation_results[
                "method"
            ]
            .isin(
                window_sensitivity_methods
            )
        )
    ]

    .copy()

    .sort_values(
        [
            "coverage_type",
            "MPIW_MW"
        ]
    )

    .reset_index(
        drop=True
    )
)


print("\n" + "=" * 120)

print(
    "90% WINDOW SENSITIVITY — DO NOT SELECT FROM TEST"
)

print("=" * 120)


display(

    window_sensitivity[
        [
            "method",

            "coverage_type",

            "Primary_Coverage_percent",

            "Coverage_Gap_percentage_points",

            "MPIW_MW",

            "Winkler_Score"
        ]
    ]

    .round(4)
)


# =====================================================================
# 33. REGIME-WISE PERFORMANCE OF PRIMARY METHOD
# =====================================================================

primary_regime_rows = []


for regime in REGIMES:

    regime_mask = (

        TEST_REGIME
        ==
        regime
    )


    if (
        regime_mask.sum()
        ==
        0
    ):

        continue


    for coverage in COVERAGES:

        coverage_int = int(
            coverage
            *
            100
        )


        for method in [

            PRIMARY_MARGINAL_METHOD,

            PRIMARY_PROFILE_METHOD
        ]:

            lower, upper, primary_type = (

                interval_predictions[
                    (
                        method,
                        coverage_int
                    )
                ]
            )


            metrics = evaluate_interval(

                Y_test[
                    regime_mask
                ],

                lower[
                    regime_mask
                ],

                upper[
                    regime_mask
                ],

                nominal_coverage=
                    coverage,

                primary_type=
                    primary_type
            )


            primary_regime_rows.append(
                {

                    "regime":
                        regime,

                    "days":
                        int(
                            regime_mask.sum()
                        ),

                    "method":
                        method,

                    "coverage_type":
                        primary_type,

                    "nominal_coverage_percent":
                        coverage_int,

                    **metrics
                }
            )


primary_regime_results = pd.DataFrame(
    primary_regime_rows
)


print("\n" + "=" * 120)

print(
    "PRIMARY METHOD PERFORMANCE BY TREND-NORMALIZED REGIME"
)

print("=" * 120)


display(

    primary_regime_results.loc[

        primary_regime_results[
            "nominal_coverage_percent"
        ]
        ==
        90

    ][
        [
            "regime",

            "days",

            "method",

            "coverage_type",

            "Primary_Coverage_percent",

            "Coverage_Gap_percentage_points",

            "MPIW_MW"
        ]
    ]

    .round(4)
)


# =====================================================================
# 34. HOUR-WISE COVERAGE FOR PRIMARY MARGINAL METHOD
# =====================================================================

hourly_rows = []


for coverage in COVERAGES:

    coverage_int = int(
        coverage
        *
        100
    )


    lower, upper, _ = (

        interval_predictions[
            (
                PRIMARY_MARGINAL_METHOD,
                coverage_int
            )
        ]
    )


    for hour_index, hour in enumerate(
        HOURS
    ):

        covered = (

            (
                Y_test[
                    :,
                    hour_index
                ]

                >=

                lower[
                    :,
                    hour_index
                ]
            )

            &

            (
                Y_test[
                    :,
                    hour_index
                ]

                <=

                upper[
                    :,
                    hour_index
                ]
            )
        )


        hourly_rows.append(
            {

                "coverage_percent":
                    coverage_int,

                "hour":
                    hour,

                "empirical_coverage_percent":
                    100
                    *
                    float(
                        covered.mean()
                    ),

                "MPIW_MW":
                    float(

                        np.mean(

                            upper[
                                :,
                                hour_index
                            ]

                            -

                            lower[
                                :,
                                hour_index
                            ]
                        )
                    )
            }
        )


hourly_primary_results = pd.DataFrame(
    hourly_rows
)


# =====================================================================
# 35. CREATE PRIMARY 90% OUTPUT FILE
# =====================================================================

PRIMARY_LOWER_90, \
PRIMARY_UPPER_90, _ = (

    interval_predictions[
        (
            PRIMARY_MARGINAL_METHOD,
            90
        )
    ]
)


PROFILE_LOWER_90, \
PROFILE_UPPER_90, _ = (

    interval_predictions[
        (
            PRIMARY_PROFILE_METHOD,
            90
        )
    ]
)


interval_output = pd.DataFrame(
    {

        "target_date":
            TEST_DATES,

        "trend_regime_signal":
            TEST_REGIME_SIGNAL,

        "trend_regime":
            TEST_REGIME
    }
)


for hour_index, hour in enumerate(
    HOURS
):

    interval_output[
        f"actual_h{hour:02d}"
    ] = (

        Y_test[
            :,
            hour_index
        ]
    )


    interval_output[
        f"ridge_pred_h{hour:02d}"
    ] = (

        TEST_PRED[
            :,
            hour_index
        ]
    )


    interval_output[
        f"adaptive_lower90_h{hour:02d}"
    ] = (

        PRIMARY_LOWER_90[
            :,
            hour_index
        ]
    )


    interval_output[
        f"adaptive_upper90_h{hour:02d}"
    ] = (

        PRIMARY_UPPER_90[
            :,
            hour_index
        ]
    )


    interval_output[
        f"profile_lower90_h{hour:02d}"
    ] = (

        PROFILE_LOWER_90[
            :,
            hour_index
        ]
    )


    interval_output[
        f"profile_upper90_h{hour:02d}"
    ] = (

        PROFILE_UPPER_90[
            :,
            hour_index
        ]
    )


# =====================================================================
# 36. SAVE RESULT TABLES
# =====================================================================

RIDGE_TUNING_FILE = (

    RESULT_DIR
    /
    "nested_ridge_tuning.csv"
)


REGIME_DISTRIBUTION_FILE = (

    RESULT_DIR
    /
    "trend_normalized_regime_distribution.csv"
)


ALL_RESULTS_FILE = (

    RESULT_DIR
    /
    "adaptive_conformal_all_results.csv"
)


PRIMARY_90_FILE = (

    RESULT_DIR
    /
    "primary_90percent_comparison.csv"
)


WINDOW_SENSITIVITY_FILE = (

    RESULT_DIR
    /
    "window_sensitivity_90percent.csv"
)


REGIME_RESULTS_FILE = (

    RESULT_DIR
    /
    "primary_method_regime_results.csv"
)


HOURLY_FILE = (

    RESULT_DIR
    /
    "primary_method_hourly_coverage.csv"
)


INTERVAL_OUTPUT_FILE = (

    RESULT_DIR
    /
    "primary_adaptive_90percent_intervals.csv"
)


ridge_tuning.to_csv(
    RIDGE_TUNING_FILE,
    index=False
)


regime_distribution.to_csv(
    REGIME_DISTRIBUTION_FILE,
    index=False
)


evaluation_results.to_csv(
    ALL_RESULTS_FILE,
    index=False
)


primary_90.to_csv(
    PRIMARY_90_FILE,
    index=False
)


window_sensitivity.to_csv(
    WINDOW_SENSITIVITY_FILE,
    index=False
)


primary_regime_results.to_csv(
    REGIME_RESULTS_FILE,
    index=False
)


hourly_primary_results.to_csv(
    HOURLY_FILE,
    index=False
)


interval_output.to_csv(
    INTERVAL_OUTPUT_FILE,
    index=False
)


# =====================================================================
# 37. FIGURE 1
# OLD VS NEW REGIME DISTRIBUTION
# =====================================================================

regime_plot = (

    regime_distribution

    .pivot(
        index="regime",
        columns="dataset",
        values="percent"
    )

    .reindex(
        REGIMES
    )
)


regime_plot.plot(
    kind="bar",
    figsize=(10, 6)
)


plt.ylabel(
    "Samples (%)"
)


plt.xlabel(
    "Trend-Normalized Regime"
)


plt.title(
    "Distribution of Forecastable Trend-Normalized Load Regimes"
)


plt.xticks(
    rotation=0
)


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG1 = (

    FIGURE_DIR
    /
    "01_trend_normalized_regime_distribution.png"
)


plt.savefig(
    FIG1,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 38. FIGURE 2
# STATIC VS ADAPTIVE COVERAGE
# =====================================================================

coverage_plot_methods = [

    "StaticHourNormalized",

    "StaticTrendRegimeHourNormalized",

    f"AdaptiveHourNormalized_W{PRIMARY_WINDOW}",

    PRIMARY_MARGINAL_METHOD
]


coverage_plot = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "method"
            ]
            .isin(
                coverage_plot_methods
            )
        )
    ]

    .copy()
)


plt.figure(
    figsize=(11, 7)
)


for method in coverage_plot_methods:

    temp = (

        coverage_plot.loc[
            coverage_plot[
                "method"
            ]
            ==
            method
        ]

        .sort_values(
            "nominal_coverage_percent"
        )
    )


    plt.plot(

        temp[
            "nominal_coverage_percent"
        ],

        temp[
            "Marginal_Coverage_percent"
        ],

        marker="o",

        linewidth=2,

        label=method
    )


plt.plot(

    [
        80,
        95
    ],

    [
        80,
        95
    ],

    linestyle="--",

    label="Ideal"
)


plt.xlabel(
    "Nominal Coverage (%)"
)


plt.ylabel(
    "Empirical Marginal Coverage (%)"
)


plt.title(
    "Static vs Prequential Adaptive Conformal Calibration"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG2 = (

    FIGURE_DIR
    /
    "02_static_vs_adaptive_coverage.png"
)


plt.savefig(
    FIG2,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 39. FIGURE 3
# 90% COVERAGE-WIDTH TRADEOFF
# =====================================================================

tradeoff = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )

        &

        (
            evaluation_results[
                "coverage_type"
            ]
            ==
            "MARGINAL"
        )
    ]

    .copy()
)


plt.figure(
    figsize=(12, 7)
)


for _, row in tradeoff.iterrows():

    plt.scatter(

        row[
            "MPIW_MW"
        ],

        row[
            "Primary_Coverage_percent"
        ],

        s=70
    )


    plt.annotate(

        row[
            "method"
        ],

        (
            row[
                "MPIW_MW"
            ],

            row[
                "Primary_Coverage_percent"
            ]
        ),

        xytext=(
            5,
            5
        ),

        textcoords="offset points",

        fontsize=8
    )


plt.axhline(
    90,
    linestyle="--"
)


plt.xlabel(
    "Mean Prediction Interval Width (MW)"
)


plt.ylabel(
    "Empirical 90% Marginal Coverage (%)"
)


plt.title(
    "Coverage-Sharpness Trade-Off of Adaptive Conformal Methods"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG3 = (

    FIGURE_DIR
    /
    "03_adaptive_coverage_width_tradeoff.png"
)


plt.savefig(
    FIG3,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 40. FIGURE 4
# PRIMARY 90% HOURLY COVERAGE
# =====================================================================

hourly_90 = (

    hourly_primary_results.loc[

        hourly_primary_results[
            "coverage_percent"
        ]
        ==
        90
    ]

    .copy()
)


plt.figure(
    figsize=(13, 6)
)


plt.plot(

    hourly_90[
        "hour"
    ],

    hourly_90[
        "empirical_coverage_percent"
    ],

    marker="o",

    linewidth=2
)


plt.axhline(

    90,

    linestyle="--",

    label="Nominal 90%"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Target Hour"
)


plt.ylabel(
    "Empirical Coverage (%)"
)


plt.title(
    "Adaptive Trend-Regime Conformal Coverage by Forecast Horizon"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG4 = (

    FIGURE_DIR
    /
    "04_primary_90coverage_by_hour.png"
)


plt.savefig(
    FIG4,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 41. FIGURE 5
# PRIMARY 90% COVERAGE BY REGIME
# =====================================================================

regime_90 = (

    primary_regime_results.loc[

        (
            primary_regime_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )

        &

        (
            primary_regime_results[
                "method"
            ]
            ==
            PRIMARY_MARGINAL_METHOD
        )
    ]

    .copy()
)


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    regime_90[
        "regime"
    ],

    regime_90[
        "Primary_Coverage_percent"
    ]
)


plt.axhline(

    90,

    linestyle="--",

    label="Nominal 90%"
)


plt.ylabel(
    "Marginal Coverage (%)"
)


plt.xlabel(
    "Trend-Normalized Load Regime"
)


plt.title(
    "Adaptive 90% Interval Reliability by Forecastable Regime"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


FIG5 = (

    FIGURE_DIR
    /
    "05_primary_90coverage_by_regime.png"
)


plt.savefig(
    FIG5,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 42. FIGURE 6
# EXAMPLE ADAPTIVE INTERVAL
# =====================================================================

positions_2024 = np.where(

    test_df[
        "target_date"
    ]
    .dt.year
    .eq(
        2024
    )
    .to_numpy()

)[0]


example_index = (

    positions_2024[

        len(
            positions_2024
        )

        //
        2
    ]
)


example_date = (

    test_df.loc[
        example_index,
        "target_date"
    ]
)


plt.figure(
    figsize=(13, 7)
)


plt.fill_between(

    HOURS,

    PRIMARY_LOWER_90[
        example_index
    ],

    PRIMARY_UPPER_90[
        example_index
    ],

    alpha=0.25,

    label="Adaptive 90% interval"
)


plt.plot(

    HOURS,

    Y_test[
        example_index
    ],

    marker="o",

    linewidth=2.5,

    label="Actual"
)


plt.plot(

    HOURS,

    TEST_PRED[
        example_index
    ],

    marker="s",

    linewidth=2,

    label="Ridge forecast"
)


plt.xticks(
    HOURS
)


plt.xlabel(
    "Hour of Day"
)


plt.ylabel(
    "Demand (MW)"
)


plt.title(

    f"Adaptive Reliable Forecast: "
    f"{example_date.date()} | "
    f"Regime={TEST_REGIME[example_index]}"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


FIG6 = (

    FIGURE_DIR
    /
    "06_example_adaptive_interval.png"
)


plt.savefig(
    FIG6,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# =====================================================================
# 43. EXTRACT PRIMARY RESULTS
# =====================================================================

primary_marginal_result = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "method"
            ]
            ==
            PRIMARY_MARGINAL_METHOD
        )

        &

        (
            evaluation_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .iloc[0]
)


primary_profile_result = (

    evaluation_results.loc[

        (
            evaluation_results[
                "split"
            ]
            ==
            "ALL_TEST_2024_2026"
        )

        &

        (
            evaluation_results[
                "method"
            ]
            ==
            PRIMARY_PROFILE_METHOD
        )

        &

        (
            evaluation_results[
                "nominal_coverage_percent"
            ]
            ==
            90
        )
    ]

    .iloc[0]
)


# =====================================================================
# 44. SUMMARY JSON
# =====================================================================

summary = {

    "best_ridge_alpha":
        float(
            BEST_ALPHA
        ),

    "ridge_test_MAE_MW":
        float(
            mae(
                Y_test,
                TEST_PRED
            )
        ),

    "ridge_test_RMSE_MW":
        float(
            rmse(
                Y_test,
                TEST_PRED
            )
        ),

    "ridge_test_MAPE_percent":
        float(
            mape(
                Y_test,
                TEST_PRED
            )
        ),

    "initial_calibration_days":
        int(
            len(
                initial_calibration_df
            )
        ),

    "primary_adaptive_window_days":
        int(
            PRIMARY_WINDOW
        ),

    "regime_signal_definition":
        (
            "lag1_daily_mean / "
            "mean(roll28mean_h01...h23)"
        ),

    "low_regime_threshold":
        float(
            LOW_THRESHOLD
        ),

    "high_regime_threshold":
        float(
            HIGH_THRESHOLD
        ),

    "primary_marginal_method":
        PRIMARY_MARGINAL_METHOD,

    "primary_90_marginal_coverage_percent":
        float(
            primary_marginal_result[
                "Marginal_Coverage_percent"
            ]
        ),

    "primary_90_marginal_coverage_gap":
        float(
            primary_marginal_result[
                "Coverage_Gap_percentage_points"
            ]
        ),

    "primary_90_marginal_MPIW_MW":
        float(
            primary_marginal_result[
                "MPIW_MW"
            ]
        ),

    "primary_90_marginal_Winkler":
        float(
            primary_marginal_result[
                "Winkler_Score"
            ]
        ),

    "primary_profile_method":
        PRIMARY_PROFILE_METHOD,

    "primary_90_profile_coverage_percent":
        float(
            primary_profile_result[
                "Profile_Coverage_percent"
            ]
        ),

    "primary_90_profile_coverage_gap":
        float(
            primary_profile_result[
                "Coverage_Gap_percentage_points"
            ]
        ),

    "primary_90_profile_MPIW_MW":
        float(
            primary_profile_result[
                "MPIW_MW"
            ]
        )
}


SUMMARY_FILE = (

    RESULT_DIR
    /
    "step5b_summary.json"
)


with open(
    SUMMARY_FILE,
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# =====================================================================
# 45. FINAL ASSERTIONS
# =====================================================================

assert (
    CAL_PRED.shape
    ==
    Y_cal.shape
)


assert (
    TEST_PRED.shape
    ==
    Y_test.shape
)


assert np.isfinite(
    TEST_PRED
).all()


assert np.all(

    PRIMARY_UPPER_90

    >=

    PRIMARY_LOWER_90
)


assert (

    initial_calibration_df[
        "target_date"
    ]
    .max()

    <

    test_df[
        "target_date"
    ]
    .min()
)


print(
    "\nAll Step 5B validation checks PASSED."
)


# =====================================================================
# 46. FINAL OUTPUT
# =====================================================================

print("\n")

print("=" * 120)

print(
    "STEP 5B COMPLETED SUCCESSFULLY"
)

print("=" * 120)


print(
    "\nRidge point MAE:",
    round(
        mae(
            Y_test,
            TEST_PRED
        ),
        3
    ),
    "MW"
)


print(
    "Ridge point MAPE:",
    round(
        mape(
            Y_test,
            TEST_PRED
        ),
        3
    ),
    "%"
)


print("\n" + "-" * 120)

print(
    "PRIMARY ADAPTIVE MARGINAL 90% INTERVAL"
)

print("-" * 120)


print(
    "Method:",
    PRIMARY_MARGINAL_METHOD
)


print(
    "Empirical marginal coverage:",
    round(
        float(
            primary_marginal_result[
                "Marginal_Coverage_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "Coverage gap:",
    round(
        float(
            primary_marginal_result[
                "Coverage_Gap_percentage_points"
            ]
        ),
        3
    ),
    "percentage points"
)


print(
    "MPIW:",
    round(
        float(
            primary_marginal_result[
                "MPIW_MW"
            ]
        ),
        3
    ),
    "MW"
)


print(
    "Normalized MPIW:",
    round(
        float(
            primary_marginal_result[
                "Normalized_MPIW_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "Winkler:",
    round(
        float(
            primary_marginal_result[
                "Winkler_Score"
            ]
        ),
        3
    )
)


print("\n" + "-" * 120)

print(
    "PRIMARY ADAPTIVE SIMULTANEOUS 90% PROFILE BAND"
)

print("-" * 120)


print(
    "Method:",
    PRIMARY_PROFILE_METHOD
)


print(
    "Whole-profile coverage:",
    round(
        float(
            primary_profile_result[
                "Profile_Coverage_percent"
            ]
        ),
        3
    ),
    "%"
)


print(
    "Coverage gap:",
    round(
        float(
            primary_profile_result[
                "Coverage_Gap_percentage_points"
            ]
        ),
        3
    ),
    "percentage points"
)


print(
    "MPIW:",
    round(
        float(
            primary_profile_result[
                "MPIW_MW"
            ]
        ),
        3
    ),
    "MW"
)


print("\n" + "-" * 120)

print(
    "IMPORTANT FILES"
)

print("-" * 120)


print(
    "\nPrimary comparison:"
)

print(
    PRIMARY_90_FILE
)


print(
    "\nWindow sensitivity:"
)

print(
    WINDOW_SENSITIVITY_FILE
)


print(
    "\nRegime results:"
)

print(
    REGIME_RESULTS_FILE
)


print(
    "\nHourly coverage:"
)

print(
    HOURLY_FILE
)


print(
    "\nFinal adaptive intervals:"
)

print(
    INTERVAL_OUTPUT_FILE
)


print(
    "\nFigures:"
)

print(
    FIGURE_DIR
)


print(
    "\nSummary:"
)

print(
    SUMMARY_FILE
)


print("\n" + "=" * 120)

print(
    "SEND THE STEP 5B FINAL SUMMARY, PRIMARY 90% TABLE,"
)

print(
    "TREND-NORMALIZED REGIME DISTRIBUTION, AND WINDOW-SENSITIVITY TABLE TO CHATGPT."
)

print(
    "DO NOT START BPDB EXTERNAL VALIDATION YET."
)

print("=" * 120)
```
